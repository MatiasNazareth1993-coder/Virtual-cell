```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MetaActivation(nn.Module):
    """
    Universal activation family
    """
    def __init__(self):
        super().__init__()
        self.alpha = nn.Parameter(torch.randn(1))
        self.beta  = nn.Parameter(torch.randn(1))
        self.gamma = nn.Parameter(torch.randn(1))
        self.delta = nn.Parameter(torch.randn(1))

    def forward(self, x):
        return (
            self.alpha * x +
            self.beta * torch.tanh(self.gamma * x) +
            self.delta * x * torch.sigmoid(x)
        )
```

```python
class MetaLoss(nn.Module):
    """
    Learns the geometry of error space
    """
    def __init__(self):
        super().__init__()
        self.w1 = nn.Parameter(torch.rand(1))
        self.w2 = nn.Parameter(torch.rand(1))
        self.w3 = nn.Parameter(torch.rand(1))

    def forward(self, pred, target):
        mse = (pred - target).pow(2).mean()
        huber = F.smooth_l1_loss(pred, target)
        kl = F.kl_div(
            F.log_softmax(pred, -1),
            F.softmax(target, -1),
            reduction="batchmean"
        )

        return self.w1*mse + self.w2*huber + self.w3*kl
```

```python
class MetaOptimizer(nn.Module):
    """
    Learns how to update parameters
    """
    def __init__(self):
        super().__init__()
        self.lr = nn.Parameter(torch.tensor(1e-3))
        self.momentum = nn.Parameter(torch.rand(1))
        self.decay = nn.Parameter(torch.rand(1))

    def step(self, params, grads, state):
        if state is None:
            state = [torch.zeros_like(p) for p in params]

        new_state = []
        for p, g, v in zip(params, grads, state):
            v_new = self.momentum * v + (1 - self.momentum) * g
            p.data -= self.lr * v_new / (torch.sqrt(g.pow(2).mean()) + self.decay)
            new_state.append(v_new)

        return new_state
```

```python
def lyapunov_meta_constraint(V, state, next_state):
    """
    Learning rules are rejected if destabilizing
    """
    return torch.relu(V(next_state) - V(state)).mean()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
```

```python
class VirtualCellWorld:
    def __init__(self):
        self.state = torch.tensor([1.0, 0.1, 0.5, 0.1])  # normalized

    def step(self, telomerase_action):
        telomere, damage, metab, stress = self.state

        telomere = torch.clamp(telomere + 0.05 * telomerase_action - 0.02, 0, 2)
        damage = torch.clamp(damage + 0.03 * stress - 0.02 * metab, 0, 1)
        metab = torch.clamp(metab + 0.02 - 0.01 * damage, 0, 1)
        stress = torch.clamp(stress + 0.03 * damage - 0.02 * telomerase_action, 0, 1)

        self.state = torch.stack([telomere, damage, metab, stress])
        return self.state
```

```python
class MetaActivation(nn.Module):
    def __init__(self):
        super().__init__()
        self.a = nn.Parameter(torch.randn(1))
        self.b = nn.Parameter(torch.randn(1))
        self.c = nn.Parameter(torch.randn(1))

    def forward(self, x):
        return self.a * x + self.b * torch.tanh(x) + self.c * x * torch.sigmoid(x)
```

```python
class MetaLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.w = nn.Parameter(torch.ones(3))

    def forward(self, pred, target):
        mse = (pred - target).pow(2).mean()
        huber = F.smooth_l1_loss(pred, target)
        energy = torch.abs(pred).mean()
        return self.w[0]*mse + self.w[1]*huber + self.w[2]*energy
```

```python
class MetaOptimizer(nn.Module):
    def __init__(self):
        super().__init__()
        self.lr = nn.Parameter(torch.tensor(1e-2))
        self.m = nn.Parameter(torch.tensor(0.9))

    def step(self, params, grads, velocity):
        if velocity is None:
            velocity = [torch.zeros_like(p) for p in params]

        new_v = []
        for p, g, v in zip(params, grads, velocity):
            v_new = self.m * v + g
            p.data -= self.lr * v_new
            new_v.append(v_new)

        return new_v
```

```python
def lyapunov(state):
    # penalize instability
    tel, dmg, metab, stress = state
    return dmg**2 + stress**2 + torch.relu(tel - 1.5)**2
```

```python
class TelomeraseController(nn.Module):
    def __init__(self):
        super().__init__()
        self.act = MetaActivation()
        self.fc1 = nn.Linear(4, 64)
        self.fc2 = nn.Linear(64, 1)

    def forward(self, x):
        x = self.act(self.fc1(x))
        return torch.sigmoid(self.fc2(x))
```

```python
def dream(world, controller, steps=50):
    states = []
    for _ in range(steps):
        action = controller(world.state)
        state = world.step(action)
        states.append(state.detach().numpy())
    return np.array(states)
```

```python
def plot_dynamics(states):
    labels = ["Telomere", "Damage", "Metabolism", "Stress"]
    plt.figure(figsize=(8,4))
    for i in range(4):
        plt.plot(states[:, i], label=labels[i])
    plt.legend()
    plt.title("Virtual Cell State Evolution")
    plt.xlabel("Time")
    plt.ylabel("Normalized Value")
    plt.show()
```

```python
def animate(states):
    fig, ax = plt.subplots()
    bars = ax.bar(["T","D","M","S"], states[0])

    def update(i):
        for b, v in zip(bars, states[i]):
            b.set_height(v)
        return bars

    ani = FuncAnimation(fig, update, frames=len(states))
    return ani
```

```python
world = VirtualCellWorld()
controller = TelomeraseController()
loss_fn = MetaLoss()
meta_opt = MetaOptimizer()

velocity = None
states_log = []

for epoch in range(200):
    prev = world.state.clone()
    action = controller(prev)
    action = action.squeeze() # Ensure action is a scalar tensor
    state = world.step(action)

    loss = loss_fn(state, torch.tensor([1.0,0.1,0.6,0.1]))
    stability = lyapunov(state) - lyapunov(prev)

    total = loss + torch.relu(stability)
    total.backward()

    grads = [p.grad for p in controller.parameters()]
    velocity = meta_opt.step(list(controller.parameters()), grads, velocity)

    controller.zero_grad()
    states_log.append(state.detach().numpy())
    world.state = world.state.detach() # Detach state to prevent graph retention across iterations
```

```python
from IPython.display import HTML
states = np.array(states_log)
plot_dynamics(states)
ani = animate(states)
HTML(ani.to_jshtml())
```

```python
class MultimodalWorldModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.state_enc = nn.Linear(STATE_DIM, 256)
        self.img_emb_fc = nn.Linear(512, 256)  # CLIP embedding size

        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(512, 8, 1024),
            6
        )

        self.state_head = nn.Linear(512, STATE_DIM)
        self.img_head = nn.Linear(512, 512)

    def forward(self, img_emb, state):
        z_img = self.img_emb_fc(img_emb)
        z_state = self.state_enc(state)
        z = torch.cat([z_img, z_state], dim=-1).unsqueeze(1)
        h = self.transformer(z).squeeze(1)
        return self.state_head(h), self.img_head(h)
```
