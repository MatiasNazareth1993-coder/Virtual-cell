```python
import tensorflow as tf
from tensorflow.keras.layers import Layer
from tensorflow.keras import activations

class BioSpectralGELU(Layer):
    def __init__(self,
                 alpha_init=1.0,
                 beta_init=0.2,
                 gamma_init=0.05,
                 delta_init=0.5,
                 **kwargs):
        super().__init__(**kwargs)

        self.alpha_init = alpha_init
        self.beta_init = beta_init
        self.gamma_init = gamma_init
        self.delta_init = delta_init

    def build(self, input_shape):
        self.alpha = self.add_weight(
            shape=(1,),
            initializer=tf.keras.initializers.Constant(self.alpha_init),
            trainable=True,
            constraint=lambda x: tf.clip_by_value(x, 0.1, 5.0)
        )
        self.beta = self.add_weight(
            shape=(1,),
            initializer=tf.keras.initializers.Constant(self.beta_init),
            trainable=True,
            constraint=lambda x: tf.clip_by_value(x, 0.0, 1.0)
        )
        self.gamma = self.add_weight(
            shape=(1,),
            initializer=tf.keras.initializers.Constant(self.gamma_init),
            trainable=True,
            constraint=lambda x: tf.clip_by_value(x, 0.0, 1.0)
        )
        self.delta = self.add_weight(
            shape=(1,),
            initializer=tf.keras.initializers.Constant(self.delta_init),
            trainable=True,
            constraint=lambda x: tf.clip_by_value(x, 0.1, 2.0)
        )

    def call(self, x):
        gelu_part = activations.gelu(x)
        elu_part = activations.elu(x)
        exp_part = tf.exp(self.delta * x) - 1.0
        gate = tf.sigmoid(self.alpha * x)

        return (
            x * gelu_part * gate
            + self.beta * elu_part
            + self.gamma * exp_part
        )

    def get_config(self):
        config = super().get_config()
        config.update({
            "alpha_init": self.alpha_init,
            "beta_init": self.beta_init,
            "gamma_init": self.gamma_init,
            "delta_init": self.delta_init
        })
        return config
```

```python
from tensorflow.keras import layers, models, optimizers, losses

def build_foundational_biocell_model(
    spectral_dim=2048,
    time_steps=128,
    channels=4
):

    spectral_input = layers.Input(
        shape=(spectral_dim, channels),
        name="spectral_input"
    )

    x = layers.Conv1D(128, 7, padding="same")(spectral_input)
    x = BioSpectralGELU()(x)
    x = layers.LayerNormalization()(x)

    x = layers.Conv1D(256, 5, padding="same")(x)
    x = BioSpectralGELU()(x)
    x = layers.LayerNormalization()(x)

    x = layers.Conv1D(512, 3, padding="same")(x)
    x = BioSpectralGELU()(x)

    x = layers.Bidirectional(
        layers.LSTM(256, return_sequences=True)
    )(x)

    x = layers.Bidirectional(
        layers.LSTM(128)
    )(x)

    latent = layers.Dense(
        256,
        activation=BioSpectralGELU(),
        name="cell_latent_state"
    )(x)

    # Uncertainty estimation (epistemic + aleatoric)
    mean_output = layers.Dense(1, name="drug_effect_mean")(latent)
    log_var_output = layers.Dense(1, name="drug_effect_logvar")(latent)

    model = models.Model(
        inputs=spectral_input,
        outputs=[mean_output, log_var_output],
        name="Foundational_BioCell_Model"
    )

    return model
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_raman_fluo_virtual_model(
    spectral_bins=2048,
    time_steps=120,
    fluo_channels=3,
    ooC_channels=5
):
    # Raman spectral input: (time_steps, spectral_bins, 1)
    # Added a channel dimension (1) to the spectral data
    raman_input = layers.Input(
        shape=(time_steps, spectral_bins, 1),
        name="raman_spectra"
    )

    # Fluorescence time series input
    fluo_input = layers.Input(
        shape=(time_steps, fluo_channels),
        name="fluorescence_ts"
    )

    # Optional OoC sensor stream
    ooc_input = layers.Input(
        shape=(time_steps, ooC_channels),
        name="ooc_sensors"
    )

    # --- SPECTRAL FEATURE EXTRACTION ---
    # Now TimeDistributed(Conv1D) will receive inputs of shape (None, spectral_bins, 1)
    x = layers.TimeDistributed(
        layers.Conv1D(128, kernel_size=7, padding="same")
    )(raman_input)
    x = layers.TimeDistributed(
        BioSpectralGELU()
    )(x)
    x = layers.TimeDistributed(
        layers.LayerNormalization()
    )(x)

    x = layers.TimeDistributed(
        layers.Conv1D(256, kernel_size=5, padding="same")
    )(x)
    x = layers.TimeDistributed(
        BioSpectralGELU()
    )(x)

    x = layers.TimeDistributed(
        layers.GlobalAveragePooling1D()
    )(x)

    # --- MULTIMODAL MERGE ---
    merged = layers.Concatenate()([x, fluo_input, ooc_input])

    # --- TEMPORAL DYNAMICS ---
    h = layers.Bidirectional(
        layers.LSTM(256, return_sequences=True)
    )(merged)
    h = layers.Bidirectional(
        layers.LSTM(128)
    )(h)

    # Latent state for drug-cell interaction
    latent = layers.Dense(
        256,
        activation=BioSpectralGELU(),
        name="latent_embedding"
    )(h)

    # Predicted drug uptake over time
    uptake_pred = layers.Dense(
        time_steps, activation="sigmoid",
        name="drug_uptake"
    )(latent)

    # Predicted spectral response curve
    spectral_resp = layers.Dense(
        spectral_bins, activation="softplus",
        name="spectral_response"
    )(latent)

    # Uncertainty estimation (variance)
    logvar_uptake = layers.Dense(
        time_steps, name="uptake_logvar"
    )(latent)

    model = models.Model(
        inputs=[raman_input, fluo_input, ooc_input],
        outputs=[uptake_pred, spectral_resp, logvar_uptake],
        name="RamanFluoVirtualCell"
    )

    return model
```

```python
import tensorflow as tf
from tensorflow.keras.optimizers import Optimizer
from tensorflow.keras import backend as K

class BioFusionOptimizer(Optimizer):
    def __init__(
        self,
        learning_rate=1e-3,
        beta_1=0.9,
        beta_2=0.999,
        weight_decay=1e-4,
        epsilon=1e-7,
        name="BioFusion",
        **kwargs
    ):
        super().__init__(learning_rate=learning_rate, name=name, **kwargs) # Corrected line
        self.beta_1 = beta_1
        self.beta_2 = beta_2
        self.weight_decay = weight_decay
        self.epsilon = epsilon

    def build(self, var_list):
        self.m = []
        self.v = []
        for var in var_list:
            self.m.append(self.add_variable_from_reference(var, "m"))
            self.v.append(self.add_variable_from_reference(var, "v"))

    def update_step(self, grad, var, index):
        lr = self._get_hyper("learning_rate")

        m = self.m[index]
        v = self.v[index]

        # First moment
        m.assign(self.beta_1 * m + (1.0 - self.beta_1) * grad)

        # Second moment
        v.assign(self.beta_2 * v + (1.0 - self.beta_2) * tf.square(grad))

        # Lion-style direction
        direction = tf.sign(m)

        # Adam / RMSProp magnitude
        denom = tf.sqrt(v) + self.epsilon

        # Weight decay (AdamW style)
        var.assign_sub(lr * self.weight_decay * var)

        # Parameter update
        var.assign_sub(lr * direction / denom)

    def get_config(self):
        return {
            "learning_rate": self._serialize_hyperparameter("learning_rate"),
            "beta_1": self.beta_1,
            "beta_2": self.beta_2,
            "weight_decay": self.weight_decay,
            "epsilon": self.epsilon,
        }
```

```python
from tensorflow.keras import layers, models

def BioFoundationModel(input_dim):
    inputs = layers.Input(shape=(input_dim,))

    x = layers.LayerNormalization()(inputs)
    x = layers.Dense(1024, activation="gelu")(x)
    x = layers.Dense(1024, activation="mish")(x)

    # Residual block
    res = x
    x = layers.Dense(1024, activation="silu")(x)
    x = layers.Dense(1024)(x)
    x = layers.Add()([x, res])

    x = layers.LayerNormalization()(x)

    latent = layers.Dense(256, activation="tanh", name="latent_state")(x)

    outputs = layers.Dense(input_dim, activation="linear")(latent)

    return models.Model(inputs, outputs)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# Encoder
def make_encoder(input_dim, latent_dim):
    inputs = layers.Input(shape=(input_dim,))
    x      = layers.Dense(256, activation='relu')(inputs)
    x      = layers.Dense(128, activation='relu')(x)
    z_mean = layers.Dense(latent_dim)(x)
    z_log_var = layers.Dense(latent_dim)(x)

    encoder = Model(inputs, [z_mean, z_log_var], name="encoder")
    return encoder

# Decoder
def make_decoder(latent_dim, output_dim):
    latent_inputs = layers.Input(shape=(latent_dim,))
    x = layers.Dense(128, activation='relu')(latent_inputs)
    x = layers.Dense(256, activation='relu')(x)
    outputs = layers.Dense(output_dim)(x)
    decoder = Model(latent_inputs, outputs, name="decoder")
    return decoder
```

```python
class LatentDynamics(tf.keras.Model):
    def __init__(self, latent_dim):
        super().__init__()
        self.rnn = layers.GRU(latent_dim, return_sequences=True, return_state=True)

    def call(self, seq):
        outputs, state = self.rnn(seq)
        return outputs, state
```

```python
def vae_loss(x, x_recon, z_log_var, z_mean):
    recon_loss = tf.reduce_mean(tf.square(x - x_recon))
    kl_loss = -0.5 * tf.reduce_mean(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var))
    return recon_loss + kl_loss
```

```python
class SelfImprovingController:
    def __init__(self, model):
        self.model = model  # world model

    def predict(self, state):
        # world model forward pass
        z_mean, z_log_var = encoder(state)
        z = z_mean + tf.exp(0.5 * z_log_var)*tf.random.normal(shape=z_mean.shape)
        next_state = decoder(z)
        return next_state

    def refine(self, dataset):
        # train model on accumulated experience
        for (x, y) in dataset:
            with tf.GradientTape() as tape:
                z_mean, z_log_var = encoder(x)
                z = z_mean + tf.exp(0.5*z_log_var)*tf.random.normal(shape=z_mean.shape)
                recon = decoder(z)
                loss = tf.reduce_mean((recon - y)**2)
            grads = tape.gradient(loss, self.model.trainable_variables)
            self.optimizer.apply_gradients(zip(grads, self.model.trainable_variables))
```

```python
def apply_cycle_drug_effect(cycle_state, drug_strength):
    """
    drug_strength: scalar in [0,1], how strongly a compound inhibits DNA synthesis
    - if drug targets S phase, more cells accumulate in G1/G2M
    """
    for i in range(len(cycle_state["G1"])):
        # S phase suppression: shift probability away from S
        suppression = drug_strength * cycle_state["S"][i]
        cycle_state["S"][i] -= suppression

        # redistribution to G1 and G2/M
        cycle_state["G1"][i] += 0.5 * suppression
        cycle_state["G2M"][i] += 0.5 * suppression

        # renormalize
        total = cycle_state["G1"][i] + cycle_state["S"][i] + cycle_state["G2M"][i]
        if total > 0:
            cycle_state["G1"][i]  /= total
            cycle_state["S"][i]   /= total
            cycle_state["G2M"][i] /= total

    return cycle_state
```

```python
import gym
import numpy as np

class VirtualCellEnv(gym.Env):
    def __init__(self, compounds):
        super().__init__()
        self.compounds = compounds
        self.state_dim = 32
        self.action_dim = len(compounds) * 5  # dose bins

        self.observation_space = gym.spaces.Box(
            low=0, high=1, shape=(self.state_dim,), dtype=np.float32
        )
        self.action_space = gym.spaces.Discrete(self.action_dim)

        self.reset()

    def reset(self):
        self.state = np.random.rand(self.state_dim)
        self.time = 0
        return self.state

    def step(self, action):
        compound_id = action // 5
        dose = (action % 5) / 4.0

        self._apply_compound(compound_id, dose)
        reward = self._compute_reward()
        self.time += 1

        done = self.time > 100
        return self.state, reward, done, {}

    def _apply_compound(self, cid, dose):
        # PK/PD + cell cycle + damage logic
        self.state[0] += 0.01 * dose          # telomeres
        self.state[5] += 0.05 * dose          # drug conc
        self.state[10] -= 0.02 * dose         # DNA damage

    def _compute_reward(self):
        return (
            + self.state[0]
            - self.state[10]
            - abs(self.state[15] - 0.5)
        )
```

```python
import torch
import torch.nn as nn

class BioZeroNet(nn.Module):
    def __init__(self, state_dim, action_dim):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(state_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 256)
        )

        self.policy_head = nn.Sequential(
            nn.Linear(256, action_dim),
            nn.Softmax(dim=-1)
        )

        self.value_head = nn.Sequential(
            nn.Linear(256, 1),
            nn.Tanh()
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.policy_head(z), self.value_head(z)
```

```python
import numpy as np

def apply_mitotic_inhibitor_effects(cycles, state, drug_type, strength):
    """
    drug_type: "taxane", "vinca", "epothilone", "combretastatin", "hdac_mod"
    strength: [0,1], relative dose effect
    """
    N = len(cycles["G1"])
    for i in range(N):
        # baseline
        g1, s_, g2m = cycles["G1"][i], cycles["S"][i], cycles["G2M"][i]

        if drug_type == "taxane":
            # taxanes stabilize microtubules -> G2/M arrest
            g2m += strength * s_
            s_ -= strength * s_

        elif drug_type == "vinca":
            # destabilize -> metaphase blockade
            g2m += strength * s_
            s_ -= strength * s_

        elif drug_type == "epothilone":
            # similar effect to taxane but also triggers cascade
            g2m += strength * s_
            s_ -= strength * s_
            state["damage"][i] += 0.03*strength

        elif drug_type == "combretastatin":
            # targets microtubule assembly and vasculature signaling
            g2m += strength * s_
            s_ -= strength * s_
            state["damage"][i] += 0.02*strength

        elif drug_type == "hdac_mod":
            # HDAC modulation -> missegregation and apoptosis signaling
            state["damage"][i] += 0.05*strength

        # renormalize
        total = g1 + s_ + g2m
        cycles["G1"][i] = g1 / total
        cycles["S"][i]  = s_ / total
        cycles["G2M"][i] = g2m / total

    return cycles, state
```

```python
# Extend action mapping
drug_types = ["taxane","vinca","epothilone","combretastatin","hdac_mod"]
dose_bins = 5

def step(self, action):
    drug_id = action // dose_bins
    drug_type = drug_types[drug_id % len(drug_types)]
    dose_level = (action % dose_bins)/ (dose_bins - 1)

    cycles, state = apply_mitotic_inhibitor_effects(
        self.cycles, self.state, drug_type, dose_level
    )

    reward = self._compute_reward()
    done = self.time > 100
    return self._get_obs(), reward, done, {}
```

```python
import numpy as np

class CellState:
    def __init__(self, n_cells):
        self.n_cells = n_cells
        self.telomere_length = np.random.normal(1000, 40, n_cells)
        self.stability = np.clip(np.random.rand(n_cells), 0, 1)
        self.damage = np.zeros(n_cells)
        self.drug_intracellular = np.zeros(n_cells)

        # Cell cycle probability vectors
        self.G1 = np.full(n_cells, 0.4)
        self.S  = np.full(n_cells, 0.3)
        self.G2M = np.full(n_cells, 0.3)

        # Multi-system observables
        self.cytoskeleton_status = np.zeros(n_cells)  # microtubule health
        self.signaling_perturb = np.zeros(n_cells)    # generic signaling error
        self.apoptosis_signal = np.zeros(n_cells)     # cell death propensity
```

```python
def apply_drug_mechanism(cell_state, drug_type, dose):
    """
    drug_type: one of {'taxane','vinca','epothilone','combretastatin'}
    dose: 0.0–1.0 scaled
    """

    for i in range(cell_state.n_cells):
        # strengthen intracellular drug
        cell_state.drug_intracellular[i] += dose

        # microtubule targeting effects
        if drug_type == "taxane":
            # stabilizes microtubules → G2/M arrest
            cell_state.G2M[i] += dose * cell_state.S[i]
            cell_state.S[i]  -= dose * cell_state.S[i]
            cell_state.cytoskeleton_status[i] += dose * 0.5

        elif drug_type == "vinca":
            # destabilizes microtubules → metaphase arrest
            cell_state.G2M[i] += dose * cell_state.S[i]
            cell_state.S[i]  -= dose * cell_state.S[i]
            cell_state.cytoskeleton_status[i] -= dose * 0.7

        elif drug_type == "epothilone":
            cell_state.G2M[i] += dose * cell_state.S[i]
            cell_state.S[i]  -= dose * cell_state.S[i]
            cell_state.cytoskeleton_status[i] += dose * 0.3
            cell_state.apoptosis_signal[i] += dose * 0.2

        elif drug_type == "combretastatin":
            cell_state.G2M[i] += dose * cell_state.S[i]
            cell_state.S[i]  -= dose * cell_state.S[i]
            cell_state.signaling_perturb[i] += dose * 0.4
            cell_state.apoptosis_signal[i] += dose * 0.3

        # normalize cycle probabilities
        total = cell_state.G1[i] + cell_state.S[i] + cell_state.G2M[i]
        if total > 0:
            cell_state.G1[i] /= total
            cell_state.S[i]  /= total
            cell_state.G2M[i]/= total

    return cell_state
```

```python
def propagate_system_interactions(cell_state):
    """
    captures interconnected effects:
      - cycle arrest increases damage
      - cytoskeletal disruption alters stability
      - signaling perturbations raise apoptosis signal
    """

    cell_state.damage += 0.05 * cell_state.G2M
    cell_state.stability -= 0.03 * np.abs(cell_state.cytoskeleton_status)
    cell_state.apoptosis_signal += 0.1 * cell_state.signaling_perturb

    return cell_state
```

```python
def compute_reward(cell_state):
    # + survival + telomere stability
    reward = np.mean(cell_state.stability)
    reward += np.mean(cell_state.telomere_length / 1200)

    # - DNA damage
    reward -= np.mean(cell_state.damage)

    # - unwanted apoptosis
    reward -= np.mean(cell_state.apoptosis_signal * 0.5)

    # disfavor cytoskeletal collapse
    reward -= np.mean(np.abs(cell_state.cytoskeleton_status) * 0.2)

    return reward
```

```python
import matplotlib.pyplot as plt

def plot_phase_distributions(cell_state, title):
    bins = [0, 1, 2, 3]
    dna_vals = []

    for i in range(cell_state.n_cells):
        phase = np.random.choice(
            ['G1','S','G2M'],
            p=[cell_state.G1[i], cell_state.S[i], cell_state.G2M[i]]
        )
        dna_vals.append(1.0 if phase=='G1' else (1.5 if phase=='S' else 2.0))

    plt.hist(dna_vals, bins=bins, color='crimson', alpha=0.7)
    plt.title(title)
    plt.xlabel('Relative DNA Content')
    plt.ylabel('Cell Count')
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_cell_cycle_distribution(cell_state, title="Cell Cycle Distribution"):
    phases = np.random.choice(
        ["G1", "S", "G2/M"],
        size=cell_state.n_cells,
        p=[
            np.mean(cell_state.G1),
            np.mean(cell_state.S),
            np.mean(cell_state.G2M),
        ],
    )

    counts = [
        np.sum(phases == "G1"),
        np.sum(phases == "S"),
        np.sum(phases == "G2/M"),
    ]

    plt.figure(figsize=(6,4))
    plt.bar(["G1", "S", "G2/M"], counts)
    plt.ylabel("Cell Count")
    plt.title(title)
    plt.tight_layout()
    plt.show()
```

```python
def plot_telomere_stability(cell_state):
    plt.figure(figsize=(6,5))
    plt.scatter(
        cell_state.telomere_length,
        cell_state.stability,
        alpha=0.6
    )
    plt.xlabel("Telomere Length")
    plt.ylabel("Cellular Stability")
    plt.title("Telomere–Stability Phase Space")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
def plot_drug_mechanism_fingerprint(cell_state, drug_name):
    metrics = {
        "Cytoskeleton Disruption": np.mean(np.abs(cell_state.cytoskeleton_status)),
        "Signaling Perturbation": np.mean(cell_state.signaling_perturb),
        "DNA Damage": np.mean(cell_state.damage),
        "Apoptosis Signal": np.mean(cell_state.apoptosis_signal),
    }

    labels = list(metrics.keys())
    values = list(metrics.values())

    plt.figure(figsize=(7,4))
    plt.barh(labels, values)
    plt.title(f"Drug Mechanism Fingerprint: {drug_name}")
    plt.xlabel("Normalized Effect Strength")
    plt.tight_layout()
    plt.show()
```

```python
def plot_population_dynamics(history):
    """
    history: list of dicts with keys
    ['stability', 'damage', 'apoptosis']
    """

    stability = [h["stability"] for h in history]
    damage = [h["damage"] for h in history]
    apoptosis = [h["apoptosis"] for h in history]

    plt.figure(figsize=(8,4))
    plt.plot(stability, label="Mean Stability")
    plt.plot(damage, label="Mean Damage")
    plt.plot(apoptosis, label="Apoptosis Signal")
    plt.xlabel("Simulation Step")
    plt.ylabel("Population Metric")
    plt.title("Emergent Population Dynamics")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
def composite_figure(cell_state):
    fig, axs = plt.subplots(2, 2, figsize=(10,8))

    axs[0,0].hist(cell_state.telomere_length, bins=30)
    axs[0,0].set_title("Telomere Length Distribution")

    axs[0,1].scatter(cell_state.telomere_length, cell_state.stability, alpha=0.5)
    axs[0,1].set_title("Telomere vs Stability")

    axs[1,0].hist(cell_state.apoptosis_signal, bins=30)
    axs[1,0].set_title("Apoptosis Signal")

    axs[1,1].hist(cell_state.damage, bins=30)
    axs[1,1].set_title("DNA Damage")

    for ax in axs.flat:
        ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_targeted_drug_phase_shift(cell_state_control, cell_state_treated):
    labels = ["G1", "S", "G2/M"]
    control_means = [
        np.mean(cell_state_control.G1),
        np.mean(cell_state_control.S),
        np.mean(cell_state_control.G2M),
    ]
    treated_means = [
        np.mean(cell_state_treated.G1),
        np.mean(cell_state_treated.S),
        np.mean(cell_state_treated.G2M),
    ]

    x = np.arange(len(labels))
    width = 0.35

    fig, ax = plt.subplots(figsize=(6,4))
    ax.bar(x - width/2, control_means, width, label='Control')
    ax.bar(x + width/2, treated_means, width, label='Targeted Drug')
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.set_ylabel("Fraction of Population")
    ax.set_title("Cell Cycle Phase Distribution\nControl vs Targeted Therapy")
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np

class VirtualCell:
    def __init__(self):
        self.phase = np.random.choice(["G1", "S", "G2M"])
        self.dna = {"G1": 1.0, "S": 1.5, "G2M": 2.0}[self.phase]
        self.cdk_activity = np.random.uniform(0.5, 1.0)
        self.apoptosis = 0.0
        self.alive = True

    def step(self, drug):
        if not self.alive:
            return

        # Drug inhibits CDK (e.g. CDK4/6 inhibitor)
        self.cdk_activity *= (1 - drug.cdk_inhibition)

        # Cell-cycle progression
        if self.cdk_activity > 0.6:
            if self.phase == "G1":
                self.phase = "S"
            elif self.phase == "S":
                self.phase = "G2M"
        else:
            # checkpoint arrest → apoptosis pressure
            self.apoptosis += 0.05

        # Apoptosis threshold
        if self.apoptosis > 1.0:
            self.alive = False

        self.dna = {"G1": 1.0, "S": 1.5, "G2M": 2.0}[self.phase]
```

```python
class Drug:
    def __init__(self, dose):
        self.dose = dose
        self.cdk_inhibition = min(0.9, dose / 10)
```

```python
class InSilicoExperiment:
    def __init__(self, n_cells=1000):
        self.cells = [VirtualCell() for _ in range(n_cells)]
        self.history = []

    def step(self, drug):
        for cell in self.cells:
            cell.step(drug)

        self.record()

    def record(self):
        alive = [c for c in self.cells if c.alive]
        self.history.append({
            "G1": sum(c.phase == "G1" for c in alive),
            "S": sum(c.phase == "S" for c in alive),
            "G2M": sum(c.phase == "G2M" for c in alive),
            "dead": len(self.cells) - len(alive)
        })
```

```python
experiment = InSilicoExperiment(n_cells=2000)
drug = Drug(dose=6.0)

for t in range(60):  # simulate 60 time steps
    experiment.step(drug)
import matplotlib.pyplot as plt

history = experiment.history

plt.figure(figsize=(7,4))
plt.plot([h["G1"] for h in history], label="G1")
plt.plot([h["S"] for h in history], label="S")
plt.plot([h["G2M"] for h in history], label="G2/M")
plt.plot([h["dead"] for h in history], label="Dead", linestyle="--")

plt.xlabel("Time")
plt.ylabel("Cell Count")
plt.title("In-Silico Drug Response: Cell Cycle & Death")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
def run_counterfactual(dose):
    exp = InSilicoExperiment(2000)
    drug = Drug(dose)
    for _ in range(60):
        exp.step(drug)
    return exp.history

treated = run_counterfactual(6.0)
control = run_counterfactual(0.0)
```

```python
plt.figure(figsize=(6,4))
plt.plot([h["dead"] for h in treated], label="Drug")
plt.plot([h["dead"] for h in control], label="No Drug")
plt.xlabel("Time")
plt.ylabel("Dead Cells")
plt.title("Counterfactual Causal Effect of Drug")
plt.legend()
plt.tight_layout()
plt.show()
```
