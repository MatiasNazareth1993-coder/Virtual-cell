```python
import numpy as np
import tensorflow as tf

class VirtualCellEnv:
    def __init__(self,
                 gene_dim,
                 drug_latent_dim,
                 tissue_latent_dim,
                 gtex_mean,
                 tcga_mean):

        self.gene_dim = gene_dim
        self.state_dim = gene_dim + drug_latent_dim + tissue_latent_dim + 2
        self.gtex_mean = gtex_mean
        self.tcga_mean = tcga_mean
        self.reset()

    def reset(self):
        self.telomere = 1.0
        self.stress = 0.1
        self.expr = np.random.normal(0, 1, self.gene_dim)
        return self._get_state()

    def _get_state(self):
        return np.concatenate([
            self.expr,
            [self.telomere, self.stress]
        ])

    def step(self, action):
        telomerase = action[0]
        phagocytosis = action[1]

        # Telomere dynamics
        self.telomere += 0.05 * telomerase
        self.telomere -= 0.01

        # Stress & cleanup
        self.stress += 0.02 - 0.05 * phagocytosis

        # Expression drift
        self.expr += np.random.normal(0, 0.05, self.gene_dim)

        reward = self._reward()
        done = self.telomere < 0.3 or self.telomere > 1.8

        return self._get_state(), reward, done

    def _reward(self):
        gtex_similarity = -np.linalg.norm(
            np.mean(self.expr) - self.gtex_mean
        )
        cancer_penalty = np.linalg.norm(
            np.mean(self.expr) - self.tcga_mean
        )

        telomere_stability = -abs(self.telomere - 1.1)

        return (
            2.0 * gtex_similarity
            + telomere_stability
            - 3.0 * cancer_penalty
        )
```

```python
def compute_returns(rewards):
    G = 0
    returns = []
    for r in reversed(rewards):
        G = r + GAMMA * G
        returns.insert(0, G)
    return np.array(returns)
```

```python
@tf.function
def train_step(states, actions, returns):
    states = tf.convert_to_tensor(states, tf.float32)
    actions = tf.convert_to_tensor(actions, tf.float32)
    returns = tf.convert_to_tensor(returns, tf.float32)

    with tf.GradientTape() as tape_p:
        pred_actions = policy(states)
        loss_p = tf.reduce_mean((pred_actions - actions) ** 2)

    grads_p = tape_p.gradient(loss_p, policy.trainable_variables)
    opt_policy.apply_gradients(zip(grads_p, policy.trainable_variables))

    with tf.GradientTape() as tape_v:
        values = value_fn(states)
        loss_v = tf.reduce_mean((values[:,0] - returns) ** 2)

    grads_v = tape_v.gradient(loss_v, value_fn.trainable_variables)
    opt_value.apply_gradients(zip(grads_v, value_fn.trainable_variables))
```

```python
class MultiOrganSystem:
    def __init__(self, organs, weights):
        self.organs = organs
        self.weights = weights  # circulation coupling

    def step(self, X, Z, I):
        Xs, Zs, Is = [], [], []

        for organ in self.organs:
            x, z, i = organ.apply(X, Z, I)
            Xs.append(x)
            Zs.append(z)
            Is.append(i)

        X_next = np.dot(self.weights, Xs)
        Z_next = np.dot(self.weights, Zs)
        I_next = np.dot(self.weights, Is)

        return X_next, Z_next, I_next
```

```python
import numpy as np

class VirtualCellSCM:
    def __init__(self):
        pass

    def sample_noise(self):
        return {
            "UT": np.random.normal(0, 0.05),
            "UX": np.random.normal(0, 0.05),
            "UI": np.random.normal(0, 0.05),
            "UZ": np.random.normal(0, 0.05)
        }

    def forward(self, D=0, E=0, intervene_T=None):
        U = self.sample_noise()

        # Structural equations
        T = intervene_T if intervene_T is not None else 0.6 + 0.4*D + U["UT"]
        X = 0.5*T - 0.3*E + 0.2*D + U["UX"]
        I = 0.6*(1 - X) + U["UI"]
        Z = 0.7*X - 0.4*I + U["UZ"]

        return dict(T=T, X=X, I=I, Z=Z)
```

```python
def run_experiment(n=1000, intervene=False):
    scm = VirtualCellSCM()
    Z_vals = []

    for _ in range(n):
        res = scm.forward(
            D=0.3,
            E=0.4,
            intervene_T=1.0 if intervene else None
        )
        Z_vals.append(res["Z"])

    return np.mean(Z_vals), np.var(Z_vals)
```

```python
import numpy as np

class CausalCellEnv:
    def __init__(self):
        self.reset()

    def reset(self):
        self.state = np.array([0.6, 0.7, 0.3, 0.7, 0.4])  # T,X,I,Z,E
        return self.state

    def scm_rollout(self, action):
        T, X, I, Z, E = self.state
        tel_up, imm_sup, drug = action

        # Structural equations
        T_new = T + 0.2*tel_up - 0.1*drug
        X_new = X + 0.5*T_new - 0.3*E
        I_new = I + 0.4*(1 - X_new) - 0.5*imm_sup
        Z_new = 0.7*X_new - 0.4*I_new

        return np.array([T_new, X_new, I_new, Z_new, E])

    def safe(self, action):
        rollout = self.scm_rollout(action)
        return rollout[3] > 0.3  # Z safety threshold

    def step(self, action):
        if not self.safe(action):
            reward = -1.0
            done = True
            return self.state, reward, done

        self.state = self.scm_rollout(action)
        T, X, I, Z, _ = self.state

        reward = Z + X - I - abs(T - 0.8)
        done = Z < 0.2

        return self.state, reward, done
```

```python
class SimplePolicy:
    def act(self, state):
        return np.clip(np.random.normal(0.5, 0.2, size=3), 0, 1)
```

```python
import torch
import torch.nn as nn

class NeuralSCM(nn.Module):
    def __init__(self, d, hidden=64):
        super().__init__()
        self.W = nn.Parameter(torch.zeros(d, d))
        self.f = nn.Sequential(
            nn.Linear(d, hidden),
            nn.ReLU(),
            nn.Linear(hidden, d)
        )

    def forward(self, X):
        return self.f(X @ self.W)

    def acyclicity(self):
        return torch.trace(torch.matrix_exp(self.W * self.W)) - self.W.shape[0]
```

```python
def scm_loss(X, model, lam=0.01, rho=10.0):
    X_hat = model(X)
    recon = ((X - X_hat) ** 2).mean()
    sparsity = lam * torch.abs(model.W).sum()
    acyclic = rho * model.acyclicity() ** 2
    return recon + sparsity + acyclic
```
