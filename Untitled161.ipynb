```python
# Virtual Cell Telomere–Telomerase–Phagocytosis In Silico Simulation
# Purely computational (no wet-lab steps)

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Simulation parameters
cells = 200
divisions = 120
desired_telomere = 10_000  # bp (virtual units)
min_safe = 7_000
max_safe = 11_500

# Initialize telomere lengths
telomeres = np.random.normal(9500, 300, size=cells)

# Tracking
history = np.zeros((divisions, cells))
phagocytosed = np.zeros((divisions, cells))

for d in range(divisions):
    # Replication erosion
    erosion = np.random.normal(50, 10, size=cells)
    telomeres -= erosion

    # Virtual telomerase controller (activates only if needed)
    need = telomeres < desired_telomere
    telomerase_pulse = need * np.clip(desired_telomere - telomeres, 0, 120)
    telomeres += telomerase_pulse * 0.6

    # Safety trimming (anti-cancer constraint)
    telomeres = np.minimum(telomeres, max_safe)

    # Phagocytosis of unsafe cells
    dead = telomeres < min_safe
    phagocytosed[d, dead] = 1
    telomeres[dead] = np.nan  # removed cells

    history[d] = telomeres

# Visualization 1: Telomere length trajectories
plt.figure()
plt.plot(history[:, :50])
plt.axhline(desired_telomere, linestyle='--')
plt.xlabel("Division count")
plt.ylabel("Telomere length (virtual bp)")
plt.title("Telomere Length vs Division Count")
plt.show()

# Visualization 2: Population mean telomere length
plt.figure()
plt.plot(np.nanmean(history, axis=1))
plt.xlabel("Division count")
plt.ylabel("Mean telomere length")
plt.title("Population Telomere Homeostasis")
plt.show()

# Visualization 3: Phagocytosis events
plt.figure()
plt.plot(phagocytosed.sum(axis=1))
plt.xlabel("Division count")
plt.ylabel("Cells removed")
plt.title("Phagocytic Clearance Over Time")
plt.show()

"Simulation complete"
```

```python
def hamiltonian(state):
    telomere, oncogenic, damage, metabolic = state
    L_star = 10000

    return (
        0.6 * (telomere - L_star)**2 +
        2.5 * oncogenic +
        1.8 * damage +
        0.8 * metabolic
    )
```

```python
def quantum_step(center, temperature=0.15):
    population = quantum_population(center, scale=temperature)
    energies = np.array([hamiltonian(s) for s in population])

    # Softmin = quantum tunneling effect
    weights = np.exp(-energies / temperature)
    sum_weights = weights.sum()

    if sum_weights == 0:
        # If all weights are effectively zero due to underflow, return the current center
        return center
    else:
        weights /= sum_weights

    new_center = (population.T @ weights)
    return new_center
```

```python
def telomerase_control(state):
    telomere, oncogenic, damage, _ = state
    if telomere < 10000 and oncogenic < 0.2 and damage < 0.3:
        return min(1.0, (10000 - telomere) / 1000)
    return 0.0
```

```python
def quantum_population(center, scale=0.15, num_samples=1000):
    # center is a 1D array representing the mean of the distribution
    # scale is the standard deviation for the normal distribution
    # num_samples is the number of "particles" in the quantum population
    return np.random.normal(center, scale, size=(num_samples, len(center)))

state = np.array([9400, 0.06, 0.12, 0.3])
history = []

for epoch in range(120):
    state = quantum_step(state)
    state[0] += telomerase_control(state) * 40  # telomere repair
    history.append(state.copy())
```

```python
import numpy as np

class Organ:
    def __init__(self, name, telomere_setpoint, regen_rate):
        self.name = name
        self.telomere = telomere_setpoint - np.random.randint(200,600)
        self.damage = 0.1
        self.oncogenic_risk = 0.05
        self.regen_rate = regen_rate
        self.cells = 1e6

    def step(self, systemic_stress):
        self.damage += systemic_stress * 0.03
        self.telomere -= 25

        if self.needs_repair():
            self.telomere += self.regen_rate * self.telomerase()
            self.cells *= 1.002

        self.phagocytosis()

    def needs_repair(self):
        return self.damage > 0.2 and self.oncogenic_risk < 0.15

    def telomerase(self):
        return max(0, 1 - self.oncogenic_risk - self.damage)

    def phagocytosis(self):
        self.damage *= 0.85
        self.oncogenic_risk *= 0.97
```

```python
class AGICoordinator:
    def __init__(self, organs):
        self.organs = organs
        self.mode = "homeostasis"

    def evaluate(self):
        avg_damage = np.mean([o.damage for o in self.organs])
        avg_risk = np.mean([o.oncogenic_risk for o in self.organs])

        if avg_damage > 0.25:
            self.mode = "repair"
        elif avg_risk > 0.15:
            self.mode = "suppression"
        else:
            self.mode = "homeostasis"

    def systemic_stress(self):
        if self.mode == "repair":
            return 0.05
        if self.mode == "suppression":
            return 0.0
        return 0.02
```

```python
organs = [
    Organ("Stem Niche", 11000, regen_rate=60),
    Organ("Liver", 10500, regen_rate=45),
    Organ("Epithelium", 10000, regen_rate=30),
    Organ("Immune", 9500, regen_rate=0),
    Organ("Brain", 11500, regen_rate=10)
]

agi = AGICoordinator(organs)

history = {o.name: [] for o in organs}

for step in range(150):
    agi.evaluate()
    stress = agi.systemic_stress()

    for o in organs:
        o.step(stress)
        history[o.name].append(o.telomere)
```

```python
import numpy as np

class VirtualCell:
    def __init__(self):
        self.L = np.random.normal(10.0, 0.5)   # telomere length
        self.D = 0.0                            # division demand
        self.S = np.random.rand()               # stemness
        self.G = 0.0                            # genomic stress
        self.alive = True

    def step(self, telomerase):
        if not self.alive:
            return

        # engineered telomere (resistant)
        erosion = 0.02 * (1 + self.G)
        trimming = 0.05 * max(0, self.L - 12.0)

        self.L += -erosion + 0.15 * telomerase - trimming
        self.G += np.random.normal(0.0, 0.01)

        # senescence / damage
        if self.L < 5.0 or self.G > 1.0:
            self.alive = False
```

```python
def phagocytosis(cells):
    return [c for c in cells if c.alive]
```

```python
import torch
import torch.nn as nn

class TelomeraseController(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(4, 64),
            nn.GELU(),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.net(x)
```

```python
import torch

def gated_telomerase(cell, u_raw):
    # Hard biological constraints
    if cell.G > 0.7 or cell.D < 0.1 or cell.S < 0.3:
        return torch.tensor(0.0, dtype=torch.float32) # Return a tensor zero
    return u_raw
```

```python
controller = TelomeraseController()
optimizer = torch.optim.Adam(controller.parameters(), lr=1e-2)

L_TARGET = 10.0

for epoch in range(2000):
    cells = [VirtualCell() for _ in range(200)]
    loss_total = torch.tensor(0.0) # Initialize as a tensor

    for _ in range(50):
        for c in cells:
            if not c.alive:
                continue

            c.D = np.random.rand()   # tissue demand
            x = torch.tensor([c.L, c.D, c.S, c.G], dtype=torch.float32)
            u_raw = controller(x) # u_raw is a tensor with requires_grad=True

            # Call gated_telomerase with u_raw (tensor)
            u_tensor_gated = gated_telomerase(c, u_raw)

            # Pass the item() value to c.step as it expects a float
            c.step(u_tensor_gated.item())

            # Calculate loss component for this cell using tensors
            current_L_tensor = torch.tensor(c.L, dtype=torch.float32) # c.L is a float

            # Ensure L_TARGET is a tensor for calculation
            L_TARGET_tensor = torch.tensor(L_TARGET, dtype=torch.float32)

            cell_loss = (current_L_tensor - L_TARGET_tensor)**2 + 0.1 * u_tensor_gated
            loss_total = loss_total + cell_loss # Accumulate tensors

        cells = phagocytosis(cells)

    # After the loops, loss_total is a tensor
    # Ensure len(cells) is not zero to avoid division by zero
    if len(cells) > 0:
        loss = loss_total / len(cells)
    else:
        loss = torch.tensor(0.0) # Handle case where all cells die

    optimizer.zero_grad()
    loss.backward() # Now loss is a tensor with a computational graph
    torch.nn.utils.clip_grad_norm_(controller.parameters(), 1.0)
    optimizer.step()

    if epoch % 200 == 0:
        print(f"Epoch {epoch} | Loss {loss.item():.4f}")
```

```python
# core/telomere.py
class Telomere:
    def __init__(self, length, repeat_stability=1.3):
        self.length = length
        self.repeat_stability = repeat_stability

    def erode(self, damage):
        return 0.02 * damage / self.repeat_stability

    def extend(self, telomerase_activity):
        return 0.15 * telomerase_activity

    def trim(self, L_max=12.0):
        return max(0.0, self.length - L_max) * 0.05
```

```python
# core/cell.py
class Cell:
    def __init__(self, stemness):
        self.telomere = Telomere(length=10.0)
        self.stemness = stemness
        self.damage = 0.0
        self.division_demand = 0.0
        self.alive = True

    def step(self, telomerase):
        erosion = self.telomere.erode(self.damage)
        extension = self.telomere.extend(telomerase)
        trimming = self.telomere.trim()

        self.telomere.length += -erosion + extension - trimming
        self.damage += 0.01 * abs(np.random.randn())

        if self.telomere.length < 5.0 or self.damage > 1.2:
            self.alive = False
```

```python
# control/safety_constraints.py
def safety_projection(cell, u):
    if cell.damage > 0.7:
        return 0.0
    if cell.stemness < 0.3:
        return 0.0
    if cell.division_demand < 0.2:
        return 0.0
    return min(u, 0.6)
```

```python
# learning/curiosity.py
def curiosity_bonus(prev_L, curr_L):
    return abs(curr_L - prev_L)
```

```python
# learning/meta_optimizer.py
def adaptive_lr(epoch):
    if epoch < 200:
        return 1e-2
    elif epoch < 1000:
        return 3e-3
    return 1e-3
```

```python
# organ/tissue.py
class Tissue:
    def __init__(self, n_cells=1000):
        self.cells = [Cell(np.random.rand()) for _ in range(n_cells)]

    def step(self, controller):
        for cell in self.cells:
            if not cell.alive:
                continue
            x = torch.tensor([
                cell.telomere.length,
                cell.division_demand,
                cell.stemness,
                cell.damage
            ])
            u_raw = controller(x)
            u = safety_projection(cell, u_raw.item())
            cell.step(u)

        self.cells = [c for c in self.cells if c.alive]
```

```python
import numpy as np
import uuid

class VirtualCellAgent:
    def __init__(self, role):
        self.id = uuid.uuid4().hex[:6]
        self.role = role

        self.alignment = np.random.uniform(0.6, 0.9)
        self.stability = np.random.uniform(0.6, 0.9)
        self.complexity = np.random.uniform(0.3, 0.6)

        self.telomere = self.compute_telomere()
        self.performance = 0.0
        self.risk = 0.0

    def compute_telomere(self):
        return self.alignment + self.stability - self.complexity

    def act(self, task_demand):
        # Task execution depletes telomere
        effort = task_demand * 0.1
        self.complexity += effort
        self.performance = self.alignment * self.stability - effort
        self.risk = max(0, self.complexity - self.stability)

        self.telomere = self.compute_telomere()
```

```python
class TelomeraseController:
    def __init__(self, T_star=1.0):
        self.T_star = T_star

    def allow_self_improvement(self, agent, demand):
        return (
            agent.telomere < self.T_star and
            demand > 0.6 and
            agent.risk < 0.3
        )

    def apply_telomerase(self, agent):
        delta = 0.1 * (self.T_star - agent.telomere)
        agent.stability += delta
        agent.complexity *= 0.98  # structural repair
        agent.telomere = agent.compute_telomere()
```

```python
def divide(agent):
    child = VirtualCellAgent(agent.role)
    child.alignment = agent.alignment
    child.stability = agent.stability * 0.95
    child.complexity = agent.complexity * 0.9
    child.telomere = child.compute_telomere()
    return child
```

```python
def phagocytosis(population):
    survivors = []
    for agent in population:
        if agent.telomere > 0.2 and agent.risk < 0.5:
            survivors.append(agent)
    return survivors
```

```python
def run_simulation(steps=100):
    population = [VirtualCellAgent("generalist") for _ in range(10)]
    telomerase = TelomeraseController(T_star=1.2)

    for step in range(steps):
        demand = np.random.uniform(0, 1)

        new_agents = []

        for agent in population:
            agent.act(demand)

            if telomerase.allow_self_improvement(agent, demand):
                telomerase.apply_telomerase(agent)

                if agent.telomere > 1.0:
                    new_agents.append(divide(agent))

        population.extend(new_agents)
        population = phagocytosis(population)

        print(f"Step {step:03d} | Pop: {len(population)} | Avg T: "
              f"{np.mean([a.telomere for a in population]):.2f}")

    return population
```

```python
TELOMERE_DATABASE = {
    "human": {
        "repeat": "TTAGGG",
        "maintenance": "telomerase",
        "stability": 0.75,
        "cancer_risk": 0.8,
        "repair_efficiency": 0.6
    },
    "mouse": {
        "repeat": "TTAGGG",
        "maintenance": "telomerase",
        "stability": 0.7,
        "cancer_risk": 0.85,
        "repair_efficiency": 0.65
    },
    "yeast": {
        "repeat": "TG(1-3)",
        "maintenance": "telomerase",
        "stability": 0.85,
        "cancer_risk": 0.3,
        "repair_efficiency": 0.8
    },
    "plants": {
        "repeat": "TTTAGGG",
        "maintenance": "telomerase",
        "stability": 0.9,
        "cancer_risk": 0.2,
        "repair_efficiency": 0.85
    },
    "drosophila": {
        "repeat": "retrotransposon-based",
        "maintenance": "ALT-like",
        "stability": 0.6,
        "cancer_risk": 0.1,
        "repair_efficiency": 0.9
    }
}
```

```python
class TelomereGenome:
    def __init__(self, species_template):
        ref = TELOMERE_DATABASE[species_template]

        self.stability = ref["stability"]
        self.repair_efficiency = ref["repair_efficiency"]
        self.cancer_risk = ref["cancer_risk"]

    def mutate(self, rate=0.05):
        self.stability += np.random.normal(0, rate)
        self.repair_efficiency += np.random.normal(0, rate)
        self.cancer_risk += np.random.normal(0, rate)

        self.stability = np.clip(self.stability, 0, 1)
        self.repair_efficiency = np.clip(self.repair_efficiency, 0, 1)
        self.cancer_risk = np.clip(self.cancer_risk, 0, 1)
```

```python
class EvolvingVirtualCell:
    def __init__(self, species="human"):
        self.telomere_genome = TelomereGenome(species)

        self.alignment = np.random.uniform(0.6, 0.9)
        self.complexity = np.random.uniform(0.3, 0.6)
        self.age = 0

        self.telomere_length = self.compute_telomere()
        self.risk = 0

    def compute_telomere(self):
        return (
            self.telomere_genome.stability +
            self.telomere_genome.repair_efficiency -
            self.complexity
        )

    def step(self, environmental_stress):
        self.age += 1

        damage = environmental_stress * (1 - self.telomere_genome.repair_efficiency)
        self.complexity += damage * 0.1

        self.risk = self.telomere_genome.cancer_risk * self.complexity
        self.telomere_length = self.compute_telomere()
```

```python
class EvolutionaryTelomeraseController:
    def __init__(self, target=1.2):
        self.target = target

    def should_activate(self, cell, demand):
        return (
            cell.telomere_length < self.target and
            demand > 0.5 and
            cell.risk < 0.4
        )

    def activate(self, cell):
        cell.telomere_genome.repair_efficiency += 0.05
        cell.telomere_genome.cancer_risk *= 0.97
        cell.complexity *= 0.98
        cell.telomere_length = cell.compute_telomere()
```

```python
def evolve(cell):
    child = EvolvingVirtualCell()
    child.telomere_genome = cell.telomere_genome
    child.telomere_genome.mutate(rate=0.03)
    return child
```

```python
def immune_selection(population):
    return [
        c for c in population
        if c.telomere_length > 0.3 and c.risk < 0.5
    ]
```

```python
def run_evolutionary_simulation(steps=200):
    population = [EvolvingVirtualCell(species=np.random.choice(list(TELOMERE_DATABASE.keys())))
                  for _ in range(20)]

    controller = EvolutionaryTelomeraseController()

    for t in range(steps):
        demand = np.random.uniform(0, 1)
        stress = np.random.uniform(0, 1)

        offspring = []

        for cell in population:
            cell.step(stress)

            if controller.should_activate(cell, demand):
                controller.activate(cell)

            if cell.telomere_length > 1.1:
                offspring.append(evolve(cell))

        population.extend(offspring)
        population = immune_selection(population)

        avg_T = np.mean([c.telomere_length for c in population])
        avg_risk = np.mean([c.risk for c in population])

        print(f"Gen {t:03d} | Pop {len(population):03d} | ⟨T⟩={avg_T:.2f} | ⟨Risk⟩={avg_risk:.2f}")

    return population
```
