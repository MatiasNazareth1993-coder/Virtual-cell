```python
"""
Quantum-inspired Variational Optimizer (simulated) for Genome Search

SAFE: purely computational. This module implements a classical, "quantum-inspired"
variational optimizer (Natural Evolution Strategies style) that can be used as a
drop-in replacement for the GA in your evolutionary pipeline. It is NOT quantum
code and contains no lab or sequence-level instructions.

Key features:
- NES-style population sampling around a parameter vector (mean genome) with adaptive
  step-size and optional covariance scaling.
- Surrogate model (optional) using PyTorch to approximate the expensive objective
  and reduce simulation calls.
- Simple API: optimizer.optimize(objective_fn, init_mean, bounds, n_iters, pop_size)
  where objective_fn(genome) -> score (higher better).
- Example integration snippet shows how to call it with the previously-defined
  evaluate_genome function from your evolve_supercells.py pipeline.

Usage example (pseudo):
    from quantum_inspired_optimizer import VariationalOptimizer

    def obj(genome):
        m = evaluate_genome(genome, cfg)
        return m['composite']

    opt = VariationalOptimizer(dim=GENOME_SIZE)
    best = opt.optimize(obj, init_mean=np.full(GENOME_SIZE,0.5), bounds=(0,1), n_iters=60)

This file is intended to be integrated into your in-silico project.
"""

import numpy as np
import math
import random
from typing import Callable, Optional, Tuple

# Optional surrogate model uses PyTorch; if not available, surrogate is disabled
try:
    import torch
    import torch.nn as nn
    HAS_TORCH = True
except Exception:
    HAS_TORCH = False


class VariationalOptimizer:
    """
    NES-like variational optimizer.

    Parameters
    ----------
    dim : int
        Dimensionality of the genome vector.
    sigma_init : float
        Initial standard deviation for sampling noise.
    lr : float
        Learning rate for mean updates.
    pop_size : int
        Default population size (can be overridden in optimize call).
    use_surrogate : bool
        If True and PyTorch is available, fit a small neural surrogate to reduce
        the number of expensive objective evaluations.
    surrogate_warmup : int
        Number of real evaluations before training surrogate.
    """

    def __init__(self, dim: int, sigma_init: float = 0.1, lr: float = 0.2, pop_size: int = 64,
                 use_surrogate: bool = True, surrogate_warmup: int = 200):
        self.dim = dim
        self.sigma = sigma_init
        self.lr = lr
        self.pop_size = pop_size
        self.mean = np.zeros(dim, dtype=float)
        self.cov_scale = 1.0
        self.use_surrogate = use_surrogate and HAS_TORCH
        self.surrogate_warmup = surrogate_warmup

        # surrogate components
        self.surrogate = None
        self.surrogate_optimizer = None
        self.surr_train_X = []
        self.surr_train_y = []

        if self.use_surrogate:
            self._init_surrogate()

    def set_mean(self, mean: np.ndarray):
        assert mean.shape[0] == self.dim
        self.mean = mean.astype(float)

    def _init_surrogate(self):
        # small MLP
        if not HAS_TORCH:
            self.use_surrogate = False
            return
        model = nn.Sequential(
            nn.Linear(self.dim, max(16, self.dim*4)),
            nn.ReLU(),
            nn.Linear(max(16, self.dim*4), max(16, self.dim*2)),
            nn.ReLU(),
            nn.Linear(max(16, self.dim*2), 1)
        )
        self.surrogate = model
        self.surrogate_optimizer = torch.optim.Adam(self.surrogate.parameters(), lr=1e-3)

    def _train_surrogate(self, epochs: int = 20):
        if not self.use_surrogate or len(self.surr_train_X) < 32:
            return
        X = np.vstack(self.surr_train_X).astype(np.float32)
        y = np.array(self.surr_train_y, dtype=np.float32).reshape(-1,1)
        import torch
        X_t = torch.from_numpy(X)
        y_t = torch.from_numpy(y)
        dataset = torch.utils.data.TensorDataset(X_t, y_t)
        loader = torch.utils.data.DataLoader(dataset, batch_size=64, shuffle=True)
        loss_fn = nn.MSELoss()
        for _ in range(epochs):
            for bx, by in loader:
                preds = self.surrogate(bx)
                loss = loss_fn(preds, by)
                self.surrogate_optimizer.zero_grad()
                loss.backward()
                self.surrogate_optimizer.step()

    def _surrogate_predict(self, X: np.ndarray) -> np.ndarray:
        if not self.use_surrogate:
            raise RuntimeError("Surrogate disabled")
        import torch
        with torch.no_grad():
            t = torch.from_numpy(X.astype(np.float32))
            out = self.surrogate(t).cpu().numpy().reshape(-1)
        return out

    def _sample_population(self, mean: np.ndarray, sigma: float, pop_size: int) -> np.ndarray:
        # isotropic Gaussian sampling scaled by cov_scale
        eps = np.random.randn(pop_size, self.dim)
        samples = mean[None,:] + eps * (sigma * self.cov_scale)
        return samples, eps

    def optimize(self,
                 objective_fn: Callable[[np.ndarray], float],
                 init_mean: Optional[np.ndarray] = None,
                 bounds: Optional[Tuple[float,float]] = (0.0,1.0),
                 n_iters: int = 100,
                 pop_size: Optional[int] = None,
                 verbose: bool = True):
        """
        Optimize the objective function (maximize) starting from init_mean.

        Returns best_genome, best_score
        """
        if pop_size is None:
            pop_size = self.pop_size
        if init_mean is not None:
            self.set_mean(init_mean)
        else:
            # initialize to 0.5 if not set
            self.mean = np.full(self.dim, 0.5, dtype=float)

        best_genome = self.mean.copy()
        best_score = -np.inf

        for it in range(n_iters):
            samples, eps = self._sample_population(self.mean, self.sigma, pop_size)
            # clip samples to bounds
            low, high = bounds
            samples_clipped = np.clip(samples, low, high)

            scores = np.zeros(pop_size, dtype=float)

            # if surrogate is enabled and we have warmup, use mixture strategy
            if self.use_surrogate and len(self.surr_train_X) >= self.surrogate_warmup:
                # predict scores with surrogate, then selectively evaluate top-K with real objective
                preds = self._surrogate_predict(samples_clipped)
                # rank by surrogate, evaluate top fraction
                k = max(4, int(pop_size * 0.25))
                top_idx = np.argsort(preds)[-k:]
                for i in range(pop_size):
                    if i in top_idx:
                        scores[i] = objective_fn(samples_clipped[i])
                        # store for surrogate training
                        self.surr_train_X.append(samples_clipped[i].copy())
                        self.surr_train_y.append(scores[i])
                    else:
                        scores[i] = preds[i]
            else:
                # full evaluation
                for i in range(pop_size):
                    scores[i] = objective_fn(samples_clipped[i])
                    # collect surrogate training data
                    if self.use_surrogate:
                        self.surr_train_X.append(samples_clipped[i].copy())
                        self.surr_train_y.append(scores[i])

            # update best
            idx = np.argmax(scores)
            if scores[idx] > best_score:
                best_score = float(scores[idx])
                best_genome = samples_clipped[idx].copy()

            # standard score normalization
            scores_norm = (scores - np.mean(scores)) / (np.std(scores) + 1e-8)
            # NES gradient estimate: update mean in direction of weighted eps
            grad = np.dot(scores_norm, eps) / pop_size
            self.mean = self.mean + self.lr * grad
            # optional: adapt sigma based on score variance
            score_var = np.var(scores)
            self.sigma = max(1e-4, self.sigma * (1.0 + 0.01 * (score_var - 1.0)))

            # keep mean in bounds
            self.mean = np.clip(self.mean, low, high)

            # train surrogate periodically
            if self.use_surrogate and it % 10 == 0 and len(self.surr_train_X) >= 64:
                try:
                    self._train_surrogate(epochs=8)
                except Exception:
                    pass

            if verbose and (it % max(1, n_iters//10) == 0 or it == n_iters-1):
                print(f"Iter {it+1}/{n_iters} best={best_score:.4f} sigma={self.sigma:.4f}")

        return best_genome, best_score


# ------------------ Example integration helper ------------------
if __name__ == '__main__':
    # Small demo using a toy objective: shifted quadratic (max at [0.8,..0.8])
    def toy_obj(x):
        center = np.full(x.shape[0], 0.8)
        return -np.sum((x - center)**2)

    dim = 6
    opt = VariationalOptimizer(dim=dim, sigma_init=0.2, lr=0.15, pop_size=80, use_surrogate=False)
    mean0 = np.full(dim, 0.5)
    best, score = opt.optimize(toy_obj, init_mean=mean0, n_iters=120, pop_size=120)
    print('Demo best score', score, 'best genome', best)
```

```python
"""
Advanced Optimizer Suite for Virtual Cell Evolution
--------------------------------------------------
SAFE: computational-only algorithms. No biological editing.

This module adds three high‑performance optimizers to integrate with the
virtual cell / organ‑on‑chip evolutionary system:

1. **CMA‑ES (Covariance Matrix Adaptation Evolution Strategy)**
   - Gold standard for continuous black‑box optimization.
   - Excellent for high‑dimensional, rugged landscapes (like stem‑cell state control,
     telomere dynamics, virtual repair efficiency metrics).

2. **POEM (Policy Optimization with Evolutionary Methods, simplified)**
   - Hybrid between policy‑gradient RL and evolutionary search.
   - Good for optimizing the deep learning controller that regulates telomerase bursts.

3. **SA‑NES (Self‑Adaptive Natural Evolution Strategies)**
   - Similar to NES, but learns its own step‑sizes and sampling distribution.
   - Works well for long training runs inside simulated environments.

You can plug these optimizers directly into evaluate_genome(...) or any other
objective from the simulation framework.
"""

import numpy as np
from typing import Callable

# -----------------------------------------------------------------------------
# 1. CMA‑ES (minimal pure‑Python version)
# -----------------------------------------------------------------------------
class CMAES:
    def __init__(self, dim, sigma=0.3, pop_size=None):
        self.dim = dim
        self.sigma = sigma
        self.mean = np.zeros(dim)
        self.pop_size = pop_size or (4 + int(3 * np.log(dim)))
        self.cov = np.eye(dim)

    def ask(self):
        A = np.linalg.cholesky(self.cov)
        z = np.random.randn(self.pop_size, self.dim)
        samples = self.mean + self.sigma * (z @ A.T)
        return samples

    def tell(self, samples, scores):
        # sort by score
        idx = np.argsort(scores)[::-1]
        samples = samples[idx]
        scores = scores[idx]

        # recombination weights
        mu = self.pop_size // 2
        weights = np.log(mu + 0.5) - np.log(np.arange(1, mu+1))
        weights = weights / np.sum(weights)

        # update mean
        new_mean = np.sum(samples[:mu] * weights[:,None], axis=0)

        # update covariance
        y = samples[:mu] - self.mean
        C_new = np.zeros_like(self.cov)
        for i in range(mu):
            C_new += weights[i] * np.outer(y[i], y[i])

        self.mean = new_mean
        self.cov = 0.9 * self.cov + 0.1 * C_new

    def run(self, objective_fn: Callable[[np.ndarray],float], n_iters=50, bounds=(0,1)):
        best, best_score = None, -np.inf
        low, high = bounds
        for _ in range(n_iters):
            samples = self.ask()
            samples = np.clip(samples, low, high)
            scores = np.array([objective_fn(s) for s in samples])
            if scores.max() > best_score:
                best_score = scores.max()
                best = samples[np.argmax(scores)].copy()
            self.tell(samples, scores)
        return best, best_score

# -----------------------------------------------------------------------------
# 2. POEM‑lite (Policy Optimization with Evolutionary Methods)
# -----------------------------------------------------------------------------
class POEM:
    """Hybrid evolutionary + gradient estimator.
    Works for optimizing neural controller parameters.
    """

    def __init__(self, dim, lr=0.05, sigma=0.15, pop_size=40):
        self.dim = dim
        self.lr = lr
        self.sigma = sigma
        self.pop_size = pop_size
        self.theta = np.zeros(dim)

    def step(self, objective_fn):
        eps = np.random.randn(self.pop_size, self.dim)
        samples = self.theta[None,:] + self.sigma * eps
        scores = np.array([objective_fn(s) for s in samples])
        adv = (scores - scores.mean()) / (scores.std() + 1e-8)
        grad = (eps.T @ adv) / self.pop_size
        self.theta += self.lr * grad
        return scores.max(), samples[np.argmax(scores)]

    def run(self, objective_fn, n_iters=60, bounds=(0,1)):
        best, best_score = None, -np.inf
        low, high = bounds
        for _ in range(n_iters):
            score, genome = self.step(objective_fn)
            genome = np.clip(genome, low, high)
            if score > best_score:
                best_score = score
                best = genome.copy()
        return best, best_score

# -----------------------------------------------------------------------------
# 3. SA‑NES (Self‑Adaptive Natural Evolution Strategy)
# -----------------------------------------------------------------------------
class SANES:
    def __init__(self, dim, lr=0.2, sigma=0.2, pop_size=50):
        self.dim = dim
        self.lr = lr
        self.sigma = sigma
        self.pop_size = pop_size
        self.mean = np.zeros(dim)
        self.alpha = 0.1  # self‑adaptive sigma multiplier

    def run(self, objective_fn, n_iters=80, bounds=(0,1)):
        low, high = bounds
        best, best_score = None, -np.inf

        for it in range(n_iters):
            eps = np.random.randn(self.pop_size, self.dim)
            samples = self.mean + self.sigma * eps
            samples = np.clip(samples, low, high)
            scores = np.array([objective_fn(s) for s in samples])

            if scores.max() > best_score:
                best_score = scores.max()
                best = samples[np.argmax(scores)].copy()

            # NES grad
            norm_scores = (scores - scores.mean()) / (scores.std() + 1e-8)
            grad_mean = (eps.T @ norm_scores) / self.pop_size

            # update mean
            self.mean += self.lr * grad_mean
            self.mean = np.clip(self.mean, low, high)

            # self-adapt sigma
            self.sigma *= np.exp(self.alpha * np.mean(norm_scores * (np.sum(eps**2,axis=1)-self.dim)))
            self.sigma = max(1e-4, min(self.sigma, 2.0))

        return best, best_score

# -----------------------------------------------------------------------------
# Utility: choose optimizer via string
# -----------------------------------------------------------------------------
def get_optimizer(name: str, dim: int):
    n = name.lower()
    if n == "cmaes":
        return CMAES(dim)
    if n == "poem":
        return POEM(dim)
    if n == "sanes":
        return SANES(dim)
    raise ValueError(f"Unknown optimizer: {name}")
```

```python
import numpy as np
from typing import Callable, Dict

# ---------------------------------------------------------------------------
# Placeholder imports — in your environment these map to canvas modules
# ---------------------------------------------------------------------------
# Removed: from advanced_optimizers import (
# Removed:     CMAESOptimizer,
# Removed:     NSGA2,
# Removed:     MAPElites,
# Removed: )
# Removed: from quantum_inspired_optimizer import QuantumInspiredNES

# Assuming CMAES, POEM, SANES (from K2hpSKqGlvn-) and VariationalOptimizer (from gNbrnGk2luKD)
# are available from previously executed cells.

# ---------------------------------------------------------------------------
# Objective wrapper — your existing simulation
# ---------------------------------------------------------------------------
def evaluate_genome(genome: np.ndarray) -> float:
    """Stub: your simulation environment plugs in here.
    Returns a scalar fitness. Replace with actual simulation call.
    """
    repair_eff = genome[1]
    longevity = genome[0]
    energy = genome[4]
    return float((repair_eff * 1.4 + longevity * 1.2 + energy * 0.6))

# ---------------------------------------------------------------------------
# Unified optimizer interface
# ---------------------------------------------------------------------------
class EvolutionEngine:
    def __init__(self, dim: int = 6, bounds=(0.0, 1.0)):
        self.dim = dim
        self.bounds = bounds

    def _clip(self, x):
        lo, hi = self.bounds
        return np.clip(x, lo, hi)

    # -------------------- GA baseline --------------------
    def run_ga(self, n_gen=50, pop=40, elite=5):
        p = np.random.uniform(*self.bounds, size=(pop, self.dim))
        for _ in range(n_gen):
            fit = np.array([evaluate_genome(g) for g in p])
            elites = p[np.argsort(fit)[-elite:]]
            children = []
            for _ in range(pop - elite):
                a, b = elites[np.random.randint(0, elite, 2)]
                c = (a + b) / 2 + np.random.normal(0, 0.05, self.dim)
                children.append(self._clip(c))
            p = np.vstack([elites] + children)
        return p[np.argmax([evaluate_genome(g) for g in p])]

    # -------------------- CMA-ES --------------------
    def run_cma(self, iters=50):
        # Changed CMAESOptimizer to CMAES and added required pop_size for CMAES constructor
        opt = CMAES(dim=self.dim, pop_size=64)
        g, f = opt.run(objective_fn=evaluate_genome, n_iters=iters, bounds=self.bounds)
        return g

    # -------------------- NSGA-II --------------------
    def run_nsga2(self, iters=40, pop=60):
        # NSGA2 class is not defined in the current notebook. Commenting out.
        # def multi_obj(g):
        #     return [evaluate_genome(g), -np.sum((g - 0.5)**2)]
        # opt = NSGA2(dim=self.dim, pop=pop, bounds=self.bounds)
        # return opt.run(multi_obj, n_gen=iters)
        print("NSGA-II optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- MAP-Elites --------------------
    def run_map_elites(self, iters=200):
        # MAPElites class is not defined in the current notebook. Commenting out.
        # opt = MAPElites(dim=self.dim, bounds=self.bounds)
        # archive = opt.run(evaluate_genome, iters)
        # return archive
        print("MAP-Elites optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- Quantum-inspired NES --------------------
    def run_qnes(self, iters=80):
        # Changed QuantumInspiredNES to VariationalOptimizer
        opt = VariationalOptimizer(dim=self.dim)
        best_genome, best_score = opt.optimize(evaluate_genome, n_iters=iters, bounds=self.bounds)
        return best_genome


if __name__ == "__main__":
    # Dummy Env for testing
    class DummyEnv:
        def __init__(self):
            self.cells = {}
            self.logger = None

    # For a standalone test, `dim` needs to be defined if not from a global context
    GENOME_SIZE = 6 # Define GENOME_SIZE for the example.
    engine = EvolutionEngine(dim=GENOME_SIZE)
    print("Running GA...")
    best_ga = engine.run_ga()
    print("Running CMA-ES...")
    best_cma = engine.run_cma()
    print("Running QNES...")
    best_q = engine.run_qnes()
    print("GA best:", best_ga)
    print("CMA-ES best:", best_cma)
    print("QNES best:", best_q)

# --- New Optimizers: Lion, Adam-Evolutionary, Hybrid Annealing ---

import torch

class LionOptimizer:
    """
    Lion-inspired evolutionary optimizer (not biological LION) — gradient-free.
    """
    def __init__(self, lr=0.1, beta=0.95):
        self.lr = lr
        self.beta = beta
        self.m = None

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
        self.m = self.beta * self.m + (1 - self.beta) * grad
        return genome - self.lr * torch.sign(self.m)


class AdamEvolutionary:
    """Adam-like optimizer but operating over sampled gradients from eval noise."""
    def __init__(self, lr=0.05):
        self.lr = lr
        self.m = None
        self.v = None
        self.t = 0

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
            self.v = torch.zeros_like(grad)
        self.t += 1
        self.m = 0.9 * self.m + 0.1 * grad
        self.v = 0.999 * self.v + 0.001 * grad**2

        m_hat = self.m / (1 - 0.9**self.t)
        v_hat = self.v / (1 - 0.999**self.t)

        return genome - self.lr * m_hat / (torch.sqrt(v_hat) + 1e-8)


class HybridAnnealing:
    def __init__(self, temp_start=1.0, temp_end=0.01, decay=0.99):
        self.temp = temp_start
        self.temp_end = temp_end
        self.decay = decay

    def step(self, genome, neighbor, score, neighbor_score):
        if neighbor_score > score:
            return neighbor
        import math, random
        prob = math.exp(-(score - neighbor_score) / max(self.temp, 1e-8))
        if random.random() < prob:
            return neighbor
        self.temp = max(self.temp * self.decay, self.temp_end)
        return genome


# --- Integration with Telomere / OoC / Stem Cell Models ---
# These hooks expect env objects from your existing simulation.

def integrate_with_telomeres(env, genome):
    env.cell.telomere_state.target_length = genome[0].item()
    env.cell.telomerase.max_activity = genome[1].item()


def integrate_with_ooc(env, genome):
    env.ooc.flow_rate = genome[2].item()
    env.ooc.ecm_response = genome[3].item()


def integrate_stem_cell_logic(env, genome):
    env.stem.division_threshold = genome[4].item()
    env.stem.quiescence_bias = genome[5].item()


# Unified registration
def apply_all_biological_mappings(env, genome):
    integrate_with_telomeres(env, genome)
    integrate_with_ooc(env, genome)
    integrate_stem_cell_logic(env, genome)


# ----------------------- Telomere Mathematical Model -----------------------
# Purely in-silico, abstract kinetic model for telomere homeostasis and risk.
# This module defines a TelomereKinetics class representing kinetics of
# telomere shortening, telomerase (TERT) action, trimming, oxidative stress
# influence, replication asymmetry and crisis probability. It provides
# convenience functions to simulate dynamics for a cell or a small population
# given a candidate genome (vector of control parameters).

import math

class TelomereKinetics:
    """
    Telomere kinetics simulator (abstract units).

    Parameters (all unitless/simulation-scaled):
      - base_shorten: baseline shortening per division (sim units)
      - tert_max: maximum telomerase activity (units per pulse)
      - tert_sensitivity: how responsive telomerase is to stress/need
      - trimming_rate: rate at which trimming reduces excess telomere
      - ox_sensitivity: how oxidative stress increases damage per tick
      - asymmetry_bias: fraction of divisions that are asymmetric (0..1)
      - crisis_threshold: mutation/damage threshold where crisis risk rises
    """

    def __init__(self,
                 base_shorten=1.0,
                 tert_max=5.0,
                 tert_sensitivity=1.0,
                 trimming_rate=0.1,
                 ox_sensitivity=0.5,
                 asymmetry_bias=0.1,
                 crisis_threshold=0.6,
                 safe_low=30.0,
                 safe_high=160.0):
        self.base_shorten = base_shorten
        self.tert_max = tert_max
        self.tert_sensitivity = tert_sensitivity
        self.trimming_rate = trimming_rate
        self.ox_sensitivity = ox_sensitivity
        self.asymmetry_bias = asymmetry_bias
        self.crisis_threshold = crisis_threshold
        self.safe_low = safe_low
        self.safe_high = safe_high

    def step(self, tel_len, damage, mutation_burden, demand_signal, oxidative_stress,
             apply_tert_fraction=0.0, allow_trimming=True):
        """
        Advance one time tick for a single cell's telomere state.

        Inputs:
          - tel_len: current telomere length (sim units)
          - damage: current damage score (0..1)
          - mutation_burden: current mutation burden (0..1)
          - demand_signal: local demand for proliferation/repair (0..1)
          - oxidative_stress: local oxidative stress (0..1)
          - apply_tert_fraction: fraction [0,1] of tert_max to apply this tick
          - allow_trimming: whether trimming is permitted

        Returns: new_tel_len, new_damage, crisis_risk
        """
        # 1) Telomere shortening due to replication-like events
        # model a small continuous shortening proportional to demand
        rep_events = demand_signal * (1.0 - self.asymmetry_bias)
        shorten = self.base_shorten * rep_events

        # 2) Oxidative stress increases effective shortening and damage
        ox_factor = 1.0 + self.ox_sensitivity * oxidative_stress
        shorten *= ox_factor
        damage += oxidative_stress * 0.01  # slow accrual from ox

        # 3) Apply telomerase pulse (fraction of tert_max)
        tert_amount = apply_tert_fraction * self.tert_max * self.tert_sensitivity
        tel_len = tel_len - shorten + tert_amount

        # 4) Trimming to avoid runaway length
        if allow_trimming and tel_len > self.safe_high:
            excess = tel_len - self.safe_high
            trim = min(excess, self.trimming_rate * excess)
            tel_len -= trim
            # trimming may produce transient damage
            damage += 0.001 * trim

        # 5) Small stochastic noise
        tel_len += np.random.normal(0.0, max(0.01, 0.001 * tel_len))
        tel_len = max(0.0, tel_len)

        # 6) Mutation/damage-driven crisis risk
        # risk increases with mutation_burden and damage and if telomeres are very long (instability)
        risk = 0.0
        risk += 0.6 * min(1.0, mutation_burden / max(1e-6, self.crisis_threshold))
        risk += 0.3 * damage
        if tel_len > self.safe_high * 1.3:
            risk += 0.2
        risk = min(1.0, risk)

        # 7) Damage decays slowly due to repair (we model some repair capacity)
        repair_capacity = max(0.0, 0.02 * (1.0 - damage))
        damage = max(0.0, damage - repair_capacity)

        return tel_len, damage, risk

    def simulate_trajectory(self, init_tel, init_damage, init_mutation, demand_series,
                            ox_series, tert_policy_fn, steps=200):
        """
        Simulate a trajectory for given series of demand and oxidative stress and a
        telomerase policy function tert_policy_fn(state)->fraction.

        Returns list of (tel_len, damage, mutation, risk) per step.
        """
        tel = init_tel
        dmg = init_damage
        mut = init_mutation
        traj = []
        for t in range(steps):
            demand = demand_series[t] if t < len(demand_series) else 0.0
            ox = ox_series[t] if t < len(ox_series) else 0.0
            # policy decides fraction to apply
            frac = tert_policy_fn({'tel': tel, 'damage': dmg, 'mutation': mut, 't': t, 'demand': demand, 'ox': ox})
            tel, dmg, risk = self.step(tel, dmg, mut, demand, ox, apply_tert_fraction=frac)
            # mutation accrual proportional to damage and oxidative stress
            mut += 0.0005 * (dmg + ox)
            mut = min(1.0, mut)
            traj.append((tel, dmg, mut, risk))
        return traj


# -------------------- Example: Genome-driven tert policy --------------------
# We define a small helper that converts a genome vector into a tert policy function.
# Genome layout (example): [target_tel, tert_sensitivity, trim_bias, ox_resilience, asymmetry, crisis_tolerance]

def genome_to_tert_policy(genome, kinetics: TelomereKinetics):
    target = float(genome[0])
    sens = float(genome[1]) * 2.0  # scale
    trim_bias = float(genome[2])

    def policy(state):
        tel = state['tel']
        dmg = state['damage']
        demand = state['demand']
        ox = state['ox']
        # base need proportional to how far below target and current demand
        need = max(0.0, (target - tel) / max(1.0, target)) * (0.4 + 0.6 * demand)
        # reduce need if damage or mutation high (safety)
        safety = max(0.0, 1.0 - 0.8 * (dmg + state['mutation']))
        frac = need * safety * sens
        # apply trimming bias to reduce pulses when trim_bias high
        frac = frac * (1.0 - 0.5 * trim_bias)
        # clip
        frac = max(0.0, min(1.0, frac))
        return frac

    return policy


# -------------------- Integration utility: evaluate telomere performance  ------

def evaluate_telomere_performance(genome, env, steps=200):
    """Run telomere kinetics for a representative subset of cells in env using genome.
    Returns a composite telomere performance score (higher is better):
      score = repair_success - risk_penalty - energy_penalty
    This is a simulation-only metric intended to be combined with other objectives.
    """
    # create kinetics using genome-inspired parameters
    kinetics = TelomereKinetics(
        base_shorten=1.0,
        tert_max=5.0 * (0.5 + genome[1] * 0.5),
        tert_sensitivity=1.0 + genome[1] * 1.5,
        trimming_rate=0.05 + genome[2] * 0.2,
        ox_sensitivity=0.2 + (1.0 - genome[3]) * 0.8,
        asymmetry_bias=0.1 + genome[4] * 0.4,
        crisis_threshold=0.4 + genome[5] * 0.6,
        safe_low=30.0,
        safe_high=160.0
    )

    # pick representative cells (if env has many, sample up to N)
    cells = list(env.cells.values()) if hasattr(env, 'cells') else []
    if len(cells) == 0:
        # no env provided: simulate a default demand/ox profile
        demand = [0.2] * steps
        ox = [0.05] * steps
        init_tel = 100.0
        init_dmg = 0.02
        init_mut = 0.01
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(init_tel, init_dmg, init_mut, demand, ox, policy, steps)
        # compute metrics
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        score = (avg_tel / 100.0) - (avg_risk * 2.0) - (energy * 0.01)
        return float(score)

    # sample up to 20 cells
    sample = random.sample(cells, min(20, len(cells)))
    scores = []
    for c in sample:
        # generate local demand/ox series from env fields if present, else defaults
        demand = []
        ox = []
        for t in range(steps):
            if hasattr(env, 'nutrient'):
                # scaled demand from nutrient at cell location with small temporal noise
                base = float(env.nutrient[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'nutrient') else 0.2
                demand.append(min(1.0, max(0.0, base + np.random.normal(0, 0.02))))
            else:
                demand.append(0.2)
            if hasattr(env, 'toxin'):
                base_ox = float(env.toxin[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'toxin') else 0.02
                ox.append(min(1.0, max(0.0, base_ox + np.random.normal(0, 0.01))))
            else:
                ox.append(0.02)
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(c.telomere if hasattr(c,'telomere') else 100.0,
                                           getattr(c,'damage',0.01),
                                           getattr(c,'mutation_burden',0.01),
                                           demand, ox, policy, steps)
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        scores.append((avg_tel, avg_risk, energy))

    # aggregate score: favor high avg_tel, penalize high risk and energy
    avg_tel_mean = np.mean([s[0] for s in scores])
    avg_risk_mean = np.mean([s[1] for s in scores])
    avg_energy = np.mean([s[2] for s in scores])
    score = (avg_tel_mean / 100.0) - (avg_risk_mean * 1.8) - (avg_energy * 0.01)
    return float(score)


# -------------------- Hook: integrate telomere performance into evaluator ----
# Replace or augment evaluate_genome in evolve_supercells_integrated.py to call
# evaluate_telomere_performance(genome, env) and combine with other objectives.


# End of appended telomere model
```

```python
import numpy as np
from typing import Callable, Dict

# ---------------------------------------------------------------------------
# Placeholder imports — in your environment these map to canvas modules
# ---------------------------------------------------------------------------
# Removed: from advanced_optimizers import (
# Removed:     CMAESOptimizer,
# Removed:     NSGA2,
# Removed:     MAPElites,
# Removed: )
# Removed: from quantum_inspired_optimizer import QuantumInspiredNES

# Assuming CMAES, POEM, SANES (from K2hpSKqGlvn-) and VariationalOptimizer (from gNbrnGk2luKD)
# are available from previously executed cells.

# ---------------------------------------------------------------------------
# Objective wrapper — your existing simulation
# ---------------------------------------------------------------------------
def evaluate_genome(genome: np.ndarray) -> float:
    """Stub: your simulation environment plugs in here.
    Returns a scalar fitness. Replace with actual simulation call.
    """
    repair_eff = genome[1]
    longevity = genome[0]
    energy = genome[4]
    return float((repair_eff * 1.4 + longevity * 1.2 + energy * 0.6))

# ---------------------------------------------------------------------------
# Unified optimizer interface
# ---------------------------------------------------------------------------
class EvolutionEngine:
    def __init__(self, dim: int = 6, bounds=(0.0, 1.0)):
        self.dim = dim
        self.bounds = bounds

    def _clip(self, x):
        lo, hi = self.bounds
        return np.clip(x, lo, hi)

    # -------------------- GA baseline --------------------
    def run_ga(self, n_gen=50, pop=40, elite=5):
        p = np.random.uniform(*self.bounds, size=(pop, self.dim))
        for _ in range(n_gen):
            fit = np.array([evaluate_genome(g) for g in p])
            elites = p[np.argsort(fit)[-elite:]]
            children = []
            for _ in range(pop - elite):
                a, b = elites[np.random.randint(0, elite, 2)]
                c = (a + b) / 2 + np.random.normal(0, 0.05, self.dim)
                children.append(self._clip(c))
            p = np.vstack([elites] + children)
        return p[np.argmax([evaluate_genome(g) for g in p])]

    # -------------------- CMA-ES --------------------
    def run_cma(self, iters=50):
        # Changed CMAESOptimizer to CMAES and added required pop_size for CMAES constructor
        opt = CMAES(dim=self.dim, pop_size=64)
        g, f = opt.run(objective_fn=evaluate_genome, n_iters=iters, bounds=self.bounds)
        return g

    # -------------------- NSGA-II --------------------
    def run_nsga2(self, iters=40, pop=60):
        # NSGA2 class is not defined in the current notebook. Commenting out.
        # def multi_obj(g):
        #     return [evaluate_genome(g), -np.sum((g - 0.5)**2)]
        # opt = NSGA2(dim=self.dim, pop=pop, bounds=self.bounds)
        # return opt.run(multi_obj, n_gen=iters)
        print("NSGA-II optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- MAP-Elites --------------------
    def run_map_elites(self, iters=200):
        # MAPElites class is not defined in the current notebook. Commenting out.
        # opt = MAPElites(dim=self.dim, bounds=self.bounds)
        # archive = opt.run(evaluate_genome, iters)
        # return archive
        print("MAP-Elites optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- Quantum-inspired NES --------------------
    def run_qnes(self, iters=80):
        # Changed QuantumInspiredNES to VariationalOptimizer
        opt = VariationalOptimizer(dim=self.dim)
        best_genome, best_score = opt.optimize(evaluate_genome, n_iters=iters, bounds=self.bounds)
        return best_genome


if __name__ == "__main__":
    # Dummy Env for testing
    class DummyEnv:
        def __init__(self):
            self.cells = {}
            self.logger = None

    # For a standalone test, `dim` needs to be defined if not from a global context
    GENOME_SIZE = 6 # Define GENOME_SIZE for the example.
    engine = EvolutionEngine(dim=GENOME_SIZE)
    print("Running GA...")
    best_ga = engine.run_ga()
    print("Running CMA-ES...")
    best_cma = engine.run_cma()
    print("Running QNES...")
    best_q = engine.run_qnes()
    print("GA best:", best_ga)
    print("CMA-ES best:", best_cma)
    print("QNES best:", best_q)

# --- New Optimizers: Lion, Adam-Evolutionary, Hybrid Annealing ---

import torch

class LionOptimizer:
    """
    Lion-inspired evolutionary optimizer (not biological LION) — gradient-free.
    """
    def __init__(self, lr=0.1, beta=0.95):
        self.lr = lr
        self.beta = beta
        self.m = None

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
        self.m = self.beta * self.m + (1 - self.beta) * grad
        return genome - self.lr * torch.sign(self.m)


class AdamEvolutionary:
    """Adam-like optimizer but operating over sampled gradients from eval noise."""
    def __init__(self, lr=0.05):
        self.lr = lr
        self.m = None
        self.v = None
        self.t = 0

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
            self.v = torch.zeros_like(grad)
        self.t += 1
        self.m = 0.9 * self.m + 0.1 * grad
        self.v = 0.999 * self.v + 0.001 * grad**2

        m_hat = self.m / (1 - 0.9**self.t)
        v_hat = self.v / (1 - 0.999**self.t)

        return genome - self.lr * m_hat / (torch.sqrt(v_hat) + 1e-8)


class HybridAnnealing:
    def __init__(self, temp_start=1.0, temp_end=0.01, decay=0.99):
        self.temp = temp_start
        self.temp_end = temp_end
        self.decay = decay

    def step(self, genome, neighbor, score, neighbor_score):
        if neighbor_score > score:
            return neighbor
        import math, random
        prob = math.exp(-(score - neighbor_score) / max(self.temp, 1e-8))
        if random.random() < prob:
            return neighbor
        self.temp = max(self.temp * self.decay, self.temp_end)
        return genome


# --- Integration with Telomere / OoC / Stem Cell Models ---
# These hooks expect env objects from your existing simulation.

def integrate_with_telomeres(env, genome):
    env.cell.telomere_state.target_length = genome[0].item()
    env.cell.telomerase.max_activity = genome[1].item()


def integrate_with_ooc(env, genome):
    env.ooc.flow_rate = genome[2].item()
    env.ooc.ecm_response = genome[3].item()


def integrate_stem_cell_logic(env, genome):
    env.stem.division_threshold = genome[4].item()
    env.stem.quiescence_bias = genome[5].item()


# Unified registration
def apply_all_biological_mappings(env, genome):
    integrate_with_telomeres(env, genome)
    integrate_with_ooc(env, genome)
    integrate_stem_cell_logic(env, genome)


# ----------------------- Telomere Mathematical Model -----------------------
# Purely in-silico, abstract kinetic model for telomere homeostasis and risk.
# This module defines a TelomereKinetics class representing kinetics of
# telomere shortening, telomerase (TERT) action, trimming, oxidative stress
# influence, replication asymmetry and crisis probability. It provides
# convenience functions to simulate dynamics for a cell or a small population
# given a candidate genome (vector of control parameters).

import math

class TelomereKinetics:
    """
    Telomere kinetics simulator (abstract units).

    Parameters (all unitless/simulation-scaled):
      - base_shorten: baseline shortening per division (sim units)
      - tert_max: maximum telomerase activity (units per pulse)
      - tert_sensitivity: how responsive telomerase is to stress/need
      - trimming_rate: rate at which trimming reduces excess telomere
      - ox_sensitivity: how oxidative stress increases damage per tick
      - asymmetry_bias: fraction of divisions that are asymmetric (0..1)
      - crisis_threshold: mutation/damage threshold where crisis risk rises
    """

    def __init__(self,
                 base_shorten=1.0,
                 tert_max=5.0,
                 tert_sensitivity=1.0,
                 trimming_rate=0.1,
                 ox_sensitivity=0.5,
                 asymmetry_bias=0.1,
                 crisis_threshold=0.6,
                 safe_low=30.0,
                 safe_high=160.0):
        self.base_shorten = base_shorten
        self.tert_max = tert_max
        self.tert_sensitivity = tert_sensitivity
        self.trimming_rate = trimming_rate
        self.ox_sensitivity = ox_sensitivity
        self.asymmetry_bias = asymmetry_bias
        self.crisis_threshold = crisis_threshold
        self.safe_low = safe_low
        self.safe_high = safe_high

    def step(self, tel_len, damage, mutation_burden, demand_signal, oxidative_stress,
             apply_tert_fraction=0.0, allow_trimming=True):
        """
        Advance one time tick for a single cell's telomere state.

        Inputs:
          - tel_len: current telomere length (sim units)
          - damage: current damage score (0..1)
          - mutation_burden: current mutation burden (0..1)
          - demand_signal: local demand for proliferation/repair (0..1)
          - oxidative_stress: local oxidative stress (0..1)
          - apply_tert_fraction: fraction [0,1] of tert_max to apply this tick
          - allow_trimming: whether trimming is permitted

        Returns: new_tel_len, new_damage, crisis_risk
        """
        # 1) Telomere shortening due to replication-like events
        # model a small continuous shortening proportional to demand
        rep_events = demand_signal * (1.0 - self.asymmetry_bias)
        shorten = self.base_shorten * rep_events

        # 2) Oxidative stress increases effective shortening and damage
        ox_factor = 1.0 + self.ox_sensitivity * oxidative_stress
        shorten *= ox_factor
        damage += oxidative_stress * 0.01  # slow accrual from ox

        # 3) Apply telomerase pulse (fraction of tert_max)
        tert_amount = apply_tert_fraction * self.tert_max * self.tert_sensitivity
        tel_len = tel_len - shorten + tert_amount

        # 4) Trimming to avoid runaway length
        if allow_trimming and tel_len > self.safe_high:
            excess = tel_len - self.safe_high
            trim = min(excess, self.trimming_rate * excess)
            tel_len -= trim
            # trimming may produce transient damage
            damage += 0.001 * trim

        # 5) Small stochastic noise
        tel_len += np.random.normal(0.0, max(0.01, 0.001 * tel_len))
        tel_len = max(0.0, tel_len)

        # 6) Mutation/damage-driven crisis risk
        # risk increases with mutation_burden and damage and if telomeres are very long (instability)
        risk = 0.0
        risk += 0.6 * min(1.0, mutation_burden / max(1e-6, self.crisis_threshold))
        risk += 0.3 * damage
        if tel_len > self.safe_high * 1.3:
            risk += 0.2
        risk = min(1.0, risk)

        # 7) Damage decays slowly due to repair (we model some repair capacity)
        repair_capacity = max(0.0, 0.02 * (1.0 - damage))
        damage = max(0.0, damage - repair_capacity)

        return tel_len, damage, risk

    def simulate_trajectory(self, init_tel, init_damage, init_mutation, demand_series,
                            ox_series, tert_policy_fn, steps=200):
        """
        Simulate a trajectory for given series of demand and oxidative stress and a
        telomerase policy function tert_policy_fn(state)->fraction.

        Returns list of (tel_len, damage, mutation, risk) per step.
        """
        tel = init_tel
        dmg = init_damage
        mut = init_mutation
        traj = []
        for t in range(steps):
            demand = demand_series[t] if t < len(demand_series) else 0.0
            ox = ox_series[t] if t < len(ox_series) else 0.0
            # policy decides fraction to apply
            frac = tert_policy_fn({'tel': tel, 'damage': dmg, 'mutation': mut, 't': t, 'demand': demand, 'ox': ox})
            tel, dmg, risk = self.step(tel, dmg, mut, demand, ox, apply_tert_fraction=frac)
            # mutation accrual proportional to damage and oxidative stress
            mut += 0.0005 * (dmg + ox)
            mut = min(1.0, mut)
            traj.append((tel, dmg, mut, risk))
        return traj


# -------------------- Example: Genome-driven tert policy --------------------
# We define a small helper that converts a genome vector into a tert policy function.
# Genome layout (example): [target_tel, tert_sensitivity, trim_bias, ox_resilience, asymmetry, crisis_tolerance]

def genome_to_tert_policy(genome, kinetics: TelomereKinetics):
    target = float(genome[0])
    sens = float(genome[1]) * 2.0  # scale
    trim_bias = float(genome[2])

    def policy(state):
        tel = state['tel']
        dmg = state['damage']
        demand = state['demand']
        ox = state['ox']
        # base need proportional to how far below target and current demand
        need = max(0.0, (target - tel) / max(1.0, target)) * (0.4 + 0.6 * demand)
        # reduce need if damage or mutation high (safety)
        safety = max(0.0, 1.0 - 0.8 * (dmg + state['mutation']))
        frac = need * safety * sens
        # apply trimming bias to reduce pulses when trim_bias high
        frac = frac * (1.0 - 0.5 * trim_bias)
        # clip
        frac = max(0.0, min(1.0, frac))
        return frac

    return policy


# -------------------- Integration utility: evaluate telomere performance  ------

def evaluate_telomere_performance(genome, env, steps=200):
    """Run telomere kinetics for a representative subset of cells in env using genome.
    Returns a composite telomere performance score (higher is better):
      score = repair_success - risk_penalty - energy_penalty
    This is a simulation-only metric intended to be combined with other objectives.
    """
    # create kinetics using genome-inspired parameters
    kinetics = TelomereKinetics(
        base_shorten=1.0,
        tert_max=5.0 * (0.5 + genome[1] * 0.5),
        tert_sensitivity=1.0 + genome[1] * 1.5,
        trimming_rate=0.05 + genome[2] * 0.2,
        ox_sensitivity=0.2 + (1.0 - genome[3]) * 0.8,
        asymmetry_bias=0.1 + genome[4] * 0.4,
        crisis_threshold=0.4 + genome[5] * 0.6,
        safe_low=30.0,
        safe_high=160.0
    )

    # pick representative cells (if env has many, sample up to N)
    cells = list(env.cells.values()) if hasattr(env, 'cells') else []
    if len(cells) == 0:
        # no env provided: simulate a default demand/ox profile
        demand = [0.2] * steps
        ox = [0.05] * steps
        init_tel = 100.0
        init_dmg = 0.02
        init_mut = 0.01
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(init_tel, init_dmg, init_mut, demand, ox, policy, steps)
        # compute metrics
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        score = (avg_tel / 100.0) - (avg_risk * 2.0) - (energy * 0.01)
        return float(score)

    # sample up to 20 cells
    sample = random.sample(cells, min(20, len(cells)))
    scores = []
    for c in sample:
        # generate local demand/ox series from env fields if present, else defaults
        demand = []
        ox = []
        for t in range(steps):
            if hasattr(env, 'nutrient'):
                # scaled demand from nutrient at cell location with small temporal noise
                base = float(env.nutrient[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'nutrient') else 0.2
                demand.append(min(1.0, max(0.0, base + np.random.normal(0, 0.02))))
            else:
                demand.append(0.2)
            if hasattr(env, 'toxin'):
                base_ox = float(env.toxin[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'toxin') else 0.02
                ox.append(min(1.0, max(0.0, base_ox + np.random.normal(0, 0.01))))
            else:
                ox.append(0.02)
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(c.telomere if hasattr(c,'telomere') else 100.0,
                                           getattr(c,'damage',0.01),
                                           getattr(c,'mutation_burden',0.01),
                                           demand, ox, policy, steps)
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        scores.append((avg_tel, avg_risk, energy))

    # aggregate score: favor high avg_tel, penalize high risk and energy
    avg_tel_mean = np.mean([s[0] for s in scores])
    avg_risk_mean = np.mean([s[1] for s in scores])
    avg_energy = np.mean([s[2] for s in scores])
    score = (avg_tel_mean / 100.0) - (avg_risk_mean * 1.8) - (avg_energy * 0.01)
    return float(score)


# -------------------- Hook: integrate telomere performance into evaluator ----
# Replace or augment evaluate_genome in evolve_supercells_integrated.py to call
# evaluate_telomere_performance(genome, env) and combine with other objectives.


# End of appended telomere model
```

```python
import numpy as np
from typing import Callable, Dict

# ---------------------------------------------------------------------------
# Placeholder imports — in your environment these map to canvas modules
# ---------------------------------------------------------------------------
# Removed: from advanced_optimizers import (
# Removed:     CMAESOptimizer,
# Removed:     NSGA2,
# Removed:     MAPElites,
# Removed: )
# Removed: from quantum_inspired_optimizer import QuantumInspiredNES

# Assuming CMAES, POEM, SANES (from K2hpSKqGlvn-) and VariationalOptimizer (from gNbrnGk2luKD)
# are available from previously executed cells.

# ---------------------------------------------------------------------------
# Objective wrapper — your existing simulation
# ---------------------------------------------------------------------------
def evaluate_genome(genome: np.ndarray) -> float:
    """Stub: your simulation environment plugs in here.
    Returns a scalar fitness. Replace with actual simulation call.
    """
    repair_eff = genome[1]
    longevity = genome[0]
    energy = genome[4]
    return float((repair_eff * 1.4 + longevity * 1.2 + energy * 0.6))

# ---------------------------------------------------------------------------
# Unified optimizer interface
# ---------------------------------------------------------------------------
class EvolutionEngine:
    def __init__(self, dim: int = 6, bounds=(0.0, 1.0)):
        self.dim = dim
        self.bounds = bounds

    def _clip(self, x):
        lo, hi = self.bounds
        return np.clip(x, lo, hi)

    # -------------------- GA baseline --------------------
    def run_ga(self, n_gen=50, pop=40, elite=5):
        p = np.random.uniform(*self.bounds, size=(pop, self.dim))
        for _ in range(n_gen):
            fit = np.array([evaluate_genome(g) for g in p])
            elites = p[np.argsort(fit)[-elite:]]
            children = []
            for _ in range(pop - elite):
                a, b = elites[np.random.randint(0, elite, 2)]
                c = (a + b) / 2 + np.random.normal(0, 0.05, self.dim)
                children.append(self._clip(c))
            p = np.vstack([elites] + children)
        return p[np.argmax([evaluate_genome(g) for g in p])]

    # -------------------- CMA-ES --------------------
    def run_cma(self, iters=50):
        # Changed CMAESOptimizer to CMAES and added required pop_size for CMAES constructor
        opt = CMAES(dim=self.dim, pop_size=64)
        g, f = opt.run(objective_fn=evaluate_genome, n_iters=iters, bounds=self.bounds)
        return g

    # -------------------- NSGA-II --------------------
    def run_nsga2(self, iters=40, pop=60):
        # NSGA2 class is not defined in the current notebook. Commenting out.
        # def multi_obj(g):
        #     return [evaluate_genome(g), -np.sum((g - 0.5)**2)]
        # opt = NSGA2(dim=self.dim, pop=pop, bounds=self.bounds)
        # return opt.run(multi_obj, n_gen=iters)
        print("NSGA-II optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- MAP-Elites --------------------
    def run_map_elites(self, iters=200):
        # MAPElites class is not defined in the current notebook. Commenting out.
        # opt = MAPElites(dim=self.dim, bounds=self.bounds)
        # archive = opt.run(evaluate_genome, iters)
        # return archive
        print("MAP-Elites optimizer is not implemented in the current environment.")
        return np.full(self.dim, 0.5), -np.inf # Return a dummy result


    # -------------------- Quantum-inspired NES --------------------
    def run_qnes(self, iters=80):
        # Changed QuantumInspiredNES to VariationalOptimizer
        opt = VariationalOptimizer(dim=self.dim)
        best_genome, best_score = opt.optimize(evaluate_genome, n_iters=iters, bounds=self.bounds)
        return best_genome


if __name__ == "__main__":
    # Dummy Env for testing
    class DummyEnv:
        def __init__(self):
            self.cells = {}
            self.logger = None

    # For a standalone test, `dim` needs to be defined if not from a global context
    GENOME_SIZE = 6 # Define GENOME_SIZE for the example.
    engine = EvolutionEngine(dim=GENOME_SIZE)
    print("Running GA...")
    best_ga = engine.run_ga()
    print("Running CMA-ES...")
    best_cma = engine.run_cma()
    print("Running QNES...")
    best_q = engine.run_qnes()
    print("GA best:", best_ga)
    print("CMA-ES best:", best_cma)
    print("QNES best:", best_q)

# --- New Optimizers: Lion, Adam-Evolutionary, Hybrid Annealing ---

import torch

class LionOptimizer:
    """
    Lion-inspired evolutionary optimizer (not biological LION) — gradient-free.
    """
    def __init__(self, lr=0.1, beta=0.95):
        self.lr = lr
        self.beta = beta
        self.m = None

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
        self.m = self.beta * self.m + (1 - self.beta) * grad
        return genome - self.lr * torch.sign(self.m)


class AdamEvolutionary:
    """Adam-like optimizer but operating over sampled gradients from eval noise."""
    def __init__(self, lr=0.05):
        self.lr = lr
        self.m = None
        self.v = None
        self.t = 0

    def step(self, genome, grad):
        if self.m is None:
            self.m = torch.zeros_like(grad)
            self.v = torch.zeros_like(grad)
        self.t += 1
        self.m = 0.9 * self.m + 0.1 * grad
        self.v = 0.999 * self.v + 0.001 * grad**2

        m_hat = self.m / (1 - 0.9**self.t)
        v_hat = self.v / (1 - 0.999**self.t)

        return genome - self.lr * m_hat / (torch.sqrt(v_hat) + 1e-8)


class HybridAnnealing:
    def __init__(self, temp_start=1.0, temp_end=0.01, decay=0.99):
        self.temp = temp_start
        self.temp_end = temp_end
        self.decay = decay

    def step(self, genome, neighbor, score, neighbor_score):
        if neighbor_score > score:
            return neighbor
        import math, random
        prob = math.exp(-(score - neighbor_score) / max(self.temp, 1e-8))
        if random.random() < prob:
            return neighbor
        self.temp = max(self.temp * self.decay, self.temp_end)
        return genome


# --- Integration with Telomere / OoC / Stem Cell Models ---
# These hooks expect env objects from your existing simulation.

def integrate_with_telomeres(env, genome):
    env.cell.telomere_state.target_length = genome[0].item()
    env.cell.telomerase.max_activity = genome[1].item()


def integrate_with_ooc(env, genome):
    env.ooc.flow_rate = genome[2].item()
    env.ooc.ecm_response = genome[3].item()


def integrate_stem_cell_logic(env, genome):
    env.stem.division_threshold = genome[4].item()
    env.stem.quiescence_bias = genome[5].item()


# Unified registration
def apply_all_biological_mappings(env, genome):
    integrate_with_telomeres(env, genome)
    integrate_with_ooc(env, genome)
    integrate_stem_cell_logic(env, genome)


# ----------------------- Telomere Mathematical Model -----------------------
# Purely in-silico, abstract kinetic model for telomere homeostasis and risk.
# This module defines a TelomereKinetics class representing kinetics of
# telomere shortening, telomerase (TERT) action, trimming, oxidative stress
# influence, replication asymmetry and crisis probability. It provides
# convenience functions to simulate dynamics for a cell or a small population
# given a candidate genome (vector of control parameters).

import math

class TelomereKinetics:
    """
    Telomere kinetics simulator (abstract units).

    Parameters (all unitless/simulation-scaled):
      - base_shorten: baseline shortening per division (sim units)
      - tert_max: maximum telomerase activity (units per pulse)
      - tert_sensitivity: how responsive telomerase is to stress/need
      - trimming_rate: rate at which trimming reduces excess telomere
      - ox_sensitivity: how oxidative stress increases damage per tick
      - asymmetry_bias: fraction of divisions that are asymmetric (0..1)
      - crisis_threshold: mutation/damage threshold where crisis risk rises
    """

    def __init__(self,
                 base_shorten=1.0,
                 tert_max=5.0,
                 tert_sensitivity=1.0,
                 trimming_rate=0.1,
                 ox_sensitivity=0.5,
                 asymmetry_bias=0.1,
                 crisis_threshold=0.6,
                 safe_low=30.0,
                 safe_high=160.0):
        self.base_shorten = base_shorten
        self.tert_max = tert_max
        self.tert_sensitivity = tert_sensitivity
        self.trimming_rate = trimming_rate
        self.ox_sensitivity = ox_sensitivity
        self.asymmetry_bias = asymmetry_bias
        self.crisis_threshold = crisis_threshold
        self.safe_low = safe_low
        self.safe_high = safe_high

    def step(self, tel_len, damage, mutation_burden, demand_signal, oxidative_stress,
             apply_tert_fraction=0.0, allow_trimming=True):
        """
        Advance one time tick for a single cell's telomere state.

        Inputs:
          - tel_len: current telomere length (sim units)
          - damage: current damage score (0..1)
          - mutation_burden: current mutation burden (0..1)
          - demand_signal: local demand for proliferation/repair (0..1)
          - oxidative_stress: local oxidative stress (0..1)
          - apply_tert_fraction: fraction [0,1] of tert_max to apply this tick
          - allow_trimming: whether trimming is permitted

        Returns: new_tel_len, new_damage, crisis_risk
        """
        # 1) Telomere shortening due to replication-like events
        # model a small continuous shortening proportional to demand
        rep_events = demand_signal * (1.0 - self.asymmetry_bias)
        shorten = self.base_shorten * rep_events

        # 2) Oxidative stress increases effective shortening and damage
        ox_factor = 1.0 + self.ox_sensitivity * oxidative_stress
        shorten *= ox_factor
        damage += oxidative_stress * 0.01  # slow accrual from ox

        # 3) Apply telomerase pulse (fraction of tert_max)
        tert_amount = apply_tert_fraction * self.tert_max * self.tert_sensitivity
        tel_len = tel_len - shorten + tert_amount

        # 4) Trimming to avoid runaway length
        if allow_trimming and tel_len > self.safe_high:
            excess = tel_len - self.safe_high
            trim = min(excess, self.trimming_rate * excess)
            tel_len -= trim
            # trimming may produce transient damage
            damage += 0.001 * trim

        # 5) Small stochastic noise
        tel_len += np.random.normal(0.0, max(0.01, 0.001 * tel_len))
        tel_len = max(0.0, tel_len)

        # 6) Mutation/damage-driven crisis risk
        # risk increases with mutation_burden and damage and if telomeres are very long (instability)
        risk = 0.0
        risk += 0.6 * min(1.0, mutation_burden / max(1e-6, self.crisis_threshold))
        risk += 0.3 * damage
        if tel_len > self.safe_high * 1.3:
            risk += 0.2
        risk = min(1.0, risk)

        # 7) Damage decays slowly due to repair (we model some repair capacity)
        repair_capacity = max(0.0, 0.02 * (1.0 - damage))
        damage = max(0.0, damage - repair_capacity)

        return tel_len, damage, risk

    def simulate_trajectory(self, init_tel, init_damage, init_mutation, demand_series,
                            ox_series, tert_policy_fn, steps=200):
        """
        Simulate a trajectory for given series of demand and oxidative stress and a
        telomerase policy function tert_policy_fn(state)->fraction.

        Returns list of (tel_len, damage, mutation, risk) per step.
        """
        tel = init_tel
        dmg = init_damage
        mut = init_mutation
        traj = []
        for t in range(steps):
            demand = demand_series[t] if t < len(demand_series) else 0.0
            ox = ox_series[t] if t < len(ox_series) else 0.0
            # policy decides fraction to apply
            frac = tert_policy_fn({'tel': tel, 'damage': dmg, 'mutation': mut, 't': t, 'demand': demand, 'ox': ox})
            tel, dmg, risk = self.step(tel, dmg, mut, demand, ox, apply_tert_fraction=frac)
            # mutation accrual proportional to damage and oxidative stress
            mut += 0.0005 * (dmg + ox)
            mut = min(1.0, mut)
            traj.append((tel, dmg, mut, risk))
        return traj


# -------------------- Example: Genome-driven tert policy --------------------
# We define a small helper that converts a genome vector into a tert policy function.
# Genome layout (example): [target_tel, tert_sensitivity, trim_bias, ox_resilience, asymmetry, crisis_tolerance]

def genome_to_tert_policy(genome, kinetics: TelomereKinetics):
    target = float(genome[0])
    sens = float(genome[1]) * 2.0  # scale
    trim_bias = float(genome[2])

    def policy(state):
        tel = state['tel']
        dmg = state['damage']
        demand = state['demand']
        ox = state['ox']
        # base need proportional to how far below target and current demand
        need = max(0.0, (target - tel) / max(1.0, target)) * (0.4 + 0.6 * demand)
        # reduce need if damage or mutation high (safety)
        safety = max(0.0, 1.0 - 0.8 * (dmg + state['mutation']))
        frac = need * safety * sens
        # apply trimming bias to reduce pulses when trim_bias high
        frac = frac * (1.0 - 0.5 * trim_bias)
        # clip
        frac = max(0.0, min(1.0, frac))
        return frac

    return policy


# -------------------- Integration utility: evaluate telomere performance  ------

def evaluate_telomere_performance(genome, env, steps=200):
    """Run telomere kinetics for a representative subset of cells in env using genome.
    Returns a composite telomere performance score (higher is better):
      score = repair_success - risk_penalty - energy_penalty
    This is a simulation-only metric intended to be combined with other objectives.
    """
    # create kinetics using genome-inspired parameters
    kinetics = TelomereKinetics(
        base_shorten=1.0,
        tert_max=5.0 * (0.5 + genome[1] * 0.5),
        tert_sensitivity=1.0 + genome[1] * 1.5,
        trimming_rate=0.05 + genome[2] * 0.2,
        ox_sensitivity=0.2 + (1.0 - genome[3]) * 0.8,
        asymmetry_bias=0.1 + genome[4] * 0.4,
        crisis_threshold=0.4 + genome[5] * 0.6,
        safe_low=30.0,
        safe_high=160.0
    )

    # pick representative cells (if env has many, sample up to N)
    cells = list(env.cells.values()) if hasattr(env, 'cells') else []
    if len(cells) == 0:
        # no env provided: simulate a default demand/ox profile
        demand = [0.2] * steps
        ox = [0.05] * steps
        init_tel = 100.0
        init_dmg = 0.02
        init_mut = 0.01
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(init_tel, init_dmg, init_mut, demand, ox, policy, steps)
        # compute metrics
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        score = (avg_tel / 100.0) - (avg_risk * 2.0) - (energy * 0.01)
        return float(score)

    # sample up to 20 cells
    sample = random.sample(cells, min(20, len(cells)))
    scores = []
    for c in sample:
        # generate local demand/ox series from env fields if present, else defaults
        demand = []
        ox = []
        for t in range(steps):
            if hasattr(env, 'nutrient'):
                # scaled demand from nutrient at cell location with small temporal noise
                base = float(env.nutrient[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'nutrient') else 0.2
                demand.append(min(1.0, max(0.0, base + np.random.normal(0, 0.02))))
            else:
                demand.append(0.2)
            if hasattr(env, 'toxin'):
                base_ox = float(env.toxin[c.x % env.grid_size, c.y % env.grid_size]) if hasattr(env, 'toxin') else 0.02
                ox.append(min(1.0, max(0.0, base_ox + np.random.normal(0, 0.01))))
            else:
                ox.append(0.02)
        policy = genome_to_tert_policy(genome, kinetics)
        traj = kinetics.simulate_trajectory(c.telomere if hasattr(c,'telomere') else 100.0,
                                           getattr(c,'damage',0.01),
                                           getattr(c,'mutation_burden',0.01),
                                           demand, ox, policy, steps)
        avg_tel = np.mean([s[0] for s in traj])
        avg_risk = np.mean([s[3] for s in traj])
        energy = np.sum([policy({'tel':s[0],'damage':s[1],'mutation':s[2],'demand':d,'ox':o,'t':i}) for i,(s,d,o) in enumerate(zip(traj,demand,ox))])
        scores.append((avg_tel, avg_risk, energy))

    # aggregate score: favor high avg_tel, penalize high risk and energy
    avg_tel_mean = np.mean([s[0] for s in scores])
    avg_risk_mean = np.mean([s[1] for s in scores])
    avg_energy = np.mean([s[2] for s in scores])
    score = (avg_tel_mean / 100.0) - (avg_risk_mean * 1.8) - (avg_energy * 0.01)
    return float(score)


# -------------------- Hook: integrate telomere performance into evaluator ----
# Replace or augment evaluate_genome in evolve_supercells_integrated.py to call
# evaluate_telomere_performance(genome, env) and combine with other objectives.


# End of appended telomere model


############################################################
# Telomere Trajectory Visualization Module (Added - Option C)
############################################################

import matplotlib.pyplot as plt
import numpy as np

class TelomereVisualizer:
    def __init__(self):
        pass

    def plot_single_trajectory(self, trajectory, title="Telomere Dynamics Trajectory"):
        steps = np.arange(len(trajectory))
        tel = [t.get("telomere", 0) for t in trajectory]
        dmg = [t.get("damage", 0) for t in trajectory]
        mut = [t.get("mutation", 0) for t in trajectory]
        risk = [t.get("risk", 0) for t in trajectory]

        fig, axs = plt.subplots(4, 1, figsize=(10, 12), sharex=True)

        axs[0].plot(steps, tel)
        axs[0].set_ylabel("Telomere Length")
        axs[0].set_title(title)

        axs[1].plot(steps, dmg)
        axs[1].set_ylabel("Damage Level")

        axs[2].plot(steps, mut)
        axs[2].set_ylabel("Mutation Load")

        axs[3].plot(steps, risk)
        axs[3].set_ylabel("Crisis Risk")
        axs[3].set_xlabel("Steps")

        plt.tight_layout()
        plt.show()

    def plot_multiple_trajectories(self, trajectories, labels=None, metric="telomere"):
        fig, ax = plt.subplots(figsize=(10, 6))
        for i, traj in enumerate(trajectories):
            values = [t.get(metric, 0) for t in traj]
            steps = np.arange(len(values))
            label = labels[i] if labels else f"Run {i}"
            ax.plot(steps, values, label=label)

        ax.set_title(f"Comparison of {metric.capitalize()} Trajectories")
        ax.set_xlabel("Steps")
        ax.set_ylabel(metric.capitalize())
        ax.legend()
        plt.tight_layout()
        plt.show()
```

```python
import numpy as np
import random
from typing import Callable, List, Tuple, Dict

# Removed: try:
# Removed:    from advanced_optimizers import CMAES, SANES, POEM
# Removed: except Exception:
# Removed:    CMAES = None
# Removed:    SANES = None
# Removed:    POEM = None

# Assuming CMAES, POEM, SANES (from K2hpSKqGlvn-) and VariationalOptimizer (from gNbrnGk2luKD)
# are available from previously executed cells.

# Fallback simple NSGA-II-lite (pareto sorting)

def pareto_front(pop_metrics: List[Tuple[float,float,float]]):
    """Return indices of Pareto front (maximize all objectives)."""
    n = len(pop_metrics)
    is_pareto = [True]*n
    for i in range(n):
        for j in range(n):
            if i==j: continue
            a = pop_metrics[i]
            b = pop_metrics[j]
            # b dominates a if all b>=a and one strictly greater
            ge = all(bk >= ak for bk,ak in zip(b,a))
            gt = any(bk > ak for bk,ak in zip(b,a))
            if ge and gt:
                is_pareto[i] = False
                break
    return [i for i,ok in enumerate(is_pareto) if ok]

# -------------------- EvoController --------------------
class EvoController:
    def __init__(self, env, genome_dim:int=6, optimizer_name:str='cmaes'):
        self.env = env
        self.genome_dim = genome_dim
        self.optimizer_name = optimizer_name
        # mutation settings: global and per-locus
        self.global_mut_rate = 0.02
        self.locus_mut_rates = np.full(genome_dim, 0.02)
        # selection weights default (for single scalar composite)
        self.weights = {'telomere':0.4, 'stability':0.3, 'fitness':0.3}
        # mode
        self.mode = 'neutral'
        # population storage
        self.population = [self.random_genome() for _ in range(40)]
        self.pop_scores = []

    def random_genome(self):
        return np.clip(np.random.rand(self.genome_dim), 0.0, 1.0)

    def set_mutation_rate(self, global_rate:float=None, locus_rates:List[float]=None):
        if global_rate is not None:
            self.global_mut_rate = float(global_rate)
            self.locus_mut_rates = np.full(self.genome_dim, self.global_mut_rate)
        if locus_rates is not None:
            arr = np.array(locus_rates, dtype=float)
            if arr.shape[0] != self.genome_dim:
                raise ValueError('locus_rates length mismatch')
            self.locus_mut_rates = np.clip(arr, 0.0, 1.0)

    def set_mode(self, mode:str):
        mode = mode.lower()
        if mode not in ('neutral','survival','aggressive_repair'):
            raise ValueError('Unknown mode')
        self.mode = mode
        # tune weights and mutation rates by mode
        if mode == 'neutral':
            self.weights = {'telomere':0.4, 'stability':0.3, 'fitness':0.3}
            self.set_mutation_rate(global_rate=0.02)
        elif mode == 'survival':
            # prioritize stability, be conservative with mutation
            self.weights = {'telomere':0.25, 'stability':0.5, 'fitness':0.25}
            # reduce mutation globally
            self.set_mutation_rate(global_rate=0.005)
        elif mode == 'aggressive_repair':
            # prioritize telomere repair and fitness, allow higher mutation for exploration
            self.weights = {'telomere':0.5, 'stability':0.15, 'fitness':0.35}
            self.set_mutation_rate(global_rate=0.08)

    def mutate_genome(self, genome:np.ndarray) -> np.ndarray:
        child = genome.copy()
        for i in range(self.genome_dim):
            if random.random() < self.locus_mut_rates[i]:
                # Gaussian perturbation scaled by locus rate
                sd = 0.1 * self.locus_mut_rates[i]
                child[i] += random.gauss(0, sd)
        return np.clip(child, 0.0, 1.0)

    # ---------------- Objective evaluations ----------------
    def telomere_objective(self, genome:np.ndarray) -> float:
        # use evaluate_telomere_performance if available in environment
        try:
            # Assumes evaluate_telomere_performance is globally available from previous cells
            return max(-1.0, evaluate_telomere_performance(genome, self.env, steps=120))
        except NameError:
            # fallback synthetic: prefer higher genome[0]
            return float(genome[0])

    def stability_objective(self, genome:np.ndarray) -> float:
        # stability = 1 - expected mutation drift; approximate via genome variance
        return 1.0 - float(np.std(genome))

    def fitness_objective(self, genome:np.ndarray) -> float:
        # call existing evaluate_genome or fallback
        try:
            # Assumes evaluate_genome is globally available from previous cells
            return float(evaluate_genome(genome))
        except NameError:
            return float(0.5*np.sum(genome)/self.genome_dim)

    def evaluate_population(self, population:List[np.ndarray]) -> List[Dict]:
        results = []
        for g in population:
            t = self.telomere_objective(g)
            s = self.stability_objective(g)
            f = self.fitness_objective(g)
            # composite
            comp = self.weights['telomere']*t + self.weights['stability']*s + self.weights['fitness']*f
            results.append({'genome':g, 'telomere':t, 'stability':s, 'fitness':f, 'composite':comp})
        return results

    # ---------------- Selection strategies ----------------
    def select_elites(self, evaluated:List[Dict], k:int=8) -> List[np.ndarray]:
        sorted_pop = sorted(evaluated, key=lambda x:x['composite'], reverse=True)
        return [x['genome'] for x in sorted_pop[:k]]

    def select_pareto(self, evaluated:List[Dict]) -> List[np.ndarray]:
        metrics = [(x['telomere'], x['stability'], x['fitness']) for x in evaluated]
        idxs = pareto_front(metrics)
        return [evaluated[i]['genome'] for i in idxs]

    # ---------------- Evolution cycle ----------------
    def run_generation_cycle(self, n_generations:int=50, pop_size:int=40, strategy:str='elite') -> Tuple[np.ndarray, float]:
        # initialize population if empty
        pop = self.population if self.population else [self.random_genome() for _ in range(pop_size)]
        best_genome = None
        best_score = -np.inf
        for gen in range(n_generations):
            evaluated = self.evaluate_population(pop)
            # logging hook (if logger exists on env)
            if hasattr(self.env, 'logger') and self.env.logger:
                for i,entry in enumerate(evaluated):
                    telstate={'length':None,'damage':None,'risk':None,'mutation':None}
                    try:
                        # if possible, run single-cell telomere sim for logging
                        tel_score = self.telomere_objective(entry['genome'])
                        telstate['length'] = tel_score
                    except Exception:
                        pass
                    self.env.logger.log_step(gen*100+i, entry['genome'].tolist(), entry['composite'], telstate, {'strategy':strategy})

            # selection
            if strategy=='elite':
                elites = self.select_elites(evaluated, k=max(4, pop_size//5))
            elif strategy=='pareto':
                elites = self.select_pareto(evaluated)
                if len(elites) < max(4, pop_size//5):
                    elites = elites + [e['genome'] for e in sorted(evaluated, key=lambda x:x['composite'], reverse=True)[:max(0, pop_size//5-len(elites))]]
            else:
                elites = self.select_elites(evaluated, k=max(4, pop_size//5))

            # reproduce
            new_pop = elites.copy()
            while len(new_pop) < pop_size:
                a,b = random.sample(elites, 2)
                child = (a + b)/2.0
                child = self.mutate_genome(child)
                new_pop.append(child)

            pop = new_pop
            # track best
            for e in evaluated:
                if e['composite'] > best_score:
                    best_score = e['composite']
                    best_genome = e['genome'].copy()
        self.population = pop
        return best_genome, best_score

# -------------------- Example run function --------------------

def sample_run():
    # create a minimal dummy env if none exists
    class DummyEnv:
        def __init__(self):
            self.cells = {}
            self.logger = None
    env = DummyEnv()
    ctrl = EvoController(env)
    print('Neutral run...')
    g,s = ctrl.run_generation_cycle(n_generations=10)
    print('Best composite (neutral):', s)
    ctrl.set_mode('survival')
    print('Survival run...')
    g2,s2 = ctrl.run_generation_cycle(n_generations=10)
    print('Best composite (survival):', s2)
    ctrl.set_mode('aggressive_repair')
    print('Aggressive repair run...')
    g3,s3 = ctrl.run_generation_cycle(n_generations=10)
    print('Best composite (aggressive):', s3)

if __name__=='__main__':
    sample_run()
```
