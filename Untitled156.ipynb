```python
class ContrastiveLearningModel(nn.Module):
    def __init__(self, state_dim):
        super(ContrastiveLearningModel, self).__init__()
        self.encoder = nn.Sequential(
            nn.Linear(state_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 64)
        )

    def forward(self, x):
        return self.encoder(x)

# Contrastive loss (for SSL)
def contrastive_loss(anchor, positive, negative, margin=1.0):
    pos_distance = F.pairwise_distance(anchor, positive, p=2)
    neg_distance = F.pairwise_distance(anchor, negative, p=2)
    loss = torch.clamp(pos_distance - neg_distance + margin, min=0.0)
    return loss.mean()

# Example usage:
ssl_model = ContrastiveLearningModel(state_dim=10)
anchor = torch.randn(32, 10)
positive = torch.randn(32, 10)
negative = torch.randn(32, 10)

loss = contrastive_loss(anchor, positive, negative)
```

```python
def total_loss(z_pred, z_true, energy):
    pred_loss = torch.nn.functional.mse_loss(z_pred, z_true)
    energy_loss = energy.mean()
    entropy_loss = -(z_pred**2).mean()
    consistency = torch.var(z_pred)

    return pred_loss + 0.01*energy_loss + 0.001*entropy_loss + 0.1*consistency
```

```python
import torch
import torch.nn as nn

class WorldModel(nn.Module):
    def __init__(self, dim, hidden=256):
        super().__init__()
        self.dynamics = nn.Sequential(
            nn.Linear(dim, hidden),
            nn.GELU(),
            nn.Linear(hidden, dim)
        )
        self.energy = nn.Sequential(
            nn.Linear(dim, hidden),
            nn.SiLU(),
            nn.Linear(hidden, 1)
        )

    def step(self, z):
        dz = self.dynamics(z)
        return z + 0.05 * dz + 0.01 * torch.randn_like(z)

    def rollout(self, z0, steps=100):
        traj = [z0]
        z = z0
        for _ in range(steps):
            z = self.step(z)
            traj.append(z)
        return torch.stack(traj)
```

```python
def evaluate_trajectory(traj):
    # Stability
    energy = (traj**2).sum(dim=-1)
    stability = -energy.var()

    # Diversity
    diversity = traj.std(dim=0).mean()

    # Predictability (smoothness)
    smoothness = -(traj[1:] - traj[:-1]).pow(2).mean()

    return stability + 0.5*diversity + 0.5*smoothness
```

```python
def curate(trajs, scores, percentile=70):
    threshold = torch.quantile(scores, percentile/100)
    selected = [t for t,s in zip(trajs, scores) if s >= threshold]
    return torch.cat(selected)
```

```python
def train_student(student, data, epochs=10):
    opt = torch.optim.AdamW(student.parameters(), lr=3e-4)
    for _ in range(epochs):
        z0 = data[0]
        pred = student.rollout(z0, steps=data.size(0)-1)
        loss = ((pred - data)**2).mean()
        opt.zero_grad()
        loss.backward()
        opt.step()
```

```python
def self_improving_loop(generations=5):
    dim = 8
    teacher = WorldModel(dim)

    for gen in range(generations):
        # Generate data
        trajs, scores = [], []
        for _ in range(50):
            z0 = torch.randn(dim)
            traj = teacher.rollout(z0)
            score = evaluate_trajectory(traj)
            trajs.append(traj)
            scores.append(score)

        scores = torch.stack(scores)
        dataset = curate(trajs, scores)

        # Train student
        student = WorldModel(dim)
        train_student(student, dataset)

        # Promote student
        teacher = student
        print(f"Generation {gen+1} complete")
```

```python
import torch
import math
from collections import Counter

def sequence_features(seq, k=3):
    L = len(seq)

    # GC content
    gc = (seq.count("G") + seq.count("C")) / L

    # k-mer entropy
    kmers = [seq[i:i+k] for i in range(L-k+1)]
    counts = Counter(kmers)
    probs = torch.tensor([v/len(kmers) for v in counts.values()])
    entropy = -torch.sum(probs * torch.log(probs + 1e-8))

    # Periodicity score
    periodicity = sum(seq[i] == seq[i % 6] for i in range(L)) / L

    return torch.tensor([gc, entropy, periodicity, L/100])
```

```python
import torch.nn as nn

class SequenceAwareWorldModel(nn.Module):
    def __init__(self, latent_dim, seq_dim):
        super().__init__()
        self.seq_encoder = nn.Linear(seq_dim, latent_dim)

        self.dynamics = nn.Sequential(
            nn.Linear(latent_dim * 2, 256),
            nn.GELU(),
            nn.Linear(256, latent_dim)
        )

        self.energy = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.SiLU(),
            nn.Linear(128, 1)
        )

    def step(self, z, seq_feat):
        s = self.seq_encoder(seq_feat)
        x = torch.cat([z, s], dim=-1)
        dz = self.dynamics(x)
        return z + 0.05 * dz
```

```python
def telomerase_score(seq_feat):
    gc, entropy, periodicity, length = seq_feat
    return (
        0.5 * periodicity +
        0.3 * gc -
        0.2 * entropy +
        0.4 * length
    )
```

```python
def degradation_noise(seq_feat):
    gc, entropy, periodicity, _ = seq_feat
    return 0.05 * (1 - gc) + 0.02 * entropy
```

```python
def simulate(z0, seq, model, steps=200):
    seq_feat = sequence_features(seq)
    z = z0.clone()
    traj = []

    for _ in range(steps):
        noise = degradation_noise(seq_feat)
        z = model.step(z, seq_feat) + noise * torch.randn_like(z)
        traj.append(z)

    return torch.stack(traj)
```

```python
ACTIVATIONS = {
    "relu": tf.keras.activations.relu,
    "gelu": tf.keras.activations.gelu,
    "silu": tf.keras.activations.silu,
    "swish": tf.keras.activations.swish,
    "mish": tf.keras.activations.mish,
    "tanh": tf.keras.activations.tanh,
    "selu": tf.keras.activations.selu,
    "softplus": tf.keras.activations.softplus,
    "leaky_relu": tf.keras.layers.LeakyReLU(alpha=0.1),
}
```

```python
def build_virtual_cell_model(
    latent_dim=16,
    seq_dim=4,
    hidden=512,
    activation="gelu"
):
    act = ACTIVATIONS[activation]

    z_in = layers.Input(shape=(latent_dim,), name="cell_state")
    seq_in = layers.Input(shape=(seq_dim,), name="sequence_features")

    x = layers.Concatenate()([z_in, seq_in])

    for _ in range(4):
        x = layers.Dense(hidden, activation=act)(x)
        x = layers.LayerNormalization()(x)

    next_state = layers.Dense(latent_dim, name="next_state")(x)
    energy = layers.Dense(1, activation="softplus", name="energy")(x)
    division = layers.Dense(1, activation="sigmoid", name="division_prob")(x)

    return models.Model(
        inputs=[z_in, seq_in],
        outputs=[next_state, energy, division]
    )
```

```python
class OptimizerSwitcher(tf.keras.callbacks.Callback):
    def __init__(self, schedule):
        self.schedule = schedule
        self.current = 0
        self.epoch_count = 0

    def on_epoch_begin(self, epoch, logs=None):
        name, duration = self.schedule[self.current]
        if self.epoch_count == 0:
            self.model.optimizer = OPTIMIZERS[name]
            print(f"\n🔄 Switched to optimizer: {name}")

        self.epoch_count += 1
        if self.epoch_count >= duration:
            self.current = min(self.current + 1, len(self.schedule) - 1)
            self.epoch_count = 0
```

```python
def exploit_explore(population, scores, mutation_rate=0.2):
    ranked = sorted(zip(population, scores), key=lambda x: x[1], reverse=True)
    top_half = ranked[:len(ranked)//2]
    bottom_half = ranked[len(ranked)//2:]

    for i, (model, genome) in enumerate(bottom_half):
        parent_model, parent_genome = random.choice(top_half)

        # Copy weights
        model.set_weights(parent_model.get_weights())

        # Mutate genome
        for k in genome:
            if random.random() < mutation_rate:
                genome[k] = random.choice(GENOME_SPACE[k])

    return population
```

```python
class LatentCore(tf.keras.Model):
    def __init__(self, dim=256, heads=8):
        super().__init__()
        self.attn = layers.MultiHeadAttention(
            num_heads=heads, key_dim=dim//heads
        )
        self.ffn = tf.keras.Sequential([
            layers.Dense(dim*4, activation="gelu"),
            layers.Dense(dim)
        ])
        self.norm1 = layers.LayerNormalization()
        self.norm2 = layers.LayerNormalization()

    def call(self, x):
        h = self.attn(x, x)
        x = self.norm1(x + h)
        h = self.ffn(x)
        return self.norm2(x + h)
```

```python
def build_encoder(input_dim, latent_dim):
    return tf.keras.Sequential([
        layers.Dense(latent_dim, activation="gelu"),
        layers.LayerNormalization(),
        layers.Dense(latent_dim)
    ])
```

```python
class GeneralWorldModel(tf.keras.Model):
    def __init__(self, latent_dim=256):
        super().__init__()
        self.encoder = build_encoder(latent_dim, latent_dim)
        self.core = LatentCore(latent_dim)
        self.heads = build_heads(latent_dim)

    def call(self, x):
        z = self.encoder(x)
        z = tf.expand_dims(z, axis=1)
        z = self.core(z)
        z = tf.squeeze(z, axis=1)

        return {k: head(z) for k, head in self.heads.items()}
```

```python
def causal_encoder(input_dim, latent_dim):
    inputs = keras.Input(shape=(input_dim,))
    x = layers.Dense(256, activation="gelu")(inputs)
    x = layers.Dense(128, activation="gelu")(x)

    # Latent causal variables
    z_mean = layers.Dense(latent_dim)(x)
    z_logvar = layers.Dense(latent_dim)(x)

    def sample_z(args):
        mean, logvar = args
        eps = tf.random.normal(shape=tf.shape(mean))
        return mean + tf.exp(0.5 * logvar) * eps

    z = layers.Lambda(sample_z)([z_mean, z_logvar])
    return keras.Model(inputs, [z, z_mean, z_logvar])
```

```python
def causal_dynamics(latent_dim, output_dim):
    z_input = keras.Input(shape=(latent_dim,))
    x = layers.Dense(128, activation="silu")(z_input)
    x = layers.Dense(128, activation="silu")(x)
    outputs = layers.Dense(output_dim)(x)
    return keras.Model(z_input, outputs)
```

```python
class CausalWorldModel(keras.Model):
    def __init__(self, input_dim, latent_dim, output_dim):
        super().__init__()
        self.encoder = causal_encoder(input_dim, latent_dim)
        self.dynamics = causal_dynamics(latent_dim, output_dim)

    def call(self, x):
        z, mean, logvar = self.encoder(x)
        next_state = self.dynamics(z)
        return next_state, mean, logvar
```

```python
def causal_loss(y_true, y_pred, mean, logvar):
    recon = tf.reduce_mean(tf.square(y_true - y_pred))
    kl = -0.5 * tf.reduce_mean(
        1 + logvar - tf.square(mean) - tf.exp(logvar)
    )
    return recon + 0.01 * kl
```

```python
class CausalTrainer:
    def __init__(self, model, optimizer):
        self.model = model
        self.optimizer = optimizer

    @tf.function
    def train_step(self, x, y):
        with tf.GradientTape() as tape:
            y_pred, mean, logvar = self.model(x)
            loss = causal_loss(y, y_pred, mean, logvar)

        grads = tape.gradient(loss, self.model.trainable_variables)
        self.optimizer.apply_gradients(
            zip(grads, self.model.trainable_variables)
        )
        return loss
```

```python
def intervene(model, x, causal_index, delta):
    z, mean, logvar = model.encoder(x)
    z_intervened = tf.identity(z)
    z_intervened = tf.tensor_scatter_nd_add(
        z_intervened,
        [[0, causal_index]],
        [delta]
    )
    return model.dynamics(z_intervened)
```

```python
class WorldModel(tf.keras.Model):
    def __init__(self, encoder, dynamics):
        super().__init__()
        self.encoder = encoder   # maps X → Z
        self.dynamics = dynamics # maps Z → X_next

    def forward(self, x):
        z, _, _ = self.encoder(x)
        return self.dynamics(z)
```

```python
def do_intervention(world_model, x, z_index, value):
    z, mean, logvar = world_model.encoder(x)

    # Replace causal variable z_i
    z_do = tf.identity(z)
    z_do = tf.tensor_scatter_nd_update(
        z_do,
        [[0, z_index]],
        [value]
    )

    return world_model.dynamics(z_do)
```

```python
def soft_intervention(world_model, x, z_index, delta, strength=0.5):
    z, _, _ = world_model.encoder(x)
    z_new = tf.identity(z)
    z_new = tf.tensor_scatter_nd_add(
        z_new,
        [[0, z_index]],
        [delta * strength]
    )
    return world_model.dynamics(z_new)
```

```python
def intervene_state(x, state_index, delta):
    x_new = tf.identity(x)
    x_new = tf.tensor_scatter_nd_add(
        x_new,
        [[0, state_index]],
        [delta]
    )
    return x_new
```

```python
def counterfactual(world_model, x, z_index, new_value):
    # Abduction
    z, _, _ = world_model.encoder(x)

    # Action
    z_cf = tf.tensor_scatter_nd_update(
        z, [[0, z_index]], [new_value]
    )

    # Prediction
    return world_model.dynamics(z_cf)
```

```python
def average_treatment_effect(world_model, x_batch, z_index, low, high):
    y_low = []
    y_high = []

    for x in x_batch:
        y_low.append(do_intervention(world_model, x[None], z_index, low))
        y_high.append(do_intervention(world_model, x[None], z_index, high))

    return tf.reduce_mean(y_high) - tf.reduce_mean(y_low)
```

```python
import numpy as np

STATE_DIM = 8  # telomere, telomerase, damage, stemness, stress, etc.

def step_world(state, action):
    telomere, ta, damage, stem, stress, debris, repair, _ = state

    # Telomerase gate (cancer-safe)
    ta = 1.0 if (stem > 0.8 and repair > 0.6) else 0.0

    # Telomere dynamics
    telomere += 0.15 * ta - 0.02 - 0.01 * damage
    telomere -= max(0.0, telomere - 1.2) * 0.1  # trimming

    # Phagocytosis
    debris = max(0.0, debris - 0.4 * repair)
    stress = max(0.0, stress - 0.3 * debris)

    damage = np.clip(damage + 0.05 * stress, 0, 1)

    next_state = np.array([
        telomere, ta, damage, stem, stress, debris, repair, action
    ])

    reward = telomere - damage - stress
    return next_state, reward
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_world_model():
    x = layers.Input(shape=(STATE_DIM + 1,))  # + action
    h = layers.Dense(512, activation="swish")(x)
    h = layers.Dense(512, activation="mish")(h)
    h = layers.Dense(256, activation="gelu")(h)

    next_state = layers.Dense(STATE_DIM)(h)
    reward = layers.Dense(1)(h)

    return models.Model(x, [next_state, reward])
```
