```python
import argparse
import json
import math
import os
import random
from dataclasses import dataclass
from typing import Tuple, Dict

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

try:
    import optuna
    OPTUNA_AVAILABLE = True
except Exception:
    OPTUNA_AVAILABLE = False

# -------------------------
# Safe synthetic dataset
# -------------------------

FEATURE_NAMES = [
    "telomere_len",        # normalized telomere length [0..1]
    "age_marker",          # synthetic age-related marker [0..1]
    "oxidative_stress",    # ROS proxy [0..1]
    "metabolic_rate",      # metabolic proxy [0..1]
    "dna_damage",          # DNA damage indicator [0..1]
    "autophagy_eff",       # autophagy efficiency [0..1]
    "division_prob",       # prior division probability [0..1]
]
CONTEXT_NAMES = [
    "nutrient_level",
    "shear_stress",
    "signal_factor",  # e.g., growth factor concentration
]

NDIMS = len(FEATURE_NAMES)
NCONTEXT = len(CONTEXT_NAMES)


class VirtualCellDataset(Dataset):
    """Generates purely synthetic virtual cell states.
    The generation rules are simple and deliberately non-biological: they produce
    correlated numerical vectors the model can learn to reconstruct and generate.
    """

    def __init__(self, n_samples: int = 10000, seed: int = 42):
        super().__init__()
        self.n_samples = n_samples
        rng = np.random.RandomState(seed)
        self.data = []
        for _ in range(n_samples):
            # context
            nutrient = rng.beta(5, 2)  # tends to be high
            shear = rng.beta(2, 5)     # tends to be low
            signal = rng.beta(3, 3)

            # base telomere length decreases with age_marker and oxidative stress
            age_marker = rng.beta(2, 5)
            oxidative = rng.beta(2, 3)
            dna_damage = np.minimum(1.0, oxidative * 0.6 + rng.normal(0, 0.05))

            telomere_len = np.clip(0.9 - 0.5 * age_marker - 0.4 * oxidative + rng.normal(0, 0.03), 0.0, 1.0)
            metabolic = np.clip(0.3 + 0.6 * nutrient - 0.2 * age_marker + rng.normal(0, 0.05), 0.0, 1.0)
            autophagy = np.clip(0.7 * nutrient - 0.3 * oxidative + rng.normal(0, 0.05), 0.0, 1.0)

            division_prob = np.clip(0.4 * signal + 0.3 * metabolic - 0.5 * dna_damage + rng.normal(0, 0.05), 0.0, 1.0)

            features = np.array([
                telomere_len,
                age_marker,
                oxidative,
                metabolic,
                dna_damage,
                autophagy,
                division_prob,
            ], dtype=np.float32)

            context = np.array([nutrient, shear, signal], dtype=np.float32)

            # desired telomerase activation decision (0/1) for supervised controller training
            # safe rule: activate if telomere_len < threshold AND dna_damage low
            thr = 0.45 + 0.15 * (1.0 - nutrient)  # nutrient-dependent threshold
            telomerase_on = float((telomere_len < thr) and (dna_damage < 0.6) and (division_prob > 0.15))

            self.data.append((features, context, telomerase_on))

    def __len__(self):
        return self.n_samples

    def __getitem__(self, idx):
        features, context, telomerase_on = self.data[idx]
        return {
            "x": torch.from_numpy(features),
            "c": torch.from_numpy(context),
            "act": torch.tensor(telomerase_on, dtype=torch.float32),
        }


# -------------------------
# Model: Conditional VAE
# -------------------------

class Encoder(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(x_dim + c_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
        )
        self.mu = nn.Linear(hidden_dim, z_dim)
        self.logvar = nn.Linear(hidden_dim, z_dim)

    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        h = self.net(h)
        return self.mu(h), self.logvar(h)


class Decoder(nn.Module):
    def __init__(self, z_dim: int, c_dim: int, hidden_dim: int, x_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(z_dim + c_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, x_dim),
            nn.Sigmoid(),  # all features normalized to [0,1]
        )

    def forward(self, z, c):
        h = torch.cat([z, c], dim=-1)
        return self.net(h)


class Controller(nn.Module):
    """A small MLP that predicts telomerase activation given reconstructed/real state and context."""

    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(x_dim + c_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(hidden_dim // 2, 1),
        )

    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        return torch.sigmoid(self.net(h)).squeeze(-1)


@dataclass
class CVAEModel(nn.Module):
    encoder: Encoder
    decoder: Decoder
    controller: Controller

    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.encoder = Encoder(x_dim, c_dim, hidden_dim, z_dim)
        self.decoder = Decoder(z_dim, c_dim, hidden_dim, x_dim)
        self.controller = Controller(x_dim, c_dim, hidden_dim)

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x, c):
        mu, logvar = self.encoder(x, c)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z, c)
        act_pred = self.controller(x_recon, c)
        return x_recon, mu, logvar, act_pred


# -------------------------
# Losses and utilities
# -------------------------

def loss_function(recon_x, x, mu, logvar, act_pred, act_true, beta=1.0):
    # Reconstruction: mean squared error
    recon_loss = F.mse_loss(recon_x, x, reduction="mean")
    # KL divergence
    kld = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
    # Controller BCE
    bce = F.binary_cross_entropy(act_pred, act_true)
    return recon_loss + beta * kld + 0.5 * bce, recon_loss, kld, bce


# -------------------------
# Training and sampling
# -------------------------

def train(model: CVAEModel, dataloader: DataLoader, optim: torch.optim.Optimizer, device: torch.device, epochs: int = 20, beta: float = 1.0, ckpt_dir: str = "checkpoints"):
    model.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    model.train()
    for epoch in range(1, epochs + 1):
        total_loss = 0.0
        total_recon = 0.0
        total_kld = 0.0
        total_bce = 0.0
        for batch in dataloader:
            x = batch["x"].to(device)
            c = batch["c"].to(device)
            act = batch["act"].to(device)
            optim.zero_grad()
            x_recon, mu, logvar, act_pred = model(x, c)
            loss, recon, kld, bce = loss_function(x_recon, x, mu, logvar, act_pred, act, beta)
            loss.backward()
            optim.step()
            total_loss += loss.item() * x.size(0)
            total_recon += recon.item() * x.size(0)
            total_kld += kld.item() * x.size(0)
            total_bce += bce.item() * x.size(0)

        n = len(dataloader.dataset)
        print(f"Epoch {epoch}/{epochs}  loss={total_loss/n:.6f} recon={total_recon/n:.6f} kld={total_kld/n:.6f} bce={total_bce/n:.6f}")
        # checkpoint
        torch.save(model.state_dict(), os.path.join(ckpt_dir, f"cvae_epoch{epoch}.pth"))


def sample(model: CVAEModel, device: torch.device, n: int = 5, context: np.ndarray = None):
    model.to(device)
    model.eval()
    with torch.no_grad():
        if context is None:
            # sample random contexts
            context = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
        else:
            if context.ndim == 1:
                context = np.tile(context[None, :], (n, 1)).astype(np.float32)
        c = torch.from_numpy(context).to(device)
        z = torch.randn(n, model.encoder.mu.out_features, device=device)
        x_gen = model.decoder(z, c)
        act_pred = model.controller(x_gen, c).cpu().numpy()
        return x_gen.cpu().numpy(), context, act_pred


# -------------------------
# Visualization helpers
# -------------------------

def plot_samples(xs: np.ndarray, ctx: np.ndarray, acts: np.ndarray, n_plot: int = 5):
    n = min(n_plot, xs.shape[0])
    fig, axes = plt.subplots(n, 1, figsize=(6, 2 * n))
    if n == 1:
        axes = [axes]
    for i in range(n):
        ax = axes[i]
        ax.bar(range(len(FEATURE_NAMES)), xs[i], tick_label=FEATURE_NAMES)
        ax.set_ylim(0, 1)
        ax.set_title(f"Context: {np.round(ctx[i],3)}  act={acts[i]:.3f}")
    plt.tight_layout()
    plt.show()


# -------------------------
# Optional Optuna sweep
# -------------------------

def objective_optuna(trial):
    # small search space for demo
    hidden_dim = trial.suggest_int("hidden_dim", 32, 256, step=32)
    z_dim = trial.suggest_int("z_dim", 4, 32, step=4)
    lr = trial.suggest_loguniform("lr", 1e-4, 1e-2)
    batch_size = trial.suggest_categorical("batch_size", [64, 128, 256])

    dataset = VirtualCellDataset(2000)
    dl = DataLoader(dataset, batch_size=batch_size, shuffle=True)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=hidden_dim, z_dim=z_dim)
    optim = torch.optim.Adam(model.parameters(), lr=lr)
    train(model, dl, optim, device, epochs=6, beta=1.0, ckpt_dir="optuna_ckpts")

    # evaluate train recon loss quickly
    model.eval()
    total = 0.0
    n = 0
    with torch.no_grad():
        for batch in dl:
            x = batch["x"].to(device)
            c = batch["c"].to(device)
            x_recon, mu, logvar, act_pred = model(x, c)
            recon = F.mse_loss(x_recon, x, reduction="sum").item()
            total += recon
            n += x.size(0)
    return total / n


# -------------------------
# CLI and main
# -------------------------

def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument("--train", action="store_true")
    p.add_argument("--sample", action="store_true")
    p.add_argument("--n-samples", type=int, default=5)
    p.add_argument("--epochs", type=int, default=20)
    p.add_argument("--batch-size", type=int, default=128)
    p.add_argument("--lr", type=float, default=1e-3)
    p.add_argument("--hidden-dim", type=int, default=128)
    p.add_argument("--z-dim", type=int, default=16)
    p.add_argument("--beta", type=float, default=1.0)
    p.add_argument("--ckpt", type=str, default="checkpoints/cvae_epoch20.pth")
    p.add_argument("--optuna", action="store_true")
    p.add_argument("--seed", type=int, default=42)
    # Fix: When running in Colab/Jupyter, the kernel passes its own arguments (like -f)
    # which argparse will not recognize. To avoid this, we explicitly pass an empty list
    # of arguments if we are not expecting command line input when running the cell directly,
    # relying on defaults.
    return p.parse_args(args=[])


def main():
    args = parse_args()
    torch.manual_seed(args.seed)
    np.random.seed(args.seed)
    random.seed(args.seed)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    if args.optuna:
        if not OPTUNA_AVAILABLE:
            print("Optuna not available. Install with `pip install optuna`.")
            return
        study = optuna.create_study(direction="minimize")
        study.optimize(objective_optuna, n_trials=12)
        print("Best trial:", study.best_trial.params)
        return

    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=args.hidden_dim, z_dim=args.z_dim)

    if args.train:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(model.parameters(), lr=args.lr)
        train(model, dl, optim, device, epochs=args.epochs, beta=args.beta)
        # final checkpoint
        os.makedirs("checkpoints", exist_ok=True)
        torch.save(model.state_dict(), "checkpoints/cvae_final.pth")
        print("Training finished. Model saved to checkpoints/cvae_final.pth")

    if args.sample:
        # load checkpoint if exists
        if os.path.exists(args.ckpt):
            model.load_state_dict(torch.load(args.ckpt, map_location=device))
            print(f"Loaded checkpoint {args.ckpt}")
        xs, ctx, acts = sample(model, device, n=args.n_samples)
        plot_samples(xs, ctx, acts, n_plot=min(8, args.n_samples))
        # save outputs to csv
        df_rows = []
        for i in range(xs.shape[0]):
            row = {**{k: float(xs[i, j]) for j, k in enumerate(FEATURE_NAMES)},
                   **{f"ctx_{j}": float(ctx[i, j]) for j in range(ctx.shape[1])},
                   "act_pred": float(acts[i])}
            df_rows.append(row)
        df = pd.DataFrame(df_rows)
        os.makedirs("samples", exist_ok=True)
        csv_path = os.path.join("samples", "generated_samples.csv")
        df.to_csv(csv_path, index=False)
        print(f"Saved generated samples to {csv_path}")


if __name__ == "__main__":
    main()
```

```python
import argparse
import os
import random
import math
from dataclasses import dataclass
from typing import Tuple, Dict, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Optional extras
try:
    import optuna
    OPTUNA_AVAILABLE = True
except Exception:
    OPTUNA_AVAILABLE = False

try:
    from fastapi import FastAPI
    import uvicorn
    FASTAPI_AVAILABLE = True
except Exception:
    FASTAPI_AVAILABLE = False

# -------------------------
# Feature definitions (abstract only)
# -------------------------
FEATURE_NAMES = [
    "telomere_len",
    "age_marker",
    "oxidative_stress",
    "metabolic_rate",
    "dna_damage",
    "autophagy_eff",
    "division_prob",
]
CONTEXT_NAMES = ["nutrient_level", "shear_stress", "signal_factor"]
NDIMS = len(FEATURE_NAMES)
NCONTEXT = len(CONTEXT_NAMES)

# -------------------------
# Synthetic dataset (unchanged, safe)
# -------------------------
class VirtualCellDataset(Dataset):
    def __init__(self, n_samples: int = 10000, seed: int = 42):
        super().__init__()
        self.n_samples = n_samples
        rng = np.random.RandomState(seed)
        self.data = []
        for _ in range(n_samples):
            nutrient = rng.beta(5, 2)
            shear = rng.beta(2, 5)
            signal = rng.beta(3, 3)

            age_marker = rng.beta(2, 5)
            oxidative = rng.beta(2, 3)
            dna_damage = np.minimum(1.0, oxidative * 0.6 + rng.normal(0, 0.05))

            telomere_len = np.clip(0.9 - 0.5 * age_marker - 0.4 * oxidative + rng.normal(0, 0.03), 0.0, 1.0)
            metabolic = np.clip(0.3 + 0.6 * nutrient - 0.2 * age_marker + rng.normal(0, 0.05), 0.0, 1.0)
            autophagy = np.clip(0.7 * nutrient - 0.3 * oxidative + rng.normal(0, 0.05), 0.0, 1.0)

            division_prob = np.clip(0.4 * signal + 0.3 * metabolic - 0.5 * dna_damage + rng.normal(0, 0.05), 0.0, 1.0)

            features = np.array([
                telomere_len,
                age_marker,
                oxidative,
                metabolic,
                dna_damage,
                autophagy,
                division_prob,
            ], dtype=np.float32)

            context = np.array([nutrient, shear, signal], dtype=np.float32)

            thr = 0.45 + 0.15 * (1.0 - nutrient)
            telomerase_on = float((telomere_len < thr) and (dna_damage < 0.6) and (division_prob > 0.15))

            self.data.append((features, context, telomerase_on))

    def __len__(self):
        return self.n_samples

    def __getitem__(self, idx):
        features, context, telomerase_on = self.data[idx]
        return {"x": torch.from_numpy(features), "c": torch.from_numpy(context), "act": torch.tensor(telomerase_on, dtype=torch.float32)}

# -------------------------
# cVAE model (unchanged)
# -------------------------
class Encoder(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU())
        self.mu = nn.Linear(hidden_dim, z_dim)
        self.logvar = nn.Linear(hidden_dim, z_dim)
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        h = self.net(h)
        return self.mu(h), self.logvar(h)

class Decoder(nn.Module):
    def __init__(self, z_dim: int, c_dim: int, hidden_dim: int, x_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(z_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, x_dim), nn.Sigmoid())
    def forward(self, z, c):
        h = torch.cat([z, c], dim=-1)
        return self.net(h)

class Controller(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim // 2), nn.ReLU(), nn.Linear(hidden_dim // 2, 1))
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        return torch.sigmoid(self.net(h)).squeeze(-1)

class CVAEModel(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.encoder = Encoder(x_dim, c_dim, hidden_dim, z_dim)
        self.decoder = Decoder(z_dim, c_dim, hidden_dim, x_dim)
        self.controller = Controller(x_dim, c_dim, hidden_dim)
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    def forward(self, x, c):
        mu, logvar = self.encoder(x, c)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z, c)
        act_pred = self.controller(x_recon, c)
        return x_recon, mu, logvar, act_pred

# -------------------------
# Diffusion model (simplified DDPM on abstract features)
# -------------------------
class SimpleMLP(nn.Module):
    def __init__(self, dim_in, dim_out, hidden=128):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(dim_in, hidden), nn.ReLU(), nn.Linear(hidden, hidden), nn.ReLU(), nn.Linear(hidden, dim_out))
    def forward(self, x, t=None, c=None):
        # t and c optionally ignored in this simple model, but acceptance of c keeps it conditional
        if c is not None:
            x = torch.cat([x, c], dim=-1)
        return self.net(x)

class SimpleDiffusion(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128, timesteps: int = 50):
        super().__init__()
        self.timesteps = timesteps
        self.model = SimpleMLP(x_dim + c_dim, x_dim, hidden)
        # linear noise schedule
        betas = torch.linspace(1e-4, 0.02, timesteps)
        alphas = 1.0 - betas
        alphas_cum = torch.cumprod(alphas, dim=0)
        self.register_buffer('betas', betas)
        self.register_buffer('alphas_cum', alphas_cum)

    def q_sample(self, x_start, t, noise=None):
        if noise is None:
            noise = torch.randn_like(x_start)
        a = self.alphas_cum[t].view(-1, 1)
        return torch.sqrt(a) * x_start + torch.sqrt(1 - a) * noise

    def p_losses(self, x_start, c):
        batch_size = x_start.size(0)
        t = torch.randint(0, self.timesteps, (batch_size,), device=x_start.device)
        noise = torch.randn_like(x_start)
        x_noisy = self.q_sample(x_start, t, noise=noise)
        pred_noise = self.model(torch.cat([x_noisy, c], dim=-1))
        return F.mse_loss(pred_noise, noise)

    def sample(self, c, device, n_samples=8):
        x = torch.randn(n_samples, NDIMS, device=device)
        for t in reversed(range(self.timesteps)):
            pred_noise = self.model(torch.cat([x, c], dim=-1))
            beta = self.betas[t]
            x = (x - beta * pred_noise) / torch.sqrt(1 - beta)
            if t > 0:
                x = x + torch.sqrt(beta) * torch.randn_like(x)
        # clamp to [0,1]
        return torch.clamp(x, 0.0, 1.0)

# -------------------------
# Losses
# -------------------------
def loss_function(recon_x, x, mu, logvar, act_pred, act_true, beta=1.0):
    recon_loss = F.mse_loss(recon_x, x, reduction="mean")
    kld = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
    bce = F.binary_cross_entropy(act_pred, act_true)
    return recon_loss + beta * kld + 0.5 * bce, recon_loss, kld, bce

# -------------------------
# Abstract reward function for RL fine-tuning (safe!)
# -------------------------
def abstract_scores(x: torch.Tensor, c: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """Compute two abstract scalar scores from feature vectors.
    - stability_score: higher when telomere_len high, dna_damage low, autophagy high
    - affinity_score: abstract proxy for 'telomerase affinity' combining division_prob and telomere_len
    All calculations are normalized to [0,1] using safe operations.
    """
    tel = x[:, 0]
    dna = x[:, 4]
    aut = x[:, 5]
    divp = x[:, 6]
    # stability: mix
    stability = 0.5 * tel + 0.3 * (1.0 - dna) + 0.2 * aut
    stability = torch.clamp(stability, 0.0, 1.0)
    # affinity: high if telomere short but division prob present (abstract)
    affinity = 0.6 * (1.0 - tel) + 0.4 * divp
    affinity = torch.clamp(affinity, 0.0, 1.0)
    return stability, affinity

# -------------------------
# Training utilities
# -------------------------

def train_cvae(model: CVAEModel, dataloader: DataLoader, optim, device, epochs: int=20, beta: float=1.0, ckpt_dir: str="checkpoints"):
    model.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    model.train()
    for epoch in range(1, epochs + 1):
        total_loss = total_recon = total_kld = total_bce = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            act = batch['act'].to(device)
            optim.zero_grad()
            x_recon, mu, logvar, act_pred = model(x, c)
            loss, recon, kld, bce = loss_function(x_recon, x, mu, logvar, act_pred, act, beta)
            loss.backward()
            optim.step()
            total_loss += loss.item() * x.size(0)
            total_recon += recon.item() * x.size(0)
            total_kld += kld.item() * x.size(0)
            total_bce += bce.item() * x.size(0)
        n = len(dataloader.dataset)
        print(f"Epoch {epoch}/{epochs} loss={total_loss/n:.6f} recon={total_recon/n:.6f} kld={total_kld/n:.6f} bce={total_bce/n:.6f}")
        torch.save(model.state_dict(), os.path.join(ckpt_dir, f"cvae_epoch{epoch}.pth"))

def train_diffusion(diffusion: SimpleDiffusion, dataloader: DataLoader, optim, device, epochs: int=20, ckpt_dir: str="checkpoints"):
    diffusion.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    diffusion.train()
    for epoch in range(1, epochs + 1):
        total = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            optim.zero_grad()
            loss = diffusion.p_losses(x, c)
            loss.backward()
            optim.step()
            total += loss.item() * x.size(0)
        print(f"Diffusion epoch {epoch}/{epochs} loss={total/len(dataloader.dataset):.6f}")
        torch.save(diffusion.state_dict(), os.path.join(ckpt_dir, f"diff_epoch{epoch}.pth"))

# -------------------------
# RL fine-tuning (policy-gradient / REINFORCE) of the Controller
# -------------------------

def rl_finetune_controller(controller: Controller, generator, device, n_iters: int = 200, batch_size: int = 64, lr: float = 1e-4):
    """controller: policy to finetune (outputs prob of activating)
    generator: callable that returns (x_generated, c) tensors given a batch size
    This will perform policy-gradient updates where the reward is a combination of
    abstract stability and affinity scores. All calculations are on abstract vectors.
    """
    controller.to(device)
    opt = torch.optim.Adam(controller.parameters(), lr=lr)
    for it in range(1, n_iters + 1):
        x_gen, c = generator(batch_size)
        x_gen = x_gen.to(device)
        c = c.to(device)
        probs = controller(x_gen, c)  # values in [0,1]
        m = torch.distributions.Bernoulli(probs)
        actions = m.sample()
        logp = m.log_prob(actions)
        # compute abstract rewards
        stability, affinity = abstract_scores(x_gen, c)
        # reward shaped: prefer high stability and moderate affinity (example)
        reward = 0.7 * stability + 0.3 * affinity
        # encourage controller to activate (action==1) when reward high
        advantage = reward - reward.mean()
        # policy gradient loss (maximize expected reward -> minimize -logp * advantage)
        loss = -(logp * advantage.detach()).mean()
        opt.zero_grad()
        loss.backward()
        opt.step()
        if it % 20 == 0:
            print(f"RL iter {it}/{n_iters} loss={loss.item():.6f} mean_reward={reward.mean().item():.4f}")
    return controller

# -------------------------
# Generator helpers used by RL: sample from cVAE or Diffusion
# -------------------------
def generator_from_cvae(model: CVAEModel, device):
    def gen(n: int):
        model.to(device)
        model.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            z = torch.randn(n, model.encoder.mu.out_features, device=device)
            x = model.decoder(z, c)
            return x, c
    return gen

def generator_from_diffusion(diffusion: SimpleDiffusion, device):
    def gen(n: int):
        diffusion.to(device)
        diffusion.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            x = diffusion.sample(c, device, n_samples=n)
            return x, c
    return gen

# -------------------------
# Visualization and sampling utilities
# -------------------------

def plot_samples(xs: np.ndarray, ctx: np.ndarray, acts: np.ndarray, n_plot: int = 5):
    n = min(n_plot, xs.shape[0])
    fig, axes = plt.subplots(n, 1, figsize=(6, 2 * n))
    if n == 1:
        axes = [axes]
    for i in range(n):
        ax = axes[i]
        ax.bar(range(len(FEATURE_NAMES)), xs[i], tick_label=FEATURE_NAMES)
        ax.set_ylim(0, 1)
        ax.set_title(f"Context: {np.round(ctx[i],3)}  act={acts[i]:.3f}")
    plt.tight_layout()
    plt.show()

# -------------------------
# FastAPI server (optional)
# -------------------------

def create_api(model_cvae: Optional[CVAEModel], diffusion: Optional[SimpleDiffusion]):
    if not FASTAPI_AVAILABLE:
        raise RuntimeError("FastAPI not installed")
    app = FastAPI()

    @app.get('/health')
    def health():
        return {'status': 'ok'}

    @app.post('/sample_cvae')
    def sample_cvae(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_cvae(model_cvae, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    @app.post('/sample_diffusion')
    def sample_diffusion(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_diffusion(diffusion, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    return app

# -------------------------
# CLI and main
# -------------------------

def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument('--train', action='store_true')
    p.add_argument('--train-diffusion', action='store_true')
    p.add_argument('--sample', action='store_true')
    p.add_argument('--sample-diffusion', action='store_true')
    p.add_argument('--rl-finetune', action='store_true')
    p.add_argument('--serve', action='store_true')
    p.add_argument('--n-samples', type=int, default=8)
    p.add_argument('--epochs', type=int, default=20)
    p.add_argument('--batch-size', type=int, default=128)
    p.add_argument('--lr', type=float, default=1e-3)
    p.add_argument('--hidden-dim', type=int, default=128)
    p.add_argument('--z-dim', type=int, default=16)
    p.add_argument('--beta', type=float, default=1.0)
    p.add_argument('--seed', type=int, default=42)
    return p.parse_args([]) # Pass an empty list to ignore kernel arguments


def main():
    args = parse_args()
    torch.manual_seed(args.seed)
    np.random.seed(args.seed)
    random.seed(args.seed)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=args.hidden_dim, z_dim=args.z_dim)
    diffusion = SimpleDiffusion(x_dim=NDIMS, c_dim=NCONTEXT, hidden=args.hidden_dim, timesteps=50)

    if args.train:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(model.parameters(), lr=args.lr)
        train_cvae(model, dl, optim, device, epochs=args.epochs, beta=args.beta)
        torch.save(model.state_dict(), 'checkpoints/cvae_final.pth')
        print('Saved cVAE to checkpoints/cvae_final.pth')

    if args.train_diffusion:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(diffusion.parameters(), lr=args.lr)
        train_diffusion(diffusion, dl, optim, device, epochs=args.epochs)
        torch.save(diffusion.state_dict(), 'checkpoints/diffusion_final.pth')
        print('Saved diffusion to checkpoints/diffusion_final.pth')

    if args.rl_finetune:
        # choose generator (diffusion if trained else cVAE)
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
            gen = generator_from_diffusion(diffusion, device)
        elif os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
            gen = generator_from_cvae(model, device)
        else:
            # quick fallback: create a small dataset-based generator
            dataset = VirtualCellDataset(2000, seed=args.seed)
            dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
            def gen_fallback(n):
                batch = next(iter(dl))
                return batch['x'], batch['c']
            gen = gen_fallback
        # finetune controller (policy)
        controller = model.controller
        controller = rl_finetune_controller(controller, gen, device, n_iters=200, batch_size=64, lr=1e-4)
        torch.save(controller.state_dict(), 'checkpoints/controller_rl_finetuned.pth')
        print('Saved finetuned controller to checkpoints/controller_rl_finetuned.pth')

    if args.sample or args.sample_diffusion:
        # load models if available
        if os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
        if args.sample:
            gen = generator_from_cvae(model, device)
        else:
            gen = generator_from_diffusion(diffusion, device)
        x, c = gen(args.n_samples)
        # controller prediction
        act = model.controller(x.to(device), c.to(device)).cpu().detach().numpy()
        plot_samples(x.cpu().numpy(), c.cpu().numpy(), act, n_plot=min(8, args.n_samples))
        # save
        rows = []
        for i in range(x.size(0)):
            d = {k: float(x[i, j].cpu().item()) for j, k in enumerate(FEATURE_NAMES)}
            for j in range(c.size(1)):
                d[f'ctx_{j}'] = float(c[i, j].cpu().item())
            d['act_pred'] = float(act[i])
            rows.append(d)
        df = pd.DataFrame(rows)
        os.makedirs('samples', exist_ok=True)
        df.to_csv('samples/generated_samples.csv', index=False)
        print('Saved samples to samples/generated_samples.csv')

    if args.serve:
        if not FASTAPI_AVAILABLE:
            print('FastAPI/uvicorn not installed. Install fastapi and uvicorn to use the serve feature.')
        else:
            # load weights if present
            if os.path.exists('checkpoints/cvae_final.pth'):
                model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location='cpu'))
            if os.path.exists('checkpoints/diffusion_final.pth'):
                diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location='cpu'))
            app = create_api(model, diffusion)
            print('Starting server on http://0.0.0.0:8000')
            uvicorn.run(app, host='0.0.0.0', port=8000)

if __name__ == '__main__':
    main()

# -------------------------
# Digital Immortal Cell Simulation (abstract, safe)
# -------------------------
class DigitalImmortalCell:
    """A purely in-silico 'digital immortal cell' simulator.

    - Models telomere length as a continuous variable in [0,1].
    - Simulates cell cycles with probabilistic division, DNA damage accumulation,
      repair processes, and an AI controller that emulates telomerase activation
      decisions based on abstract state and context.
    - Implements safety checks to prevent runaway unchecked proliferation by
      simulating "genomic instability" and triggering virtual apoptosis.

    This class operates on abstract numerical features only. No biological
    sequences, wet-lab protocols, or real-world actionable steps are produced.
    """

    def __init__(self,
                 initial_telomere: float = 0.9,
                 autophagy_eff: float = 0.7,
                 repair_eff: float = 0.8,
                 mutation_rate: float = 0.01,
                 division_base_prob: float = 0.3,
                 seed: int = 42):
        self.rng = np.random.RandomState(seed)
        # abstract state
        self.telomere = float(np.clip(initial_telomere, 0.0, 1.0))
        self.autophagy = float(np.clip(autophagy_eff, 0.0, 1.0))
        self.repair_eff = float(np.clip(repair_eff, 0.0, 1.0))
        self.mutation_rate = float(np.clip(mutation_rate, 0.0, 1.0))
        self.division_base_prob = float(np.clip(division_base_prob, 0.0, 1.0))
        # metadata
        self.age = 0
        self.alive = True
        self.history = []

    def step(self, context: Optional[np.ndarray], controller_callable) -> Dict:
        """One simulation timestep (one potential cell cycle).

        context: array-like of context features (nutrient, shear, signal)
        controller_callable: function(state_tensor, context_tensor) -> prob[T]
            it should return a telomerase activation probability in [0,1]

        Returns a dict with new state and flags.
        """
        if not self.alive:
            return {"alive": False}

        # environmental/context influence
        if context is None:
            nutrient = float(self.rng.beta(5, 2))
            shear = float(self.rng.beta(2, 5))
            signal = float(self.rng.beta(3, 3))
            context = np.array([nutrient, shear, signal], dtype=np.float32)
        else:
            context = np.array(context, dtype=np.float32)

        # form state vector
        state = np.array([
            self.telomere,
            min(1.0, 0.5 + 0.5 * (self.age / 100.0)),  # abstract age_marker increasing weakly
            float(self.rng.beta(2, 3)),  # oxidative proxy sample
            float(np.clip(0.3 + 0.6 * context[0] - 0.2 * (self.age / 100.0), 0.0, 1.0)),  # metabolic
            float(np.clip(0.2 + 0.4 * (1 - self.repair_eff) + self.rng.normal(0, 0.05), 0.0, 1.0)),  # dna damage
            self.autophagy,
            float(self.division_base_prob),
        ], dtype=np.float32)

        # controller decision (telomerase activation probability)
        with torch.no_grad():
            inp_state = torch.from_numpy(state).unsqueeze(0).float()
            inp_context = torch.from_numpy(context).unsqueeze(0).float()
            prob = 0.0
            try:
                p = controller_callable(inp_state, inp_context)
                # controller might return tensor or numpy
                if isinstance(p, torch.Tensor):
                    prob = float(p.squeeze().cpu().numpy())
                else:
                    prob = float(np.array(p).squeeze())
            except Exception:
                # fallback heuristic if controller not available
                prob = float(0.2 * (1.0 - self.telomere) + 0.1 * (1.0 - state[4]))
                prob = float(np.clip(prob, 0.0, 1.0))

        # sample activation
        activate = self.rng.rand() < prob

        # telomere update: activation increases telomere by synthetic amount, otherwise natural shortening
        if activate:
            self.telomere = float(np.clip(self.telomere + 0.06 * self.repair_eff, 0.0, 1.0))
        else:
            # natural erosion influenced by oxidative stress and age
            erosion = 0.02 + 0.05 * state[2] + 0.002 * (self.age / 10.0)
            self.telomere = float(np.clip(self.telomere - erosion, 0.0, 1.0))

        # DNA damage accumulation and repair
        dna_damage = state[4]
        repair = self.repair_eff * (0.5 + 0.5 * self.autophagy)
        dna_damage = float(np.clip(dna_damage - repair * 0.1 + self.rng.normal(0, 0.02), 0.0, 1.0))

        # mutation/genomic instability modeled as rising probability with damage and short telomeres
        instability_prob = float(np.clip(0.02 + 0.4 * dna_damage + 0.4 * (1.0 - self.telomere), 0.0, 1.0))
        mutated = self.rng.rand() < instability_prob * self.mutation_rate

        # division decision
        div_prob = float(np.clip(self.division_base_prob + 0.3 * context[2] - 0.5 * dna_damage, 0.0, 1.0))
        divided = self.rng.rand() < div_prob

        # safety: if instability too high, trigger apoptosis (virtual)
        if instability_prob > 0.85 and dna_damage > 0.8:
            self.alive = False
            reason = 'apoptosis_due_to_instability'
        elif self.telomere <= 0.02 and dna_damage > 0.7:
            # senescence-like virtual endpoint (not immortal)
            self.alive = False
            reason = 'senescence'
        else:
            reason = 'alive'

        # age increment
        self.age += 1

        # record
        record = {
            'age': self.age,
            'telomere': self.telomere,
            'dna_damage': dna_damage,
            'instability_prob': instability_prob,
            'mutated': bool(mutated),
            'divided': bool(divided),
            'activated_telomerase': bool(activate),
            'alive': bool(self.alive),
            'reason': reason,
            'context': context.tolist(),
        }
        self.history.append(record)
        return record

    def run(self, controller_callable, steps: int = 1000, context_schedule=None):
        """Run simulation for given steps. context_schedule can be:
        - None: random contexts each step
        - callable(step)->context array
        - fixed array of length steps
        """
        self.history = []
        for t in range(steps):
            if callable(context_schedule):
                ctx = context_schedule(t)
            elif isinstance(context_schedule, (list, np.ndarray)):
                ctx = context_schedule[t] if t < len(context_schedule) else None
            else:
                ctx = None
            rec = self.step(ctx, controller_callable)
            if not rec['alive']:
                break
        return self.history

# -------------------------
# CLI helper to run a demo digital immortal cell simulation
# -------------------------

def demo_immortal_cell(steps: int = 500):
    device = torch.device('cpu')
    # create a default model.controller (loaded if checkpoints exist)
    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=128, z_dim=16)
    if os.path.exists('checkpoints/controller_rl_finetuned.pth'):
        model.controller.load_state_dict(torch.load('checkpoints/controller_rl_finetuned.pth', map_location=device))
    elif os.path.exists('checkpoints/cvae_final.pth'):
        model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))

    controller = model.controller

    cell = DigitalImmortalCell(initial_telomere=0.95, autophagy_eff=0.8, repair_eff=0.85, mutation_rate=0.002)

    def ctrl_callable(s_tensor, c_tensor):
        # simple wrapper to call controller in eval mode
        controller.eval()
        with torch.no_grad():
            return controller(s_tensor, c_tensor)

    history = cell.run(ctrl_callable, steps=steps)
    # convert history to DataFrame for inspection
    df = pd.DataFrame(history)
    return df

# -------------------------
# If invoked with a special flag, run demo and save results
# -------------------------

def run_demo_and_save(steps: int = 500):
    df = demo_immortal_cell(steps=steps)
    os.makedirs('simulations', exist_ok=True)
    path = os.path.join('simulations', 'digital_immortal_cell_demo.csv')
    df.to_csv(path, index=False)
    print(f"Saved simulation to {path}. Steps simulated: {len(df)}. Final state alive={df.iloc[-1]['alive']}")

# -------------------------
# Advanced RL: PPO-style training of Controller within DigitalImmortalCell environment
# -------------------------

class PPOMemory:
    def __init__(self):
        self.states = []
        self.contexts = []
        self.actions = []
        self.logprobs = []
        self.rewards = []
        self.is_terminals = []

    def clear(self):
        self.__init__()


class PPOController(nn.Module):
    """Actor-Critic style network for PPO. Policy outputs Bernoulli probability to 'activate'.
    Critic outputs state-value for advantage estimation."""

    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128):
        super().__init__()
        self.actor = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1), nn.Sigmoid())
        self.critic = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1))

    def forward(self, x, c):
        inp = torch.cat([x, c], dim=-1)
        prob = self.actor(inp).squeeze(-1)
        value = self.critic(inp).squeeze(-1)
        return prob, value

    def act(self, x, c):
        prob, _ = self.forward(x, c)
        dist = torch.distributions.Bernoulli(probs=prob)
        action = dist.sample()
        return action, dist.log_prob(action)


def compute_gae(rewards, values, is_terminals, gamma=0.99, lam=0.95):
    advantages = []
    gae = 0
    values = values + [0]
    for step in reversed(range(len(rewards))):
        delta = rewards[step] + gamma * values[step + 1] * (1 - is_terminals[step]) - values[step]
        gae = delta + gamma * lam * (1 - is_terminals[step]) * gae
        advantages.insert(0, gae)
    return advantages


def train_ppo(controller: PPOController,
              env_constructor,
              device,
              epochs: int = 200,
              steps_per_epoch: int = 2048,
              mini_batch_size: int = 64,
              ppo_epochs: int = 8,
              clip_epsilon: float = 0.2,
              gamma: float = 0.99,
              lam: float = 0.95,
              lr: float = 3e-4):
    """Train a PPO controller interacting with repeated DigitalImmortalCell episodes.

    env_constructor: callable() -> new DigitalImmortalCell instance (fresh env per episode)
    The controller policy is updated to maximize long-term simulated rewards (longevity, low instability)
    while avoiding unsafe behaviors (penalized).
    """
    controller.to(device)
    optimizer = torch.optim.Adam(controller.parameters(), lr=lr)

    def rollout():
        memory = PPOMemory()
        env = env_constructor()
        state = None
        # we'll run until we collect steps_per_epoch transitions
        steps = 0
        while steps < steps_per_epoch:
            # prepare state and context
            # DigitalImmortalCell.step expects controller_callable that accepts tensors, but for sampling we pass policy
            s_vec = np.array([
                env.telomere,
                min(1.0, 0.5 + 0.5 * (env.age / 100.0)),
                float(env.rng.beta(2, 3)),
                float(np.clip(0.3 + 0.6 * env.rng.beta(5, 2) - 0.2 * (env.age / 100.0), 0.0, 1.0)),
                float(np.clip(0.2 + 0.4 * (1 - env.repair_eff) + env.rng.normal(0, 0.05), 0.0, 1.0)),
                env.autophagy,
                float(env.division_base_prob),
            ], dtype=np.float32)
            ctx = np.array([env.rng.beta(5, 2), env.rng.beta(2, 5), env.rng.beta(3, 3)], dtype=np.float32)
            s_t = torch.from_numpy(s_vec).unsqueeze(0).float().to(device)
            c_t = torch.from_numpy(ctx).unsqueeze(0).float().to(device)

            with torch.no_grad():
                prob, value = controller(s_t, c_t)
                dist = torch.distributions.Bernoulli(probs=prob)
                action = dist.sample()
                logp = dist.log_prob(action)
            # apply action via env.step using a thin wrapper
            def ctrl_callable(inner_s, inner_c):
                # inner tensors correspond to state/context; policy expects same shape
                with torch.no_grad():
                    p, _ = controller(inner_s.to(device), inner_c.to(device))
                    return p
            rec = env.step(ctx, ctrl_callable)
            # reward design: encourage staying alive, penalize instability and mutations
            reward = 0.0
            if rec['alive']:
                reward += 1.0  # survival bonus
            # small penalty proportional to instability
            reward += -1.0 * rec['instability_prob']
            if rec['mutated']:
                reward += -2.0
            if rec['activated_telomerase']:
                # small cost to activating to avoid trivial always-on policy
                reward += -0.02
            memory.states.append(s_t.squeeze(0).cpu())
            memory.contexts.append(c_t.squeeze(0).cpu())
            memory.actions.append(action.squeeze(0).cpu())
            memory.logprobs.append(logp.squeeze(0).cpu())
            memory.rewards.append(reward)
            memory.is_terminals.append(not rec['alive'])

            steps += 1
            if not rec['alive']:
                # restart env when cell dies
                env = env_constructor()
        return memory

    for epoch in range(1, epochs + 1):
        memory = rollout()
        # convert memory lists to tensors
        states = torch.stack(memory.states).to(device)
        contexts = torch.stack(memory.contexts).to(device)
        actions = torch.stack(memory.actions).to(device)
        old_logprobs = torch.stack(memory.logprobs).to(device)
        rewards = memory.rewards
        is_term = memory.is_terminals

        # compute values for states
        with torch.no_grad():
            _, values = controller(states, contexts)
            values = values.cpu().numpy().tolist()
        # GAE advantages
        advantages = compute_gae(rewards, values, is_term, gamma=gamma, lam=lam)
        advantages = torch.tensor(advantages, dtype=torch.float32).to(device)
        returns = advantages + torch.tensor(values[:-1] if len(values) > len(advantages) else values[:len(advantages)], dtype=torch.float32).to(device)

        # normalize advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO updates
        dataset_size = states.size(0)
        for _ in range(ppo_epochs):
            # minibatch sampling
            idxs = torch.randperm(dataset_size)
            for start in range(0, dataset_size, mini_batch_size):
                idx = idxs[start:start + mini_batch_size]
                batch_states = states[idx]
                batch_contexts = contexts[idx]
                batch_actions = actions[idx]
                batch_oldlogp = old_logprobs[idx]
                batch_adv = advantages[idx]
                batch_returns = returns[idx]

                probs, values = controller(batch_states, batch_contexts)
                dist = torch.distributions.Bernoulli(probs=probs)
                newlogp = dist.log_prob(batch_actions)
                entropy = dist.entropy().mean()

                ratio = torch.exp(newlogp - batch_oldlogp)
                surr1 = ratio * batch_adv
                surr2 = torch.clamp(ratio, 1.0 - clip_epsilon, 1.0 + clip_epsilon) * batch_adv
                policy_loss = -torch.min(surr1, surr2).mean()
                value_loss = F.mse_loss(values, batch_returns)
                loss = policy_loss + 0.5 * value_loss - 0.01 * entropy

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

        # end epoch logging
        print(f"PPO Epoch {epoch}/{epochs} completed. Mean reward per step: {np.mean(memory.rewards):.4f}")
        # optional checkpoint
        torch.save(controller.state_dict(), f'checkpoints/ppo_controller_epoch{epoch}.pth')

    return controller


# CLI: add flag to run PPO training
import inspect
def add_ppo_flag_and_run():
    import sys
    if '--train-ppo' in sys.argv:
        # default simple env constructor
        def env_ctor():
            return DigitalImmortalCell(initial_telomere=0.9, autophagy_eff=0.7, repair_eff=0.8, mutation_rate=0.005)
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        ppo_ctrl = PPOController(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
        trained = train_ppo(ppo_ctrl, env_ctor, device, epochs=50, steps_per_epoch=1024)
        torch.save(trained.state_dict(), 'checkpoints/ppo_final.pth')
        print('Saved PPO controller to checkpoints/ppo_final.pth')

# Try to run if flag present
add_ppo_flag_and_run()

# -------------------------
# Discovery Engine: latent-space and controller search for "discoveries" (abstract-only)
# -------------------------

class DiscoveryEngine:
    """Performs algorithmic discovery in the virtual system by searching latent
    spaces (cVAE/diffusion) and controller parameter spaces using safe,
    purely computational optimization strategies. Outputs are abstract feature
    vectors and controller parameter snapshots — not biological sequences.
    """

    def __init__(self, cvae: Optional[CVAEModel] = None, diffusion: Optional[SimpleDiffusion] = None, device: Optional[torch.device] = None):
        self.cvae = cvae
        self.diffusion = diffusion
        self.device = device or torch.device('cpu')

    def score_features(self, x: np.ndarray, c: np.ndarray) -> Dict[str, float]:
        # wrapper to compute abstract stability and affinity for numpy arrays
        xt = torch.from_numpy(x).float()
        ct = torch.from_numpy(c).float()
        s, a = abstract_scores(xt, ct)
        return {'stability': float(s.mean().item()), 'affinity': float(a.mean().item())}

    def random_latent_search_cvae(self, n_samples: int = 2000, context: Optional[np.ndarray] = None, top_k: int = 20):
        """Randomly samples latent z vectors from standard normal, decodes via cVAE
        and returns the top_k candidates ranked by a composite score (e.g., stability - alpha * affinity).
        """
        assert self.cvae is not None, 'cVAE model required for this search'
        self.cvae.to(self.device)
        self.cvae.eval()
        if context is None:
            context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        else:
            if context.ndim == 1:
                context = np.tile(context[None, :], (n_samples, 1)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        z_dim = self.cvae.encoder.mu.out_features
        zs = torch.randn(n_samples, z_dim, device=self.device)
        with torch.no_grad():
            x_gen = self.cvae.decoder(zs, c_t)
        x_np = x_gen.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        # save
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'cvae_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} cVAE latent discoveries to {path}')
        return df

    def random_latent_search_diffusion(self, n_samples: int = 1000, top_k: int = 20):
        assert self.diffusion is not None, 'Diffusion model required for this search'
        self.diffusion.to(self.device)
        self.diffusion.eval()
        context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        xs = self.diffusion.sample(c_t, self.device, n_samples=n_samples)
        x_np = xs.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'diffusion_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} diffusion latent discoveries to {path}')
        return df

    def evo_search_controller(self, base_controller: nn.Module, env_constructor, generations: int = 50, pop_size: int = 20, sigma: float = 0.02, elite_frac: float = 0.2):
        """Evolutionary strategy to search controller parameter perturbations that maximize
        long-term simulated rewards in the DigitalImmortalCell environment.
        Returns a DataFrame of top parameter sets (serialized as flattened arrays) and their scores.
        """
        # flatten params
        params = np.concatenate([p.detach().cpu().numpy().ravel() for p in base_controller.parameters()])
        n_params = params.size
        population = np.tile(params[None, :], (pop_size, 1)) + np.random.randn(pop_size, n_params) * sigma

        def set_params_from_flat(model, flat):
            idx = 0
            for p in model.parameters():
                num = p.numel()
                arr = flat[idx:idx+num].reshape(p.shape)
                p.data = torch.from_numpy(arr).float()
                idx += num

        scores = []
        for g in range(generations):
            pop_scores = []
            for i in range(pop_size):
                indiv = population[i]
                # evaluate
                model_copy = type(base_controller)(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
                set_params_from_flat(model_copy, indiv)
                # quick rollout: average cumulative reward over few episodes
                total = 0.0
                n_eps = 3
                for e in range(n_eps):
                    env = env_constructor()
                    steps = 0
                    while steps < 200 and env.alive:
                        # prepare current state/context
                        ctx = [env.rng.beta(5,2), env.rng.beta(2,5), env.rng.beta(3,3)]
                        def ctrl_call(s,c):
                            with torch.no_grad():
                                s_t = torch.from_numpy(s).unsqueeze(0).float()
                                c_t = torch.from_numpy(c).unsqueeze(0).float()
                                p, _ = model_copy(s_t, c_t)
                                return p
                        rec = env.step(ctx, ctrl_call)
                        r = 1.0 if rec['alive'] else -5.0
                        r -= rec['instability_prob']
                        if rec['mutated']:
                            r -= 2.0
                        total += r
                        steps += 1
                avg_score = total / n_eps
                pop_scores.append(avg_score)
            pop_scores = np.array(pop_scores)
            # select elites
            elites_idx = pop_scores.argsort()[::-1][:max(1, int(pop_size*elite_frac))]
            elites = population[elites_idx]
            elite_mean = elites.mean(axis=0)
            # generate new population around elites mean
            population = elite_mean[None, :] + np.random.randn(pop_size, n_params) * sigma
            best_score = float(pop_scores.max())
            print(f'Generation {g+1}/{generations} best_score={best_score:.4f}')
            scores.append((g, best_score))
        # save best individual
        best_idx = int(np.argmax(pop_scores))
        best_flat = population[best_idx]
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'controller_evo_best.npy')
        np.save(path, best_flat)
        print(f'Saved best controller flat params to {path}')
        # return simple summary DataFrame
        df = pd.DataFrame([{'generation': g, 'best_score': sc} for g, sc in scores])
        path2 = os.path.join('discoveries', 'controller_evo_scores.csv')
        df.to_csv(path2, index=False)
        print(f'Saved controller evolution scores to {path2}')
        return df

# End of file
```

```python
import argparse
import os
import random
import math
from dataclasses import dataclass
from typing import Tuple, Dict, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Optional extras
try:
    import optuna
    OPTUNA_AVAILABLE = True
except Exception:
    OPTUNA_AVAILABLE = False

try:
    from fastapi import FastAPI
    import uvicorn
    FASTAPI_AVAILABLE = True
except Exception:
    FASTAPI_AVAILABLE = False

# -------------------------
# Feature definitions (abstract only)
# -------------------------
FEATURE_NAMES = [
    "telomere_len",
    "age_marker",
    "oxidative_stress",
    "metabolic_rate",
    "dna_damage",
    "autophagy_eff",
    "division_prob",
]
CONTEXT_NAMES = ["nutrient_level", "shear_stress", "signal_factor"]
NDIMS = len(FEATURE_NAMES)
NCONTEXT = len(CONTEXT_NAMES)

# -------------------------
# Synthetic dataset (unchanged, safe)
# -------------------------
class VirtualCellDataset(Dataset):
    def __init__(self, n_samples: int = 10000, seed: int = 42):
        super().__init__()
        self.n_samples = n_samples
        rng = np.random.RandomState(seed)
        self.data = []
        for _ in range(n_samples):
            nutrient = rng.beta(5, 2)
            shear = rng.beta(2, 5)
            signal = rng.beta(3, 3)

            age_marker = rng.beta(2, 5)
            oxidative = rng.beta(2, 3)
            dna_damage = np.minimum(1.0, oxidative * 0.6 + rng.normal(0, 0.05))

            telomere_len = np.clip(0.9 - 0.5 * age_marker - 0.4 * oxidative + rng.normal(0, 0.03), 0.0, 1.0)
            metabolic = np.clip(0.3 + 0.6 * nutrient - 0.2 * age_marker + rng.normal(0, 0.05), 0.0, 1.0)
            autophagy = np.clip(0.7 * nutrient - 0.3 * oxidative + rng.normal(0, 0.05), 0.0, 1.0)

            division_prob = np.clip(0.4 * signal + 0.3 * metabolic - 0.5 * dna_damage + rng.normal(0, 0.05), 0.0, 1.0)

            features = np.array([
                telomere_len,
                age_marker,
                oxidative,
                metabolic,
                dna_damage,
                autophagy,
                division_prob,
            ], dtype=np.float32)

            context = np.array([nutrient, shear, signal], dtype=np.float32)

            thr = 0.45 + 0.15 * (1.0 - nutrient)
            telomerase_on = float((telomere_len < thr) and (dna_damage < 0.6) and (division_prob > 0.15))

            self.data.append((features, context, telomerase_on))

    def __len__(self):
        return self.n_samples

    def __getitem__(self, idx):
        features, context, telomerase_on = self.data[idx]
        return {"x": torch.from_numpy(features), "c": torch.from_numpy(context), "act": torch.tensor(telomerase_on, dtype=torch.float32)}

# -------------------------
# cVAE model (unchanged)
# -------------------------
class Encoder(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU())
        self.mu = nn.Linear(hidden_dim, z_dim)
        self.logvar = nn.Linear(hidden_dim, z_dim)
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        h = self.net(h)
        return self.mu(h), self.logvar(h)

class Decoder(nn.Module):
    def __init__(self, z_dim: int, c_dim: int, hidden_dim: int, x_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(z_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, x_dim), nn.Sigmoid())
    def forward(self, z, c):
        h = torch.cat([z, c], dim=-1)
        return self.net(h)

class Controller(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim // 2), nn.ReLU(), nn.Linear(hidden_dim // 2, 1))
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        return torch.sigmoid(self.net(h)).squeeze(-1)

class CVAEModel(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.encoder = Encoder(x_dim, c_dim, hidden_dim, z_dim)
        self.decoder = Decoder(z_dim, c_dim, hidden_dim, x_dim)
        self.controller = Controller(x_dim, c_dim, hidden_dim)
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    def forward(self, x, c):
        mu, logvar = self.encoder(x, c)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z, c)
        act_pred = self.controller(x_recon, c)
        return x_recon, mu, logvar, act_pred

# -------------------------
# Diffusion model (simplified DDPM on abstract features)
# -------------------------
class SimpleMLP(nn.Module):
    def __init__(self, dim_in, dim_out, hidden=128):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(dim_in, hidden), nn.ReLU(), nn.Linear(hidden, hidden), nn.ReLU(), nn.Linear(hidden, dim_out))
    def forward(self, x, t=None, c=None):
        # t and c optionally ignored in this simple model, but acceptance of c keeps it conditional
        if c is not None:
            x = torch.cat([x, c], dim=-1)
        return self.net(x)

class SimpleDiffusion(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128, timesteps: int = 50):
        super().__init__()
        self.timesteps = timesteps
        self.model = SimpleMLP(x_dim + c_dim, x_dim, hidden)
        # linear noise schedule
        betas = torch.linspace(1e-4, 0.02, timesteps)
        alphas = 1.0 - betas
        alphas_cum = torch.cumprod(alphas, dim=0)
        self.register_buffer('betas', betas)
        self.register_buffer('alphas_cum', alphas_cum)

    def q_sample(self, x_start, t, noise=None):
        if noise is None:
            noise = torch.randn_like(x_start)
        a = self.alphas_cum[t].view(-1, 1)
        return torch.sqrt(a) * x_start + torch.sqrt(1 - a) * noise

    def p_losses(self, x_start, c):
        batch_size = x_start.size(0)
        t = torch.randint(0, self.timesteps, (batch_size,), device=x_start.device)
        noise = torch.randn_like(x_start)
        x_noisy = self.q_sample(x_start, t, noise=noise)
        pred_noise = self.model(torch.cat([x_noisy, c], dim=-1))
        return F.mse_loss(pred_noise, noise)

    def sample(self, c, device, n_samples=8):
        x = torch.randn(n_samples, NDIMS, device=device)
        for t in reversed(range(self.timesteps)):
            pred_noise = self.model(torch.cat([x, c], dim=-1))
            beta = self.betas[t]
            x = (x - beta * pred_noise) / torch.sqrt(1 - beta)
            if t > 0:
                x = x + torch.sqrt(beta) * torch.randn_like(x)
        # clamp to [0,1]
        return torch.clamp(x, 0.0, 1.0)

# -------------------------
# Losses
# -------------------------
def loss_function(recon_x, x, mu, logvar, act_pred, act_true, beta=1.0):
    recon_loss = F.mse_loss(recon_x, x, reduction="mean")
    kld = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
    bce = F.binary_cross_entropy(act_pred, act_true)
    return recon_loss + beta * kld + 0.5 * bce, recon_loss, kld, bce

# -------------------------
# Abstract reward function for RL fine-tuning (safe!)
# -------------------------
def abstract_scores(x: torch.Tensor, c: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """Compute two abstract scalar scores from feature vectors.
    - stability_score: higher when telomere_len high, dna_damage low, autophagy high
    - affinity_score: abstract proxy for 'telomerase affinity' combining division_prob and telomere_len
    All calculations are normalized to [0,1] using safe operations.
    """
    tel = x[:, 0]
    dna = x[:, 4]
    aut = x[:, 5]
    divp = x[:, 6]
    # stability: mix
    stability = 0.5 * tel + 0.3 * (1.0 - dna) + 0.2 * aut
    stability = torch.clamp(stability, 0.0, 1.0)
    # affinity: high if telomere short but division prob present (abstract)
    affinity = 0.6 * (1.0 - tel) + 0.4 * divp
    affinity = torch.clamp(affinity, 0.0, 1.0)
    return stability, affinity

# -------------------------
# Training utilities
# -------------------------

def train_cvae(model: CVAEModel, dataloader: DataLoader, optim, device, epochs: int=20, beta: float=1.0, ckpt_dir: str="checkpoints"):
    model.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    model.train()
    for epoch in range(1, epochs + 1):
        total_loss = total_recon = total_kld = total_bce = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            act = batch['act'].to(device)
            optim.zero_grad()
            x_recon, mu, logvar, act_pred = model(x, c)
            loss, recon, kld, bce = loss_function(x_recon, x, mu, logvar, act_pred, act, beta)
            loss.backward()
            optim.step()
            total_loss += loss.item() * x.size(0)
            total_recon += recon.item() * x.size(0)
            total_kld += kld.item() * x.size(0)
            total_bce += bce.item() * x.size(0)
        n = len(dataloader.dataset)
        print(f"Epoch {epoch}/{epochs} loss={total_loss/n:.6f} recon={total_recon/n:.6f} kld={total_kld/n:.6f} bce={total_bce/n:.6f}")
        torch.save(model.state_dict(), os.path.join(ckpt_dir, f"cvae_epoch{epoch}.pth"))

def train_diffusion(diffusion: SimpleDiffusion, dataloader: DataLoader, optim, device, epochs: int=20, ckpt_dir: str="checkpoints"):
    diffusion.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    diffusion.train()
    for epoch in range(1, epochs + 1):
        total = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            optim.zero_grad()
            loss = diffusion.p_losses(x, c)
            loss.backward()
            optim.step()
            total += loss.item() * x.size(0)
        print(f"Diffusion epoch {epoch}/{epochs} loss={total/len(dataloader.dataset):.6f}")
        torch.save(diffusion.state_dict(), os.path.join(ckpt_dir, f"diff_epoch{epoch}.pth"))

# -------------------------
# RL fine-tuning (policy-gradient / REINFORCE) of the Controller
# -------------------------

def rl_finetune_controller(controller: Controller, generator, device, n_iters: int = 200, batch_size: int = 64, lr: float = 1e-4):
    """controller: policy to finetune (outputs prob of activating)
    generator: callable that returns (x_generated, c) tensors given a batch size
    This will perform policy-gradient updates where the reward is a combination of
    abstract stability and affinity scores. All calculations are on abstract vectors.
    """
    controller.to(device)
    opt = torch.optim.Adam(controller.parameters(), lr=lr)
    for it in range(1, n_iters + 1):
        x_gen, c = generator(batch_size)
        x_gen = x_gen.to(device)
        c = c.to(device)
        probs = controller(x_gen, c)  # values in [0,1]
        m = torch.distributions.Bernoulli(probs)
        actions = m.sample()
        logp = m.log_prob(actions)
        # compute abstract rewards
        stability, affinity = abstract_scores(x_gen, c)
        # reward shaped: prefer high stability and moderate affinity (example)
        reward = 0.7 * stability + 0.3 * affinity
        # encourage controller to activate (action==1) when reward high
        advantage = reward - reward.mean()
        # policy gradient loss (maximize expected reward -> minimize -logp * advantage)
        loss = -(logp * advantage.detach()).mean()
        opt.zero_grad()
        loss.backward()
        opt.step()
        if it % 20 == 0:
            print(f"RL iter {it}/{n_iters} loss={loss.item():.6f} mean_reward={reward.mean().item():.4f}")
    return controller

# -------------------------
# Generator helpers used by RL: sample from cVAE or Diffusion
# -------------------------
def generator_from_cvae(model: CVAEModel, device):
    def gen(n: int):
        model.to(device)
        model.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            z = torch.randn(n, model.encoder.mu.out_features, device=device)
            x = model.decoder(z, c)
            return x, c
    return gen

def generator_from_diffusion(diffusion: SimpleDiffusion, device):
    def gen(n: int):
        diffusion.to(device)
        diffusion.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            x = diffusion.sample(c, device, n_samples=n)
            return x, c
    return gen

# -------------------------
# Visualization and sampling utilities
# -------------------------

def plot_samples(xs: np.ndarray, ctx: np.ndarray, acts: np.ndarray, n_plot: int = 5):
    n = min(n_plot, xs.shape[0])
    fig, axes = plt.subplots(n, 1, figsize=(6, 2 * n))
    if n == 1:
        axes = [axes]
    for i in range(n):
        ax = axes[i]
        ax.bar(range(len(FEATURE_NAMES)), xs[i], tick_label=FEATURE_NAMES)
        ax.set_ylim(0, 1)
        ax.set_title(f"Context: {np.round(ctx[i],3)}  act={acts[i]:.3f}")
    plt.tight_layout()
    plt.show()

# -------------------------
# FastAPI server (optional)
# -------------------------

def create_api(model_cvae: Optional[CVAEModel], diffusion: Optional[SimpleDiffusion]):
    if not FASTAPI_AVAILABLE:
        raise RuntimeError("FastAPI not installed")
    app = FastAPI()

    @app.get('/health')
    def health():
        return {'status': 'ok'}

    @app.post('/sample_cvae')
    def sample_cvae(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_cvae(model_cvae, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    @app.post('/sample_diffusion')
    def sample_diffusion(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_diffusion(diffusion, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    return app

# -------------------------
# CLI and main
# -------------------------

def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument('--train', action='store_true')
    p.add_argument('--train-diffusion', action='store_true')
    p.add_argument('--sample', action='store_true')
    p.add_argument('--sample-diffusion', action='store_true')
    p.add_argument('--rl-finetune', action='store_true')
    p.add_argument('--serve', action='store_true')
    p.add_argument('--n-samples', type=int, default=8)
    p.add_argument('--epochs', type=int, default=20)
    p.add_argument('--batch-size', type=int, default=128)
    p.add_argument('--lr', type=float, default=1e-3)
    p.add_argument('--hidden-dim', type=int, default=128)
    p.add_argument('--z-dim', type=int, default=16)
    p.add_argument('--beta', type=float, default=1.0)
    p.add_argument('--seed', type=int, default=42)
    return p.parse_args([]) # Pass an empty list to ignore kernel arguments


def main():
    args = parse_args()
    torch.manual_seed(args.seed)
    np.random.seed(args.seed)
    random.seed(args.seed)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=args.hidden_dim, z_dim=args.z_dim)
    diffusion = SimpleDiffusion(x_dim=NDIMS, c_dim=NCONTEXT, hidden=args.hidden_dim, timesteps=50)

    if args.train:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(model.parameters(), lr=args.lr)
        train_cvae(model, dl, optim, device, epochs=args.epochs, beta=args.beta)
        torch.save(model.state_dict(), 'checkpoints/cvae_final.pth')
        print('Saved cVAE to checkpoints/cvae_final.pth')

    if args.train_diffusion:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(diffusion.parameters(), lr=args.lr)
        train_diffusion(diffusion, dl, optim, device, epochs=args.epochs)
        torch.save(diffusion.state_dict(), 'checkpoints/diffusion_final.pth')
        print('Saved diffusion to checkpoints/diffusion_final.pth')

    if args.rl_finetune:
        # choose generator (diffusion if trained else cVAE)
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
            gen = generator_from_diffusion(diffusion, device)
        elif os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
            gen = generator_from_cvae(model, device)
        else:
            # quick fallback: create a small dataset-based generator
            dataset = VirtualCellDataset(2000, seed=args.seed)
            dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
            def gen_fallback(n):
                batch = next(iter(dl))
                return batch['x'], batch['c']
            gen = gen_fallback
        # finetune controller (policy)
        controller = model.controller
        controller = rl_finetune_controller(controller, gen, device, n_iters=200, batch_size=64, lr=1e-4)
        torch.save(controller.state_dict(), 'checkpoints/controller_rl_finetuned.pth')
        print('Saved finetuned controller to checkpoints/controller_rl_finetuned.pth')

    if args.sample or args.sample_diffusion:
        # load models if available
        if os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
        if args.sample:
            gen = generator_from_cvae(model, device)
        else:
            gen = generator_from_diffusion(diffusion, device)
        x, c = gen(args.n_samples)
        # controller prediction
        act = model.controller(x.to(device), c.to(device)).cpu().detach().numpy()
        plot_samples(x.cpu().numpy(), c.cpu().numpy(), act, n_plot=min(8, args.n_samples))
        # save
        rows = []
        for i in range(x.size(0)):
            d = {k: float(x[i, j].cpu().item()) for j, k in enumerate(FEATURE_NAMES)}
            for j in range(c.size(1)):
                d[f'ctx_{j}'] = float(c[i, j].cpu().item())
            d['act_pred'] = float(act[i])
            rows.append(d)
        df = pd.DataFrame(rows)
        os.makedirs('samples', exist_ok=True)
        df.to_csv('samples/generated_samples.csv', index=False)
        print('Saved samples to samples/generated_samples.csv')

    if args.serve:
        if not FASTAPI_AVAILABLE:
            print('FastAPI/uvicorn not installed. Install fastapi and uvicorn to use the serve feature.')
        else:
            # load weights if present
            if os.path.exists('checkpoints/cvae_final.pth'):
                model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location='cpu'))
            if os.path.exists('checkpoints/diffusion_final.pth'):
                diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location='cpu'))
            app = create_api(model, diffusion)
            print('Starting server on http://0.0.0.0:8000')
            uvicorn.run(app, host='0.0.0.0', port=8000)

if __name__ == '__main__':
    main()

# -------------------------
# Digital Immortal Cell Simulation (abstract, safe)
# -------------------------
class DigitalImmortalCell:
    """A purely in-silico 'digital immortal cell' simulator.

    - Models telomere length as a continuous variable in [0,1].
    - Simulates cell cycles with probabilistic division, DNA damage accumulation,
      repair processes, and an AI controller that emulates telomerase activation
      decisions based on abstract state and context.
    - Implements safety checks to prevent runaway unchecked proliferation by
      simulating "genomic instability" and triggering virtual apoptosis.

    This class operates on abstract numerical features only. No biological
    sequences, wet-lab protocols, or real-world actionable steps are produced.
    """

    def __init__(self,
                 initial_telomere: float = 0.9,
                 autophagy_eff: float = 0.7,
                 repair_eff: float = 0.8,
                 mutation_rate: float = 0.01,
                 division_base_prob: float = 0.3,
                 seed: int = 42):
        self.rng = np.random.RandomState(seed)
        # abstract state
        self.telomere = float(np.clip(initial_telomere, 0.0, 1.0))
        self.autophagy = float(np.clip(autophagy_eff, 0.0, 1.0))
        self.repair_eff = float(np.clip(repair_eff, 0.0, 1.0))
        self.mutation_rate = float(np.clip(mutation_rate, 0.0, 1.0))
        self.division_base_prob = float(np.clip(division_base_prob, 0.0, 1.0))
        # metadata
        self.age = 0
        self.alive = True
        self.history = []

    def step(self, context: Optional[np.ndarray], controller_callable) -> Dict:
        """One simulation timestep (one potential cell cycle).

        context: array-like of context features (nutrient, shear, signal)
        controller_callable: function(state_tensor, context_tensor) -> prob[T]
            it should return a telomerase activation probability in [0,1]

        Returns a dict with new state and flags.
        """
        if not self.alive:
            return {"alive": False}

        # environmental/context influence
        if context is None:
            nutrient = float(self.rng.beta(5, 2))
            shear = float(self.rng.beta(2, 5))
            signal = float(self.rng.beta(3, 3))
            context = np.array([nutrient, shear, signal], dtype=np.float32)
        else:
            context = np.array(context, dtype=np.float32)

        # form state vector
        state = np.array([
            self.telomere,
            min(1.0, 0.5 + 0.5 * (self.age / 100.0)),  # abstract age_marker increasing weakly
            float(self.rng.beta(2, 3)),  # oxidative proxy sample
            float(np.clip(0.3 + 0.6 * context[0] - 0.2 * (self.age / 100.0), 0.0, 1.0)),  # metabolic
            float(np.clip(0.2 + 0.4 * (1 - self.repair_eff) + self.rng.normal(0, 0.05), 0.0, 1.0)),  # dna damage
            self.autophagy,
            float(self.division_base_prob),
        ], dtype=np.float32)

        # controller decision (telomerase activation probability)
        with torch.no_grad():
            inp_state = torch.from_numpy(state).unsqueeze(0).float()
            inp_context = torch.from_numpy(context).unsqueeze(0).float()
            prob = 0.0
            try:
                p = controller_callable(inp_state, inp_context)
                # controller might return tensor or numpy
                if isinstance(p, torch.Tensor):
                    prob = float(p.squeeze().cpu().numpy())
                else:
                    prob = float(np.array(p).squeeze())
            except Exception:
                # fallback heuristic if controller not available
                prob = float(0.2 * (1.0 - self.telomere) + 0.1 * (1.0 - state[4]))
                prob = float(np.clip(prob, 0.0, 1.0))

        # sample activation
        activate = self.rng.rand() < prob

        # telomere update: activation increases telomere by synthetic amount, otherwise natural shortening
        if activate:
            self.telomere = float(np.clip(self.telomere + 0.06 * self.repair_eff, 0.0, 1.0))
        else:
            # natural erosion influenced by oxidative stress and age
            erosion = 0.02 + 0.05 * state[2] + 0.002 * (self.age / 10.0)
            self.telomere = float(np.clip(self.telomere - erosion, 0.0, 1.0))

        # DNA damage accumulation and repair
        dna_damage = state[4]
        repair = self.repair_eff * (0.5 + 0.5 * self.autophagy)
        dna_damage = float(np.clip(dna_damage - repair * 0.1 + self.rng.normal(0, 0.02), 0.0, 1.0))

        # mutation/genomic instability modeled as rising probability with damage and short telomeres
        instability_prob = float(np.clip(0.02 + 0.4 * dna_damage + 0.4 * (1.0 - self.telomere), 0.0, 1.0))
        mutated = self.rng.rand() < instability_prob * self.mutation_rate

        # division decision
        div_prob = float(np.clip(self.division_base_prob + 0.3 * context[2] - 0.5 * dna_damage, 0.0, 1.0))
        divided = self.rng.rand() < div_prob

        # safety: if instability too high, trigger apoptosis (virtual)
        if instability_prob > 0.85 and dna_damage > 0.8:
            self.alive = False
            reason = 'apoptosis_due_to_instability'
        elif self.telomere <= 0.02 and dna_damage > 0.7:
            # senescence-like virtual endpoint (not immortal)
            self.alive = False
            reason = 'senescence'
        else:
            reason = 'alive'

        # age increment
        self.age += 1

        # record
        record = {
            'age': self.age,
            'telomere': self.telomere,
            'dna_damage': dna_damage,
            'instability_prob': instability_prob,
            'mutated': bool(mutated),
            'divided': bool(divided),
            'activated_telomerase': bool(activate),
            'alive': bool(self.alive),
            'reason': reason,
            'context': context.tolist(),
        }
        self.history.append(record)
        return record

    def run(self, controller_callable, steps: int = 1000, context_schedule=None):
        """Run simulation for given steps. context_schedule can be:
        - None: random contexts each step
        - callable(step)->context array
        - fixed array of length steps
        """
        self.history = []
        for t in range(steps):
            if callable(context_schedule):
                ctx = context_schedule(t)
            elif isinstance(context_schedule, (list, np.ndarray)):
                ctx = context_schedule[t] if t < len(context_schedule) else None
            else:
                ctx = None
            rec = self.step(ctx, controller_callable)
            if not rec['alive']:
                break
        return self.history

# -------------------------
# CLI helper to run a demo digital immortal cell simulation
# -------------------------

def demo_immortal_cell(steps: int = 500):
    device = torch.device('cpu')
    # create a default model.controller (loaded if checkpoints exist)
    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=128, z_dim=16)
    if os.path.exists('checkpoints/controller_rl_finetuned.pth'):
        model.controller.load_state_dict(torch.load('checkpoints/controller_rl_finetuned.pth', map_location=device))
    elif os.path.exists('checkpoints/cvae_final.pth'):
        model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))

    controller = model.controller

    cell = DigitalImmortalCell(initial_telomere=0.95, autophagy_eff=0.8, repair_eff=0.85, mutation_rate=0.002)

    def ctrl_callable(s_tensor, c_tensor):
        # simple wrapper to call controller in eval mode
        controller.eval()
        with torch.no_grad():
            return controller(s_tensor, c_tensor)

    history = cell.run(ctrl_callable, steps=steps)
    # convert history to DataFrame for inspection
    df = pd.DataFrame(history)
    return df

# -------------------------
# If invoked with a special flag, run demo and save results
# -------------------------

def run_demo_and_save(steps: int = 500):
    df = demo_immortal_cell(steps=steps)
    os.makedirs('simulations', exist_ok=True)
    path = os.path.join('simulations', 'digital_immortal_cell_demo.csv')
    df.to_csv(path, index=False)
    print(f"Saved simulation to {path}. Steps simulated: {len(df)}. Final state alive={df.iloc[-1]['alive']}")

# -------------------------
# Advanced RL: PPO-style training of Controller within DigitalImmortalCell environment
# -------------------------

class PPOMemory:
    def __init__(self):
        self.states = []
        self.contexts = []
        self.actions = []
        self.logprobs = []
        self.rewards = []
        self.is_terminals = []

    def clear(self):
        self.__init__()


class PPOController(nn.Module):
    """Actor-Critic style network for PPO. Policy outputs Bernoulli probability to 'activate'.
    Critic outputs state-value for advantage estimation."""

    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128):
        super().__init__()
        self.actor = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1), nn.Sigmoid())
        self.critic = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1))

    def forward(self, x, c):
        inp = torch.cat([x, c], dim=-1)
        prob = self.actor(inp).squeeze(-1)
        value = self.critic(inp).squeeze(-1)
        return prob, value

    def act(self, x, c):
        prob, _ = self.forward(x, c)
        dist = torch.distributions.Bernoulli(probs=prob)
        action = dist.sample()
        return action, dist.log_prob(action)


def compute_gae(rewards, values, is_terminals, gamma=0.99, lam=0.95):
    advantages = []
    gae = 0
    values = values + [0]
    for step in reversed(range(len(rewards))):
        delta = rewards[step] + gamma * values[step + 1] * (1 - is_terminals[step]) - values[step]
        gae = delta + gamma * lam * (1 - is_terminals[step]) * gae
        advantages.insert(0, gae)
    return advantages


def train_ppo(controller: PPOController,
              env_constructor,
              device,
              epochs: int = 200,
              steps_per_epoch: int = 2048,
              mini_batch_size: int = 64,
              ppo_epochs: int = 8,
              clip_epsilon: float = 0.2,
              gamma: float = 0.99,
              lam: float = 0.95,
              lr: float = 3e-4):
    """Train a PPO controller interacting with repeated DigitalImmortalCell episodes.

    env_constructor: callable() -> new DigitalImmortalCell instance (fresh env per episode)
    The controller policy is updated to maximize long-term simulated rewards (longevity, low instability)
    while avoiding unsafe behaviors (penalized).
    """
    controller.to(device)
    optimizer = torch.optim.Adam(controller.parameters(), lr=lr)

    def rollout():
        memory = PPOMemory()
        env = env_constructor()
        state = None
        # we'll run until we collect steps_per_epoch transitions
        steps = 0
        while steps < steps_per_epoch:
            # prepare state and context
            # DigitalImmortalCell.step expects controller_callable that accepts tensors, but for sampling we pass policy
            s_vec = np.array([
                env.telomere,
                min(1.0, 0.5 + 0.5 * (env.age / 100.0)),
                float(env.rng.beta(2, 3)),
                float(np.clip(0.3 + 0.6 * env.rng.beta(5, 2) - 0.2 * (env.age / 100.0), 0.0, 1.0)),
                float(np.clip(0.2 + 0.4 * (1 - env.repair_eff) + env.rng.normal(0, 0.05), 0.0, 1.0)),
                env.autophagy,
                float(env.division_base_prob),
            ], dtype=np.float32)
            ctx = np.array([env.rng.beta(5, 2), env.rng.beta(2, 5), env.rng.beta(3, 3)], dtype=np.float32)
            s_t = torch.from_numpy(s_vec).unsqueeze(0).float().to(device)
            c_t = torch.from_numpy(ctx).unsqueeze(0).float().to(device)

            with torch.no_grad():
                prob, value = controller(s_t, c_t)
                dist = torch.distributions.Bernoulli(probs=prob)
                action = dist.sample()
                logp = dist.log_prob(action)
            # apply action via env.step using a thin wrapper
            def ctrl_callable(inner_s, inner_c):
                # inner tensors correspond to state/context; policy expects same shape
                with torch.no_grad():
                    p, _ = controller(inner_s.to(device), inner_c.to(device))
                    return p.cpu()
            rec = env.step(ctx, ctrl_callable)
            # reward design: encourage staying alive, penalize instability and mutations
            reward = 0.0
            if rec['alive']:
                reward += 1.0  # survival bonus
            # small penalty proportional to instability
            reward += -1.0 * rec['instability_prob']
            if rec['mutated']:
                reward += -2.0
            if rec['activated_telomerase']:
                # small cost to activating to avoid trivial always-on policy
                reward += -0.02
            memory.states.append(s_t.squeeze(0).cpu())
            memory.contexts.append(c_t.squeeze(0).cpu())
            memory.actions.append(action.squeeze(0).cpu())
            memory.logprobs.append(logp.squeeze(0).cpu())
            memory.rewards.append(reward)
            memory.is_terminals.append(not rec['alive'])

            steps += 1
            if not rec['alive']:
                # restart env when cell dies
                env = env_constructor()
        return memory

    for epoch in range(1, epochs + 1):
        memory = rollout()
        # convert memory lists to tensors
        states = torch.stack(memory.states).to(device)
        contexts = torch.stack(memory.contexts).to(device)
        actions = torch.stack(memory.actions).to(device)
        old_logprobs = torch.stack(memory.logprobs).to(device)
        rewards = memory.rewards
        is_term = memory.is_terminals

        # compute values for states
        with torch.no_grad():
            _, values = controller(states, contexts)
            values = values.cpu().numpy().tolist()
        # GAE advantages
        advantages = compute_gae(rewards, values, is_term, gamma=gamma, lam=lam)
        advantages = torch.tensor(advantages, dtype=torch.float32).to(device)
        returns = advantages + torch.tensor(values[:-1] if len(values) > len(advantages) else values[:len(advantages)], dtype=torch.float32).to(device)

        # normalize advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO updates
        dataset_size = states.size(0)
        for _ in range(ppo_epochs):
            # minibatch sampling
            idxs = torch.randperm(dataset_size)
            for start in range(0, dataset_size, mini_batch_size):
                idx = idxs[start:start + mini_batch_size]
                batch_states = states[idx]
                batch_contexts = contexts[idx]
                batch_actions = actions[idx]
                batch_oldlogp = old_logprobs[idx]
                batch_adv = advantages[idx]
                batch_returns = returns[idx]

                probs, values = controller(batch_states, batch_contexts)
                dist = torch.distributions.Bernoulli(probs=probs)
                newlogp = dist.log_prob(batch_actions)
                entropy = dist.entropy().mean()

                ratio = torch.exp(newlogp - batch_oldlogp)
                surr1 = ratio * batch_adv
                surr2 = torch.clamp(ratio, 1.0 - clip_epsilon, 1.0 + clip_epsilon) * batch_adv
                policy_loss = -torch.min(surr1, surr2).mean()
                value_loss = F.mse_loss(values, batch_returns)
                loss = policy_loss + 0.5 * value_loss - 0.01 * entropy

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

        # end epoch logging
        print(f"PPO Epoch {epoch}/{epochs} completed. Mean reward per step: {np.mean(memory.rewards):.4f}")
        # optional checkpoint
        torch.save(controller.state_dict(), f'checkpoints/ppo_controller_epoch{epoch}.pth')

    return controller


# CLI: add flag to run PPO training
import inspect
def add_ppo_flag_and_run():
    import sys
    if '--train-ppo' in sys.argv:
        # default simple env constructor
        def env_ctor():
            return DigitalImmortalCell(initial_telomere=0.9, autophagy_eff=0.7, repair_eff=0.8, mutation_rate=0.005)
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        ppo_ctrl = PPOController(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
        trained = train_ppo(ppo_ctrl, env_ctor, device, epochs=50, steps_per_epoch=1024)
        torch.save(trained.state_dict(), 'checkpoints/ppo_final.pth')
        print('Saved PPO controller to checkpoints/ppo_final.pth')

# Try to run if flag present
add_ppo_flag_and_run()

# -------------------------
# Discovery Engine: latent-space and controller search for "discoveries" (abstract-only)
# -------------------------

class DiscoveryEngine:
    """Performs algorithmic discovery in the virtual system by searching latent
    spaces (cVAE/diffusion) and controller parameter spaces using safe,
    purely computational optimization strategies. Outputs are abstract feature
    vectors and controller parameter snapshots - not biological sequences.
    """

    def __init__(self, cvae: Optional[CVAEModel] = None, diffusion: Optional[SimpleDiffusion] = None, device: Optional[torch.device] = None):
        self.cvae = cvae
        self.diffusion = diffusion
        self.device = device or torch.device('cpu')

    def score_features(self, x: np.ndarray, c: np.ndarray) -> Dict[str, float]:
        # wrapper to compute abstract stability and affinity for numpy arrays
        xt = torch.from_numpy(x).float()
        ct = torch.from_numpy(c).float()
        s, a = abstract_scores(xt, ct)
        return {'stability': float(s.mean().item()), 'affinity': float(a.mean().item())}

    def random_latent_search_cvae(self, n_samples: int = 2000, context: Optional[np.ndarray] = None, top_k: int = 20):
        """Randomly samples latent z vectors from standard normal, decodes via cVAE
        and returns the top_k candidates ranked by a composite score (e.g., stability - alpha * affinity).
        """
        assert self.cvae is not None, 'cVAE model required for this search'
        self.cvae.to(self.device)
        self.cvae.eval()
        if context is None:
            context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        else:
            if context.ndim == 1:
                context = np.tile(context[None, :], (n_samples, 1)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        z_dim = self.cvae.encoder.mu.out_features
        zs = torch.randn(n_samples, z_dim, device=self.device)
        with torch.no_grad():
            x_gen = self.cvae.decoder(zs, c_t)
        x_np = x_gen.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        # save
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'cvae_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} cVAE latent discoveries to {path}')
        return df

    def random_latent_search_diffusion(self, n_samples: int = 1000, top_k: int = 20):
        assert self.diffusion is not None, 'Diffusion model required for this search'
        self.diffusion.to(self.device)
        self.diffusion.eval()
        context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        xs = self.diffusion.sample(c_t, self.device, n_samples=n_samples)
        x_np = xs.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'diffusion_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} diffusion latent discoveries to {path}')
        return df

    def evo_search_controller(self, base_controller: nn.Module, env_constructor, generations: int = 50, pop_size: int = 20, sigma: float = 0.02, elite_frac: float = 0.2):
        """Evolutionary strategy to search controller parameter perturbations that maximize
        long-term simulated rewards in the DigitalImmortalCell environment.
        Returns a DataFrame of top parameter sets (serialized as flattened arrays) and their scores.
        """
        # flatten params
        params = np.concatenate([p.detach().cpu().numpy().ravel() for p in base_controller.parameters()])
        n_params = params.size
        population = np.tile(params[None, :], (pop_size, 1)) + np.random.randn(pop_size, n_params) * sigma

        def set_params_from_flat(model, flat):
            idx = 0
            for p in model.parameters():
                num = p.numel()
                arr = flat[idx:idx+num].reshape(p.shape)
                p.data = torch.from_numpy(arr).float()
                idx += num

        scores = []
        for g in range(generations):
            pop_scores = []
            for i in range(pop_size):
                indiv = population[i]
                # evaluate
                model_copy = type(base_controller)(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
                set_params_from_flat(model_copy, indiv)
                # quick rollout: average cumulative reward over few episodes
                total = 0.0
                n_eps = 3
                for e in range(n_eps):
                    env = env_constructor()
                    steps = 0
                    while steps < 200 and env.alive:
                        # prepare current state/context
                        ctx = [env.rng.beta(5,2), env.rng.beta(2,5), env.rng.beta(3,3)]
                        def ctrl_call(s,c):
                            with torch.no_grad():
                                s_t = torch.from_numpy(s).unsqueeze(0).float()
                                c_t = torch.from_numpy(c).unsqueeze(0).float()
                                p, _ = model_copy(s_t, c_t)
                                return p
                        rec = env.step(ctx, ctrl_call)
                        r = 1.0 if rec['alive'] else -5.0
                        r -= rec['instability_prob']
                        if rec['mutated']:
                            r -= 2.0
                        total += r
                        steps += 1
                avg_score = total / n_eps
                pop_scores.append(avg_score)
            pop_scores = np.array(pop_scores)
            # select elites
            elites_idx = pop_scores.argsort()[::-1][:max(1, int(pop_size*elite_frac))]
            elites = population[elites_idx]
            elite_mean = elites.mean(axis=0)
            # generate new population around elites mean
            population = elite_mean[None, :] + np.random.randn(pop_size, n_params) * sigma
            best_score = float(pop_scores.max())
            print(f'Generation {g+1}/{generations} best_score={best_score:.4f}')
            scores.append((g, best_score))
        # save best individual
        best_idx = int(np.argmax(pop_scores))
        best_flat = population[best_idx]
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'controller_evo_best.npy')
        np.save(path, best_flat)
        print(f'Saved best controller flat params to {path}')
        # return simple summary DataFrame
        df = pd.DataFrame([{'generation': g, 'best_score': sc} for g, sc in scores])
        path2 = os.path.join('discoveries', 'controller_evo_scores.csv')
        df.to_csv(path2, index=False)
        print(f'Saved controller evolution scores to {path2}')
        return df

# End of file
```

```python
%%writefile virtual_cell_generative_model.py
"""
virtual_cell_generative_model.py

Expanded version: includes safe, in-silico generative AI enhancements (purely abstract features).
Additions in this file compared to the previous version:
 - A simplified conditional Diffusion model (DDPM-style) for generating abstract virtual cell states
 - An RL-style fine-tuning loop (policy-gradient / REINFORCE) that treats the Controller as a policy and
   optimizes it for an abstract numerical reward ("stability_score" and "affinity_score") computed from
   generated feature vectors. This is *not* biological sequence design — it operates on abstract vectors.
 - An optional FastAPI server (if FastAPI is installed) exposing endpoints for sampling and explainability
 - Additional utilities and CLI flags to run diffusion training, RL fine-tuning, and start the API.

Safety: This file only manipulates numerical vectors and abstract scores. It contains no nucleotide
sequences, no wet-lab protocols, and no actionable biological design instructions.

Requirements (additional):
 - fastapi, uvicorn (optional, only for serving)
 - torch
 - numpy, pandas, matplotlib
 - optuna (optional)

Usage examples:
  python virtual_cell_generative_model.py --train (trains cVAE)
  python virtual_cell_generative_model.py --train-diffusion (trains diffusion on same synthetic data)
  python virtual_cell_generative_model.py --rl-finetune (runs RL fine-tuning of controller)
  python virtual_cell_generative_model.py --serve  (starts FastAPI server if fastapi installed)

"""

import argparse
import os
import random
import math
from dataclasses import dataclass
from typing import Tuple, Dict, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Optional extras
try:
    import optuna
    OPTUNA_AVAILABLE = True
except Exception:
    OPTUNA_AVAILABLE = False

try:
    from fastapi import FastAPI
    import uvicorn
    FASTAPI_AVAILABLE = True
except Exception:
    FASTAPI_AVAILABLE = False

# -------------------------
# Feature definitions (abstract only)
# -------------------------
FEATURE_NAMES = [
    "telomere_len",
    "age_marker",
    "oxidative_stress",
    "metabolic_rate",
    "dna_damage",
    "autophagy_eff",
    "division_prob",
]
CONTEXT_NAMES = ["nutrient_level", "shear_stress", "signal_factor"]
NDIMS = len(FEATURE_NAMES)
NCONTEXT = len(CONTEXT_NAMES)

# -------------------------
# Synthetic dataset (unchanged, safe)
# -------------------------
class VirtualCellDataset(Dataset):
    def __init__(self, n_samples: int = 10000, seed: int = 42):
        super().__init__()
        self.n_samples = n_samples
        rng = np.random.RandomState(seed)
        self.data = []
        for _ in range(n_samples):
            nutrient = rng.beta(5, 2)
            shear = rng.beta(2, 5)
            signal = rng.beta(3, 3)

            age_marker = rng.beta(2, 5)
            oxidative = rng.beta(2, 3)
            dna_damage = np.minimum(1.0, oxidative * 0.6 + rng.normal(0, 0.05))

            telomere_len = np.clip(0.9 - 0.5 * age_marker - 0.4 * oxidative + rng.normal(0, 0.03), 0.0, 1.0)
            metabolic = np.clip(0.3 + 0.6 * nutrient - 0.2 * age_marker + rng.normal(0, 0.05), 0.0, 1.0)
            autophagy = np.clip(0.7 * nutrient - 0.3 * oxidative + rng.normal(0, 0.05), 0.0, 1.0)

            division_prob = np.clip(0.4 * signal + 0.3 * metabolic - 0.5 * dna_damage + rng.normal(0, 0.05), 0.0, 1.0)

            features = np.array([
                telomere_len,
                age_marker,
                oxidative,
                metabolic,
                dna_damage,
                autophagy,
                division_prob,
            ], dtype=np.float32)

            context = np.array([nutrient, shear, signal], dtype=np.float32)

            thr = 0.45 + 0.15 * (1.0 - nutrient)
            telomerase_on = float((telomere_len < thr) and (dna_damage < 0.6) and (division_prob > 0.15))

            self.data.append((features, context, telomerase_on))

    def __len__(self):
        return self.n_samples

    def __getitem__(self, idx):
        features, context, telomerase_on = self.data[idx]
        return {"x": torch.from_numpy(features), "c": torch.from_numpy(context), "act": torch.tensor(telomerase_on, dtype=torch.float32)}

# -------------------------
# cVAE model (unchanged)
# -------------------------
class Encoder(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU())
        self.mu = nn.Linear(hidden_dim, z_dim)
        self.logvar = nn.Linear(hidden_dim, z_dim)
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        h = self.net(h)
        return self.mu(h), self.logvar(h)

class Decoder(nn.Module):
    def __init__(self, z_dim: int, c_dim: int, hidden_dim: int, x_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(z_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, x_dim), nn.Sigmoid())
    def forward(self, z, c):
        h = torch.cat([z, c], dim=-1)
        return self.net(h)

class Controller(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(x_dim + c_dim, hidden_dim), nn.ReLU(), nn.Linear(hidden_dim, hidden_dim // 2), nn.ReLU(), nn.Linear(hidden_dim // 2, 1))
    def forward(self, x, c):
        h = torch.cat([x, c], dim=-1)
        return torch.sigmoid(self.net(h)).squeeze(-1)

class CVAEModel(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden_dim: int, z_dim: int):
        super().__init__()
        self.encoder = Encoder(x_dim, c_dim, hidden_dim, z_dim)
        self.decoder = Decoder(z_dim, c_dim, hidden_dim, x_dim)
        self.controller = Controller(x_dim, c_dim, hidden_dim)
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    def forward(self, x, c):
        mu, logvar = self.encoder(x, c)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z, c)
        act_pred = self.controller(x_recon, c)
        return x_recon, mu, logvar, act_pred

# -------------------------
# Diffusion model (simplified DDPM on abstract features)
# -------------------------
class SimpleMLP(nn.Module):
    def __init__(self, dim_in, dim_out, hidden=128):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(dim_in, hidden), nn.ReLU(), nn.Linear(hidden, hidden), nn.ReLU(), nn.Linear(hidden, dim_out))
    def forward(self, x, t=None, c=None):
        # t and c optionally ignored in this simple model, but acceptance of c keeps it conditional
        if c is not None:
            x = torch.cat([x, c], dim=-1)
        return self.net(x)

class SimpleDiffusion(nn.Module):
    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128, timesteps: int = 50):
        super().__init__()
        self.timesteps = timesteps
        self.model = SimpleMLP(x_dim + c_dim, x_dim, hidden)
        # linear noise schedule
        betas = torch.linspace(1e-4, 0.02, timesteps)
        alphas = 1.0 - betas
        alphas_cum = torch.cumprod(alphas, dim=0)
        self.register_buffer('betas', betas)
        self.register_buffer('alphas_cum', alphas_cum)

    def q_sample(self, x_start, t, noise=None):
        if noise is None:
            noise = torch.randn_like(x_start)
        a = self.alphas_cum[t].view(-1, 1)
        return torch.sqrt(a) * x_start + torch.sqrt(1 - a) * noise

    def p_losses(self, x_start, c):
        batch_size = x_start.size(0)
        t = torch.randint(0, self.timesteps, (batch_size,), device=x_start.device)
        noise = torch.randn_like(x_start)
        x_noisy = self.q_sample(x_start, t, noise=noise)
        pred_noise = self.model(torch.cat([x_noisy, c], dim=-1))
        return F.mse_loss(pred_noise, noise)

    def sample(self, c, device, n_samples=8):
        x = torch.randn(n_samples, NDIMS, device=device)
        for t in reversed(range(self.timesteps)):
            pred_noise = self.model(torch.cat([x, c], dim=-1))
            beta = self.betas[t]
            x = (x - beta * pred_noise) / torch.sqrt(1 - beta)
            if t > 0:
                x = x + torch.sqrt(beta) * torch.randn_like(x)
        # clamp to [0,1]
        return torch.clamp(x, 0.0, 1.0)

# -------------------------
# Losses
# -------------------------
def loss_function(recon_x, x, mu, logvar, act_pred, act_true, beta=1.0):
    recon_loss = F.mse_loss(recon_x, x, reduction="mean")
    kld = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
    bce = F.binary_cross_entropy(act_pred, act_true)
    return recon_loss + beta * kld + 0.5 * bce, recon_loss, kld, bce

# -------------------------
# Abstract reward function for RL fine-tuning (safe!)
# -------------------------
def abstract_scores(x: torch.Tensor, c: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """Compute two abstract scalar scores from feature vectors.
    - stability_score: higher when telomere_len high, dna_damage low, autophagy high
    - affinity_score: abstract proxy for 'telomerase affinity' combining division_prob and telomere_len
    All calculations are normalized to [0,1] using safe operations.
    """
    tel = x[:, 0]
    dna = x[:, 4]
    aut = x[:, 5]
    divp = x[:, 6]
    # stability: mix
    stability = 0.5 * tel + 0.3 * (1.0 - dna) + 0.2 * aut
    stability = torch.clamp(stability, 0.0, 1.0)
    # affinity: high if telomere short but division prob present (abstract)
    affinity = 0.6 * (1.0 - tel) + 0.4 * divp
    affinity = torch.clamp(affinity, 0.0, 1.0)
    return stability, affinity

# -------------------------
# Training utilities
# -------------------------

def train_cvae(model: CVAEModel, dataloader: DataLoader, optim, device, epochs: int=20, beta: float=1.0, ckpt_dir: str="checkpoints"):
    model.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    model.train()
    for epoch in range(1, epochs + 1):
        total_loss = total_recon = total_kld = total_bce = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            act = batch['act'].to(device)
            optim.zero_grad()
            x_recon, mu, logvar, act_pred = model(x, c)
            loss, recon, kld, bce = loss_function(x_recon, x, mu, logvar, act_pred, act, beta)
            loss.backward()
            optim.step()
            total_loss += loss.item() * x.size(0)
            total_recon += recon.item() * x.size(0)
            total_kld += kld.item() * x.size(0)
            total_bce += bce.item() * x.size(0)
        n = len(dataloader.dataset)
        print(f"Epoch {epoch}/{epochs} loss={total_loss/n:.6f} recon={total_recon/n:.6f} kld={total_kld/n:.6f} bce={total_bce/n:.6f}")
        torch.save(model.state_dict(), os.path.join(ckpt_dir, f"cvae_epoch{epoch}.pth"))

def train_diffusion(diffusion: SimpleDiffusion, dataloader: DataLoader, optim, device, epochs: int=20, ckpt_dir: str="checkpoints"):
    diffusion.to(device)
    os.makedirs(ckpt_dir, exist_ok=True)
    diffusion.train()
    for epoch in range(1, epochs + 1):
        total = 0.0
        for batch in dataloader:
            x = batch['x'].to(device)
            c = batch['c'].to(device)
            optim.zero_grad()
            loss = diffusion.p_losses(x, c)
            loss.backward()
            optim.step()
            total += loss.item() * x.size(0)
        print(f"Diffusion epoch {epoch}/{epochs} loss={total/len(dataloader.dataset):.6f}")
        torch.save(diffusion.state_dict(), os.path.join(ckpt_dir, f"diff_epoch{epoch}.pth"))

# -------------------------
# RL fine-tuning (policy-gradient / REINFORCE) of the Controller
# -------------------------

def rl_finetune_controller(controller: Controller, generator, device, n_iters: int = 200, batch_size: int = 64, lr: float = 1e-4):
    """controller: policy to finetune (outputs prob of activating)
    generator: callable that returns (x_generated, c) tensors given a batch size
    This will perform policy-gradient updates where the reward is a combination of
    abstract stability and affinity scores. All calculations are on abstract vectors.
    """
    controller.to(device)
    opt = torch.optim.Adam(controller.parameters(), lr=lr)
    for it in range(1, n_iters + 1):
        x_gen, c = generator(batch_size)
        x_gen = x_gen.to(device)
        c = c.to(device)
        probs = controller(x_gen, c)  # values in [0,1]
        m = torch.distributions.Bernoulli(probs)
        actions = m.sample()
        logp = m.log_prob(actions)
        # compute abstract rewards
        stability, affinity = abstract_scores(x_gen, c)
        # reward shaped: prefer high stability and moderate affinity (example)
        reward = 0.7 * stability + 0.3 * affinity
        # encourage controller to activate (action==1) when reward high
        advantage = reward - reward.mean()
        # policy gradient loss (maximize expected reward -> minimize -logp * advantage)
        loss = -(logp * advantage.detach()).mean()
        opt.zero_grad()
        loss.backward()
        opt.step()
        if it % 20 == 0:
            print(f"RL iter {it}/{n_iters} loss={loss.item():.6f} mean_reward={reward.mean().item():.4f}")
    return controller

# -------------------------
# Generator helpers used by RL: sample from cVAE or Diffusion
# -------------------------
def generator_from_cvae(model: CVAEModel, device):
    def gen(n: int):
        model.to(device)
        model.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            z = torch.randn(n, model.encoder.mu.out_features, device=device)
            x = model.decoder(z, c)
            return x, c
    return gen

def generator_from_diffusion(diffusion: SimpleDiffusion, device):
    def gen(n: int):
        diffusion.to(device)
        diffusion.eval()
        with torch.no_grad():
            c_np = np.random.beta(2, 2, size=(n, NCONTEXT)).astype(np.float32)
            c = torch.from_numpy(c_np).to(device)
            x = diffusion.sample(c, device, n_samples=n)
            return x, c
    return gen

# -------------------------
# Visualization and sampling utilities
# -------------------------

def plot_samples(xs: np.ndarray, ctx: np.ndarray, acts: np.ndarray, n_plot: int = 5):
    n = min(n_plot, xs.shape[0])
    fig, axes = plt.subplots(n, 1, figsize=(6, 2 * n))
    if n == 1:
        axes = [axes]
    for i in range(n):
        ax = axes[i]
        ax.bar(range(len(FEATURE_NAMES)), xs[i], tick_label=FEATURE_NAMES)
        ax.set_ylim(0, 1)
        ax.set_title(f"Context: {np.round(ctx[i],3)}  act={acts[i]:.3f}")
    plt.tight_layout()
    plt.show()

# -------------------------
# FastAPI server (optional)
# -------------------------

def create_api(model_cvae: Optional[CVAEModel], diffusion: Optional[SimpleDiffusion]):
    if not FASTAPI_AVAILABLE:
        raise RuntimeError("FastAPI not installed")
    app = FastAPI()

    @app.get('/health')
    def health():
        return {'status': 'ok'}

    @app.post('/sample_cvae')
    def sample_cvae(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_cvae(model_cvae, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    @app.post('/sample_diffusion')
    def sample_diffusion(n: int = 4):
        device = torch.device('cpu')
        gen = generator_from_diffusion(diffusion, device)
        x, c = gen(n)
        x_np = x.cpu().numpy()
        c_np = c.cpu().numpy()
        stability, affinity = abstract_scores(x, c)
        return {'features': x_np.tolist(), 'context': c_np.tolist(), 'stability': stability.cpu().numpy().tolist(), 'affinity': affinity.cpu().numpy().tolist()}

    return app

# -------------------------
# CLI and main
# -------------------------

def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument('--train', action='store_true')
    p.add_argument('--train-diffusion', action='store_true')
    p.add_argument('--sample', action='store_true')
    p.add_argument('--sample-diffusion', action='store_true')
    p.add_argument('--rl-finetune', action='store_true')
    p.add_argument('--serve', action='store_true')
    p.add_argument('--n-samples', type=int, default=8)
    p.add_argument('--epochs', type=int, default=20)
    p.add_argument('--batch-size', type=int, default=128)
    p.add_argument('--lr', type=float, default=1e-3)
    p.add_argument('--hidden-dim', type=int, default=128)
    p.add_argument('--z-dim', type=int, default=16)
    p.add_argument('--beta', type=float, default=1.0)
    p.add_argument('--seed', type=int, default=42)
    return p.parse_args([])


def main():
    args = parse_args()
    torch.manual_seed(args.seed)
    np.random.seed(args.seed)
    random.seed(args.seed)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=args.hidden_dim, z_dim=args.z_dim)
    diffusion = SimpleDiffusion(x_dim=NDIMS, c_dim=NCONTEXT, hidden=args.hidden_dim, timesteps=50)

    if args.train:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(model.parameters(), lr=args.lr)
        train_cvae(model, dl, optim, device, epochs=args.epochs, beta=args.beta)
        torch.save(model.state_dict(), 'checkpoints/cvae_final.pth')
        print('Saved cVAE to checkpoints/cvae_final.pth')

    if args.train_diffusion:
        dataset = VirtualCellDataset(8000, seed=args.seed)
        dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
        optim = torch.optim.Adam(diffusion.parameters(), lr=args.lr)
        train_diffusion(diffusion, dl, optim, device, epochs=args.epochs)
        torch.save(diffusion.state_dict(), 'checkpoints/diffusion_final.pth')
        print('Saved diffusion to checkpoints/diffusion_final.pth')

    if args.rl_finetune:
        # choose generator (diffusion if trained else cVAE)
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
            gen = generator_from_diffusion(diffusion, device)
        elif os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
            gen = generator_from_cvae(model, device)
        else:
            # quick fallback: create a small dataset-based generator
            dataset = VirtualCellDataset(2000, seed=args.seed)
            dl = DataLoader(dataset, batch_size=args.batch_size, shuffle=True)
            def gen_fallback(n):
                batch = next(iter(dl))
                return batch['x'], batch['c']
            gen = gen_fallback
        # finetune controller (policy)
        controller = model.controller
        controller = rl_finetune_controller(controller, gen, device, n_iters=200, batch_size=64, lr=1e-4)
        torch.save(controller.state_dict(), 'checkpoints/controller_rl_finetuned.pth')
        print('Saved finetuned controller to checkpoints/controller_rl_finetuned.pth')

    if args.sample or args.sample_diffusion:
        # load models if available
        if os.path.exists('checkpoints/cvae_final.pth'):
            model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))
        if os.path.exists('checkpoints/diffusion_final.pth'):
            diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location=device))
        if args.sample:
            gen = generator_from_cvae(model, device)
        else:
            gen = generator_from_diffusion(diffusion, device)
        x, c = gen(args.n_samples)
        # controller prediction
        act = model.controller(x.to(device), c.to(device)).cpu().detach().numpy()
        plot_samples(x.cpu().numpy(), c.cpu().numpy(), act, n_plot=min(8, args.n_samples))
        # save
        rows = []
        for i in range(x.size(0)):
            d = {k: float(x[i, j].cpu().item()) for j, k in enumerate(FEATURE_NAMES)}
            for j in range(c.size(1)):
                d[f'ctx_{j}'] = float(c[i, j].cpu().item())
            d['act_pred'] = float(act[i])
            rows.append(d)
        df = pd.DataFrame(rows)
        os.makedirs('samples', exist_ok=True)
        df.to_csv('samples/generated_samples.csv', index=False)
        print('Saved samples to samples/generated_samples.csv')

    if args.serve:
        if not FASTAPI_AVAILABLE:
            print('FastAPI/uvicorn not installed. Install fastapi and uvicorn to use the serve feature.')
        else:
            # load weights if present
            if os.path.exists('checkpoints/cvae_final.pth'):
                model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location='cpu'))
            if os.path.exists('checkpoints/diffusion_final.pth'):
                diffusion.load_state_dict(torch.load('checkpoints/diffusion_final.pth', map_location='cpu'))
            app = create_api(model, diffusion)
            print('Starting server on http://0.0.0.0:8000')
            uvicorn.run(app, host='0.0.0.0', port=8000)

if __name__ == '__main__':
    main()

# -------------------------
# Digital Immortal Cell Simulation (abstract, safe)
# -------------------------
class DigitalImmortalCell:
    """A purely in-silico 'digital immortal cell' simulator.

    - Models telomere length as a continuous variable in [0,1].
    - Simulates cell cycles with probabilistic division, DNA damage accumulation,
      repair processes, and an AI controller that emulates telomerase activation
      decisions based on abstract state and context.
    - Implements safety checks to prevent runaway unchecked proliferation by
      simulating "genomic instability" and triggering virtual apoptosis.

    This class operates on abstract numerical features only. No biological
    sequences, wet-lab protocols, or real-world actionable steps are produced.
    """

    def __init__(self,
                 initial_telomere: float = 0.9,
                 autophagy_eff: float = 0.7,
                 repair_eff: float = 0.8,
                 mutation_rate: float = 0.01,
                 division_base_prob: float = 0.3,
                 seed: int = 42):
        self.rng = np.random.RandomState(seed)
        # abstract state
        self.telomere = float(np.clip(initial_telomere, 0.0, 1.0))
        self.autophagy = float(np.clip(autophagy_eff, 0.0, 1.0))
        self.repair_eff = float(np.clip(repair_eff, 0.0, 1.0))
        self.mutation_rate = float(np.clip(mutation_rate, 0.0, 1.0))
        self.division_base_prob = float(np.clip(division_base_prob, 0.0, 1.0))
        # metadata
        self.age = 0
        self.alive = True
        self.history = []

    def step(self, context: Optional[np.ndarray], controller_callable) -> Dict:
        """One simulation timestep (one potential cell cycle).

        context: array-like of context features (nutrient, shear, signal)
        controller_callable: function(state_tensor, context_tensor) -> prob[T]
            it should return a telomerase activation probability in [0,1]

        Returns a dict with new state and flags.
        """
        if not self.alive:
            return {"alive": False}

        # environmental/context influence
        if context is None:
            nutrient = float(self.rng.beta(5, 2))
            shear = float(self.rng.beta(2, 5))
            signal = float(self.rng.beta(3, 3))
            context = np.array([nutrient, shear, signal], dtype=np.float32)
        else:
            context = np.array(context, dtype=np.float32)

        # form state vector
        state = np.array([
            self.telomere,
            min(1.0, 0.5 + 0.5 * (self.age / 100.0)),  # abstract age_marker increasing weakly
            float(self.rng.beta(2, 3)),  # oxidative proxy sample
            float(np.clip(0.3 + 0.6 * context[0] - 0.2 * (self.age / 100.0), 0.0, 1.0)),  # metabolic
            float(np.clip(0.2 + 0.4 * (1 - self.repair_eff) + self.rng.normal(0, 0.05), 0.0, 1.0)),  # dna damage
            self.autophagy,
            float(self.division_base_prob),
        ], dtype=np.float32)

        # controller decision (telomerase activation probability)
        with torch.no_grad():
            inp_state = torch.from_numpy(state).unsqueeze(0).float()
            inp_context = torch.from_numpy(context).unsqueeze(0).float()
            prob = 0.0
            try:
                p = controller_callable(inp_state, inp_context)
                # controller might return tensor or numpy
                if isinstance(p, torch.Tensor):
                    prob = float(p.squeeze().cpu().numpy())
                else:
                    prob = float(np.array(p).squeeze())
            except Exception:
                # fallback heuristic if controller not available
                prob = float(0.2 * (1.0 - self.telomere) + 0.1 * (1.0 - state[4]))
                prob = float(np.clip(prob, 0.0, 1.0))

        # sample activation
        activate = self.rng.rand() < prob

        # telomere update: activation increases telomere by synthetic amount, otherwise natural shortening
        if activate:
            self.telomere = float(np.clip(self.telomere + 0.06 * self.repair_eff, 0.0, 1.0))
        else:
            # natural erosion influenced by oxidative stress and age
            erosion = 0.02 + 0.05 * state[2] + 0.002 * (self.age / 10.0)
            self.telomere = float(np.clip(self.telomere - erosion, 0.0, 1.0))

        # DNA damage accumulation and repair
        dna_damage = state[4]
        repair = self.repair_eff * (0.5 + 0.5 * self.autophagy)
        dna_damage = float(np.clip(dna_damage - repair * 0.1 + self.rng.normal(0, 0.02), 0.0, 1.0))

        # mutation/genomic instability modeled as rising probability with damage and short telomeres
        instability_prob = float(np.clip(0.02 + 0.4 * dna_damage + 0.4 * (1.0 - self.telomere), 0.0, 1.0))
        mutated = self.rng.rand() < instability_prob * self.mutation_rate

        # division decision
        div_prob = float(np.clip(self.division_base_prob + 0.3 * context[2] - 0.5 * dna_damage, 0.0, 1.0))
        divided = self.rng.rand() < div_prob

        # safety: if instability too high, trigger apoptosis (virtual)
        if instability_prob > 0.85 and dna_damage > 0.8:
            self.alive = False
            reason = 'apoptosis_due_to_instability'
        elif self.telomere <= 0.02 and dna_damage > 0.7:
            # senescence-like virtual endpoint (not immortal)
            self.alive = False
            reason = 'senescence'
        else:
            reason = 'alive'

        # age increment
        self.age += 1

        # record
        record = {
            'age': self.age,
            'telomere': self.telomere,
            'dna_damage': dna_damage,
            'instability_prob': instability_prob,
            'mutated': bool(mutated),
            'divided': bool(divided),
            'activated_telomerase': bool(activate),
            'alive': bool(self.alive),
            'reason': reason,
            'context': context.tolist(),
        }
        self.history.append(record)
        return record

    def run(self, controller_callable, steps: int = 1000, context_schedule=None):
        """Run simulation for given steps. context_schedule can be:
        - None: random contexts each step
        - callable(step)->context array
        - fixed array of length steps
        """
        self.history = []
        for t in range(steps):
            if callable(context_schedule):
                ctx = context_schedule(t)
            elif isinstance(context_schedule, (list, np.ndarray)):
                ctx = context_schedule[t] if t < len(context_schedule) else None
            else:
                ctx = None
            rec = self.step(ctx, controller_callable)
            if not rec['alive']:
                break
        return self.history

# -------------------------
# CLI helper to run a demo digital immortal cell simulation
# -------------------------

def demo_immortal_cell(steps: int = 500):
    device = torch.device('cpu')
    # create a default model.controller (loaded if checkpoints exist)
    model = CVAEModel(x_dim=NDIMS, c_dim=NCONTEXT, hidden_dim=128, z_dim=16)
    if os.path.exists('checkpoints/controller_rl_finetuned.pth'):
        model.controller.load_state_dict(torch.load('checkpoints/controller_rl_finetuned.pth', map_location=device))
    elif os.path.exists('checkpoints/cvae_final.pth'):
        model.load_state_dict(torch.load('checkpoints/cvae_final.pth', map_location=device))

    controller = model.controller

    cell = DigitalImmortalCell(initial_telomere=0.95, autophagy_eff=0.8, repair_eff=0.85, mutation_rate=0.002)

    def ctrl_callable(s_tensor, c_tensor):
        # simple wrapper to call controller in eval mode
        controller.eval()
        with torch.no_grad():
            return controller(s_tensor, c_tensor)

    history = cell.run(ctrl_callable, steps=steps)
    # convert history to DataFrame for inspection
    df = pd.DataFrame(history)
    return df

# -------------------------
# If invoked with a special flag, run demo and save results
# -------------------------

def run_demo_and_save(steps: int = 500):
    df = demo_immortal_cell(steps=steps)
    os.makedirs('simulations', exist_ok=True)
    path = os.path.join('simulations', 'digital_immortal_cell_demo.csv')
    df.to_csv(path, index=False)
    print(f"Saved simulation to {path}. Steps simulated: {len(df)}. Final state alive={df.iloc[-1]['alive']}")

# -------------------------
# Advanced RL: PPO-style training of Controller within DigitalImmortalCell environment
# -------------------------

class PPOMemory:
    def __init__(self):
        self.states = []
        self.contexts = []
        self.actions = []
        self.logprobs = []
        self.rewards = []
        self.is_terminals = []

    def clear(self):
        self.__init__()


class PPOController(nn.Module):
    """Actor-Critic style network for PPO. Policy outputs Bernoulli probability to 'activate'.
    Critic outputs state-value for advantage estimation."""

    def __init__(self, x_dim: int, c_dim: int, hidden: int = 128):
        super().__init__()
        self.actor = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1), nn.Sigmoid())
        self.critic = nn.Sequential(nn.Linear(x_dim + c_dim, hidden), nn.Tanh(), nn.Linear(hidden, hidden), nn.Tanh(), nn.Linear(hidden, 1))

    def forward(self, x, c):
        inp = torch.cat([x, c], dim=-1)
        prob = self.actor(inp).squeeze(-1)
        value = self.critic(inp).squeeze(-1)
        return prob, value

    def act(self, x, c):
        prob, _ = self.forward(x, c)
        dist = torch.distributions.Bernoulli(probs=prob)
        action = dist.sample()
        return action, dist.log_prob(action)


def compute_gae(rewards, values, is_terminals, gamma=0.99, lam=0.95):
    advantages = []
    gae = 0
    values = values + [0]
    for step in reversed(range(len(rewards))):
        delta = rewards[step] + gamma * values[step + 1] * (1 - is_terminals[step]) - values[step]
        gae = delta + gamma * lam * (1 - is_terminals[step]) * gae
        advantages.insert(0, gae)
    return advantages


def train_ppo(controller: PPOController,
              env_constructor,
              device,
              epochs: int = 200,
              steps_per_epoch: int = 2048,
              mini_batch_size: int = 64,
              ppo_epochs: int = 8,
              clip_epsilon: float = 0.2,
              gamma: float = 0.99,
              lam: float = 0.95,
              lr: float = 3e-4):
    """Train a PPO controller interacting with repeated DigitalImmortalCell episodes.

    env_constructor: callable() -> new DigitalImmortalCell instance (fresh env per episode)
    The controller policy is updated to maximize long-term simulated rewards (longevity, low instability)
    while avoiding unsafe behaviors (penalized).
    """
    controller.to(device)
    optimizer = torch.optim.Adam(controller.parameters(), lr=lr)

    def rollout():
        memory = PPOMemory()
        env = env_constructor()
        state = None
        # we'll run until we collect steps_per_epoch transitions
        steps = 0
        while steps < steps_per_epoch:
            # prepare state and context
            # DigitalImmortalCell.step expects controller_callable that accepts tensors, but for sampling we pass policy
            s_vec = np.array([
                env.telomere,
                min(1.0, 0.5 + 0.5 * (env.age / 100.0)),
                float(env.rng.beta(2, 3)),
                float(np.clip(0.3 + 0.6 * env.rng.beta(5, 2) - 0.2 * (env.age / 100.0), 0.0, 1.0)),
                float(np.clip(0.2 + 0.4 * (1 - env.repair_eff) + env.rng.normal(0, 0.05), 0.0, 1.0)),
                env.autophagy,
                float(env.division_base_prob),
            ], dtype=np.float32)
            ctx = np.array([env.rng.beta(5, 2), env.rng.beta(2, 5), env.rng.beta(3, 3)], dtype=np.float32)
            s_t = torch.from_numpy(s_vec).unsqueeze(0).float().to(device)
            c_t = torch.from_numpy(ctx).unsqueeze(0).float().to(device)

            with torch.no_grad():
                prob, value = controller(s_t, c_t)
                dist = torch.distributions.Bernoulli(probs=prob)
                action = dist.sample()
                logp = dist.log_prob(action)
            # apply action via env.step using a thin wrapper
            def ctrl_callable(inner_s, inner_c):
                # inner tensors correspond to state/context; policy expects same shape
                with torch.no_grad():
                    p, _ = controller(inner_s.to(device), inner_c.to(device))
                    return p.cpu()
            rec = env.step(ctx, ctrl_callable)
            # reward design: encourage staying alive, penalize instability and mutations
            reward = 0.0
            if rec['alive']:
                reward += 1.0  # survival bonus
            # small penalty proportional to instability
            reward += -1.0 * rec['instability_prob']
            if rec['mutated']:
                reward += -2.0
            if rec['activated_telomerase']:
                # small cost to activating to avoid trivial always-on policy
                reward += -0.02
            memory.states.append(s_t.squeeze(0).cpu())
            memory.contexts.append(c_t.squeeze(0).cpu())
            memory.actions.append(action.squeeze(0).cpu())
            memory.logprobs.append(logp.squeeze(0).cpu())
            memory.rewards.append(reward)
            memory.is_terminals.append(not rec['alive'])

            steps += 1
            if not rec['alive']:
                # restart env when cell dies
                env = env_constructor()
        return memory

    for epoch in range(1, epochs + 1):
        memory = rollout()
        # convert memory lists to tensors
        states = torch.stack(memory.states).to(device)
        contexts = torch.stack(memory.contexts).to(device)
        actions = torch.stack(memory.actions).to(device)
        old_logprobs = torch.stack(memory.logprobs).to(device)
        rewards = memory.rewards
        is_term = memory.is_terminals

        # compute values for states
        with torch.no_grad():
            _, values = controller(states, contexts)
            values = values.cpu().numpy().tolist()
        # GAE advantages
        advantages = compute_gae(rewards, values, is_term, gamma=gamma, lam=lam)
        advantages = torch.tensor(advantages, dtype=torch.float32).to(device)
        returns = advantages + torch.tensor(values[:-1] if len(values) > len(advantages) else values[:len(advantages)], dtype=torch.float32).to(device)

        # normalize advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO updates
        dataset_size = states.size(0)
        for _ in range(ppo_epochs):
            # minibatch sampling
            idxs = torch.randperm(dataset_size)
            for start in range(0, dataset_size, mini_batch_size):
                idx = idxs[start:start + mini_batch_size]
                batch_states = states[idx]
                batch_contexts = contexts[idx]
                batch_actions = actions[idx]
                batch_oldlogp = old_logprobs[idx]
                batch_adv = advantages[idx]
                batch_returns = returns[idx]

                probs, values = controller(batch_states, batch_contexts)
                dist = torch.distributions.Bernoulli(probs=probs)
                newlogp = dist.log_prob(batch_actions)
                entropy = dist.entropy().mean()

                ratio = torch.exp(newlogp - batch_oldlogp)
                surr1 = ratio * batch_adv
                surr2 = torch.clamp(ratio, 1.0 - clip_epsilon, 1.0 + clip_epsilon) * batch_adv
                policy_loss = -torch.min(surr1, surr2).mean()
                value_loss = F.mse_loss(values, batch_returns)
                loss = policy_loss + 0.5 * value_loss - 0.01 * entropy

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

        # end epoch logging
        print(f"PPO Epoch {epoch}/{epochs} completed. Mean reward per step: {np.mean(memory.rewards):.4f}")
        # optional checkpoint
        torch.save(controller.state_dict(), f'checkpoints/ppo_controller_epoch{epoch}.pth')

    return controller


# CLI: add flag to run PPO training
import inspect
def add_ppo_flag_and_run():
    import sys
    if '--train-ppo' in sys.argv:
        # default simple env constructor
        def env_ctor():
            return DigitalImmortalCell(initial_telomere=0.9, autophagy_eff=0.7, repair_eff=0.8, mutation_rate=0.005)
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        ppo_ctrl = PPOController(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
        trained = train_ppo(ppo_ctrl, env_ctor, device, epochs=50, steps_per_epoch=1024)
        torch.save(trained.state_dict(), 'checkpoints/ppo_final.pth')
        print('Saved PPO controller to checkpoints/ppo_final.pth')

# Try to run if flag present
add_ppo_flag_and_run()

# -------------------------
# Discovery Engine: latent-space and controller search for "discoveries" (abstract-only)
# -------------------------

class DiscoveryEngine:
    """Performs algorithmic discovery in the virtual system by searching latent
    spaces (cVAE/diffusion) and controller parameter spaces using safe,
    purely computational optimization strategies. Outputs are abstract feature
    vectors and controller parameter snapshots — not biological sequences.
    """

    def __init__(self, cvae: Optional[CVAEModel] = None, diffusion: Optional[SimpleDiffusion] = None, device: Optional[torch.device] = None):
        self.cvae = cvae
        self.diffusion = diffusion
        self.device = device or torch.device('cpu')

    def score_features(self, x: np.ndarray, c: np.ndarray) -> Dict[str, float]:
        # wrapper to compute abstract stability and affinity for numpy arrays
        xt = torch.from_numpy(x).float()
        ct = torch.from_numpy(c).float()
        s, a = abstract_scores(xt, ct)
        return {'stability': float(s.mean().item()), 'affinity': float(a.mean().item())}

    def random_latent_search_cvae(self, n_samples: int = 2000, context: Optional[np.ndarray] = None, top_k: int = 20):
        """Randomly samples latent z vectors from standard normal, decodes via cVAE
        and returns the top_k candidates ranked by a composite score (e.g., stability - alpha * affinity).
        """
        assert self.cvae is not None, 'cVAE model required for this search'
        self.cvae.to(self.device)
        self.cvae.eval()
        if context is None:
            context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        else:
            if context.ndim == 1:
                context = np.tile(context[None, :], (n_samples, 1)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        z_dim = self.cvae.encoder.mu.out_features
        zs = torch.randn(n_samples, z_dim, device=self.device)
        with torch.no_grad():
            x_gen = self.cvae.decoder(zs, c_t)
        x_np = x_gen.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        # save
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'cvae_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} cVAE latent discoveries to {path}')
        return df

    def random_latent_search_diffusion(self, n_samples: int = 1000, top_k: int = 20):
        assert self.diffusion is not None, 'Diffusion model required for this search'
        self.diffusion.to(self.device)
        self.diffusion.eval()
        context = np.random.beta(2, 2, size=(n_samples, NCONTEXT)).astype(np.float32)
        c_t = torch.from_numpy(context).to(self.device)
        xs = self.diffusion.sample(c_t, self.device, n_samples=n_samples)
        x_np = xs.cpu().numpy()
        scores = []
        for i in range(n_samples):
            s,a = abstract_scores(torch.from_numpy(x_np[i:i+1]).float(), torch.from_numpy(context[i:i+1]).float())
            composite = float(s.item() - 0.3 * a.item())
            scores.append((composite, float(s.item()), float(a.item()), x_np[i], context[i]))
        scores.sort(key=lambda t: t[0], reverse=True)
        top = scores[:top_k]
        rows = []
        for comp, s, a, xvec, cvec in top:
            row = {f: float(xvec[j]) for j,f in enumerate(FEATURE_NAMES)}
            row.update({f'ctx_{j}': float(cvec[j]) for j in range(len(cvec))})
            row.update({'composite': float(comp), 'stability': float(s), 'affinity': float(a)})
            rows.append(row)
        df = pd.DataFrame(rows)
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'diffusion_latent_topk.csv')
        df.to_csv(path, index=False)
        print(f'Saved top {top_k} diffusion latent discoveries to {path}')
        return df

    def evo_search_controller(self, base_controller: nn.Module, env_constructor, generations: int = 50, pop_size: int = 20, sigma: float = 0.02, elite_frac: float = 0.2):
        """Evolutionary strategy to search controller parameter perturbations that maximize
        long-term simulated rewards in the DigitalImmortalCell environment.
        Returns a DataFrame of top parameter sets (serialized as flattened arrays) and their scores.
        """
        # flatten params
        params = np.concatenate([p.detach().cpu().numpy().ravel() for p in base_controller.parameters()])
        n_params = params.size
        population = np.tile(params[None, :], (pop_size, 1)) + np.random.randn(pop_size, n_params) * sigma

        def set_params_from_flat(model, flat):
            idx = 0
            for p in model.parameters():
                num = p.numel()
                arr = flat[idx:idx+num].reshape(p.shape)
                p.data = torch.from_numpy(arr).float()
                idx += num

        scores = []
        for g in range(generations):
            pop_scores = []
            for i in range(pop_size):
                indiv = population[i]
                # evaluate
                model_copy = type(base_controller)(x_dim=NDIMS, c_dim=NCONTEXT, hidden=128)
                set_params_from_flat(model_copy, indiv)
                # quick rollout: average cumulative reward over few episodes
                total = 0.0
                n_eps = 3
                for e in range(n_eps):
                    env = env_constructor()
                    steps = 0
                    while steps < 200 and env.alive:
                        # prepare current state/context
                        ctx = [env.rng.beta(5,2), env.rng.beta(2,5), env.rng.beta(3,3)]
                        def ctrl_call(s,c):
                            with torch.no_grad():
                                s_t = torch.from_numpy(s).unsqueeze(0).float()
                                c_t = torch.from_numpy(c).unsqueeze(0).float()
                                p, _ = model_copy(s_t, c_t)
                                return p
                        rec = env.step(ctx, ctrl_call)
                        r = 1.0 if rec['alive'] else -5.0
                        r -= rec['instability_prob']
                        if rec['mutated']:
                            r -= 2.0
                        total += r
                        steps += 1
                avg_score = total / n_eps
                pop_scores.append(avg_score)
            pop_scores = np.array(pop_scores)
            # select elites
            elites_idx = pop_scores.argsort()[::-1][:max(1, int(pop_size*elite_frac))]
            elites = population[elites_idx]
            elite_mean = elites.mean(axis=0)
            # generate new population around elites mean
            population = elite_mean[None, :] + np.random.randn(pop_size, n_params) * sigma
            best_score = float(pop_scores.max())
            print(f'Generation {g+1}/{generations} best_score={best_score:.4f}')
            scores.append((g, best_score))
        # save best individual
        best_idx = int(np.argmax(pop_scores))
        best_flat = population[best_idx]
        os.makedirs('discoveries', exist_ok=True)
        path = os.path.join('discoveries', 'controller_evo_best.npy')
        np.save(path, best_flat)
        print(f'Saved best controller flat params to {path}')
        # return simple summary DataFrame
        df = pd.DataFrame([{'generation': g, 'best_score': sc} for g, sc in scores])
        path2 = os.path.join('discoveries', 'controller_evo_scores.csv')
        df.to_csv(path2, index=False)
        print(f'Saved controller evolution scores to {path2}')
        return df

# End of file


# === 11. Real Data Integration Layer
# Placeholder module for loading SAFE, PUBLIC, NON-BIOHAZARD real datasets
# such as gene-expression matrices (e.g., GTEx summary vectors), proteomics stats,
# or generic timeseries for benchmarking AI models. No raw, actionable wet-lab
# sequences or protocols are loaded.

class RealDataLoader:
    def __init__(self, base_path="real_data"):
        self.base_path = Path(base_path)
        self.base_path.mkdir(exist_ok=True)

    def load_tabular(self, name: str):
        """
        Load SAFE PUBLIC tabular datasets (e.g., expression summaries, PCA embeddings,
        generic biological statistics). Returns pandas DataFrame.
        """
        path = self.base_path / f"{name}.csv"
        if not path.exists():
            raise FileNotFoundError(f"Dataset not found: {path}")
        return pd.read_csv(path)

    def save_tabular(self, name: str, df):
        path = self.base_path / f"{name}.csv"
        df.to_csv(path, index=False)
        return path

class RealDataIntegrator:
    def __init__(self, real_data_loader: RealDataLoader):
        self.loader = real_data_loader

    def condition_latent_space(self, cvae, dataset_name: str, k: int = 1):
        """
        Use real summary data vectors to guide sampling. Still abstract and safe.
        """
        df = self.loader.load_tabular(dataset_name)
        vec = torch.tensor(df.values.mean(axis=0), dtype=torch.float32)
        z = cvae.encoder(vec.unsqueeze(0))[0]
        return cvae.decode(z.unsqueeze(0)).detach()
```
