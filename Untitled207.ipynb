```python
# ======= EVOLV-X SuperScience Setup =======
!pip install --quiet torch torchvision transformers diffusers accelerate matplotlib pandas pillow streamlit seaborn

# ======= Imports =======
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from transformers import GPT2Tokenizer, GPT2LMHeadModel
from diffusers import StableDiffusionPipeline
from io import BytesIO
import streamlit as st

# ======= Device =======
device = "cuda" if torch.cuda.is_available() else "cpu"
print("EVOLV-X running on:", device)

# ======= Multimodal Encoder =======
class MultimodalEncoder(nn.Module):
    """Encodes tabular, textual, and image data into a joint latent space"""
    def __init__(self, tabular_dim=10, latent_dim=512):
        super().__init__()
        self.tabular_fc = nn.Sequential(
            nn.Linear(tabular_dim, latent_dim),
            nn.ReLU(),
            nn.Linear(latent_dim, latent_dim),
        )
        self.text_fc = nn.Linear(768, latent_dim)
        # FIX: Change the input dimension of img_fc from 512 to 3
        # because encode_image currently returns a 3-dimensional embedding
        self.img_fc = nn.Linear(3, latent_dim) # Original: self.img_fc = nn.Linear(512, latent_dim)

    def forward(self, tabular, text_emb, img_emb):
        tab_latent = self.tabular_fc(tabular)
        text_latent = self.text_fc(text_emb)
        img_latent = self.img_fc(img_emb)
        joint = tab_latent + text_latent + img_latent
        return joint

# ======= EVOLV-X Core Lab =======
class EVOLVXSuperScience:
    def __init__(self, num_cells=50, latent_dim=512):
        self.num_cells = num_cells
        self.latent_dim = latent_dim
        self.iteration = 0
        self.history = []
        self.global_feedback = None

        # Text and Image Models
        self.tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
        self.tokenizer.pad_token = self.tokenizer.eos_token
        self.text_model = GPT2LMHeadModel.from_pretrained("gpt2").to(device)
        self.sd_pipe = StableDiffusionPipeline.from_pretrained(
            "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
        ).to(device)

        # Multimodal Encoder
        self.encoder = MultimodalEncoder(tabular_dim=3, latent_dim=latent_dim).to(device)

    # ====== Simulate Virtual Cells ======
    def simulate_cells(self):
        df = pd.DataFrame({
            "telomere_length": np.random.rand(self.num_cells),
            "TERT_expression": np.random.rand(self.num_cells),
            "phagocytosis_score": np.random.rand(self.num_cells)
        })
        sequences = ["".join(np.random.choice(list("ATCG"), 20)) for _ in range(self.num_cells)]
        return df, sequences

    # ====== Encode Data ======
    def encode_tabular(self, df):
        tabular_tensor = torch.tensor(df.values, dtype=torch.float32).to(device)
        # Average over cells
        return tabular_tensor.mean(dim=0, keepdim=True)

    def encode_text(self, prompt):
        inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(device)
        with torch.no_grad():
            outputs = self.text_model(**inputs, output_hidden_states=True)
        return outputs.hidden_states[-1][:,0,:]  # CLS token approximation

    def encode_image(self, image):
        img_tensor = torch.tensor(np.array(image)/255.0, dtype=torch.float32).permute(2,0,1).unsqueeze(0).to(device)
        img_emb = img_tensor.mean(dim=[2,3])  # simple global average pooling
        return img_emb

    # ====== Generate Hypotheses / Instructions ======
    def generate_hypotheses(self, df, sequences):
        mean_tel = df["telomere_length"].mean()
        prompt = f"""
        AI Scientific Research Engine.
        Cells have mean telomere length {mean_tel:.2f} and sample sequences {sequences[:3]}.
        Generate actionable scientific instructions, discoveries, and hidden pattern insights.
        """
        if self.global_feedback is not None:
            prompt += f"\nIncorporate previous feedback: {str(self.global_feedback)[:300]}..."

        text_emb = self.encode_text(prompt)
        # Dummy tabular + image embedding
        tab_emb = self.encode_tabular(df)
        sim_img = Image.fromarray((np.random.rand(256,256,3)*255).astype(np.uint8))
        img_emb = self.encode_image(sim_img)

        joint_latent = self.encoder(tab_emb, text_emb, img_emb)
        self.global_feedback = joint_latent.mean().item()  # feedback loop for self-improvement

        # Generate textual hypotheses
        inputs = self.tokenizer(prompt, return_tensors="pt").to(device)
        outputs = self.text_model.generate(**inputs, max_length=350, num_beams=5)
        hypotheses = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
        return hypotheses

    # ====== Generate Figures ======
    def generate_figures(self, df):
        # Telomere Distribution
        plt.figure(figsize=(5,3))
        sns.histplot(df["telomere_length"], bins=10, kde=True, color='skyblue')
        plt.title("Telomere Length Distribution")
        plt.xlabel("Length")
        plt.ylabel("Frequency")
        buf = BytesIO()
        plt.savefig(buf, format="png")
        plt.close()
        buf.seek(0)
        hist_img = Image.open(buf)

        # AI microscopy
        img_prompt = "In-silico enhanced telomere cells, telomerase active, organ-on-a-chip, microscopy style"
        ai_img = self.sd_pipe(img_prompt, guidance_scale=7.5).images[0].resize((256,256))
        return hist_img, ai_img

    # ====== Self-Improving Iteration ======
    def run_iteration(self):
        df, seqs = self.simulate_cells()
        hypotheses = self.generate_hypotheses(df, seqs)
        hist_img, ai_img = self.generate_figures(df)
        self.history.append({"df": df, "seqs": seqs, "hypotheses": hypotheses,
                             "hist_img": hist_img, "ai_img": ai_img})
        self.iteration += 1
        return df, seqs, hypotheses, hist_img, ai_img

    # ====== Generate Autonomous Paper ======
    def generate_paper(self, iteration=-1):
        if iteration < 0: iteration = self.iteration-1
        data = self.history[iteration]
        paper = f"# EVOLV-X SuperScience Autonomous Research Paper\n\n"
        paper += "## Abstract\nGenerated by EVOLV-X, global multimodal scientific engine.\n\n"
        paper += "## Methods\nSimulation of virtual cells with enhanced telomeres, AI-driven iterative experiments, and self-improving cross-modal reasoning.\n\n"
        paper += "## Results & Instructions\n"
        paper += data["hypotheses"] + "\n\n"
        paper += "## Figures\n1. Telomere histogram\n2. AI-generated microscopy image\n"
        return paper

# ====== Streamlit Interface ======
st.set_page_config(layout="wide", page_title="EVOLV-X SuperScience")
st.title("EVOLV-X SuperScience | Global Autonomous Scientific Engine")

st.sidebar.header("Simulation Settings")
num_cells = st.sidebar.slider("Number of Virtual Cells", 10, 200, 50)
iterations = st.sidebar.slider("Number of Iterations", 1, 5, 3)

lab = EVOLVXSuperScience(num_cells=num_cells)

for i in range(iterations):
    st.subheader(f"Iteration {i+1}")
    df, seqs, hypotheses, hist_img, ai_img = lab.run_iteration()
    st.write("**Hypotheses / Instructions:**")
    st.write(hypotheses[:400]+"...")
    st.write("**Telomere Length Distribution:**")
    st.image(hist_img)
    st.write("**AI-Generated Microscopy Image:**")
    st.image(ai_img)

st.subheader("Autonomous Research Paper")
paper = lab.generate_paper()
st.text_area("Generated Paper", paper, height=600)
```

```python
# ======= EVOLV-X Global SuperIntelligence Setup =======
!pip install --quiet torch torchvision transformers diffusers accelerate matplotlib pandas pillow streamlit seaborn

# ======= Imports =======
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from transformers import GPT2Tokenizer, GPT2LMHeadModel
from diffusers import StableDiffusionPipeline
from io import BytesIO
import streamlit as st
import random

# ======= Device Configuration =======
device = "cuda" if torch.cuda.is_available() else "cpu"
print("EVOLV-X SuperIntelligence running on:", device)

# ======= Multimodal Latent Encoder =======
class MultimodalLatentEncoder(nn.Module):
    """
    Encodes tabular, textual, and image data into a joint latent space.
    Supports self-improvement by stacking latent representations.
    """
    def __init__(self, tabular_dim=10, latent_dim=1024):
        super().__init__()
        self.tabular_fc = nn.Sequential(
            nn.Linear(tabular_dim, latent_dim),
            nn.ReLU(),
            nn.Linear(latent_dim, latent_dim),
            nn.LayerNorm(latent_dim)
        )
        self.text_fc = nn.Linear(768, latent_dim)
        # Fix: Change img_fc input dimension from 512 to 3, as encode_image returns a 3-dimensional embedding
        self.img_fc = nn.Linear(3, latent_dim)
        self.self_improve_factor = nn.Parameter(torch.tensor(1.0))

    def forward(self, tabular, text_emb, img_emb, prev_latent=None):
        tab_latent = self.tabular_fc(tabular)
        text_latent = self.text_fc(text_emb)
        img_latent = self.img_fc(img_emb)
        joint = tab_latent + text_latent + img_latent
        if prev_latent is not None:
            joint = joint + self.self_improve_factor * prev_latent
        return joint

# ======= EVOLV-X SuperIntelligence Core =======
class EVOLVXGlobalAI:
    def __init__(self, domains=["biology","chemistry","physics","medicine","environment"], latent_dim=1024):
        self.domains = domains
        self.latent_dim = latent_dim
        self.iteration = 0
        self.history = []
        self.global_feedback = None

        # Multimodal Models
        self.tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
        self.tokenizer.pad_token = self.tokenizer.eos_token
        self.text_model = GPT2LMHeadModel.from_pretrained("gpt2").to(device)
        self.sd_pipe = StableDiffusionPipeline.from_pretrained(
            "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
        ).to(device)

        self.encoder = MultimodalLatentEncoder(tabular_dim=5, latent_dim=latent_dim).to(device)
        self.prev_latent = None

    # ====== Simulate Domain Data ======
    def simulate_domain_data(self):
        """
        Generates tabular + textual + image data for multiple scientific domains.
        """
        data_dict = {}
        for domain in self.domains:
            df = pd.DataFrame({
                "feature1": np.random.rand(50),
                "feature2": np.random.rand(50),
                "feature3": np.random.rand(50),
                "feature4": np.random.rand(50),
                "feature5": np.random.rand(50)
            })
            sequences = ["".join(random.choices("ATCG", k=20)) for _ in range(50)]
            data_dict[domain] = {"df": df, "sequences": sequences}
        return data_dict

    # ====== Encode Data ======
    def encode_tabular(self, df):
        tensor = torch.tensor(df.values, dtype=torch.float32).mean(dim=0, keepdim=True).to(device)
        return tensor

    def encode_text(self, prompt):
        inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(device)
        with torch.no_grad():
            outputs = self.text_model(**inputs, output_hidden_states=True)
        return outputs.hidden_states[-1][:,0,:]

    def encode_image(self, image):
        tensor = torch.tensor(np.array(image)/255.0, dtype=torch.float32).permute(2,0,1).unsqueeze(0).to(device)
        emb = tensor.mean(dim=[2,3])
        return emb

    # ====== Generate Hypotheses ======
    def generate_hypotheses(self, domain_data):
        joint_latents = []
        hypotheses_dict = {}
        for domain, data in domain_data.items():
            df, seqs = data["df"], data["sequences"]
            mean_feat = df.mean().mean()
            prompt = f"Domain: {domain}\nMean feature: {mean_feat:.2f}\nSequences sample: {seqs[:3]}\nGenerate actionable research hypotheses, instructions, hidden patterns."
            if self.global_feedback:
                prompt += f"\nIncorporate previous global feedback: {self.global_feedback:.4f}"

            text_emb = self.encode_text(prompt)
            tab_emb = self.encode_tabular(df)
            sim_img = Image.fromarray((np.random.rand(256,256,3)*255).astype(np.uint8))
            img_emb = self.encode_image(sim_img)

            joint_latent = self.encoder(tab_emb, text_emb, img_emb, prev_latent=self.prev_latent)
            self.prev_latent = joint_latent.detach()

            self.global_feedback = joint_latent.mean().item()
            joint_latents.append(joint_latent)

            # Generate textual hypotheses
            inputs = self.tokenizer(prompt, return_tensors="pt").to(device)
            outputs = self.text_model.generate(**inputs, max_length=350, num_beams=5)
            hypotheses = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
            hypotheses_dict[domain] = hypotheses

        return hypotheses_dict, joint_latents

    # ====== Generate Figures ======
    def generate_figures(self, domain_data):
        figures = {}
        for domain, data in domain_data.items():
            df = data["df"]
            plt.figure(figsize=(4,3))
            sns.histplot(df["feature1"], bins=10, kde=True, color='skyblue')
            plt.title(f"{domain.capitalize()} Feature1 Distribution")
            buf = BytesIO()
            plt.savefig(buf, format="png")
            plt.close()
            buf.seek(0)
            hist_img = Image.open(buf)

            img_prompt = f"In-silico AI visualization for {domain}, experimental simulation, microscopy style"
            ai_img = self.sd_pipe(img_prompt, guidance_scale=7.5).images[0].resize((256,256))

            figures[domain] = {"hist_img": hist_img, "ai_img": ai_img}
        return figures

    # ====== Run Global Iteration ======
    def run_iteration(self):
        domain_data = self.simulate_domain_data()
        hypotheses, _ = self.generate_hypotheses(domain_data)
        figures = self.generate_figures(domain_data)
        self.history.append({"domain_data": domain_data, "hypotheses": hypotheses, "figures": figures})
        self.iteration += 1
        return hypotheses, figures

    # ====== Generate Autonomous Global Paper ======
    def generate_paper(self, iteration=-1):
        if iteration < 0: iteration = self.iteration-1
        data = self.history[iteration]
        paper = "# EVOLV-X Global SuperIntelligence Autonomous Research Paper\n\n"
        paper += "## Abstract\nGenerated by EVOLV-X, a multimodal global autonomous scientific engine.\n\n"
        paper += "## Methods\nMulti-domain data simulation, AI-driven iterative experiments, self-improving cross-modal reasoning.\n\n"
        paper += "## Results & Instructions\n"
        for domain, hypothesis in data["hypotheses"].items():
            paper += f"### Domain: {domain.capitalize()}\n{hypothesis[:400]}...\n\n"
        paper += "## Figures\n"
        for domain, figs in data["figures"].items():
            paper += f"### {domain.capitalize()}\n1. Feature histogram\n2. AI-generated simulation image\n\n"
        return paper

# ====== Streamlit Interface ======
st.set_page_config(layout="wide", page_title="EVOLV-X Global SuperIntelligence")
st.title("EVOLV-X Global SuperIntelligence | Multi-Domain Autonomous AI Engine")

domains = ["biology","chemistry","physics","medicine","environment"]
iterations = st.sidebar.slider("Number of Iterations", 1, 5, 2)

lab = EVOLVXGlobalAI(domains=domains)

for i in range(iterations):
    st.subheader(f"Iteration {i+1}")
    hypotheses, figures = lab.run_iteration()
    for domain in domains:
        st.write(f"**Domain: {domain.capitalize()} Hypotheses / Instructions:**")
        st.write(hypotheses[domain][:400]+"...")
        st.write(f"**Domain: {domain.capitalize()} Feature Distribution:**")
        st.image(figures[domain]["hist_img"])
        st.write(f"**Domain: {domain.capitalize()} AI Simulation Image:**")
        st.image(figures[domain]["ai_img"])

st.subheader("Autonomous Global Research Paper")
paper = lab.generate_paper()
st.text_area("Generated Paper", paper, height=600)
```

```python
# ======= Multi-Agent EVOLV-X Network Setup =======
!pip install --quiet torch torchvision transformers diffusers accelerate matplotlib pandas pillow streamlit

import torch, torch.nn as nn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from transformers import GPT2Tokenizer, GPT2LMHeadModel
from diffusers import StableDiffusionPipeline
from io import BytesIO
import streamlit as st
import random

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Multi-Agent EVOLV-X running on:", device)

# ======= Multi-Agent Class =======
class EVOLVXAgent:
    """Single EVOLV-X agent capable of generating hypotheses, figures, and latent feedback"""
    def __init__(self, agent_id, sd_pipe, latent_dim=1024, domains=["biology","chemistry","physics","medicine","environment"]):
        self.agent_id = agent_id
        self.latent_dim = latent_dim
        self.domains = domains
        self.iteration = 0
        self.history = []
        self.prev_latent = None
        self.global_feedback = 0.0

        self.tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
        self.tokenizer.pad_token = self.tokenizer.eos_token
        self.text_model = GPT2LMHeadModel.from_pretrained("gpt2").to(device)
        self.sd_pipe = sd_pipe # Use the shared Stable Diffusion pipeline

        # Correct the input dimension of the encoder: 768 (text) + 5 (tabular) + 3 (image)
        self.encoder = nn.Linear(768 + 5 + 3, latent_dim).to(device)  # simplified multimodal

    def simulate_domain_data(self):
        data_dict = {}
        for domain in self.domains:
            df = pd.DataFrame({
                "feature1": np.random.rand(50),
                "feature2": np.random.rand(50),
                "feature3": np.random.rand(50),
                "feature4": np.random.rand(50),
                "feature5": np.random.rand(50)
            })
            sequences = ["".join(random.choices("ATCG", k=20)) for _ in range(50)]
            data_dict[domain] = {"df": df, "sequences": sequences}
        return data_dict

    def encode_text(self, prompt):
        inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(device)
        with torch.no_grad():
            outputs = self.text_model(**inputs, output_hidden_states=True)
        return outputs.hidden_states[-1][:,0,:]

    def encode_tabular(self, df):
        tensor = torch.tensor(df.mean().values, dtype=torch.float32).unsqueeze(0).to(device)
        return tensor

    def encode_image(self, img):
        tensor = torch.tensor(np.array(img)/255.0, dtype=torch.float32).permute(2,0,1).unsqueeze(0).to(device)
        return tensor.mean(dim=[2,3])

    def generate_hypotheses(self, domain_data, network_feedback):
        hypotheses_dict = {}
        for domain, data in domain_data.items():
            df, seqs = data["df"], data["sequences"]
            prompt = f"Agent {self.agent_id} | Domain {domain} | Mean feature: {df.mean().mean():.2f}\nSequences: {seqs[:3]}\nNetwork feedback: {network_feedback:.4f}\nGenerate actionable hypotheses and hidden patterns."

            text_emb = self.encode_text(prompt)
            tab_emb = self.encode_tabular(df)
            img_emb = self.encode_image(Image.fromarray((np.random.rand(256,256,3)*255).astype(np.uint8)))
            latent = self.encoder(torch.cat([text_emb, tab_emb, img_emb], dim=1))
            if self.prev_latent is not None:
                latent += self.prev_latent * 0.5
            self.prev_latent = latent.detach()

            # Generate textual hypothesis
            inputs = self.tokenizer(prompt, return_tensors="pt").to(device)
            outputs = self.text_model.generate(**inputs, max_length=250, num_beams=5)
            hypotheses_dict[domain] = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
        return hypotheses_dict, latent.mean().item()

    def generate_figures(self, domain_data):
        figures = {}
        for domain, data in domain_data.items():
            df = data["df"]
            plt.figure(figsize=(4,3))
            plt.hist(df["feature1"], bins=10, color='skyblue')
            buf = BytesIO()
            plt.savefig(buf, format="png")
            plt.close()
            buf.seek(0)
            hist_img = Image.open(buf)
            ai_img = self.sd_pipe(f"Agent {self.agent_id} in-silico AI visualization for {domain}", guidance_scale=7.5).images[0].resize((256,256))
            figures[domain] = {"hist_img": hist_img, "ai_img": ai_img}
        return figures

# ======= Multi-Agent Network =======
class EVOLVXNetwork:
    """Network of multiple EVOLV-X agents collaborating and sharing feedback"""
    def __init__(self, num_agents=3):
        # Initialize Stable Diffusion pipeline once and share it
        self.sd_pipe = StableDiffusionPipeline.from_pretrained(
            "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
        ).to(device)
        self.agents = [EVOLVXAgent(agent_id=i, sd_pipe=self.sd_pipe) for i in range(num_agents)]
        self.network_feedback = 0.0
        self.global_history = []

    def run_iteration(self):
        iteration_data = {}
        all_latents = []
        for agent in self.agents:
            domain_data = agent.simulate_domain_data()
            hypotheses, latent_feedback = agent.generate_hypotheses(domain_data, self.network_feedback)
            figures = agent.generate_figures(domain_data)
            iteration_data[agent.agent_id] = {"hypotheses": hypotheses, "figures": figures}
            all_latents.append(latent_feedback)
        # Update network feedback as average latent feedback
        self.network_feedback = np.mean(all_latents)
        self.global_history.append(iteration_data)
        return iteration_data

# ======= Example Multi-Agent Run =======
network = EVOLVXNetwork(num_agents=3)
for iter_num in range(2):  # multiple iterations
    results = network.run_iteration()
    print(f"\n--- Iteration {iter_num+1} ---")
    for agent_id, data in results.items():
        print(f"\nAgent {agent_id} Hypotheses (sample):")
        for domain, hypo in data["hypotheses"].items():
            print(f"{domain}: {hypo[:200]}...")
```

```python
class GlobalMemory(nn.Module):
    """Persistent cross-iteration scientific memory"""
    def __init__(self, dim=1024):
        super().__init__()
        self.state = torch.zeros(dim, device=device)
        self.update_layer = nn.Linear(dim, dim).to(device)

    def update(self, new_latent):
        self.state = 0.8 * self.state + 0.2 * self.update_layer(new_latent)
        return self.state
```

```python
class Encoder(nn.Module):
    def __init__(self, obs_dim, latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(obs_dim, 256),
            nn.GELU(),
            nn.Linear(256, latent_dim)
        )

    def forward(self, obs):
        return self.net(obs)
```

```python
class WorldModel(nn.Module):
    def __init__(self, latent_dim, action_dim):
        super().__init__()
        self.transition = nn.Sequential(
            nn.Linear(latent_dim + action_dim, 256),
            nn.GELU(),
            nn.Linear(256, latent_dim)
        )

    def forward(self, latent, action):
        x = torch.cat([latent, action], dim=-1)
        return self.transition(x)
```

```python
class RewardModel(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.GELU(),
            nn.Linear(128, 1)
        )

    def forward(self, latent):
        return self.net(latent)
```

```python
def imagine_rollout(world_model, reward_model, policy, latent, horizon=10):
    total_reward = 0
    for _ in range(horizon):
        action = policy(latent)
        latent = world_model(latent, action)
        reward = reward_model(latent)
        total_reward += reward
    return total_reward
```

```python
class AgentWorldModel(nn.Module):
    def __init__(self, latent_dim=64, action_dim=8):
        super().__init__()
        self.transition = nn.Sequential(
            nn.Linear(latent_dim + action_dim, 256),
            nn.GELU(),
            nn.Linear(256, latent_dim)
        )
        self.reward = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.GELU(),
            nn.Linear(128, 1)
        )

    def forward(self, latent, action):
        x = torch.cat([latent, action], dim=-1)
        next_latent = self.transition(x)
        reward = self.reward(next_latent)
        return next_latent, reward
```

```python
class AgentPolicy(nn.Module):
    def __init__(self, latent_dim=64, action_dim=8):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.GELU(),
            nn.Linear(256, action_dim),
            nn.Tanh()
        )

    def forward(self, latent):
        return self.net(latent)
```

```python
class EvolvXMultiAgent:
    def __init__(self, num_agents=4):
        self.encoder = SharedEncoder().to(device)
        self.agents = []

        for _ in range(num_agents):
            agent = {
                "world": AgentWorldModel().to(device),
                "policy": AgentPolicy().to(device)
            }
            self.agents.append(agent)

        self.optimizer = torch.optim.AdamW(
            self.encoder.parameters(), lr=3e-4
        )

        for agent in self.agents:
            self.optimizer.add_param_group(
                {"params": agent["world"].parameters()}
            )
            self.optimizer.add_param_group(
                {"params": agent["policy"].parameters()}
            )
```

```python
def imagine(agent, latent, horizon=8):
    total_reward = 0
    for _ in range(horizon):
        action = agent["policy"](latent)
        latent, reward = agent["world"](latent, action)
        total_reward += reward
    return total_reward
```

```python
def societal_consensus(rewards):
    rewards = torch.stack(rewards)
    mean = rewards.mean()
    disagreement = rewards.std()
    return mean, disagreement
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
class CausalEncoder(nn.Module):
    def __init__(self, obs_dim=7, latent_dim=16):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(obs_dim, 128),
            nn.GELU(),
            nn.Linear(128, latent_dim)
        )

    def forward(self, x):
        return self.net(x)
```

```python
class CausalScoreNet(nn.Module):
    def __init__(self, latent_dim=16):
        super().__init__()
        self.score = nn.Sequential(
            nn.Linear(2, 128), # Changed input dimension from latent_dim * 2 to 2
            nn.GELU(),
            nn.Linear(128, 1)
        )

    def forward(self, zi, zj):
        return self.score(torch.cat([zi.unsqueeze(1), zj.unsqueeze(1)], dim=-1)) # Unsqueeze to make it (batch_size, 1) before concatenating
```

```python
class EvolvXCausalEngine:
    def __init__(self, variables):
        self.variables = variables
        self.encoder = CausalEncoder().to(device)
        self.score_net = CausalScoreNet().to(device)

        self.optimizer = torch.optim.AdamW(
            list(self.encoder.parameters()) +
            list(self.score_net.parameters()),
            lr=3e-4
        )

    def compute_adjacency(self, z):
        n = z.shape[1] # n is latent_dim (16)
        adj = torch.zeros(n, n, device=device)

        for i in range(n):
            for j in range(n):
                if i != j:
                    # z[:, i] is (batch_size,), z[:, j] is (batch_size,)
                    # score_net expects (batch_size, 2) from its current definition
                    adj[i, j] = self.score_net(z[:, i], z[:, j]).mean()
        return adj

    def train_step(self, x):
        z = self.encoder(x)
        adj = self.compute_adjacency(z)

        # Acyclicity penalty (NOTEARS-inspired)
        h = torch.trace(torch.matrix_exp(adj * adj)) - adj.shape[0]

        loss = h + 0.01 * torch.sum(adj ** 2)

        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

        return adj.detach().cpu()
```

```python
def generate_virtual_cell_data(batch=512):
    tel_len = np.random.normal(1.0, 0.1, batch)
    tert = tel_len + np.random.normal(0, 0.05, batch)
    dna_damage = 1.2 - tel_len + np.random.normal(0, 0.1, batch)
    immune = dna_damage + np.random.normal(0, 0.1, batch)
    cycle = tert - immune + np.random.normal(0, 0.1, batch)
    ooc_stress = np.random.normal(0.5, 0.2, batch)
    longevity = tel_len + cycle - immune + np.random.normal(0, 0.1, batch)

    data = np.stack([
        tel_len, tert, dna_damage,
        immune, cycle, ooc_stress, longevity
    ], axis=1)

    return torch.tensor(data, dtype=torch.float32, device=device)
```

```python
variables = [
    "TelomereLength",
    "TelomeraseActivity",
    "DNA_Damage",
    "ImmuneClearance",
    "CellCycleRate",
    "OoC_Stress",
    "LongevityScore"
]

engine = EvolvXCausalEngine(variables)

for step in range(2000):
    x = generate_virtual_cell_data()
    adj = engine.train_step(x)

    if step % 400 == 0:
        print(f"Causal learning step {step}")
```

```python
def plot_causal_graph(adj, variables, threshold=0.1):
    G = nx.DiGraph()

    for i, v in enumerate(variables):
        G.add_node(v)

    for i in range(len(variables)):
        for j in range(len(variables)):
            if abs(adj[i, j]) > threshold:
                G.add_edge(variables[i], variables[j],
                           weight=float(adj[i, j]))

    plt.figure(figsize=(10, 8))
    pos = nx.spring_layout(G, seed=42)
    nx.draw(G, pos, with_labels=True,
            node_size=3000, font_size=9,
            arrowsize=20)
    plt.show()

plot_causal_graph(adj.numpy(), variables)
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from transformers import GPT2LMHeadModel, GPT2Tokenizer

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
class VirtualCellGenerator:
    def generate(self, n=512):
        telomere = np.random.normal(1.0, 0.1, n)
        telomerase = telomere + np.random.normal(0, 0.05, n)
        dna_damage = 1.2 - telomere + np.random.normal(0, 0.1, n)
        immune = dna_damage + np.random.normal(0, 0.1, n)
        cell_cycle = telomerase - immune + np.random.normal(0, 0.1, n)
        longevity = telomere + cell_cycle - immune + np.random.normal(0, 0.1, n)

        return torch.tensor(
            np.stack([telomere, telomerase, dna_damage,
                      immune, cell_cycle, longevity], axis=1),
            dtype=torch.float32, device=device
        )
```

```python
class ScientificEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(6, 128),
            nn.GELU(),
            nn.Linear(128, 32)
        )

    def forward(self, x):
        return self.net(x)
```

```python
class CausalDiscovery(nn.Module):
    def __init__(self):
        super().__init__()
        self.score = nn.Sequential(
            nn.Linear(64, 128),
            nn.GELU(),
            nn.Linear(128, 1)
        )

    def forward(self, zi, zj):
        return self.score(torch.cat([zi, zj], dim=-1))
```

```python
class EvolvX:
    def __init__(self):
        self.encoder = ScientificEncoder().to(device)
        self.causal = CausalDiscovery().to(device)
        self.opt = torch.optim.AdamW(
            list(self.encoder.parameters()) + list(self.causal.parameters()),
            lr=3e-4
        )

        self.variables = [
            "TelomereLength",
            "TelomeraseActivity",
            "DNA_Damage",
            "ImmuneClearance",
            "CellCycleRate",
            "Longevity"
        ]

    def discover(self, data):
        z = self.encoder(data)
        n = z.shape[1]
        adj = torch.zeros(n, n, device=device)

        for i in range(n):
            for j in range(n):
                if i != j:
                    adj[i, j] = self.causal(z[:, i], z[:, j]).mean()

        loss = torch.sum(adj ** 2)
        self.opt.zero_grad()
        loss.backward()
        self.opt.step()

        return adj.detach().cpu()
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2").to(device)

def generate_instructions(insights):
    prompt = f"""
You are an autonomous scientific researcher.
Based on the following causal discoveries, produce
clear step-by-step experimental instructions.

Discoveries:
{insights}

Instructions:
"""
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    output = model.generate(**inputs, max_length=350)
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
def plot_causal_graph(adj, variables):
    G = nx.DiGraph()
    for i, v in enumerate(variables):
        for j, u in enumerate(variables):
            if abs(adj[i, j]) > 0.1:
                G.add_edge(v, u)

    plt.figure(figsize=(8,6))
    nx.draw(G, with_labels=True, node_size=3000)
    plt.show()
```

```python
class CellGenome:
    def __init__(self, telomere, repair, stress, metabolism):
        self.telomere = telomere
        self.repair = repair
        self.stress = stress
        self.metabolism = metabolism

    def mutate(self, rate=0.05):
        def m(x):
            return np.clip(x + np.random.normal(0, rate), 0, 1)
        return CellGenome(
            m(self.telomere),
            m(self.repair),
            m(self.stress),
            m(self.metabolism)
        )
```

```python
def individual_fitness(cell):
    return (
        0.4 * cell.telomere +
        0.3 * cell.repair +
        0.2 * cell.metabolism -
        0.3 * cell.stress
    )
```

```python
def organism_fitness(population):
    tel = np.mean([c.telomere for c in population])
    stress = np.mean([c.stress for c in population])
    repair = np.mean([c.repair for c in population])
    return tel + repair - stress
```

```python
def evolve_population(population, max_size=50):
    fitnesses = np.array([individual_fitness(c) for c in population])
    probs = fitnesses / (fitnesses.sum() + 1e-8)

    new_population = []
    for _ in range(len(population)):
        parent = np.random.choice(population, p=probs)
        child = parent.mutate()
        new_population.append(child)

    # Organism-level pruning
    if organism_fitness(new_population) < organism_fitness(population):
        new_population = population

    return new_population[:max_size]
```

```python
def init_population(n=20):
    return [
        CellGenome(
            telomere=np.random.rand(),
            repair=np.random.rand(),
            stress=np.random.rand(),
            metabolism=np.random.rand()
        ) for _ in range(n)
    ]
```

```python
population = init_population()
history = []

for generation in range(30):
    population = evolve_population(population)
    history.append(organism_fitness(population))
```

```python
plt.figure(figsize=(6,4))
plt.plot(history, marker="o")
plt.xlabel("Generation")
plt.ylabel("Organism Fitness")
plt.title("Emergent Optimization Through Cell Division")
plt.grid(True)
plt.show()
```

```python
class DiscoveryEngine:
    def discover(self, results):
        correlation = np.corrcoef(results[:-1], results[1:])[0,1]
        entropy = np.std(results)
        return {
            "emergent_correlation": correlation,
            "system_entropy": entropy
        }
```

```python
class Falsifier:
    def reject(self, hypothesis, metrics):
        return metrics["system_entropy"] > 0.5
```
