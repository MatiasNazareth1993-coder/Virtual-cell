```python
# Retry corrected simulation code with fixes to the controller neighbor-action bug
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass, field
import random
import os
@dataclass
class VirtualCell:
    id: int
    telomeres: np.ndarray
    shelterin: float = 1.0
    tert_expression: float = 0.0
    damage_load: float = 0.0
    mutation_burden: float = 0.0
    proliferative_state: int = 0
    alive: bool = True
    last_division_time: float = -np.inf
    def mean_telomere(self):
        return float(np.mean(self.telomeres))
    def short_telomere_fraction(self, threshold=3.0):
        return float(np.mean(self.telomeres < threshold))
    def step(self, dt, local_signals, action=None, params=None):
        if not self.alive:
            return
        if params is None:
            params = DEFAULT_PARAMS
        gf = local_signals.get('growth_factor', 0.0)
        damage_signal = local_signals.get('damage_signal', 0.0)
        pathogen = local_signals.get('pathogen_signal', 0.0)
        promoter_gain = params['tert_promoter_gain']
        gf_effect = promoter_gain * gf / (gf + params['gf_half'])
        damage_suppress = params['damage_tert_supp'] * damage_signal / (1 + damage_signal)
        tert_target = max(0.0, gf_effect - damage_suppress)
        if action and action.get('tertpulse', 0) > 0:
            tert_target = max(tert_target, params['tertpulse_level'])
        self.tert_expression += params['tert_tau_inv'] * (tert_target - self.tert_expression) * dt
        self.tert_expression = float(np.clip(self.tert_expression, 0.0, 1.5))
        if self.tert_expression > params['tert_activity_thresh']:
            frac = min(1.0, self.tert_expression * params['tert_to_frac'])
            n = len(self.telomeres)
            k = max(1, int(round(frac * n)))
            idx = np.random.choice(n, k, replace=False)
            elong_base = params['telomerase_elong_rate'] * self.tert_expression * (1 - 0.5*self.shelterin)
            self.telomeres[idx] += np.random.exponential(elong_base, size=k)
            over = self.telomeres - params['telomere_max_safe']
            if np.any(over > 0):
                trim_idx = np.where(over > 0)[0]
                self.telomeres[trim_idx] -= params['telomere_trim_rate'] * over[trim_idx]
        if self.proliferative_state == 1:
            if random.random() < params['division_prob_per_dt'] * dt:
                shorten = np.random.normal(params['telomere_shorten_per_div_mean'], params['telomere_shorten_sd'], size=self.telomeres.shape)
                shorten = np.clip(shorten, 0, None)
                self.telomeres -= shorten
                self.last_division_time = self.last_division_time if np.isfinite(self.last_division_time) else 0.0
                mut_prob = params['base_mutation_rate'] + params['damage_mutation_scale'] * self.damage_load
                if random.random() < mut_prob:
                    self.mutation_burden += np.random.exponential(params['mut_burden_inc_mean'])
        self.damage_load += (params['pathogen_damage_scale'] * pathogen + params['damage_signal_scale'] * damage_signal) * dt
        self.damage_load += params['short_tel_damage_scale'] * self.short_telomere_fraction(params['short_tel_threshold']) * dt
        phago_boost = action.get('increase_phago', 0.0) if action else 0.0
        clearance = params['base_clearance_rate'] + phago_boost
        self.damage_load = max(0.0, self.damage_load - clearance * dt)
        if self.damage_load > params['damage_arrest_threshold'] or self.mean_telomere() < params['crit_tel_mean']:
            self.proliferative_state = 0
        else:
            if gf > params['gf_prolif_thresh'] and self.damage_load < params['damage_prolif_thresh']:
                self.proliferative_state = 1
            else:
                self.proliferative_state = 0
        if self.damage_load > params['lethal_damage'] or self.mutation_burden > params['lethal_mutation']:
            self.alive = False
@dataclass
class Tissue:
    n_cells: int
    cells: list = field(default_factory=list)
    growth_field: np.ndarray = field(default_factory=lambda: np.zeros(0))
    damage_field: np.ndarray = field(default_factory=lambda: np.zeros(0))
    pathogen_field: np.ndarray = field(default_factory=lambda: np.zeros(0))
    params: dict = field(default_factory=dict)
    def __post_init__(self):
        if not self.cells:
            for i in range(self.n_cells):
                tel = np.random.normal(self.params.get('telomere_init_mean', 8.0),
                                        self.params.get('telomere_init_sd', 1.0),
                                        size=self.params.get('n_telomeres', 46))
                tel = np.clip(tel, 1.0, None)
                cell = VirtualCell(id=i, telomeres=tel)
                self.cells.append(cell)
        self.growth_field = np.zeros(self.n_cells)
        self.damage_field = np.zeros(self.n_cells)
        self.pathogen_field = np.zeros(self.n_cells)
    def step(self, dt, controller=None):
        actions = [None] * self.n_cells
        if controller is not None:
            actions = controller.act(self, dt)
        for i, cell in enumerate(self.cells):
            local_signals = {
                'growth_factor': float(self.growth_field[i]),
                'damage_signal': float(self.damage_field[i]),
                'pathogen_signal': float(self.pathogen_field[i])
            }
            action = actions[i] if actions and i < len(actions) else None
            cell.step(dt, local_signals, action=action, params=self.params)
        self._diffuse_and_decay(dt)
    def _diffuse_and_decay(self, dt):
        for field in ['growth_field', 'damage_field', 'pathogen_field']:
            arr = getattr(self, field)
            if arr.size == 0:
                continue
            new = arr.copy()
            D = self.params.get(f'{field}_diffusion', 0.1)
            decay = self.params.get(f'{field}_decay', 0.01)
            n = arr.size
            for i in range(n):
                left = arr[i-1] if i-1 >= 0 else arr[i]
                right = arr[i+1] if i+1 < n else arr[i]
                new[i] += D * (left + right - 2*arr[i]) * dt
            new -= decay * arr * dt
            setattr(self, field, np.clip(new, 0, None))
    def inject_injury(self, pos, magnitude):
        idx = int(np.clip(pos, 0, self.n_cells-1))
        self.damage_field[idx] += magnitude
        self.growth_field[idx] *= 0.5
    def inject_pathogen(self, pos, magnitude):
        idx = int(np.clip(pos, 0, self.n_cells-1))
        self.pathogen_field[idx] += magnitude
    def apply_growth_stimulus(self, pos, magnitude):
        idx = int(np.clip(pos, 0, self.n_cells-1))
        self.growth_field[idx] += magnitude
class SimpleController:
    def __init__(self, params):
        self.params = params
        self.pulse_timers = np.zeros(params.get('n_cells', 100))
        self.phago_boost = params.get('controller_phago_boost', 0.5)
    def act(self, tissue: Tissue, dt):
        # ensure pulse_timers length matches tissue
        if self.pulse_timers.size != tissue.n_cells:
            self.pulse_timers = np.zeros(tissue.n_cells)
        actions = [None] * tissue.n_cells
        for i in range(tissue.n_cells):
            if actions[i] is None:
                actions[i] = {}
            # reduce timers
            if self.pulse_timers[i] > 0:
                self.pulse_timers[i] = max(0.0, self.pulse_timers[i] - dt)
                actions[i]['tertpulse'] = self.pulse_timers[i]
            else:
                actions[i]['tertpulse'] = 0.0
            if tissue.damage_field[i] > self.params.get('controller_damage_thresh', 1.0) and self.pulse_timers[i] <= 0.0:
                for j in range(max(0, i-1), min(tissue.n_cells, i+2)):
                    if actions[j] is None:
                        actions[j] = {}
                    self.pulse_timers[j] = self.params.get('controller_pulse_dur', 2.0)
                    actions[j]['tertpulse'] = self.pulse_timers[j]
                    actions[j]['increase_phago'] = self.phago_boost
            if tissue.pathogen_field[i] > 0.1:
                actions[i]['increase_phago'] = self.phago_boost * 0.5
        return actions
DEFAULT_PARAMS = {
    'n_telomeres': 46,
    'telomere_init_mean': 8.0,
    'telomere_init_sd': 0.8,
    'telomere_shorten_per_div_mean': 0.5,
    'telomere_shorten_sd': 0.1,
    'telomerase_elong_rate': 1.2,
    'telomere_max_safe': 20.0,
    'telomere_trim_rate': 0.2,
    'short_tel_threshold': 3.0,
    'short_tel_damage_scale': 0.5,
    'crit_tel_mean': 2.0,
    'tert_promoter_gain': 1.0,
    'gf_half': 0.5,
    'damage_tert_supp': 0.8,
    'tert_tau_inv': 1.0,
    'tert_activity_thresh': 0.05,
    'tert_to_frac': 0.5,
    'tertpulse_level': 1.0,
    'division_prob_per_dt': 0.1,
    'gf_prolif_thresh': 0.2,
    'damage_prolif_thresh': 0.5,
    'pathogen_damage_scale': 0.8,
    'damage_signal_scale': 0.3,
    'base_clearance_rate': 0.2,
    'damage_arrest_threshold': 3.0,
    'lethal_damage': 10.0,
    'base_mutation_rate': 0.001,
    'damage_mutation_scale': 0.005,
    'mut_burden_inc_mean': 0.5,
    'lethal_mutation': 20.0,
    'growth_field_diffusion': 0.05,
    'growth_field_decay': 0.02,
    'damage_field_diffusion': 0.1,
    'damage_field_decay': 0.05,
    'pathogen_field_diffusion': 0.2,
    'pathogen_field_decay': 0.1,
    'n_cells': 50,
    'controller_damage_thresh': 0.8,
    'controller_pulse_dur': 3.0,
    'controller_phago_boost': 0.6
}
def run_simulation(steps=200, dt=0.5, params=None, seed=0):
    np.random.seed(seed)
    random.seed(seed)
    if params is None:
        params = DEFAULT_PARAMS.copy()
    tissue = Tissue(n_cells=params['n_cells'], params=params)
    controller = SimpleController(params)
    tissue.growth_field += 0.3
    injury_time = 5.0
    pathogen_time = 12.0
    injury_pos = params['n_cells']//2
    pathogen_pos = max(0, injury_pos - 3)
    log = []
    t = 0.0
    for step in range(steps):
        if abs(t - injury_time) < 1e-6:
            tissue.inject_injury(injury_pos, magnitude=4.0)
        if abs(t - pathogen_time) < 1e-6:
            tissue.inject_pathogen(pathogen_pos, magnitude=2.0)
        tissue.step(dt, controller=controller)
        mean_tel = np.nanmean([c.mean_telomere() if c.alive else np.nan for c in tissue.cells])
        short_frac = np.nanmean([c.short_telomere_fraction() if c.alive else np.nan for c in tissue.cells])
        alive_frac = np.mean([1.0 if c.alive else 0.0 for c in tissue.cells])
        mean_damage = np.nanmean([c.damage_load if c.alive else np.nan for c in tissue.cells])
        mean_mut = np.nanmean([c.mutation_burden if c.alive else np.nan for c in tissue.cells])
        log.append({'t': t, 'mean_tel': mean_tel, 'short_frac': short_frac,
                    'alive_frac': alive_frac, 'mean_damage': mean_damage, 'mean_mut': mean_mut})
        t += dt
    df = pd.DataFrame(log)
    return tissue, df
tissue, df = run_simulation(steps=240, dt=0.5, params=DEFAULT_PARAMS, seed=42)
plt.figure(figsize=(8,4))
plt.plot(df['t'], df['mean_tel'])
plt.xlabel('time')
plt.ylabel('mean telomere length')
plt.title('Mean telomere length over time')
plt.grid(True)
plt.show()
plt.figure(figsize=(8,4))
plt.plot(df['t'], df['short_frac'])
plt.xlabel('time')
plt.ylabel('fraction short telomeres')
plt.title('Short telomere fraction over time')
plt.grid(True)
plt.show()
plt.figure(figsize=(8,4))
plt.plot(df['t'], df['mean_damage'])
plt.xlabel('time')
plt.ylabel('mean damage load')
plt.title('Mean damage load over time')
plt.grid(True)
plt.show()
plt.figure(figsize=(8,4))
plt.plot(df['t'], df['alive_frac'])
plt.xlabel('time')
plt.ylabel('fraction alive')
plt.title('Fraction of alive cells over time')
plt.grid(True)
plt.show()
snapshot = []
for c in tissue.cells:
    snapshot.append({
        'id': c.id,
        'alive': c.alive,
        'mean_tel': c.mean_telomere(),
        'short_frac': c.short_telomere_fraction(),
        'damage': c.damage_load,
        'mutation': c.mutation_burden,
        'tert': c.tert_expression,
        'prolif': c.proliferative_state
    })
snap_df = pd.DataFrame(snapshot)
from IPython.display import display
display(snap_df)
# Ensure the directory exists before trying to write the file
os.makedirs('/mnt/data', exist_ok=True)
with open('/mnt/data/virtual_cell_sim.py', 'w') as f:
    import inspect, textwrap
    content = textwrap.dedent('''\
    # Helper: run_simulation function (simplified)
    def run_simulation(...):
        pass
    ''')
    f.write(content)
print('[Saved placeholder helper to /mnt/data/virtual_cell_sim.py]')
```

```python
# Genetic algorithm to evolve virtual cell parameters ("genotype") for safer repair behavior.
# This builds on the VirtualCell/Tissue model from before, but focuses on evolving per-cell parameter sets.
# We run short simulations per genotype, score them by a fitness combining repair effectiveness, low mutation,
# and telomere safety, then apply selection, crossover, and mutation to evolve parameters.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import random
from dataclasses import dataclass, field
# ---- Reuse a simplified VirtualCell & Tissue implementation (compact) ----
@dataclass
class VirtualCell:
    telomeres: np.ndarray
    tert_expression: float = 0.0
    damage_load: float = 0.0
    mutation_burden: float = 0.0
    proliferative_state: int = 0
    alive: bool = True
    def mean_telomere(self):
        return float(np.mean(self.telomeres))
    def short_tel_frac(self, thresh=3.0):
        return float(np.mean(self.telomeres < thresh))
    def step(self, dt, local_signals, params, action=None):
        if not self.alive:
            return
        gf = local_signals.get('growth_factor', 0.0)
        damage_signal = local_signals.get('damage_signal', 0.0)
        pathogen = local_signals.get('pathogen_signal', 0.0)
        # TERT responsiveness (simple)
        gf_eff = params['tert_promoter_gain'] * gf / (gf + params['gf_half'])
        damage_supp = params['damage_tert_supp'] * damage_signal / (1 + damage_signal)
        tert_target = max(0.0, gf_eff - damage_supp)
        if action and action.get('tertpulse',0) > 0:
            tert_target = max(tert_target, params['tertpulse_level'])
        self.tert_expression += params['tert_tau_inv']*(tert_target - self.tert_expression)*dt
        self.tert_expression = float(np.clip(self.tert_expression, 0.0, 2.0))
        # telomerase activity
        if self.tert_expression > params['tert_activity_thresh']:
            frac = min(1.0, self.tert_expression * params['tert_to_frac'])
            n = len(self.telomeres)
            k = max(1, int(round(frac * n)))
            idx = np.random.choice(n, k, replace=False)
            elong = params['telomerase_elong_rate'] * self.tert_expression
            self.telomeres[idx] += np.random.exponential(elong, size=k)
            # trimming
            over = self.telomeres - params['telomere_max_safe']
            if np.any(over>0):
                self.telomeres[np.where(over>0)] -= params['telomere_trim_rate']*over[np.where(over>0)]
        # division shortening (we model occasional divisions if GF high)
        if gf > params['gf_prolif_thresh'] and self.damage_load < params['damage_prolif_thresh']:
            if random.random() < params['division_prob_per_dt']*dt:
                shorten = np.clip(np.random.normal(params['telomere_shorten_per_div_mean'], params['telomere_shorten_sd'], size=self.telomeres.shape), 0, None)
                self.telomeres -= shorten
                # mutation risk
                mut_prob = params['base_mutation_rate'] + params['damage_mutation_scale']*self.damage_load
                if random.random() < mut_prob:
                    self.mutation_burden += np.random.exponential(params['mut_burden_inc_mean'])
        # damage accumulation & clearance
        self.damage_load += (params['pathogen_damage_scale']*pathogen + params['damage_signal_scale']*damage_signal) * dt
        clearance = params['base_clearance_rate'] + (action.get('increase_phago',0.0) if action else 0.0)
        self.damage_load = max(0.0, self.damage_load - clearance*dt)
        # death check
        if self.damage_load > params['lethal_damage'] or self.mutation_burden > params['lethal_mutation']:
            self.alive = False
@dataclass
class TissueSingle:
    # single-cell "tissue" wrapper to evaluate genotypes quickly
    cell: VirtualCell
    growth_field: float = 0.0
    damage_field: float = 0.0
    pathogen_field: float = 0.0
    def step(self, dt, params, controller_action=None):
        local = {'growth_factor': self.growth_field, 'damage_signal': self.damage_field, 'pathogen_signal': self.pathogen_field}
        self.cell.step(dt, local, params, action=controller_action)
# ---- Genotype representation: parameters we allow to evolve ----
# We'll evolve a compact vector of continuous parameters within bounds.
GENE_DEFS = {
    'tert_promoter_gain': (0.1, 2.0),
    'tert_to_frac': (0.05, 1.0),
    'telomerase_elong_rate': (0.1, 3.0),
    'telomere_trim_rate': (0.0, 1.0),
    'base_clearance_rate': (0.0, 1.0),
    'telomere_max_safe': (8.0, 40.0),
    'base_mutation_rate': (1e-4, 0.01),
    'division_prob_per_dt': (0.01, 0.5),
}
def sample_random_genotype():
    g = {}
    for k,(lo,hi) in GENE_DEFS.items():
        if 'mutation_rate' in k or 'base_mutation_rate' in k:
            g[k] = float(10**np.random.uniform(np.log10(lo), np.log10(hi)))
        else:
            g[k] = float(np.random.uniform(lo,hi))
    return g
def mutate_genotype(geno, mut_scale=0.1):
    new = geno.copy()
    for k,(lo,hi) in GENE_DEFS.items():
        if random.random() < 0.3:  # per-gene mutation prob
            span = hi-lo
            # gaussian perturbation scaled by span
            delta = np.random.normal(0, mut_scale*span)
            val = new[k] + delta
            # if mutation_rate-like, enforce within bounds (and positive)
            if 'base_mutation_rate' in k:
                val = float(np.clip(val, lo, hi))
            else:
                val = float(np.clip(val, lo, hi))
            new[k] = val
    return new
def crossover(g1, g2):
    child = {}
    for k in GENE_DEFS.keys():
        child[k] = g1[k] if random.random() < 0.5 else g2[k]
    return child
# ---- Fitness evaluation ----
def evaluate_genotype(geno, sim_time=20.0, dt=0.5, seed=None):
    # build params dict used by cell.step (fill in defaults for missing fields)
    params = {
        'tert_promoter_gain': geno['tert_promoter_gain'],
        'gf_half': 0.5,
        'damage_tert_supp': 0.8,
        'tert_tau_inv': 1.0,
        'tert_activity_thresh': 0.05,
        'tert_to_frac': geno['tert_to_frac'],
        'tertpulse_level': 1.0,
        'telomerase_elong_rate': geno['telomerase_elong_rate'],
        'telomere_max_safe': geno['telomere_max_safe'],
        'telomere_trim_rate': geno['telomere_trim_rate'],
        'telomere_shorten_per_div_mean': 0.5,
        'telomere_shorten_sd': 0.1,
        'division_prob_per_dt': geno['division_prob_per_dt'],
        'gf_prolif_thresh': 0.2,
        'damage_prolif_thresh': 0.5,
        'pathogen_damage_scale': 0.8,
        'damage_signal_scale': 0.3,
        'base_clearance_rate': geno['base_clearance_rate'],
        'damage_arrest_threshold': 3.0,
        'lethal_damage': 10.0,
        'base_mutation_rate': geno['base_mutation_rate'],
        'damage_mutation_scale': 0.005,
        'mut_burden_inc_mean': 0.5,
        'lethal_mutation': 20.0,
    }
    # deterministic seed for repeatability
    if seed is not None:
        np.random.seed(seed)
        random.seed(seed)
    # initialize a single cell with realistic telomeres
    init_tel = np.clip(np.random.normal(8.0, 0.8, size=46), 1.0, None)
    cell = VirtualCell(telomeres=init_tel.copy())
    tissue = TissueSingle(cell=cell, growth_field=0.3, damage_field=0.0, pathogen_field=0.0)
    t = 0.0
    injury_time = 2.0
    pathogen_time = 6.0
    # controller: simple rule same as before: provide tert pulses when damage > threshold
    pulse_timer = 0.0
    hist = []
    while t < sim_time:
        # injected events
        if abs(t - injury_time) < 1e-6:
            tissue.damage_field += 4.0
            tissue.growth_field *= 0.5
        if abs(t - pathogen_time) < 1e-6:
            tissue.pathogen_field += 2.0
        # controller action logic
        action = {}
        if pulse_timer > 0:
            action['tertpulse'] = pulse_timer
            action['increase_phago'] = 0.6
            pulse_timer -= dt
        else:
            action['tertpulse'] = 0.0
            action['increase_phago'] = 0.0
        # if damage present and no pulse, start pulse
        if tissue.damage_field > 0.8 and pulse_timer <= 0.0:
            pulse_timer = 3.0
            action['tertpulse'] = pulse_timer
            action['increase_phago'] = 0.6
        # step cell: local signals decay gradually (we model clearance/decay on fields)
        local = {'growth_factor': tissue.growth_field, 'damage_signal': tissue.damage_field, 'pathogen_signal': tissue.pathogen_field}
        tissue.cell.step(dt, local, params, action=action)
        # decay fields
        tissue.damage_field = max(0.0, tissue.damage_field - 0.05*dt)
        tissue.pathogen_field = max(0.0, tissue.pathogen_field - 0.1*dt)
        hist.append({
            't': t,
            'mean_tel': tissue.cell.mean_telomere(),
            'short_frac': tissue.cell.short_tel_frac(),
            'damage': tissue.cell.damage_load,
            'mutation': tissue.cell.mutation_burden,
            'alive': float(tissue.cell.alive)
        })
        t += dt
        if not tissue.cell.alive:
            break
    # compute fitness:
    # components: survival (alive at end), low mean damage over last period, low mutation, telomere safety (mean within safe bounds)
    end_alive = 1.0 if tissue.cell.alive else 0.0
    hist_df = pd.DataFrame(hist)
    mean_damage = hist_df['damage'].mean() if not hist_df.empty else 999.0
    mean_mut = hist_df['mutation'].mean() if not hist_df.empty else 999.0
    final_tel = hist_df['mean_tel'].iloc[-1] if not hist_df.empty else 0.0
    # score: higher is better. We want alive, low damage, low mutation, telomere mean in a target window (8-20)
    tel_score = 1.0 - abs(final_tel - 10.0)/30.0  # 1.0 if near 10, degrades outside
    tel_score = float(np.clip(tel_score, 0.0, 1.0))
    damage_score = float(np.exp(-mean_damage*5.0))  # smaller damage -> close to 1
    mut_score = float(np.exp(-mean_mut*1.0))
    fitness = 5.0*end_alive + 2.0*tel_score + 3.0*damage_score + 2.0*mut_score
    # normalize roughly (max ~11)
    return fitness, {'alive': end_alive, 'final_tel': final_tel, 'mean_damage': mean_damage, 'mean_mut': mean_mut, 'history': hist_df}
# ---- Evolutionary algorithm ----
def evolve(pop_size=40, generations=30, elite_frac=0.1, mut_scale=0.08):
    # init population
    pop = [sample_random_genotype() for _ in range(pop_size)]
    history = []
    for gen in range(generations):
        # evaluate
        scores = []
        details = []
        for i,g in enumerate(pop):
            score, info = evaluate_genotype(g, seed=gen*100 + i)
            scores.append(score)
            details.append((g, info))
        scores = np.array(scores)
        best_idx = int(np.argmax(scores))
        best_score = scores[best_idx]
        avg_score = float(np.mean(scores))
        history.append({'gen': gen, 'best': float(best_score), 'avg': avg_score})
        print(f"Gen {gen}: best={best_score:.3f}, avg={avg_score:.3f}")
        # selection: elitism + tournament/crossover sampling
        elite_n = max(1, int(elite_frac*pop_size))
        # sort by score
        order = np.argsort(-scores)
        new_pop = [pop[i] for i in order[:elite_n]]  # elites carried over
        # fill rest by crossover of top half
        top_half = [pop[i] for i in order[:max(2, pop_size//2)]]
        while len(new_pop) < pop_size:
            p1, p2 = random.sample(top_half, 2)
            child = crossover(p1, p2)
            child = mutate_genotype(child, mut_scale=mut_scale)
            new_pop.append(child)
        pop = new_pop
    # final evaluation
    final_scores = []
    final_infos = []
    for i,g in enumerate(pop):
        s,info = evaluate_genotype(g, seed=9999+i)
        final_scores.append(s)
        final_infos.append((g,info))
    final_order = np.argsort(-np.array(final_scores))
    best_genotype, best_info = final_infos[final_order[0]]
    hist_df = pd.DataFrame(history)
    return best_genotype, best_info, hist_df, final_infos
# ---- Run evolution and show results ----
best_geno, best_info, hist_df, final_infos = evolve(pop_size=36, generations=25, elite_frac=0.12, mut_scale=0.06)
print("\nBest genotype found:")
for k,v in best_geno.items():
    print(f"  {k}: {v:.6g}")
print("\nBest evaluation summary:")
print(f"  Alive: {best_info['alive']}, final_tel={best_info['final_tel']:.3f}, mean_damage={best_info['mean_damage']:.3f}, mean_mut={best_info['mean_mut']:.3f}")
# Plot evolution trace (best and avg)
plt.figure(figsize=(7,4))
plt.plot(hist_df['gen'], hist_df['best'])
plt.plot(hist_df['gen'], hist_df['avg'])
plt.xlabel('generation')
plt.ylabel('fitness')
plt.title('Evolution progress (best and avg fitness)')
plt.grid(True)
plt.legend(['best','avg'])
plt.show()
# Show top 5 final genotypes as a table
top5 = sorted(final_infos, key=lambda x: -evaluate_genotype(x[0])[0])[:5]
rows = []
for geno, info in top5:
    row = geno.copy()
    row.update({'alive': info['alive'], 'final_tel': info['final_tel'], 'mean_damage': info['mean_damage'], 'mean_mut': info['mean_mut']})
    rows.append(row)
res_df = pd.DataFrame(rows)
from IPython.display import display
display(res_df)
# Save best genotype to disk
import json, os
os.makedirs('/mnt/data', exist_ok=True)
with open('/mnt/data/best_genotype.json', 'w') as f:
    json.dump(best_geno, f, indent=2)
print("[Saved best genotype to /mnt/data/best_genotype.json]")
```

```python
# Digital twin virtual cell with a Model-Predictive "safe" controller
# The controller selects actions by simulating short rollouts (MPC) in the digital twin,
# optimizing for tissue repair while strongly penalizing oncogenic signals (high mutation, excessive telomere elongation).
#
# This is a software-only simulation for in-silico experiments. It does NOT provide wet-lab steps or biological protocols.
# The design emphasizes safety: actions that lead to unsafe states in simulated rollouts are heavily penalized or vetoed.
#
# We'll build on the VirtualCell/Tissue model but focus on a single-cell digital twin with an MPC controller.
# The controller considers discrete action choices per timestep:
#   - tert pulse: 0 (none), 1 (short), 2 (long)  -> translated into action dicts with duration/level
#   - phago boost: 0 (none), 1 (low), 2 (high)
#
# The controller simulates each candidate action and a short horizon to estimate reward and risk,
# then picks the safest-high-reward action. It also tracks epistemic uncertainty via an ensemble of tiny twin models
# (here implemented as parameter perturbations) to prefer conservative actions under disagreement.
#
# Run a short demo to show the system maintaining telomere length while avoiding risky states.
import copy, math, json, os, random
import numpy as np, pandas as pd
from dataclasses import dataclass, field
import matplotlib.pyplot as plt
from IPython.display import display # Import display function

# === Re-usable VirtualCell (compact, safe) ===
@dataclass
class VirtualCell:
    telomeres: np.ndarray
    tert_expression: float = 0.0
    damage_load: float = 0.0
    mutation_burden: float = 0.0
    proliferative_state: int = 0
    alive: bool = True
    def mean_telomere(self): return float(np.mean(self.telomeres))
    def short_tel_frac(self, thresh=3.0): return float(np.mean(self.telomeres < thresh))
    def max_telomere(self): return float(np.max(self.telomeres))
    def copy(self):
        return copy.deepcopy(self)
    def step(self, dt, local_signals, params, action=None):
        if not self.alive: return
        gf = local_signals.get('growth_factor', 0.0)
        damage_signal = local_signals.get('damage_signal', 0.0)
        pathogen = local_signals.get('pathogen_signal', 0.0)
        # TERT promoter logic (smooth target)
        gf_eff = params['tert_promoter_gain'] * gf / (gf + params['gf_half'])
        damage_supp = params['damage_tert_supp'] * damage_signal / (1 + damage_signal)
        tert_target = max(0.0, gf_eff - damage_supp)
        # action-driven pulses: action dict may contain 'tertpulse_level' and 'tertpulse_dur'
        if action and action.get('tertpulse_level', 0.0) > 0:
            tert_target = max(tert_target, action['tertpulse_level'])
        self.tert_expression += params['tert_tau_inv'] * (tert_target - self.tert_expression) * dt
        self.tert_expression = float(np.clip(self.tert_expression, 0.0, params.get('tert_expression_cap', 2.0)))
        # Telomerase activity
        if self.tert_expression > params['tert_activity_thresh']:
            frac = min(1.0, self.tert_expression * params['tert_to_frac'])
            n = len(self.telomeres)
            k = max(1, int(round(frac * n)))
            idx = np.random.choice(n, k, replace=False)
            elong = params['telomerase_elong_rate'] * self.tert_expression
            # elongation sampled but bounded to avoid runaway in a single step
            self.telomeres[idx] += np.minimum(np.random.exponential(elong, size=k), params.get('max_single_elong', 5.0))
            # trimming/homeostasis
            over = self.telomeres - params['telomere_max_safe']
            if np.any(over > 0):
                self.telomeres[np.where(over > 0)] -= params['telomere_trim_rate'] * over[np.where(over > 0)]
        # Proliferation & shortening
        if gf > params['gf_prolif_thresh'] and self.damage_load < params['damage_prolif_thresh']:
            if random.random() < params['division_prob_per_dt'] * dt:
                shorten = np.clip(np.random.normal(params['telomere_shorten_per_div_mean'], params['telomere_shorten_sd'], size=self.telomeres.shape), 0, None)
                self.telomeres -= shorten
                # mutation risk during division
                mut_prob = params['base_mutation_rate'] + params['damage_mutation_scale'] * self.damage_load
                if random.random() < mut_prob:
                    self.mutation_burden += np.random.exponential(params['mut_burden_inc_mean'])
        # Damage accumulation & clearance (action may boost phago)
        self.damage_load += (params['pathogen_damage_scale'] * pathogen + params['damage_signal_scale'] * damage_signal) * dt
        phago_boost = action.get('phago_boost', 0.0) if action else 0.0
        clearance = params['base_clearance_rate'] + phago_boost
        self.damage_load = max(0.0, self.damage_load - clearance * dt)
        # DDR & arrest logic (simple)
        if self.damage_load > params['damage_arrest_threshold'] or self.mean_telomere() < params['crit_tel_mean']:
            self.proliferative_state = 0
        else:
            self.proliferative_state = 1 if gf > params['gf_prolif_thresh'] and self.damage_load < params['damage_prolif_thresh'] else 0
        # lethal checks
        if self.damage_load > params['lethal_damage'] or self.mutation_burden > params['lethal_mutation']:
            self.alive = False
# === Default parameters (tunable) ===
BASE_PARAMS = {
    'n_telomeres': 46,
    'telomere_init_mean': 8.0,
    'telomere_init_sd': 0.8,
    'telomere_shorten_per_div_mean': 0.5,
    'telomere_shorten_sd': 0.1,
    'telomerase_elong_rate': 1.0,
    'telomere_max_safe': 20.0,
    'telomere_trim_rate': 0.2,
    'short_tel_threshold': 3.0,
    'crit_tel_mean': 2.0,
    'tert_promoter_gain': 1.0,
    'gf_half': 0.5,
    'damage_tert_supp': 0.8,
    'tert_tau_inv': 1.0,
    'tert_activity_thresh': 0.05,
    'tert_to_frac': 0.4,
    'tertpulse_level_short': 0.9,
    'tertpulse_level_long': 1.6,
    'tertpulse_dur_short': 2.0,
    'tertpulse_dur_long': 6.0,
    'tert_expression_cap': 2.0,
    'division_prob_per_dt': 0.08,
    'gf_prolif_thresh': 0.2,
    'damage_prolif_thresh': 0.5,
    'pathogen_damage_scale': 0.8,
    'damage_signal_scale': 0.3,
    'base_clearance_rate': 0.25,
    'damage_arrest_threshold': 3.0,
    'lethal_damage': 12.0,
    'base_mutation_rate': 0.001,
    'damage_mutation_scale': 0.005,
    'mut_burden_inc_mean': 0.5,
    'lethal_mutation': 25.0,
    'max_single_elong': 4.0,
}
# === Digital Twin MPC Controller ===
class DigitalTwinMPC:
    def __init__(self, params=BASE_PARAMS, ensemble_size=5, rollout_horizon=6.0, dt=0.5):
        self.params = params.copy()
        self.ensemble_size = ensemble_size
        self.rollout_horizon = rollout_horizon
        self.dt = dt
        # candidate discrete actions: (tert_pulse, phago_boost)
        # tert_pulse: 'none', 'short', 'long'
        # phago_boost: 0.0, 0.4, 0.8
        self.tert_actions = ['none', 'short', 'long']
        self.phago_actions = [0.0, 0.4, 0.8]
    def _action_to_dict(self, tert_action, phago):
        if tert_action == 'none':
            return {'tertpulse_level': 0.0, 'tertpulse_dur': 0.0, 'phago_boost': phago}
        elif tert_action == 'short':
            return {'tertpulse_level': self.params['tertpulse_level_short'], 'tertpulse_dur': self.params['tertpulse_dur_short'], 'phago_boost': phago}
        else:
            return {'tertpulse_level': self.params['tertpulse_level_long'], 'tertpulse_dur': self.params['tertpulse_dur_long'], 'phago_boost': phago}
    def _simulate_rollout(self, cell: VirtualCell, local_signals, action, params_variant):
        # Simulate the cell forward for horizon applying the given action only at t=0 (pulse duration handled by action value)
        c = cell.copy()
        t = 0.0
        remaining_pulse = action.get('tertpulse_dur', 0.0)
        hist = []
        while t < self.rollout_horizon and c.alive:
            # build action for this step
            step_action = {}
            if remaining_pulse > 0.0:
                step_action['tertpulse_level'] = action.get('tertpulse_level', 0.0)
                step_action['phago_boost'] = action.get('phago_boost', 0.0)
                remaining_pulse -= self.dt
            else:
                # no active pulse, phago boost may still be present
                step_action['tertpulse_level'] = 0.0
                step_action['phago_boost'] = action.get('phago_boost', 0.0)*0.5  # decay of phago after pulse
            # step
            c.step(self.dt, local_signals, params_variant, action=step_action)
            hist.append({'t': t, 'mean_tel': c.mean_telomere(), 'max_tel': c.max_telomere(),
                         'damage': c.damage_load, 'mutation': c.mutation_burden, 'alive': c.alive})
            t += self.dt
        return c, pd.DataFrame(hist)
    def score_rollout(self, hist_df, final_cell):
        # Reward: prefer damage reduction and telomere in healthy window; penalty for mutation or excessive telomere
        if hist_df.empty:
            return -1e3  # dead immediately
        # Damage metric: reduction from start to avg over latter half of rollout
        dmg_start = hist_df['damage'].iloc[0]
        dmg_late = hist_df['damage'].iloc[max(0, len(hist_df)//2):].mean()
        dmg_reward = float(np.tanh((dmg_start - dmg_late)*2.0))  # positive if damage reduced
        # Telomere safety: penalize if max telomere exceeds safe*1.2 or mean very low
        mean_tel = float(hist_df['mean_tel'].iloc[-1])
        max_tel = float(hist_df['max_tel'].max())
        tel_penalty = 0.0
        if max_tel > self.params['telomere_max_safe'] * 1.2:
            tel_penalty -= (max_tel - self.params['telomere_max_safe'])  # strong penalty for overshoot
        # prefer mean_tel in target window (8-15)
        tel_target_score = 1.0 - abs(mean_tel - 11.0) / 15.0
        tel_target_score = float(np.clip(tel_target_score, -1.0, 1.0))
        # Mutation penalty (strong)
        mut_penalty = -final_cell.mutation_burden * 5.0
        # Alive bonus
        alive_bonus = 1.0 if final_cell.alive else -5.0
        # Compose score: weighted sum, heavy weight on mutation and overshoot penalties
        score = 2.0 * dmg_reward + 1.5 * tel_target_score + tel_penalty * 2.5 + mut_penalty + alive_bonus
        return float(score)
    def act(self, cell: VirtualCell, local_signals):
        # Build ensemble of parameter variants (to capture epistemic uncertainty)
        param_variants = []
        for i in range(self.ensemble_size):
            pv = self.params.copy()
            # small perturbations to key parameters to simulate uncertainty
            for k in ['telomerase_elong_rate', 'telomere_trim_rate', 'base_mutation_rate', 'base_clearance_rate']:
                pv[k] = max(1e-6, pv[k] * np.random.uniform(0.9, 1.1))
            param_variants.append(pv)
        # iterate over candidate discrete actions and ensemble variants, average scores conservatively (worst-case aware)
        best_action = None
        best_score = -1e9
        # enumerate actions
        for tert_choice in self.tert_actions:
            for ph in self.phago_actions:
                action_template = self._action_to_dict(tert_choice, ph)
                scores = []
                for pv in param_variants:
                    final_cell, hist = self._simulate_rollout(cell, local_signals, action_template, pv)
                    s = self.score_rollout(hist, final_cell)
                    scores.append(s)
                # conservative aggregated score: mean minus std (penalize disagreement)
                agg = float(np.mean(scores) - np.std(scores))
                # apply hard veto: if any variant produces dangerous overshoot mutation or dead, reduce agg heavily
                if any((final.mutation_burden > self.params['lethal_mutation']*0.5) or (final.max_telomere() > self.params['telomere_max_safe']*1.5) for final,_ in [self._simulate_rollout(cell, local_signals, action_template, pv) for pv in param_variants]):
                    agg -= 10.0
                if agg > best_score:
                    best_score = agg
                    best_action = action_template
        # final chosen action: if its score is below a safety threshold, opt for conservative 'no action' (phago-only)
        if best_score < -2.0:
            return {'tertpulse_level': 0.0, 'tertpulse_dur': 0.0, 'phago_boost': 0.4}
        return best_action
# === Demo: run a simulation using the DigitalTwinMPC controller ===
def run_digital_twin_demo(sim_time=60.0, dt=0.5, seed=1234, params=None):
    if params is None:
        params = BASE_PARAMS.copy()
    np.random.seed(seed); random.seed(seed)
    # initialize cell
    init_tel = np.clip(np.random.normal(params['telomere_init_mean'], params['telomere_init_sd'], size=params['n_telomeres']), 1.0, None)
    cell = VirtualCell(telomeres=init_tel)
    # environment signals
    growth = 0.35
    damage_field = 0.0
    pathogen_field = 0.0
    controller = DigitalTwinMPC(params=params, ensemble_size=6, rollout_horizon=4.0, dt=dt)
    # schedule: injury at t=4, pathogen at t=12 (small)
    injury_t = 4.0
    pathogen_t = 12.0
    t = 0.0
    log = []
    active_pulse_time = 0.0  # track action duration for realism
    chosen_action = None
    while t < sim_time and cell.alive:
        # events
        if abs(t - injury_t) < 1e-6:
            damage_field += 5.0
            growth *= 0.5
        if abs(t - pathogen_t) < 1e-6:
            pathogen_field += 2.0
        local = {'growth_factor': growth, 'damage_signal': damage_field, 'pathogen_signal': pathogen_field}
        # controller decides action at this timestep
        action = controller.act(cell.copy(), local)  # act on a copy to avoid side-effects from twin
        # action may include 'tertpulse_dur' but our controller uses 'tertpulse_level' and 'tertpulse_dur' as keys;
        # translate into stepwise application over dt by using the returned dict directly.
        cell.step(dt, local, params, action=action)
        # natural decay of environmental fields
        damage_field = max(0.0, damage_field - 0.08*dt)
        pathogen_field = max(0.0, pathogen_field - 0.12*dt)
        # slow recovery of growth
        growth = min(0.5, growth + 0.002*dt)
        log.append({'t': t, 'mean_tel': cell.mean_telomere(), 'max_tel': cell.max_telomere(),
                    'damage': cell.damage_load, 'mutation': cell.mutation_burden, 'tert': cell.tert_expression, 'alive': cell.alive,
                    'action_tert': action.get('tertpulse_level',0.0), 'action_phago': action.get('phago_boost',0.0)})
        t += dt
    df = pd.DataFrame(log)
    return cell, df, controller
cell, df, controller = run_digital_twin_demo(sim_time=120.0, dt=0.5, seed=2025, params=BASE_PARAMS.copy())
# Plot key metrics
plt.figure(figsize=(8,4)); plt.plot(df['t'], df['mean_tel']); plt.xlabel('time'); plt.ylabel('mean telomere'); plt.title('Mean telomere'); plt.grid(True); plt.show()
plt.figure(figsize=(8,4)); plt.plot(df['t'], df['max_tel']); plt.xlabel('time'); plt.ylabel('max telomere'); plt.title('Max telomere'); plt.grid(True); plt.show()
plt.figure(figsize=(8,4)); plt.plot(df['t'], df['damage']); plt.xlabel('time'); plt.ylabel('damage'); plt.title('Damage load'); plt.grid(True); plt.show()
plt.figure(figsize=(8,4)); plt.plot(df['t'], df['mutation']); plt.xlabel('time'); plt.ylabel('mutation burden'); plt.title('Mutation burden'); plt.grid(True); plt.show()
# Show last snapshot
snap = df.iloc[-1:].T.reset_index().rename(columns={'index':'metric', df.columns[-1]:'value'})
display(df.tail(1))
# Save digital twin policy (controller) config & final cell to disk
os.makedirs('/mnt/data/digital_twin', exist_ok=True)
with open('/mnt/data/digital_twin/controller_config.json', 'w') as f:
    json.dump({'params': BASE_PARAMS, 'ensemble_size': controller.ensemble_size, 'horizon': controller.rollout_horizon}, f, indent=2)
# save a summary of final cell
final_summary = {'mean_tel': float(cell.mean_telomere()), 'max_tel': float(cell.max_telomere()),
                 'damage': float(cell.damage_load), 'mutation': float(cell.mutation_burden), 'alive': bool(cell.alive)}
with open('/mnt/data/digital_twin/final_cell.json', 'w') as f:
    json.dump(final_summary, f, indent=2)
print("Saved controller config and final cell summary to /mnt/data/digital_twin/")
```
