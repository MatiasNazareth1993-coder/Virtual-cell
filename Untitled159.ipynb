```python
class SyntheticTelomere:
    def __init__(self, initial_length=1000.0, sequence_vector=None):
        self.L = float(initial_length)
        self.sequence_vector = sequence_vector or np.random.rand(8)
        self.resistance = np.clip(np.mean(self.sequence_vector), 0.2, 0.9)

    def shorten(self, divisions):
        loss = divisions * (1.0 - self.resistance)
        self.L = max(self.L - loss, 0.0)

    def extend(self, telomerase_signal):
        gain = telomerase_signal * (0.5 + self.resistance)
        self.L += gain
```

```python
class PhaseAwareOptimizer:
    def __init__(self, base_lr=1e-3):
        self.base_lr = base_lr
        self.phase = "homeostasis"

    def select(self, damage, telomere, instability):
        if instability > 0.7:
            self.phase = "stabilization"
            return OPTIMIZERS["rmsprop"](self.base_lr * 0.3)

        if telomere < 300:
            self.phase = "rejuvenation"
            return OPTIMIZERS["adam"](self.base_lr)

        if damage < 50:
            self.phase = "maintenance"
            return OPTIMIZERS["lion"](self.base_lr * 0.5)

        self.phase = "repair"
        return OPTIMIZERS["adamw"](self.base_lr * 0.8)
```

```python
def build_world_model(input_dim=32):
    x = tf.keras.Input(shape=(input_dim,))
    h = tf.keras.layers.Dense(256, activation="selu")(x)
    h = tf.keras.layers.Dense(256, activation="elu")(h)
    h = tf.keras.layers.Dense(128, activation="tanh")(h)

    tel_pred = tf.keras.layers.Dense(1)(h)
    dmg_pred = tf.keras.layers.Dense(1)(h)
    div_pred = tf.keras.layers.Dense(1, activation="sigmoid")(h)

    return tf.keras.Model(x, [tel_pred, dmg_pred, div_pred])
```

```python
import tensorflow as tf

class PolyActivation(tf.keras.layers.Layer):
    def __init__(self):
        super().__init__()
        self.weights_mix = tf.Variable(
            initial_value=tf.ones(8) / 8,
            trainable=True,
            dtype=tf.float32
        )

    def call(self, x):
        acts = tf.stack([
            tf.nn.relu(x),
            tf.nn.gelu(x),
            tf.nn.silu(x),
            tf.nn.mish(x),
            tf.nn.selu(x),
            tf.nn.tanh(x),
            tf.nn.softplus(x),
            tf.nn.softsign(x)
        ], axis=-1)

        mix = tf.nn.softmax(self.weights_mix)
        return tf.reduce_sum(acts * mix, axis=-1)
```

```python
class PolyOptimizer(tf.keras.optimizers.Optimizer):
    def __init__(self, lr=1e-3, name="PolyOptimizer", **kwargs):
        super().__init__(name, **kwargs)

        self.opt_fast = tf.keras.optimizers.AdamW(lr)
        self.opt_general = tf.keras.optimizers.Lion(lr * 0.5)
        self.opt_sparse = tf.keras.optimizers.Adagrad(lr * 0.2)
        self.opt_stable = tf.keras.optimizers.RMSprop(lr * 0.3)
        self.opt_long = tf.keras.optimizers.Ftrl(lr * 0.1)

        self.alpha = tf.Variable(0.5, trainable=False)  # aging ↔ rejuvenation

    def apply_gradients(self, grads_and_vars, **kwargs):
        for g, v in grads_and_vars:
            if g is None:
                continue

            # Phase-aware optimizer blending
            if self.alpha < 0.3:
                self.opt_fast.apply_gradients([(g, v)])
            elif self.alpha < 0.6:
                self.opt_general.apply_gradients([(g, v)])
            else:
                self.opt_stable.apply_gradients([(g, v)])
```

```python
import tensorflow as tf

class PolyOptimizer(tf.keras.optimizers.Optimizer):
    def __init__(self, lr=3e-4, name="PolyOptimizer", **kwargs):
        super().__init__(name, **kwargs)

        self.opts = {
            "adam": tf.keras.optimizers.Adam(lr),
            "adamw": tf.keras.optimizers.AdamW(lr),
            "lion": tf.keras.optimizers.Lion(lr*0.5),
            "nadam": tf.keras.optimizers.Nadam(lr),
            "rms": tf.keras.optimizers.RMSprop(lr),
            "sgd": tf.keras.optimizers.SGD(lr, momentum=0.9),
            "adagrad": tf.keras.optimizers.Adagrad(lr),
            "adadelta": tf.keras.optimizers.Adadelta(),
            "ftrl": tf.keras.optimizers.Ftrl(lr),
        }

        # Learnable optimizer weights
        self.mix_logits = tf.Variable(
            tf.zeros(len(self.opts)), trainable=True
        )

    def apply_gradients(self, grads_and_vars, **kwargs):
        weights = tf.nn.softmax(self.mix_logits)
        for (grad, var) in grads_and_vars:
            if grad is None:
                continue
            for w, opt in zip(weights, self.opts.values()):
                opt.apply_gradients([(w * grad, var)])
```

```python
class PolyActivation(tf.keras.layers.Layer):
    def __init__(self):
        super().__init__()
        self.alpha = tf.Variable(tf.ones(12)/12, trainable=True)

    def call(self, x):
        acts = tf.stack([
            tf.nn.relu(x),
            tf.nn.relu6(x),
            tf.nn.gelu(x),
            tf.nn.silu(x),
            tf.nn.swish(x),
            tf.nn.mish(x),
            tf.nn.selu(x),
            tf.nn.tanh(x),
            tf.nn.sigmoid(x),
            tf.nn.softplus(x),
            tf.nn.softsign(x),
            tf.nn.leaky_relu(x),
        ], axis=-1)

        w = tf.nn.softmax(self.alpha)
        return tf.reduce_sum(acts * w, axis=-1)
```

```python
def self_improvement_loop(model, data_fn, epochs=50):
    for e in range(epochs):
        x, y = data_fn()
        model.train_on_batch(x, y)

        # Encourage exploration early, stability later
        temp = max(0.1, 1.0 - e/epochs)
        model.optimizer.mix_logits.assign_add(
            tf.random.normal(model.optimizer.mix_logits.shape) * temp
        )
```

```python
import tensorflow as tf

class OptimizerIntelligence(tf.keras.optimizers.Optimizer):
    def __init__(self, lr=3e-4, name="OptimizerIntelligence", **kwargs):
        super().__init__(name, **kwargs)

        # Core optimizers
        self.optims = [
            tf.keras.optimizers.Adam(lr),
            tf.keras.optimizers.AdamW(lr),
            tf.keras.optimizers.Lion(lr*0.5),
            tf.keras.optimizers.RMSprop(lr),
            tf.keras.optimizers.SGD(lr, momentum=0.9),
            tf.keras.optimizers.Adagrad(lr),
            tf.keras.optimizers.Adadelta(),
            tf.keras.optimizers.Ftrl(lr),
        ]

        # Learnable optimizer weights (constant intelligence)
        self.logits = tf.Variable(tf.zeros(len(self.optims)), trainable=True)

        # Online statistics
        self.grad_ema = tf.Variable(0.0, trainable=False)
        self.var_ema  = tf.Variable(0.0, trainable=False)
        self.loss_ema = tf.Variable(0.0, trainable=False)

    def update_stats(self, grad, loss):
        gnorm = tf.norm(grad)
        self.grad_ema.assign(0.95 * self.grad_ema + 0.05 * gnorm)
        self.var_ema.assign(0.95 * self.var_ema + 0.05 * tf.square(gnorm))
        self.loss_ema.assign(0.95 * self.loss_ema + 0.05 * loss)

    def apply_gradients(self, grads_and_vars, loss=None, **kwargs):
        weights = tf.nn.softmax(self.logits)

        for grad, var in grads_and_vars:
            if grad is None:
                continue

            if loss is not None:
                self.update_stats(grad, loss)

            # Curvature & noise adaptation
            noise_factor = tf.tanh(self.var_ema)
            adapt = tf.stop_gradient(0.5 + noise_factor)

            for w, opt in zip(weights, self.optims):
                opt.apply_gradients([(w * adapt * grad, var)])
```

```python
class ActivationIntelligence(tf.keras.layers.Layer):
    def __init__(self):
        super().__init__()
        self.mix = tf.Variable(tf.ones(10)/10, trainable=True)

    def call(self, x):
        activations = tf.stack([
            tf.nn.relu(x),
            tf.nn.gelu(x),
            tf.nn.silu(x),
            tf.nn.mish(x),
            tf.nn.selu(x),
            tf.nn.tanh(x),
            tf.nn.sigmoid(x),
            tf.nn.softplus(x),
            tf.nn.softsign(x),
            tf.nn.leaky_relu(x),
        ], axis=-1)

        weights = tf.nn.softmax(self.mix)
        return tf.reduce_sum(activations * weights, axis=-1)
```

```python
class OrganOptimizerBrain:
    def __init__(self, base_optimizer, lr, safety_projection=True):
        self.optimizer = base_optimizer
        self.lr = lr
        self.safety_projection = safety_projection

    def step(self, gradients, state):
        update = -self.lr * gradients

        if self.safety_projection:
            update = self.project_safe(update)

        return state + update

    def project_safe(self, update):
        # Cancer-safety / telomere-stability constraint
        return np.clip(update, -0.05, 0.05)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MetaOptimizer(nn.Module):
    def __init__(self, state_dim, hidden_dim=128):
        super().__init__()
        self.fc1 = nn.Linear(state_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)

        # Outputs optimizer parameters
        self.lr_head = nn.Linear(hidden_dim, 1)
        self.momentum_head = nn.Linear(hidden_dim, 1)
        self.optimizer_gate = nn.Linear(hidden_dim, 5)  # Adam, SGD, Lion, RMSprop, Adafactor

    def forward(self, state):
        x = F.silu(self.fc1(state))
        x = F.silu(self.fc2(x))

        lr = torch.sigmoid(self.lr_head(x)) * 1e-2
        momentum = torch.sigmoid(self.momentum_head(x))

        optimizer_choice = torch.softmax(self.optimizer_gate(x), dim=-1)

        return lr, momentum, optimizer_choice
```

```python
class OptimizerGenome:
    def __init__(self, lr, momentum, optimizer_type):
        self.lr = lr
        self.momentum = momentum
        self.optimizer_type = optimizer_type

    def mutate(self, scale=0.1):
        self.lr *= (1 + torch.randn(1).item() * scale)
        self.momentum = torch.clamp(
            self.momentum + torch.randn(1).item() * scale, 0, 1
        )
```

```python
def meta_training_step(meta_optimizer, system_state, reward):
    lr, momentum, opt_probs = meta_optimizer(system_state)

    # Reward shapes optimizer behavior
    loss = -reward * (lr.mean() + momentum.mean())

    loss.backward()
```

```python
import tensorflow as tf
from tensorflow.keras import layers

class BioWorldModel(tf.keras.Model):
    def __init__(self, d_model=256, num_heads=8):
        super().__init__()
        self.embedding = layers.Dense(d_model)
        self.transformer = layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads # key_dim should be for each head
        )

        # Biological control heads
        self.telomerase = layers.Dense(1, activation="sigmoid")
        self.division = layers.Dense(1, activation="sigmoid")
        self.phagocytosis = layers.Dense(1, activation="sigmoid")

    def call(self, x):
        x = self.embedding(x)
        # Add a sequence dimension (treat as sequence of length 1)
        x = tf.expand_dims(x, axis=1) # Shape becomes (batch_size, 1, d_model)
        z = self.transformer(x, x) # z shape (batch_size, 1, d_model)
        # Remove the sequence dimension before passing to dense layers
        z = tf.squeeze(z, axis=1) # Shape becomes (batch_size, d_model)

        return {
            "telomerase": self.telomerase(z),
            "division": self.division(z),
            "phagocytosis": self.phagocytosis(z),
        }
```

```python
import tensorflow as tf
from tensorflow.keras import layers

class BioAgent(tf.keras.Model):
    def __init__(self, out_dim):
        super().__init__()
        self.net = tf.keras.Sequential([
            layers.Dense(128, activation="silu"),
            layers.Dense(64, activation="mish"),
            layers.Dense(out_dim, activation="sigmoid")
        ])

    def call(self, state):
        return self.net(state)
```

```python
class Governor(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.policy = tf.keras.Sequential([
            layers.Dense(128, activation="gelu"),
            layers.Dense(5, activation="softmax")  # weight agents
        ])

    def call(self, agent_outputs):
        # agent_outputs is expected to be a list of tensors from BioAgents
        concatenated_outputs = tf.concat(agent_outputs, axis=-1) # This will be (batch_size, 5)
        weights = self.policy(concatenated_outputs) # Weights will be (batch_size, 5)
        # Return the element-wise product, which is the weighted agent decisions.
        # Do not reduce_sum here, so the output is still (batch_size, 5).
        decision = weights * concatenated_outputs
        return decision
```

```python
def governance_step(state):
    stem_out     = stem_agent(state)      # (batch_size, 2) for divide/differentiate
    telo_out     = telomere_agent(state)  # (batch_size, 1) for telomerase pulse
    safety_out   = safety_agent(state)    # (batch_size, 1) for cancer risk
    repair_out   = repair_agent(state)    # (batch_size, 1) for regeneration signal
    cleanup_out  = cleanup_agent(state)   # (batch_size, 1) for phagocytosis

    # Safety gate - check the scalar value of safety_out
    if safety_out[0, 0] > 0.8:
        # Return zero actions if safety threshold is exceeded
        zero_action = tf.zeros((1, 3), dtype=tf.float32) # For divide, telomerase, phagocytosis
        return {"divide": 0.0, "telomerase": 0.0, "phagocytosis": 0.0, "action_tensor": zero_action}

    # Governor expects a list of tensors
    # The governor will return a (batch_size, 5) tensor if Governor.call is modified as suggested
    decision_vector = governor([
        stem_out, telo_out, repair_out, cleanup_out
    ]) # Shape (batch_size, 5)

    # Assuming batch_size = 1 (from sample_virtual_cell)
    # Map decision_vector elements to actions based on concatenation order:
    # [stem_out_0, stem_out_1, telo_out_0, repair_out_0, cleanup_out_0]
    # The governance_step only returns divide, telomerase, phagocytosis
    divide_action = decision_vector[0, 0]
    # differentiate_action = decision_vector[0, 1] # Not returned by governance_step
    telomerase_action = decision_vector[0, 2]
    # repair_action = decision_vector[0, 3] # Not returned by governance_step
    phagocytosis_action = decision_vector[0, 4]

    # Concatenate the relevant actions into a single tensor for curiosity_step
    action_for_curiosity = tf.concat([
        tf.expand_dims(divide_action, axis=0),
        tf.expand_dims(telomerase_action, axis=0),
        tf.expand_dims(phagocytosis_action, axis=0)
    ], axis=0) # Shape (3,)
    action_for_curiosity = tf.expand_dims(action_for_curiosity, axis=0) # Shape (1, 3)

    return {
        "divide": divide_action,
        "telomerase": telomerase_action,
        "phagocytosis": phagocytosis_action,
        "action_tensor": action_for_curiosity
    }
```

```python
import tensorflow as tf
from tensorflow.keras import layers

class ForwardModel(tf.keras.Model):
    def __init__(self, state_dim):
        super().__init__()
        self.net = tf.keras.Sequential([
            layers.Dense(256, activation="gelu"),
            layers.Dense(256, activation="silu"),
            layers.Dense(state_dim)
        ])

    def call(self, state, action):
        x = tf.concat([state, action], axis=-1)
        return self.net(x)
```

```python
class CuriosityModule:
    def __init__(self, forward_model):
        self.forward_model = forward_model

    def intrinsic_reward(self, state, action, next_state):
        pred_next = self.forward_model(state, action)
        error = tf.reduce_mean(tf.square(pred_next - next_state))
        return error
```

```python
def curiosity_step(state, action, next_state):
    r_int = curiosity.intrinsic_reward(state, action, next_state)

    # Cancer safety clamp
    r_int = tf.clip_by_value(r_int, 0.0, 1.0)

    total_reward = (
        0.7 * r_int
        - 0.2 * genomic_instability(state)
        - 0.1 * excessive_division(action)
    )

    return total_reward
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import random

# -------------------------
# Adaptive Activation
# -------------------------
class AdaptiveActivation(nn.Module):
    def __init__(self):
        super().__init__()
        self.alpha = nn.Parameter(torch.zeros(4))  # mixing weights

    def forward(self, x):
        w = torch.softmax(self.alpha, dim=0)
        return (
            w[0] * F.relu(x) +
            w[1] * torch.tanh(x) +
            w[2] * F.silu(x) +
            w[3] * F.elu(x)
        )

# -------------------------
# Core Model
# -------------------------
class CoreModel(nn.Module):
    def __init__(self, dim=32):
        super().__init__()
        self.fc1 = nn.Linear(dim, 64)
        self.act1 = AdaptiveActivation()
        self.fc2 = nn.Linear(64, 64)
        self.act2 = AdaptiveActivation()
        self.out = nn.Linear(64, 1)

    def forward(self, x):
        x = self.act1(self.fc1(x))
        x = self.act2(self.fc2(x))
        return self.out(x)

# -------------------------
# Meta Optimizer (learns LR)
# -------------------------
class MetaOptimizer:
    def __init__(self, model):
        self.model = model
        self.lr = 1e-3
        self.opt = torch.optim.Adam(model.parameters(), lr=self.lr)

    def step(self, loss):
        loss.backward()
        grad_norm = 0.0
        for p in self.model.parameters():
            if p.grad is not None:
                grad_norm += p.grad.norm().item()

        # self-regulation (homeostasis)
        if grad_norm > 10:
            self.lr *= 0.5
        elif grad_norm < 1:
            self.lr *= 1.05

        self.lr = float(max(1e-5, min(self.lr, 1e-2)))
        for g in self.opt.param_groups:
            g['lr'] = self.lr

        self.opt.step()
        self.opt.zero_grad()
        return grad_norm, self.lr

# -------------------------
# Task Generator (Generality)
# -------------------------
def sample_task(x, task_id):
    if task_id == 0:   # linear
        return x.sum(dim=1, keepdim=True)
    if task_id == 1:   # nonlinear
        return torch.sin(x.sum(dim=1, keepdim=True))
    if task_id == 2:   # multiplicative
        return x.prod(dim=1, keepdim=True)
    if task_id == 3:   # threshold
        return (x.sum(dim=1, keepdim=True) > 0).float()

# -------------------------
# Training Loop
# -------------------------
model = CoreModel()
meta_opt = MetaOptimizer(model)

losses, lrs, grads = [], [], []

for step in range(2000):
    x = torch.randn(64, 32)
    task = random.randint(0, 3)   # multiple domains
    y = sample_task(x, task)

    y_hat = model(x)
    loss = F.mse_loss(y_hat, y)

    grad_norm, lr = meta_opt.step(loss)

    losses.append(loss.item())
    lrs.append(lr)
    grads.append(grad_norm)

# -------------------------
# Visuals
# -------------------------
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.plot(losses)
plt.title("Loss (multi-task)")

plt.subplot(1,3,2)
plt.plot(lrs)
plt.title("Self-adjusted Learning Rate")

plt.subplot(1,3,3)
plt.plot(grads)
plt.title("Gradient Norm (stability)")
plt.tight_layout()
plt.show()
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import random

# -------------------------
# Adaptive Activation
# -------------------------
class AdaptiveActivation(nn.Module):
    def __init__(self):
        super().__init__()
        self.alpha = nn.Parameter(torch.zeros(4))  # mixing weights

    def forward(self, x):
        w = torch.softmax(self.alpha, dim=0)
        return (
            w[0] * F.relu(x) +
            w[1] * torch.tanh(x) +
            w[2] * F.silu(x) +
            w[3] * F.elu(x)
        )

# -------------------------
# Core Model
# -------------------------
class CoreModel(nn.Module):
    def __init__(self, dim=32):
        super().__init__()
        self.fc1 = nn.Linear(dim, 64)
        self.act1 = AdaptiveActivation()
        self.fc2 = nn.Linear(64, 64)
        self.act2 = AdaptiveActivation()
        self.out = nn.Linear(64, 1)

    def forward(self, x):
        x = self.act1(self.fc1(x))
        x = self.act2(self.fc2(x))
        return self.out(x)

# -------------------------
# Meta Optimizer (learns LR)
# -------------------------
class MetaOptimizer:
    def __init__(self, model):
        self.model = model
        self.lr = 1e-3
        self.opt = torch.optim.Adam(model.parameters(), lr=self.lr)

    def step(self, loss):
        loss.backward()
        grad_norm = 0.0
        for p in self.model.parameters():
            if p.grad is not None:
                grad_norm += p.grad.norm().item()

        # self-regulation (homeostasis)
        if grad_norm > 10:
            self.lr *= 0.5
        elif grad_norm < 1:
            self.lr *= 1.05

        self.lr = float(max(1e-5, min(self.lr, 1e-2)))
        for g in self.opt.param_groups:
            g['lr'] = self.lr

        self.opt.step()
        self.opt.zero_grad()
        return grad_norm, self.lr

# -------------------------
# Task Generator (Generality)
# -------------------------
def sample_task(x, task_id):
    if task_id == 0:   # linear
        return x.sum(dim=1, keepdim=True)
    if task_id == 1:   # nonlinear
        return torch.sin(x.sum(dim=1, keepdim=True))
    if task_id == 2:   # multiplicative
        return x.prod(dim=1, keepdim=True)
    if task_id == 3:   # threshold
        return (x.sum(dim=1, keepdim=True) > 0).float()

# -------------------------
# Training Loop
# -------------------------
model = CoreModel()
meta_opt = MetaOptimizer(model)

losses, lrs, grads = [], [], []

for step in range(2000):
    x = torch.randn(64, 32)
    task = random.randint(0, 3)   # multiple domains
    y = sample_task(x, task)

    y_hat = model(x)
    loss = F.mse_loss(y_hat, y)

    grad_norm, lr = meta_opt.step(loss)

    losses.append(loss.item())
    lrs.append(lr)
    grads.append(grad_norm)

# -------------------------
# Visuals
# -------------------------
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.plot(losses)
plt.title("Loss (multi-task)")

plt.subplot(1,3,2)
plt.plot(lrs)
plt.title("Self-adjusted Learning Rate")

plt.subplot(1,3,3)
plt.plot(grads)
plt.title("Gradient Norm (stability)")
plt.tight_layout()
plt.show()
```

```python
# Digital Cell / Organism Simulation with Self-Regulation
# This code builds on the previous general intelligence substrate
# and turns it into a minimal "digital cell" with metabolism,
# telomere dynamics, division control, and self-regulation visuals.

import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Digital Cell Definition
# -------------------------
class DigitalCell:
    def __init__(self):
        self.energy = 1.0
        self.damage = 0.0
        self.telomere = 1.0
        self.age = 0
        self.alive = True

    def metabolize(self, nutrient):
        # energy gain with diminishing returns
        self.energy += 0.1 * nutrient
        self.energy = min(self.energy, 2.0)

    def accumulate_damage(self, stress):
        self.damage += 0.05 * stress
        self.damage = min(self.damage, 1.0)

    def repair(self):
        # repair costs energy, reduces damage
        if self.energy > 0.3:
            self.damage -= 0.1
            self.energy -= 0.1
        self.damage = max(self.damage, 0.0)

    def divide(self):
        # division only if safe
        if self.energy > 1.2 and self.damage < 0.3 and self.telomere > 0.3:
            self.energy *= 0.5
            self.telomere -= 0.1
            return DigitalCell()
        return None

    def step(self, environment):
        if not self.alive:
            return None

        nutrient, stress = environment
        self.metabolize(nutrient)
        self.accumulate_damage(stress)

        # telomere erosion with age
        self.telomere -= 0.01
        self.age += 1

        # conditional repair
        self.repair()

        # death condition
        if self.telomere <= 0 or self.damage >= 1.0 or self.energy <= 0:
            self.alive = False

        return self.divide()

# -------------------------
# Simulation Loop
# -------------------------
population = [DigitalCell()]
history = {
    "population": [],
    "avg_energy": [],
    "avg_damage": [],
    "avg_telomere": []
}

for t in range(200):
    new_cells = []
    for cell in population:
        offspring = cell.step(environment=(np.random.rand(), np.random.rand()))
        if offspring:
            new_cells.append(offspring)

    population.extend(new_cells)
    population = [c for c in population if c.alive]

    if population:
        history["population"].append(len(population))
        history["avg_energy"].append(np.mean([c.energy for c in population]))
        history["avg_damage"].append(np.mean([c.damage for c in population]))
        history["avg_telomere"].append(np.mean([c.telomere for c in population]))
    else:
        break

# -------------------------
# Visualizations
# -------------------------
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.plot(history["population"])
plt.title("Population Size")

plt.subplot(2,2,2)
plt.plot(history["avg_energy"], label="Energy")
plt.plot(history["avg_damage"], label="Damage")
plt.legend()
plt.title("Metabolic State")

plt.subplot(2,2,3)
plt.plot(history["avg_telomere"])
plt.title("Average Telomere Length")

plt.subplot(2,2,4)
plt.plot(history["population"], history["avg_telomere"], '.')
plt.xlabel("Population")
plt.ylabel("Telomere Length")
plt.title("Population vs Telomere")

plt.tight_layout()
plt.show()
```
