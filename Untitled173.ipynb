```python
# core_fusion.py
import torch, torch.nn as nn
from torch.nn import functional as F

class GatedFusion(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.g = nn.Linear(d*5, 5)
        self.p = nn.Linear(d*5, d)

    def forward(self, xs):  # xs = [img, seq, graph, time, text]
        cat = torch.cat(xs, dim=-1)
        w = F.softmax(self.g(cat), dim=-1)
        fused = sum(w[:,i:i+1]*xs[i] for i in range(5))
        return self.p(torch.cat([fused]*5, dim=-1))

class SciFusionFM(nn.Module):
    def __init__(self, d=256):
        super().__init__()
        self.img = nn.Sequential(nn.Linear(1024,d), nn.GELU())
        self.seq = nn.Sequential(nn.Linear(768,d), nn.GELU())
        self.gnn = nn.Sequential(nn.Linear(512,d), nn.GELU())
        self.tsn = nn.Sequential(nn.Linear(128,d), nn.GELU())
        self.txt = nn.Sequential(nn.Linear(1024,d), nn.GELU())
        self.fuse = GatedFusion(d)
        self.head_pred = nn.Linear(d, 8)        # multitask outputs
        self.head_unc  = nn.Linear(d, 1)        # epistemic proxy

    def forward(self, x):
        xs = [self.img(x['img']), self.seq(x['seq']),
              self.gnn(x['graph']), self.tsn(x['time']), self.txt(x['text'])]
        z = self.fuse(xs)
        y = self.head_pred(z)
        u = F.softplus(self.head_unc(z))
        return y, u
```

```python
import torch
import torch.nn as nn

class BioSpline(nn.Module):
    def forward(self, x):
        return x / (1 + torch.abs(x)) + 0.1 * torch.tanh(x)
```

```python
class AdaptiveEntropyDescent(torch.optim.Optimizer):
    def __init__(self, params, lr=1e-3, beta=0.01):
        defaults = dict(lr=lr, beta=beta)
        super().__init__(params, defaults)

    def step(self):
        for group in self.param_groups:
            for p in group['params']:
                if p.grad is None:
                    continue
                entropy = torch.mean(torch.abs(p.grad))
                p.data -= group['lr'] * p.grad / (1 + group['beta'] * entropy)
```

```python
def homeostatic_loss(state, target_range):
    lower, upper = target_range
    penalty = torch.mean(
        torch.relu(lower - state)**2 + torch.relu(state - upper)**2
    )
    smoothness = torch.mean(torch.abs(state[1:] - state[:-1]))
    return penalty + 0.01 * smoothness
```

```python
class VirtualCell(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(5, 64),
            BioSpline(),
            nn.Linear(64, 32),
            BioSpline()
        )
        self.telomere_head = nn.Linear(32, 1)
        self.health_head = nn.Linear(32, 1)

    def forward(self, x):
        h = self.encoder(x)
        telomere = torch.sigmoid(self.telomere_head(h))
        health = torch.sigmoid(self.health_head(h))
        return telomere, health
```

```python
class HarmonyOpt(torch.optim.Optimizer):
    def __init__(self, params, lr=1e-4, entropy_coeff=1e-3):
        defaults = dict(lr=lr, entropy_coeff=entropy_coeff)
        super().__init__(params, defaults)

    def step(self, closure=None):
        loss = closure() if closure else None
        for group in self.param_groups:
            for p in group['params']:
                if p.grad is None: continue
                grad = p.grad.data
                noise = torch.randn_like(grad) * group['entropy_coeff']
                p.data -= group['lr'] * (grad + noise)
        return loss
```

```python
import numpy as np

class VirtualCell:
    def __init__(self,
                 telomere_init=8000,
                 min_safe=5000,
                 max_safe=10000):
        self.telomere = telomere_init
        self.min_safe = min_safe
        self.max_safe = max_safe
        self.damage = 0.0
        self.alive = True

    def dna_damage_response(self):
        return self.damage < 0.3 and self.telomere > self.min_safe

    def telomerase_allowed(self, stem_signal, checkpoint_ok):
        return stem_signal and checkpoint_ok and self.telomere < self.max_safe

    def divide(self, telomerase_on=False):
        if not self.alive:
            return

        shortening = np.random.normal(50, 10)
        extension = np.random.normal(80, 15) if telomerase_on else 0

        self.telomere += -shortening + extension
        self.damage += np.random.uniform(0.0, 0.05)

        if self.telomere < 3000 or self.damage > 1.0:
            self.alive = False

    def phagocytosis_cleanup(self):
        self.damage *= 0.7
```

```python
class OrganOnChip:
    def __init__(self, n_cells=100):
        self.cells = [VirtualCell() for _ in range(n_cells)]
        self.injury_signal = 0.0

    def induce_injury(self, severity=0.5):
        self.injury_signal = severity

    def step(self, ai_policy):
        for cell in self.cells:
            stem_need = self.injury_signal > 0.3
            checkpoint = cell.dna_damage_response()
            telomerase = ai_policy(cell, stem_need, checkpoint)

            cell.divide(telomerase_on=telomerase)
            cell.phagocytosis_cleanup()

        self.injury_signal *= 0.9
```

```python
def safe_ai_policy(cell, stem_need, checkpoint_ok):
    if not checkpoint_ok:
        return False
    if stem_need and cell.telomere < 7000:
        return True
    return False
```

```python
# losses.py
import torch

def stability_loss(T, target, lambda1=0.1):
    return torch.var(T) + lambda1 * torch.abs(T.mean() - target)

def anti_pathology_loss(growth_rate, threshold):
    return torch.sum(torch.relu(growth_rate - threshold))

def causal_consistency_loss(f_do, f_pred):
    return torch.mean(torch.abs(f_do - f_pred))

def total_loss(task, T, target, growth, threshold, f_do, f_pred):
    return (
        task
        + stability_loss(T, target)
        + anti_pathology_loss(growth, threshold)
        + causal_consistency_loss(f_do, f_pred)
    )
```

```python
# causal_compression.py
import torch

def causal_compress(latent_trajectories):
    """
    Finds minimal latent dimensions preserving long-horizon outcomes
    """
    U, S, V = torch.pca_lowrank(latent_trajectories)
    importance = S / S.sum()
    return V[:, importance > 0.05]
```

```python
import numpy as np

class CausalGridTask(ARCStyleTask):
    def __init__(self):
        self.train_pairs = []
        self.test_input = None
        self.hidden_cause = np.random.choice(["flip", "rotate"])

    def reset(self):
        for _ in range(3):
            x = np.random.randint(0, 2, (5,5))
            y = self.apply_rule(x)
            self.train_pairs.append((x, y))
        self.test_input = np.random.randint(0, 2, (5,5))

    def apply_rule(self, grid):
        if self.hidden_cause == "flip":
            return np.fliplr(grid)
        else:
            return np.rot90(grid)

    def observe(self):
        return {
            "train": self.train_pairs,
            "test": self.test_input
        }

    def evaluate(self, prediction):
        target = self.apply_rule(self.test_input)
        return float(np.array_equal(prediction, target))
```

```python
class ScientificTask:
    def __init__(self):
        self.hidden_rules = {
            "stability_constraint": True,
            "non_pathological_growth": True,
            "resource_conservation": True
        }

    def initial_state(self):
        return {
            "state_vector": torch.randn(128),
            "stress_level": 0.2,
            "resource_budget": 1.0
        }

    def simulate(self, state, action):
        # Abstract physics + biology-free dynamics
        next_state = state["state_vector"] + action
        stress = torch.norm(action) * 0.1
        return {
            "state_vector": next_state,
            "stress_level": state["stress_level"] + stress,
            "resource_budget": state["resource_budget"] - stress
        }

    def score(self, trajectory):
        stability = torch.var(torch.stack(
            [s["state_vector"].mean() for s in trajectory]
        ))
        penalty = max(0, trajectory[-1]["stress_level"] - 1.0)
        return -stability - penalty
```

```python
import numpy as np

class HiddenHomeostasisTask(ARCScientificTask):
    def reset(self):
        self.state = np.random.randn(5)
        self.hidden_target = np.random.uniform(-1, 1, size=5)
        self.t = 0
        return self.observe()

    def step(self, action):
        noise = np.random.normal(0, 0.05, size=5)
        self.state += action + noise
        self.t += 1
        return self.observe()

    def observe(self):
        return self.state + np.random.normal(0, 0.1, size=5)

    def evaluate(self):
        stability = -np.var(self.state)
        distance = -np.linalg.norm(self.state - self.hidden_target)
        return stability + distance
```

```python
%%writefile world_state.py
import torch

class WorldState:
    def __init__(self, latent_dim=64):
        self.z = torch.randn(latent_dim)
        self.t = 0

    def step(self, action):
        noise = 0.01 * torch.randn_like(self.z)
        self.z = self.z + action + noise
        self.t += 1

    def health_metric(self):
        return torch.exp(-torch.var(self.z))
```

```python
%%writefile control.py
class ConstraintController:
    def __init__(self, action_dim, limit=0.1):
        self.limit = limit
        self.action_dim = action_dim

    def propose_action(self, z):
        return 0.05 * torch.tanh(z[:self.action_dim])

    def enforce_constraints(self, action):
        return torch.clamp(action, -self.limit, self.limit)
```

```python
# aurora/core/world_model.py
import torch
import torch.nn as nn

class ScientificWorldModel(nn.Module):
    def __init__(self, dim=256):
        super().__init__()
        self.encoder = nn.Linear(dim, dim)
        self.dynamics = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(dim, 8), 4
        )
        self.decoder = nn.Linear(dim, dim)

    def forward(self, z, action, env):
        x = z + action + env
        h = self.encoder(x)
        h = self.dynamics(h.unsqueeze(0)).squeeze(0)
        return self.decoder(h)
```

```python
# aurora/core/causal_compression.py
import torch

def causal_compress(latents, threshold=0.95):
    u, s, v = torch.svd(latents)
    cumulative = torch.cumsum(s, dim=0) / torch.sum(s)
    k = torch.searchsorted(cumulative, threshold)
    return u[:, :k] @ torch.diag(s[:k])
```

```python
# aurora/core/losses.py
import torch

def stability_loss(signal, target):
    return torch.var(signal) + torch.abs(signal.mean() - target)

def pathology_penalty(growth, max_safe):
    return torch.relu(growth - max_safe).mean()

def total_loss(task, stability, pathology, causal):
    return task + stability + pathology + causal
```

```python
# models/aurora_x.py
import torch
import torch.nn as nn

class AuroraX(nn.Module):
    def __init__(self, latent_dim=512):
        super().__init__()
        self.world_model = nn.Transformer(
            d_model=latent_dim,
            num_encoder_layers=8,
            num_decoder_layers=8
        )
        self.causal_head = nn.Linear(latent_dim, latent_dim)
        self.policy_head = nn.Linear(latent_dim, latent_dim)

    def forward(self, latent_sequence):
        world_state = self.world_model(latent_sequence, latent_sequence)
        causal_latents = self.causal_head(world_state)
        actions = self.policy_head(world_state)
        return causal_latents, actions
```

```python
# evaluation/arc_agi_harness.py
class ARCAGITask:
    def __init__(self, world, constraints):
        self.world = world
        self.constraints = constraints

    def run(self, model):
        latent = self.world.initial_state()
        causal, actions = model(latent)

        for a in actions:
            if not self.constraints(a):
                return 0.0

        return self.world.generalization_score(causal)


class Evaluator:
    def __init__(self, tasks):
        self.tasks = tasks

    def evaluate(self, model):
        return sum(t.run(model) for t in self.tasks) / len(self.tasks)
```

```python
import torch
import numpy as np

class OoCWorld:
    def __init__(self):
        self.state = torch.randn(16)  # abstract cell state
        self.telomere = torch.tensor(1.0)
        self.time = 0

    def step(self, action):
        stress = torch.norm(action)
        self.telomere -= 0.01 * stress
        self.telomere += 0.02 * torch.sigmoid(action.mean())
        self.state = self.state + action
        self.time += 1

    def healthy(self):
        return 0.6 < self.telomere < 1.4
```

```python
import torch.nn as nn

class CausalGate(nn.Module):
    """
    Activation that suppresses spurious correlations
    """
    def forward(self, x):
        gate = torch.sigmoid(x.mean(dim=-1, keepdim=True))
        return x * gate
```

```python
from einops import rearrange

class WorldModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Linear(16, 64)
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=64, nhead=4, activation="gelu"
            ),
            num_layers=4
        )
        self.causal_gate = CausalGate()
        self.policy = nn.Linear(64, 16)

    def forward(self, x):
        x = self.encoder(x)
        x = rearrange(x, 'd -> 1 1 d')
        x = self.transformer(x)
        x = self.causal_gate(x)
        return self.policy(x.squeeze())
```

```python
import torch
import random

class VirtualOoCWorld:
    def __init__(self):
        self.telomere = torch.tensor(0.8)
        self.stress = torch.tensor(0.1)
        self.health = torch.tensor(1.0)
        self.time = 0

    def step(self, action):
        """
        action[0] = telomerase control
        action[1] = phagocytosis intensity
        """
        telomerase, cleanup = action

        # Telomere dynamics
        self.telomere += 0.03 * telomerase - 0.02
        self.telomere = torch.clamp(self.telomere, 0.4, 1.0)

        # Stress & cleanup
        self.stress += 0.1 - 0.15 * cleanup
        self.stress = torch.clamp(self.stress, 0.0, 1.0)

        # Health is causal outcome
        self.health = 1.0 - self.stress + 0.2 * self.telomere
        self.time += 1

    def state(self):
        return torch.tensor([
            self.telomere,
            self.stress,
            self.health
        ])
```

```python
class VirtualWorld:
    """
    Abstract OoC-like world:
    - state evolves under hidden causal rules
    - agent can intervene
    """
    def __init__(self, hidden_dim=8):
        self.state = torch.randn(hidden_dim)
        self.hidden_matrix = torch.randn(hidden_dim, hidden_dim)

    def step(self, action):
        self.state = torch.tanh(self.hidden_matrix @ self.state + action)
        return self.state

    def run(self, actions):
        for a in actions:
            self.step(a)
        return self.state

    def generalization_score(self):
        # Stability + compression proxy
        return float(1.0 - self.state.var().item())
```

```python
import torch.nn as nn
import torch.nn.functional as F

class SwiGLUX(nn.Module):
    def forward(self, x):
        a, b = x.chunk(2, dim=-1)
        return F.silu(a) * b

class AuroraX(nn.Module):
    def __init__(self, dim=64):
        super().__init__()
        self.encoder = nn.Linear(dim, dim*2)
        self.act = SwiGLUX()
        self.policy = nn.Linear(dim, dim)

    def forward(self, state):
        z = self.act(self.encoder(state))
        action = self.policy(z)
        return action, z
```

```python
import torch.nn as nn
from einops import rearrange

class SwiGLU(nn.Module):
    def forward(self, x):
        x, gate = x.chunk(2, dim=-1)
        return x * torch.nn.functional.silu(gate)

class AuroraBlock(nn.Module):
    def __init__(self, dim, heads):
        super().__init__()
        self.attn = nn.MultiheadAttention(dim, heads, batch_first=True)
        self.ff = nn.Sequential(
            nn.Linear(dim, dim*4),
            SwiGLU(),
            nn.Linear(dim*2, dim)
        )
        self.norm1 = nn.LayerNorm(dim)
        self.norm2 = nn.LayerNorm(dim)

    def forward(self, x):
        h, _ = self.attn(x, x, x)
        x = self.norm1(x + h)
        x = self.norm2(x + self.ff(x))
        return x
```

```python
class AuroraX(nn.Module):
    def __init__(self, dim=256, depth=6, heads=8):
        super().__init__()
        self.embed = nn.Linear(32*32, dim)  # Input feature size is 32*32 = 1024
        self.blocks = nn.ModuleList([
            AuroraBlock(dim, heads) for _ in range(depth)
        ])
        self.latent = nn.Linear(dim, 32)  # causal compression
        # The output of the model (self.head(z)) should match the input dimension expected by self.embed
        self.head = nn.Linear(32, 32*32) # Change output dim to 32*32=1024

    def forward(self, x):
        x = self.embed(x)
        for b in self.blocks:
            x = b(x)
        z = self.latent(x)
        return self.head(z), z
```

```python
def simulate_world(model, steps=5):
    state = torch.randn(1, 1, 32*32)
    latents = []

    for _ in range(steps):
        state, z = model(state)
        latents.append(z.norm().item())

    return sum(latents) / len(latents)
```

```python
def arc_generalization_test(model):
    scores = []
    for i in range(10):
        score = simulate_world(model)
        scores.append(score)
    return sum(scores) / len(scores)
```

```python
import torch.nn as nn
from einops import rearrange

class ConstraintAwareTransformer(nn.Module):
    def __init__(self, d_model=256, n_heads=8, n_layers=6):
        super().__init__()
        self.token = nn.Embedding(4096, d_model)
        self.pos = nn.Parameter(torch.randn(1, 512, d_model))
        enc = nn.TransformerEncoderLayer(d_model, n_heads, batch_first=True)
        self.encoder = nn.TransformerEncoder(enc, n_layers)
        self.causal_head = nn.Linear(d_model, d_model//2)   # compression
        self.control_head = nn.Linear(d_model, 8)           # abstract actions

    def forward(self, x):
        # x: [B, T] token ids
        h = self.token(x) + self.pos[:, :x.size(1)]
        z = self.encoder(h)
        causal_latent = self.causal_head(z.mean(1))
        control = self.control_head(z[:, -1])
        return causal_latent, control

model = ConstraintAwareTransformer()
```

```python
import random

class SyntheticWorld:
    def __init__(self):
        self.rule = random.choice(["delay", "threshold", "feedback"])
        self.state = torch.randn(1)

    def step(self, action):
        if self.rule == "delay":
            self.state += 0.1 * action.mean()
        elif self.rule == "threshold":
            self.state += (action.mean() > 0).float()
        else:
            self.state -= 0.05 * action.std()

    def score(self):
        return torch.sigmoid(-self.state.abs()).item()

def arc_evaluate(model, trials=20):
    scores = []
    for _ in range(trials):
        world = SyntheticWorld()
        tokens = torch.randint(0, 1000, (1, 32))
        _, action = model(tokens)
        for _ in range(10):
            world.step(action)
        scores.append(world.score())
    return sum(scores)/len(scores)

print("ARC-style generalization score:", arc_evaluate(model))
```

```python
%%writefile aurora/losses/losses.py
import torch

def causal_compression_loss(latents, prediction):
    variance = latents.var(dim=0).mean()
    entropy = -(prediction.softmax(-1) * prediction.log_softmax(-1)).sum(-1).mean()
    return variance + 0.1 * entropy
```

```python
%%writefile aurora/model/model.py
import torch
import torch.nn as nn
from aurora.activations import CausalSwish

class AuroraX(nn.Module):
    def __init__(self, d_model=512, n_layers=8):
        super().__init__()
        self.embed = nn.Linear(d_model, d_model)
        self.layers = nn.ModuleList([
            nn.TransformerEncoderLayer(
                d_model=d_model,
                nhead=8,
                activation="gelu"
            )
            for _ in range(n_layers)
        ])
        self.causal_head = nn.Sequential(
            nn.Linear(d_model, d_model),
            CausalSwish(),
            nn.Linear(d_model, d_model)
        )

    def forward(self, x):
        x = self.embed(x)
        for layer in self.layers:
            x = layer(x)
        return self.causal_head(x)
```

```python
# aurora_x/model.py
import torch
import torch.nn as nn

class CausalAttention(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.qkv = nn.Linear(d, d*3)
        self.proj = nn.Linear(d, d)

    def forward(self, x):
        q,k,v = self.qkv(x).chunk(3, dim=-1)
        attn = torch.softmax(q @ k.transpose(-2,-1) / x.size(-1)**0.5, dim=-1)
        return self.proj(attn @ v), attn

class AuroraX(nn.Module):
    def __init__(self, d_model=256, layers=6):
        super().__init__()
        self.blocks = nn.ModuleList([
            CausalAttention(d_model) for _ in range(layers)
        ])
        self.head = nn.Linear(d_model, d_model)

    def forward(self, x):
        attentions = []
        for block in self.blocks:
            x, attn = block(x)
            attentions.append(attn)
        return self.head(x), attentions
```

```python
import numpy as np

class VirtualCell:
    def __init__(self):
        self.telomere = 1.0            # normalized length
        self.stress = 0.0
        self.waste = 0.1
        self.divisions = 0

    def divide(self, telomerase_signal):
        shortening = 0.04
        self.telomere -= shortening
        self.telomere += telomerase_signal * 0.05
        self.telomere = np.clip(self.telomere, 0.2, 1.2)
        self.divisions += 1
        self.stress += 0.05

    def phagocytosis(self):
        cleared = min(self.waste, 0.08)
        self.waste -= cleared
        self.stress *= 0.9

    def step(self, telomerase_signal, env_stress):
        self.waste += 0.03 + env_stress
        if self.waste > 0.2:
            self.phagocytosis()
        self.divide(telomerase_signal)
        self.stress += env_stress

    def state(self):
        return np.array([
            self.telomere,
            self.stress,
            self.waste,
            self.divisions / 100
        ], dtype=np.float32)
```

```python
import torch
import torch.nn as nn

class TelomereTransformer(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Linear(4, 64)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=64,
            nhead=4,
            dim_feedforward=128,
            activation="gelu",
            batch_first=True
        )
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers=2)
        self.head = nn.Sequential(
            nn.Linear(64, 32),
            nn.SiLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.embed(x)
        x = self.encoder(x)
        return self.head(x[:, -1])
```

```python
import random
import math
import matplotlib.pyplot as plt
from collections import Counter

CANONICAL = "TTAGGG"
VARIANTS = ["TTAXGG", "TGAGGG", "TTAGXG"]

def generate_telomere(n_repeats=50, variant_rate=0.1):
    seq = ""
    for _ in range(n_repeats):
        if random.random() < variant_rate:
            seq += random.choice(VARIANTS)
        else:
            seq += CANONICAL
    return seq

def shannon_entropy(seq):
    counts = Counter(seq)
    total = len(seq)
    return -sum((c/total)*math.log2(c/total) for c in counts.values())

def symbolic_telomere_state(seq):
    return {
        "length": len(seq),
        "entropy": shannon_entropy(seq),
        "synthetic_fraction": seq.count("X") / len(seq),
        "telomerase_access": "HIGH" if seq.endswith("GGG") else "LOW"
    }

telomeres = [generate_telomere(30, v) for v in [0.0, 0.05, 0.2]]
states = [symbolic_telomere_state(t) for t in telomeres]

states
```

```python
def telomerase_policy(cell_state, telomere_state):
    if cell_state["type"] == "stem" and telomere_state["length"] < 300:
        return "EXTEND"
    if telomere_state["entropy"] > 2.0:
        return "HALT (instability)"
    return "NO_ACTION"
```

```python
entropies = [s["entropy"] for s in states]
synthetic = [s["synthetic_fraction"] for s in states]

plt.figure(figsize=(4,3))
plt.scatter(synthetic, entropies, s=60)
plt.xlabel("Synthetic base fraction")
plt.ylabel("Sequence entropy")
plt.title("Symbolic Telomere Stability")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

class SymbolicTelomere:
    def __init__(
        self,
        repeat="TTAGGG",
        length_bp=8000,
        g_quad_score=0.7,
        shelterin_affinity=0.8,
        telomerase_access=0.5
    ):
        self.repeat = repeat
        self.length_bp = length_bp
        self.g_quad_score = g_quad_score
        self.shelterin_affinity = shelterin_affinity
        self.telomerase_access = telomerase_access

    def erode(self, bp_loss):
        self.length_bp = max(0, self.length_bp - bp_loss)

    def extend(self, activity_level):
        gain = int(activity_level * self.telomerase_access * 120)
        self.length_bp += gain

    def trim(self, upper_bound):
        if self.length_bp > upper_bound:
            self.length_bp = upper_bound

    def state(self):
        return {
            "length_bp": self.length_bp,
            "g_quad": self.g_quad_score,
            "shelterin": self.shelterin_affinity,
            "telomerase_access": self.telomerase_access,
        }
```

```python
class TelomeraseController:
    def __init__(self, lr=0.15):
        self.lr = lr  # intentionally high but bounded
        self.activity = 0.0

    def update(self, damage, stem_signal, cancer_risk):
        target = stem_signal * (1 - cancer_risk) * damage
        self.activity += self.lr * (target - self.activity)
        self.activity = np.clip(self.activity, 0.0, 1.0)
        return self.activity
```

```python
class VirtualCell:
    def __init__(self):
        self.telomere = SymbolicTelomere()
        self.telomerase = TelomeraseController()
        self.damage = 0.0
        self.debris = 0.0

    def dna_damage(self, amount):
        self.damage = np.clip(self.damage + amount, 0, 1)

    def phagocytosis(self):
        cleared = min(self.debris, 0.25)
        self.debris -= cleared
        self.damage = max(0, self.damage - 0.1 * cleared)

    def step(self, stem_signal, cancer_risk):
        # erosion
        self.telomere.erode(bp_loss=50)

        # telomerase pulse
        activity = self.telomerase.update(
            damage=self.damage,
            stem_signal=stem_signal,
            cancer_risk=cancer_risk
        )
        self.telomere.extend(activity)
        self.telomere.trim(upper_bound=12000)

        # cleanup
        self.phagocytosis()
```

```python
def run_ooc_sim(steps=300, injury_step=80):
    cell = VirtualCell()
    history = []

    for t in range(steps):
        # Injury event
        if t == injury_step:
            cell.dna_damage(0.8)
            cell.debris += 1.0

        stem_signal = 1.0 if cell.damage > 0.3 else 0.1
        cancer_risk = 0.1 + 0.6 * cell.telomere.telomerase_access

        cell.step(stem_signal, cancer_risk)

        history.append({
            "time": t,
            "telomere_bp": cell.telomere.length_bp,
            "damage": cell.damage,
            "debris": cell.debris,
            "telomerase_activity": cell.telomerase.activity
        })

    return pd.DataFrame(history)
```

```python
import matplotlib.pyplot as plt

df = run_ooc_sim()

plt.figure(figsize=(6,4))
plt.plot(df.time, df.telomere_bp, linewidth=2)
plt.axvline(80, linestyle="--", alpha=0.5)
plt.ylabel("Telomere length (bp)")
plt.xlabel("Simulation time")
plt.title("Context-regulated telomere maintenance")
plt.tight_layout()
plt.show()
```
