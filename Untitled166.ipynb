```python
import torch
import torch.nn as nn

class WorldModel(nn.Module):
    def __init__(self, state_dim=3, latent=64):
        super().__init__()
        self.encoder = nn.Linear(state_dim, latent)
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(latent, 4, batch_first=True), 3
        )
        self.decoder = nn.Linear(latent, state_dim)

    def forward(self, x):
        z = torch.relu(self.encoder(x)) # z has shape (latent,)
        # Reshape z to (batch_size=1, sequence_length=1, embedding_dim=latent)
        z_transformed = self.transformer(z.unsqueeze(0).unsqueeze(0))
        # Squeeze back to (latent,) for the decoder
        z_output = z_transformed.squeeze(0).squeeze(0)
        return self.decoder(z_output), z_output
```

```python
import numpy as np
from dataclasses import dataclass

@dataclass
class TelomereGenome:
    repair: float        # telomerase efficiency
    resistance: float    # degradation resistance
    stability: float     # chromosomal stability gain
    risk: float          # oncogenic risk proxy

    def mutate(self, rate=0.05):
        def m(x):
            return np.clip(x + np.random.normal(0, rate), 0.0, 2.0)
        return TelomereGenome(
            repair=m(self.repair),
            resistance=m(self.resistance),
            stability=m(self.stability),
            risk=m(self.risk)
        )
```

```python
class EvoCell:
    def __init__(self, genome):
        self.genome = genome
        self.telomere = 1.0
        self.age = 0

    def step(self, telomerase_signal):
        self.age += 1
        self.telomere -= 0.02 / self.genome.resistance
        self.telomere += telomerase_signal * self.genome.repair
        self.telomere += 0.01 * self.genome.stability
        self.telomere -= 0.03 * self.genome.risk
```

```python
class EvoPopulation:
    def __init__(self, n=100):
        self.cells = [
            EvoCell(TelomereGenome(
                repair=np.random.uniform(0.5,1.5),
                resistance=np.random.uniform(0.5,1.5),
                stability=np.random.uniform(0.5,1.5),
                risk=np.random.uniform(0.0,0.5)
            )) for _ in range(n)
        ]

    def step(self, telomerase_signal):
        next_gen = []

        for c in self.cells:
            c.step(telomerase_signal)
            if c.telomere > 0:
                next_gen.append(c)

                if c.telomere > 1.2:  # division
                    child_genome = c.genome.mutate()
                    next_gen.append(EvoCell(child_genome))

        self.cells = next_gen[:500]  # cap explosion
```

```python
import numpy as np

class ECMOptimizer:
    def __init__(
        self,
        lr=1e-3,
        beta1=0.9,
        beta2=0.999,
        eps=1e-8,
        weight_decay=1e-4,
        sign_beta=0.95,
        causal_penalty=0.1,
        evo_pressure=0.05
    ):
        self.lr = lr
        self.beta1 = beta1
        self.beta2 = beta2
        self.eps = eps
        self.wd = weight_decay
        self.sign_beta = sign_beta
        self.causal_penalty = causal_penalty
        self.evo_pressure = evo_pressure
        self.state = {}

    def step(self, params, grads, causal_signal=1.0, fitness_signal=1.0):
        for k in params:
            if k not in self.state:
                self.state[k] = {
                    "m": 0.0,
                    "v": 0.0,
                    "g2": 0.0,
                    "sign": 0.0,
                    "instability": 0.0
                }

            s = self.state[k]
            g = grads[k]

            # Momentum (Adam / Nadam)
            s["m"] = self.beta1 * s["m"] + (1 - self.beta1) * g
            s["v"] = self.beta2 * s["v"] + (1 - self.beta2) * (g ** 2)
            s["g2"] += g ** 2

            # Lion-style sign memory
            s["sign"] = self.sign_beta * s["sign"] + (1 - self.sign_beta) * np.sign(g)

            # Instability estimate (chaos proxy)
            s["instability"] = 0.9 * s["instability"] + 0.1 * abs(g - s["m"])

            # Adaptive denominator (Adam + Adagrad hybrid)
            denom = np.sqrt(s["v"] + s["g2"]) + self.eps

            # Causal trust modulation
            causal_weight = np.exp(-self.causal_penalty * s["instability"]) * causal_signal

            # Evolutionary pressure
            evo_weight = np.tanh(fitness_signal * self.evo_pressure)

            # Final update
            update = (
                self.lr
                * causal_weight
                * evo_weight
                * (s["m"] / denom + 0.1 * s["sign"])
            )

            # AdamW-style decoupled decay
            params[k] -= update + self.wd * params[k]

        return params
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class LearnedOptimizer(nn.Module):
    def __init__(self, hidden_size=64):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(7, hidden_size),
            nn.SiLU(),
            nn.LayerNorm(hidden_size)
        )

        self.rnn = nn.GRU(
            input_size=hidden_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.decoder = nn.Sequential(
            nn.Linear(hidden_size, hidden_size),
            nn.GELU(),
            nn.Linear(hidden_size, 1)
        )

    def forward(self, features, hidden):
        """
        features: [B, T, 7]
        hidden: optimizer hidden state
        """
        x = self.encoder(features)
        x, hidden = self.rnn(x, hidden)
        update = self.decoder(x)
        return update, hidden
```

```python
class MetaECMO:
    def __init__(self, learned_optimizer, base_lr=1e-3):
        self.learned_opt = learned_optimizer
        self.base_lr = base_lr
        self.hidden = None

    def step(self, param, grad, state, causal, fitness):
        features = torch.tensor([[
            grad,
            state["m"],
            state["v"],
            torch.sign(torch.tensor(grad)),
            state["instability"],
            causal,
            fitness
        ]], dtype=torch.float32).unsqueeze(0)

        update, self.hidden = self.learned_opt(features, self.hidden)

        # ECMO safety gate
        stability_gate = torch.exp(-state["instability"])
        final_update = self.base_lr * stability_gate * update.item()

        return param - final_update
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class BioFusionAct(nn.Module):
    def __init__(self, alpha=0.7, beta=0.3):
        super().__init__()
        self.alpha = alpha
        self.beta = beta

    def forward(self, x):
        # smooth + bounded + gated
        return (
            self.alpha * (x * torch.sigmoid(x)) +           # Swish
            self.beta  * torch.tanh(x) +                     # Stability
            (1 - self.alpha - self.beta) * F.gelu(x)         # Expressivity
        )
```

```python
class ECMOLionW(torch.optim.Optimizer):
    def __init__(self, params, lr=1e-3, betas=(0.9, 0.99), weight_decay=1e-2, eps=1e-8):
        defaults = dict(lr=lr, betas=betas, wd=weight_decay, eps=eps)
        super().__init__(params, defaults)

    @torch.no_grad()
    def step(self, closure=None):
        for group in self.param_groups:
            lr, (b1, b2), wd, eps = group['lr'], group['betas'], group['wd'], group['eps']
            for p in group['params']:
                if p.grad is None: continue
                g = p.grad

                state = self.state.setdefault(p, {})
                if not state:
                    state['m'] = torch.zeros_like(p)
                    state['v'] = torch.zeros_like(p)

                m, v = state['m'], state['v']
                m.mul_(b1).add_(g, alpha=1-b1)
                v.mul_(b2).addcmul_(g, g, value=1-b2)

                rms = (v.sqrt() + eps)
                stability_gate = torch.exp(-torch.mean(rms))  # chaos suppression

                update = torch.sign(m) / rms
                p.add_(update, alpha=-lr * stability_gate)

                # decoupled weight decay
                p.add_(p, alpha=-lr * wd)
```

```python
def foundation_loss(pred, target, causal_logits, scm_targets, instability):
    recon = F.huber_loss(pred, target, delta=1.0)
    causal = F.binary_cross_entropy_with_logits(causal_logits, scm_targets)
    stability = torch.mean(instability**2)                 # Lyapunov-like
    info = -torch.mean(F.log_softmax(pred, dim=-1))         # entropy shaping
    return recon + 0.5*causal + 0.1*stability + 0.01*info
```

```python
class WorldModel(nn.Module):
    def __init__(self, d=256, heads=8, layers=6):
        super().__init__()
        self.embed = nn.Linear(d, d)
        self.blocks = nn.ModuleList([
            nn.TransformerEncoderLayer(
                d_model=d, nhead=heads, activation='gelu', batch_first=True
            ) for _ in range(layers)
        ])
        self.act = BioFusionAct()
        self.predict = nn.Linear(d, d)
        self.causal = nn.Linear(d, 1)   # SCM edge probability

    def forward(self, x):
        x = self.act(self.embed(x))
        for b in self.blocks:
            x = b(x)
        return self.predict(x), self.causal(x)
```

```python
def train_step(model, opt, batch):
    x, y, scm_y = batch
    pred, causal = model(x)
    instability = pred.var(dim=1)
    loss = foundation_loss(pred, y, causal, scm_y, instability)
    opt.zero_grad()
    loss.backward()
    opt.step()
    return loss.item()
```

```python
import numpy as np

ALPHABET = ["A", "B", "C", "D", "E"]  # abstract symbols

def random_sequence(length=32):
    return [np.random.choice(ALPHABET) for _ in range(length)]

def encode_sequence(seq):
    # map symbols to integers
    mapping = {s:i for i,s in enumerate(ALPHABET)}
    return np.array([mapping[s] for s in seq], dtype=np.float32)
```

```python
import torch

def sequence_to_modulation(seq_vec, dim=256):
    torch.manual_seed(int(seq_vec.sum()))
    return torch.randn(dim) * 0.05
```

```python
def apply_sequence_modulation(world_model, modulation):
    with torch.no_grad():
        for p in world_model.parameters():
            p.add_(modulation.mean() * 1e-3)
```

```python
def fitness_function(model, env_rollout_fn, seq):
    seq_vec = encode_sequence(seq)
    modulation = sequence_to_modulation(seq_vec)
    apply_sequence_modulation(model, modulation)

    rewards, instability = env_rollout_fn(model)

    return (
        np.mean(rewards)
        - 0.5 * np.mean(instability)
        - 0.01 * np.var(rewards)
    )
```

```python
def mutate(seq, rate=0.1):
    new_seq = seq.copy()
    for i in range(len(seq)):
        if np.random.rand() < rate:
            new_seq[i] = np.random.choice(ALPHABET)
    return new_seq

def crossover(a, b):
    point = np.random.randint(1, len(a)-1)
    return a[:point] + b[point:]
```

```python
def evolutionary_search(
    model,
    env_rollout_fn,
    population_size=40,
    generations=30,
    elite_frac=0.2
):
    population = [random_sequence() for _ in range(population_size)]
    history = []

    for gen in range(generations):
        scores = [
            fitness_function(model, env_rollout_fn, seq)
            for seq in population
        ]

        ranked = sorted(zip(population, scores), key=lambda x: x[1], reverse=True)
        elites = [s for s,_ in ranked[:int(elite_frac * population_size)]]

        new_pop = elites.copy()
        while len(new_pop) < population_size:
            a, b = np.random.choice(elites, 2, replace=False)
            child = mutate(crossover(a, b))
            new_pop.append(child)

        population = new_pop
        history.append(np.max(scores))

        print(f"Gen {gen:02d} | Best fitness: {np.max(scores):.4f}")

    return population[0], history
```

```python
import asyncio
import websockets
import json

CONNECTED_CLIENTS = set()

async def handler(websocket, path):
    CONNECTED_CLIENTS.add(websocket)
    print(f"Client connected: {websocket.remote_address}")
    try:
        async for message in websocket:
            print(f"Received message from {websocket.remote_address}: {message[:100]}...") # Print first 100 chars
            # Re-send message to all connected clients (including the sender for now)
            for client in CONNECTED_CLIENTS:
                if client != websocket: # Optional: don't send back to sender
                    await client.send(message)
    except websockets.exceptions.ConnectionClosed as e:
        print(f"Client disconnected: {websocket.remote_address} - {e}")
    finally:
        CONNECTED_CLIENTS.remove(websocket)

async def main():
    print("Starting WebSocket server on ws://0.0.0.0:8765")
    async with websockets.serve(handler, "0.0.0.0", 8765):
        await asyncio.Future()  # Run forever

# Note: This will block the Colab cell. To run it, you would typically execute it in a separate process or terminal.
# asyncio.run(main())
print("WebSocket server setup code defined. To run, uncomment asyncio.run(main()) and execute in a separate environment.")
```
