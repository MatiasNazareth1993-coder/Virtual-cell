```python
# Immortality stress test simulation with visuals
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

steps = 600
population = 20

energy = np.random.uniform(5, 9, population)
damage = np.random.uniform(0.5, 2.0, population)
telomere = np.random.uniform(40, 70, population)

pop_hist, tel_hist, dmg_hist, eng_hist = [], [], [], []

for t in range(steps):
    # record
    current_population_size = len(telomere)
    pop_hist.append(current_population_size)

    if current_population_size == 0:
        tel_hist.append(np.nan) # Use NaN for mean of empty population
        dmg_hist.append(np.nan)
        eng_hist.append(np.nan)
        break # End simulation if population is extinct
    else:
        tel_hist.append(np.mean(telomere))
        dmg_hist.append(np.mean(damage))
        eng_hist.append(np.mean(energy))

    # environmental stress pulses
    if t % 80 == 0 and t > 0:
        damage += np.random.uniform(0.5, 1.5, len(damage))
        energy -= np.random.uniform(0.5, 1.0, len(energy))

    # basal dynamics
    energy += np.random.normal(0.2, 0.1, len(energy))
    damage += np.random.normal(0.15, 0.05, len(damage))
    telomere -= 0.25

    # rejuvenation pulse (regulated telomerase)
    repair_mask = (damage > 1.8) & (energy > 6)
    telomere[repair_mask] += 0.35
    damage[repair_mask] -= 0.6
    energy[repair_mask] -= 0.4

    # division (strict safety)
    divide_mask = (energy > 8.5) & (damage < 2.2) & (telomere > 25)
    if np.any(divide_mask):
        # Store values for new offspring before modifying parent cells
        new_energy_cells = energy[divide_mask] / 2
        new_damage_cells = damage[divide_mask]
        new_telomere_cells = telomere[divide_mask] - 1.0

        # Halve the energy of the parent cells *before* concatenation
        energy[divide_mask] /= 2

        # Concatenate the new cells
        energy = np.concatenate([energy, new_energy_cells])
        damage = np.concatenate([damage, new_damage_cells])
        telomere = np.concatenate([telomere, new_telomere_cells])

    # death
    alive = (energy > 0.5) & (damage < 4.5) & (telomere > 5)
    energy, damage, telomere = energy[alive], damage[alive], telomere[alive]

# -------- VISUALS --------

plt.figure()
plt.plot(pop_hist)
plt.xlabel("Time")
plt.ylabel("Population")
plt.title("Immortality Stress Test – Population Stability")
plt.show()

plt.figure()
plt.plot(tel_hist)
plt.xlabel("Time")
plt.ylabel("Mean Telomere Length")
plt.title("Telomere Homeostasis Under Stress")
plt.show()

plt.figure()
plt.plot(eng_hist, label="Energy")
plt.plot(dmg_hist, label="Damage")
plt.legend()
plt.xlabel("Time")
plt.ylabel("Mean State")
plt.title("Energy vs Damage (Aging ↔ Rejuvenation)")
plt.show()
```

```python
import numpy as np

class DigitalGenome:
    def __init__(self, dna=None):
        # genes are continuous, bounded
        self.dna = dna if dna is not None else np.array([
            np.random.uniform(0.3, 0.8),  # telomerase sensitivity
            np.random.uniform(0.2, 0.7),  # repair efficiency
            np.random.uniform(0.1, 0.5),  # mutation resistance
            np.random.uniform(0.2, 0.6),  # division restraint
        ])

    def mutate(self, rate=0.05):
        noise = np.random.normal(0, rate, size=len(self.dna))
        self.dna = np.clip(self.dna + noise, 0.0, 1.0)
        return self
```

```python
def fitness(cell):
    if not cell.alive:
        return 0.0

    tel_score = np.exp(-abs(cell.telomere.L - 1000) / 300)
    damage_penalty = np.exp(-cell.age * 0.001)
    division_penalty = np.exp(-cell.divisions * 0.05)

    return tel_score * damage_penalty * division_penalty
```

```python
# Adversarial Stress Generator & Immortality Stress-Test (Visuals)

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Adversarial Stress Generator
# -----------------------------
class StressGenerator:
    def __init__(self, modes=("oxidative","nutrient","replication"), intensity=1.0, adversarial=True):
        self.modes = modes
        self.intensity = intensity
        self.adversarial = adversarial

    def sample(self, step):
        # curriculum + adversarial bursts
        base = self.intensity * (1 + 0.01*step)
        spikes = np.random.rand() < (0.05 if self.adversarial else 0.01)
        spike = base * (3.0 if spikes else 1.0)

        stress = {
            "oxidative": spike * np.random.gamma(1.2, 0.8),
            "nutrient":   spike * np.random.gamma(1.0, 0.6),
            "replication":spike * np.random.gamma(1.4, 0.7),
        }
        return stress

# -----------------------------
# Minimal Digital Cell (stress-aware)
# -----------------------------
class Cell:
    def __init__(self, tel=1000.0, repair=0.6, restraint=0.4):
        self.tel = tel
        self.repair = repair
        self.restraint = restraint
        self.age = 0
        self.divs = 0
        self.alive = True
        self.damage = 0.0

    def step(self, stress):
        if not self.alive:
            return
        # accumulate damage
        self.damage += stress["oxidative"]*0.6 + stress["replication"]*0.4
        # repair & cleanup
        self.damage = max(0.0, self.damage - self.repair*(0.8+0.2*np.tanh(self.age/200)))
        # division (restrained)
        p_div = max(0.0, 0.6*(1-self.restraint) * np.exp(-self.damage/120))
        if np.random.rand() < p_div:
            self.divs += 1
            self.tel -= 1.0
            self.tel += 4.0*(1-self.damage/200)  # controlled telomerase
        # senescence / death checks
        if self.tel < 80 or self.damage > 220:
            self.alive = False
        self.age += 1

# -----------------------------
# Run Immortality Stress Test
# -----------------------------
np.random.seed(3)
N = 40
cells = [Cell(tel=1000, repair=np.random.uniform(0.5,0.8), restraint=np.random.uniform(0.3,0.6)) for _ in range(N)]
stress_gen = StressGenerator(intensity=1.0, adversarial=True)

T = 600
alive_hist, tel_hist, dmg_hist = [], [], []
stress_hist = []

for t in range(T):
    stress = stress_gen.sample(t)
    stress_mag = sum(stress.values())
    stress_hist.append(stress_mag)

    for c in cells:
        c.step(stress)

    alive = sum(c.alive for c in cells)
    alive_hist.append(alive)
    tel_hist.append(np.mean([c.tel for c in cells if c.alive] or [0]))
    dmg_hist.append(np.mean([c.damage for c in cells if c.alive] or [0]))

# -----------------------------
# Visuals
# -----------------------------
plt.figure(figsize=(10,6))

plt.subplot(3,1,1)
plt.plot(alive_hist)
plt.ylabel("Alive Cells")
plt.title("Adversarial Immortality Stress Test")

plt.subplot(3,1,2)
plt.plot(tel_hist, label="Telomere")
plt.plot(dmg_hist, label="Damage")
plt.legend()
plt.ylabel("State")

plt.subplot(3,1,3)
plt.plot(stress_hist)
plt.xlabel("Time")
plt.ylabel("Stress Magnitude")

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf

class AdaptiveNesterovRMSAdam(tf.keras.optimizers.Optimizer):
    def __init__(
        self,
        learning_rate=0.01,        # Higher than Adam
        beta_1=0.9,                # Momentum
        beta_2=0.999,              # Variance
        rms_decay=0.9,             # RMSprop smoothing
        epsilon=1e-7,
        name="AdaptiveNesterovRMSAdam",
        **kwargs
    ):
        super().__init__(name, **kwargs)
        self._set_hyper("learning_rate", learning_rate)
        self.beta_1 = beta_1
        self.beta_2 = beta_2
        self.rms_decay = rms_decay
        self.epsilon = epsilon

    def build(self, var_list):
        for var in var_list:
            self.add_variable_from_reference(var, "m")   # momentum
            self.add_variable_from_reference(var, "v")   # variance
            self.add_variable_from_reference(var, "rms") # RMS accumulator

    @tf.function
    def update_step(self, grad, var):
        lr = self._get_hyper("learning_rate")

        m = self.get_slot(var, "m")
        v = self.get_slot(var, "v")
        rms = self.get_slot(var, "rms")

        # Adam moments
        m.assign(self.beta_1 * m + (1. - self.beta_1) * grad)
        v.assign(self.beta_2 * v + (1. - self.beta_2) * tf.square(grad))

        # RMSprop normalization
        rms.assign(self.rms_decay * rms + (1. - self.rms_decay) * tf.square(grad))

        # Bias correction
        t = tf.cast(self.iterations + 1, tf.float32)
        m_hat = m / (1. - tf.pow(self.beta_1, t))
        v_hat = v / (1. - tf.pow(self.beta_2, t))

        # Nesterov-style lookahead
        nesterov_grad = self.beta_1 * m_hat + (1. - self.beta_1) * grad

        # Final update (high-LR but stabilized)
        denom = tf.sqrt(v_hat + rms) + self.epsilon
        update = lr * nesterov_grad / denom

        var.assign_sub(update)

    def get_config(self):
        return {
            "learning_rate": self._serialize_hyperparameter("learning_rate"),
            "beta_1": self.beta_1,
            "beta_2": self.beta_2,
            "rms_decay": self.rms_decay,
            "epsilon": self.epsilon,
        }
```

```python
import tensorflow as tf

class CancerSafeOptimizer(tf.keras.optimizers.Optimizer):
    def __init__(
        self,
        learning_rate=0.02,
        beta_1=0.9,
        beta_2=0.999,
        max_grad_norm=1.0,            # Proliferation limit
        telomere_safe_min=0.8,        # normalized bounds
        telomere_safe_max=1.2,
        epsilon=1e-7,
        name="CancerSafeOptimizer",
        **kwargs
    ):
        super().__init__(name, **kwargs)
        self._set_hyper("learning_rate", learning_rate)
        self.beta_1 = beta_1
        self.beta_2 = beta_2
        self.max_grad_norm = max_grad_norm
        self.telomere_safe_min = telomere_safe_min
        self.telomere_safe_max = telomere_safe_max
        self.epsilon = epsilon

    def build(self, var_list):
        for var in var_list:
            self.add_variable_from_reference(var, "m")
            self.add_variable_from_reference(var, "v")

    @tf.function
    def update_step(self, grad, var):
        lr = self._get_hyper("learning_rate")

        m = self.get_slot(var, "m")
        v = self.get_slot(var, "v")

        # Adam-style moments
        m.assign(self.beta_1 * m + (1. - self.beta_1) * grad)
        v.assign(self.beta_2 * v + (1. - self.beta_2) * tf.square(grad))

        # Bias correction
        t = tf.cast(self.iterations + 1, tf.float32)
        m_hat = m / (1. - tf.pow(self.beta_1, t))
        v_hat = v / (1. - tf.pow(self.beta_2, t))

        # --- Cancer Safety Layer ---

        # 1. Gradient norm projection (anti-tumor)
        grad_norm = tf.norm(m_hat)
        safe_scale = tf.minimum(1.0, self.max_grad_norm / (grad_norm + self.epsilon))
        safe_grad = m_hat * safe_scale

        # 2. Telomere safety constraint (soft clamp)
        is_telomere = tf.strings.regex_full_match(var.name, ".*telomere.*")

        def telomere_guard():
            return tf.where(
                var < self.telomere_safe_min,
                tf.abs(safe_grad),
                tf.where(
                    var > self.telomere_safe_max,
                    -tf.abs(safe_grad),
                    safe_grad
                )
            )

        safe_grad = tf.cond(is_telomere, telomere_guard, lambda: safe_grad)

        # 3. Final stabilized update
        update = lr * safe_grad / (tf.sqrt(v_hat) + self.epsilon)
        var.assign_sub(update)

    def get_config(self):
        return {
            "learning_rate": self._serialize_hyperparameter("learning_rate"),
            "beta_1": self.beta_1,
            "beta_2": self.beta_2,
            "max_grad_norm": self.max_grad_norm,
            "telomere_safe_min": self.telomere_safe_min,
            "telomere_safe_max": self.telomere_safe_max,
            "epsilon": self.epsilon,
        }
```

```python
import tensorflow as tf

class PhaseAwareOptimizer(tf.keras.optimizers.Optimizer):
    def __init__(
        self,
        base_lr=0.01,
        epsilon=1e-7,
        name="PhaseAwareOptimizer",
        **kwargs
    ):
        super().__init__(name, **kwargs)
        self._set_hyper("base_lr", base_lr)
        self.epsilon = epsilon

        # Phase configs
        self.phase_config = {
            "AGING":        {"lr": 0.002, "beta1": 0.95, "beta2": 0.9995},
            "REGENERATION": {"lr": 0.05,  "beta1": 0.85, "beta2": 0.99},
            "IMMORTALITY":  {"lr": 0.01,  "beta1": 0.9,  "beta2": 0.999},
        }

        self.current_phase = tf.Variable(
            initial_value="IMMORTALITY",
            trainable=False,
            dtype=tf.string
        )

    def build(self, var_list):
        for var in var_list:
            self.add_variable_from_reference(var, "m")
            self.add_variable_from_reference(var, "v")

    def set_phase(self, phase):
        self.current_phase.assign(phase)

    @tf.function
    def update_step(self, grad, var):
        cfg = self.phase_config[self.current_phase.numpy().decode()]

        lr = cfg["lr"]
        beta1 = cfg["beta1"]
        beta2 = cfg["beta2"]

        m = self.get_slot(var, "m")
        v = self.get_slot(var, "v")

        # Adam/RMSprop hybrid
        m.assign(beta1 * m + (1. - beta1) * grad)
        v.assign(beta2 * v + (1. - beta2) * tf.square(grad))

        # Stability constraint (anti-cancer)
        grad_norm = tf.norm(m)
        safe_grad = m / (tf.sqrt(v) + self.epsilon)
        safe_grad *= tf.minimum(1.0, 1.0 / (grad_norm + self.epsilon))

        var.assign_sub(lr * safe_grad)

    def get_config(self):
        return {"base_lr": self._serialize_hyperparameter("base_lr")}
```

```python
import tensorflow as tf

class PhaseAwareActivation(tf.keras.layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.phase = tf.Variable(
            initial_value="IMMORTALITY",
            trainable=False,
            dtype=tf.string
        )

    def set_phase(self, phase):
        self.phase.assign(phase)

    def call(self, x):
        phase = self.phase.numpy().decode()

        if phase == "AGING":
            return (
                0.4 * tf.nn.tanh(x) +
                0.3 * tf.nn.softsign(x) +
                0.3 * tf.nn.sigmoid(x)
            )

        elif phase == "REGENERATION":
            return (
                0.4 * tf.nn.relu(x) +
                0.3 * tf.nn.leaky_relu(x, alpha=0.1) +
                0.3 * tf.nn.mish(x)
            )

        else:  # IMMORTALITY
            return (
                0.4 * tf.nn.selu(x) +
                0.3 * tf.nn.elu(x) +
                0.3 * tf.nn.gelu(x)
            )
```

```python
class PhaseAwareDense(tf.keras.layers.Layer):
    def __init__(self, units):
        super().__init__()
        self.dense = tf.keras.layers.Dense(units)
        self.activation = PhaseAwareActivation()

    def set_phase(self, phase):
        self.activation.set_phase(phase)

    def call(self, x):
        x = self.dense(x)
        return self.activation(x)
```

```python
import tensorflow as tf

class CellularHamiltonian:
    def __init__(self, telomere_target=1.0):
        self.telomere_target = telomere_target

    def potential(self, telomere, damage):
        # Stability cost
        return (
            tf.square(telomere - self.telomere_target) +
            2.0 * tf.square(damage)
        )

    def kinetic(self, grad_norm):
        # Growth / signaling intensity
        return 0.5 * tf.square(grad_norm)

    def total(self, telomere, damage, grad_norm):
        return self.kinetic(grad_norm) + self.potential(telomere, damage)
```

```python
import tensorflow as tf

class TelomereAdaptiveActivation(tf.keras.layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)

    def call(self, x, telomere, damage, stem_signal):
        """
        x: pre-activation
        telomere: [0,1] normalized
        damage: [0,1]
        stem_signal: [0,1]
        """

        # --- Activation experts ---
        growth = (
            0.5 * tf.nn.relu(x) +
            0.5 * tf.nn.leaky_relu(x, alpha=0.1)
        )

        repair = (
            0.6 * tf.nn.mish(x) +
            0.4 * tf.nn.gelu(x)
        )

        stability = (
            0.6 * tf.nn.selu(x) +
            0.4 * tf.nn.elu(x)
        )

        suppression = (
            0.5 * tf.nn.tanh(x) +
            0.5 * tf.nn.softsign(x)
        )

        # --- Gating logic (biology-inspired) ---
        regen_gate = tf.nn.sigmoid(
            4.0 * stem_signal - 3.0 * damage
        )

        immortality_gate = tf.nn.sigmoid(
            6.0 * telomere - 3.0 * damage
        )

        safety_gate = tf.nn.sigmoid(
            4.0 * damage
        )

        # --- Combine ---
        y = (
            regen_gate * repair +
            (1.0 - regen_gate) * stability +
            0.3 * growth
        )

        # --- Cancer safety envelope ---
        y = (1.0 - safety_gate) * y + safety_gate * suppression

        # --- Final bounded output (prevents runaway) ---
        return tf.nn.softplus(y)
```

```python
from dataclasses import dataclass
import numpy as np

@dataclass
class CellState:
    telomere_length: float
    dna_damage: float
    metabolic_load: float
    division_pressure: float
    stemness: float           # 0 = somatic, 1 = stem
    debris_load: float
    alive: bool = True
```

```python
class TelomereModel:
    def __init__(self, target=1.0, min_len=0.5, max_len=1.3):
        self.target = target
        self.min = min_len
        self.max = max_len

    def erosion(self, cell: CellState):
        return 0.01 * cell.division_pressure + 0.005 * cell.dna_damage

    def apply(self, cell: CellState, telomerase_signal):
        delta = telomerase_signal - self.erosion(cell)
        cell.telomere_length = np.clip(
            cell.telomere_length + delta,
            self.min,
            self.max
        )
```

```python
class PhagocytosisSystem:
    def clean(self, cell: CellState):
        clearance = 0.3 * cell.debris_load
        cell.debris_load -= clearance
        cell.metabolic_load *= 0.95
        cell.dna_damage *= 0.97
```

```python
def simulate(cells, env, steps=1000):
    telomeres = TelomereModel()
    phago = PhagocytosisSystem()
    ctrl = TelomeraseController(target=1.0)

    history = []

    for t in range(steps):
        env.step()
        for cell in cells:
            if not cell.alive:
                continue

            signal = ctrl.act(cell, env)
            telomeres.apply(cell, signal)
            phago.clean(cell)

            if cell.telomere_length < telomeres.min:
                cell.alive = False

            history.append(cell.telomere_length)

    return history
```

```python
class SymbolicState:
    def __init__(self, cell):
        self.telomere = self.bucket(cell.telomere_length)
        self.damage = self.bucket(cell.dna_damage)
        self.stem = cell.stemness > 0.7
        self.dividing = cell.division_pressure > 0.6
        self.debris = self.bucket(cell.debris_load)

    def bucket(self, x):
        if x < 0.3: return "LOW"
        if x < 0.7: return "MEDIUM"
        return "HIGH"
```

```python
BIO_RULES = [
    # Telomerase safety
    lambda s: not (s.telomere == "HIGH" and s.dividing and not s.stem),

    # Cancer prevention
    lambda s: not (s.damage == "HIGH" and s.dividing),

    # Phagocytosis necessity
    lambda s: not (s.debris == "HIGH" and s.dividing),

    # Stem cell exclusivity
    lambda s: s.stem or not s.dividing
]
```

```python
class TemporalMemory:
    def __init__(self, window=10):
        self.history = []
        self.window = window

    def update(self, symbolic_state):
        self.history.append(symbolic_state)
        if len(self.history) > self.window:
            self.history.pop(0)

    def aging_trend(self):
        lows = sum(1 for s in self.history if s.telomere == "LOW")
        return lows > self.window * 0.6
```

```python
class NeurosymbolicController:
    def __init__(self):
        self.memory = TemporalMemory()

    def validate(self, cell, proposed_telomerase):
        sym = SymbolicState(cell)
        self.memory.update(sym)

        # Rule checking
        for rule in BIO_RULES:
            if not rule(sym):
                return 0.0  # hard veto

        # Temporal reasoning
        if self.memory.aging_trend() and sym.stem:
            return min(proposed_telomerase, 0.02)

        return proposed_telomerase
```

```python
import numpy as np

class TelomereDynamics:
    def __init__(self,
                 base_shortening=50.0,
                 noise_std=10.0,
                 max_extension=120.0):
        self.base_shortening = base_shortening
        self.noise_std = noise_std
        self.max_extension = max_extension

    def step(self, telomere_length, telomerase_activity):
        """
        Mechanistic update:
        L(t+1) = L(t) - Δ + E + noise
        """
        shortening = self.base_shortening
        extension = np.clip(
            telomerase_activity * self.max_extension,
            0.0,
            self.max_extension
        )
        noise = np.random.normal(0, self.noise_std)

        return telomere_length - shortening + extension + noise
```

```python
class TelomereBudget:
    def __init__(self, target=1.0):
        self.length = target
        self.target = target

    def shorten(self, error):
        self.length -= 0.01 * error

    def extend(self, repair_signal):
        self.length += 0.02 * repair_signal

    def clamp(self):
        self.length = max(0.2, min(self.length, 1.2))
```

```python
class VirtualTelomerase:
    def activate(self, damage, task_need, safety_ok):
        if not safety_ok:
            return 0.0
        if damage > 0.3 and task_need > 0.4:
            return min(1.0, damage * task_need)
        return 0.0
```

```python
def agi_step(agent, environment):
    # Measure state
    damage = environment.error_signal()
    task_need = environment.repair_demand()
    safety_ok = agent.symbolic_core.verify()

    # Telomerase decision
    telomerase = agent.telomerase.activate(
        damage, task_need, safety_ok
    )

    # Update telomere budget
    agent.telomeres.extend(telomerase)
    agent.telomeres.shorten(damage)
    agent.telomeres.clamp()

    # Learning gate
    if telomerase > 0 and agent.telomeres.length > 0.6:
        agent.learn(rate=telomerase)
    else:
        agent.freeze_or_prune()
```

```python
class CellLatentEncoder(nn.Module):
    """
    Compresses high-dimensional cell state into a stable latent vector.
    """
    def __init__(self, latent_dim=16):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(6, 64),
            nn.GELU(),
            nn.Linear(64, 32),
            nn.GELU(),
            nn.Linear(32, latent_dim)
        )

    def forward(self, tel, dmg, pha, prolif, gf, nutrient):
        x = torch.stack([tel, dmg, pha, prolif, gf, nutrient], dim=-1)
        return self.encoder(x)
```

```python
class CellWorldModel(nn.Module):
    """
    Predicts next latent state given current latent + action.
    """
    def __init__(self, latent_dim=16):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(latent_dim + 1, 128),
            nn.GELU(),
            nn.Linear(128, latent_dim)
        )

    def forward(self, z, pulse):
        pulse = pulse.unsqueeze(-1)
        return self.model(torch.cat([z, pulse], dim=-1))
```

```python
import math
import random
from dataclasses import dataclass
from typing import Tuple, List

import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim

# -----------------------------
# Simulation primitives (safe)
# -----------------------------

@dataclass
class CellState:
    telomere_length: float  # abstract scalar (arbitrary units)
    proliferation_state: float  # 0 = quiescent, >0 = proliferative drive
    phagocytosis_activity: float  # 0..1 activity level
    damage: float  # abstract damage score, 0..1


class VirtualCell:
    """Abstract virtual cell: no biology, only numerical dynamics.

    The update rules are intentionally high-level and safe: they operate on
    abstract scalars and deterministic rules. The 'telomerase_pulse' input is
    normalized and purely numerical.
    """

    def __init__(self,
                 init_telomere: float = 10.0,
                 min_telomere: float = 2.0,
                 max_telomere: float = 50.0,
                 telomere_trim_rate: float = 0.01,
                 division_cost: float = 0.5):
        self.min_telomere = min_telomere
        self.max_telomere = max_telomere
        self.telomere_trim_rate = telomere_trim_rate
        self.division_cost = division_cost
        self.state = CellState(
            telomere_length=init_telomere,
            proliferation_state=0.0,
            phagocytosis_activity=0.5,
            damage=0.0,
        )

    def step(self, env_inputs: dict, telomerase_pulse: float, dt: float = 1.0):
        """Advance cell state by one timestep.

        env_inputs can include 'growth_factor', 'stress', and 'nutrient'.
        telomerase_pulse is a scalar in [0, 1] (model output) controlling
        telomere elongation.
        """
        gf = float(env_inputs.get('growth_factor', 0.0))
        stress = float(env_inputs.get('stress', 0.0))
        nutrient = float(env_inputs.get('nutrient', 1.0))

        # Proliferative drive increases with growth factor, decreases with low nutrient
        prolift = self.state.proliferation_state
        prolift += dt * (0.5 * gf - 0.3 * (1.0 - nutrient) - 0.2 * stress)
        prolift = max(0.0, prolift)

        # Telomere dynamics (abstract): length increases with telomerase pulse,
        # decreases with divisions (proliferation) and a small passive shortening.
        tel = self.state.telomere_length
        tel += dt * (telomerase_pulse * 0.8 - 0.02)  # passive shortening -0.02

        # If proliferation_state above threshold -> division event
        if prolift > 0.8:
            # division consumes telomere length (abstract cost)
            tel -= self.division_cost
            # regeneration of damage via phagocytosis if active
            self.state.damage = max(0.0, self.state.damage - 0.1 * self.state.phagocytosis_activity)
            # small random reset of proliferation
            prolift *= 0.4

        # Telomere trimming if above safe max
        if tel > self.max_telomere:
            tel -= self.telomere_trim_rate * (tel - self.max_telomere)

        # Damage accrues with higher stress or extreme telomere shortening
        if stress > 0.6:
            self.state.damage = min(1.0, self.state.damage + 0.05 * stress)
        if tel < self.min_telomere + 1.0:
            self.state.damage = min(1.0, self.state.damage + 0.02)

        # Phagocytosis activity adapts to damage and nutrient availability
        pha = self.state.phagocytosis_activity
        pha += dt * (0.2 * self.state.damage - 0.1 * (1.0 - nutrient))
        pha = float(np.clip(pha, 0.0, 1.0))

        # Save back
        self.state.telomere_length = float(np.clip(tel, 0.0, 1000.0))
        self.state.proliferation_state = float(prolift)
        self.state.phagocytosis_activity = pha

        # Return a snapshot
        return CellState(
            telomere_length=self.state.telomere_length,
            proliferation_state=self.state.proliferation_state,
            phagocytosis_activity=self.state.phagocytosis_activity,
            damage=self.state.damage,
        )


# -----------------------------
# Virtual Organ-on-a-Chip (abstract)
# -----------------------------

class VirtualOOC:
    """Simple multi-cell environment. No real biology; purely numerical.

    - Tracks global nutrient / flow / growth factor.
    - Can introduce injury (localized increase in damage) or pathogen (increase stress).
    """

    def __init__(self, n_cells: int = 64):
        self.n_cells = n_cells
        self.cells = [VirtualCell(init_telomere=10.0 + random.random()*2.0) for _ in range(n_cells)]
        # environment fields
        self.nutrient = 1.0
        self.growth_factor = 0.1
        self.flow = 1.0

    def step(self, control_signals: List[float], dt: float = 1.0, perturb: dict = None):
        """Advance environment and all cells.

        control_signals: per-cell telomerase pulses (length = n_cells)
        perturb: optional dict: {'injury_indices':[...], 'pathogen': strength}
        """
        if perturb is None:
            perturb = {}
        injury_indices = perturb.get('injury_indices', [])
        pathogen = float(perturb.get('pathogen', 0.0))

        # Ambient effects
        # Pathogen increases stress globally
        stress_base = pathogen * 0.5

        snapshots = []
        for i, cell in enumerate(self.cells):
            env_inputs = {
                'growth_factor': self.growth_factor,
                'stress': stress_base,
                'nutrient': self.nutrient,
            }
            if i in injury_indices:
                env_inputs['stress'] += 0.4  # local injury stress
                # local damage bump
                cell.state.damage = min(1.0, cell.state.damage + 0.2)

            pulse = float(np.clip(control_signals[i], 0.0, 1.0))
            snapshot = cell.step(env_inputs, telomerase_pulse=pulse, dt=dt)
            snapshots.append(snapshot)

        # Update global fields with simple ad-hoc rules
        # nutrient is consumed proportional to proliferative drive
        avg_prolif = np.mean([s.proliferation_state for s in snapshots])
        self.nutrient = float(np.clip(self.nutrient - 0.01 * avg_prolif + 0.001 * self.flow, 0.0, 1.0))

        return snapshots

    def get_observation(self):
        """Return an observation vector for the AI regulator.

        We'll return per-cell telomere lengths, damage, and phagocytosis activity.
        """
        tel = np.array([c.state.telomere_length for c in self.cells], dtype=np.float32)
        dmg = np.array([c.state.damage for c in self.cells], dtype=np.float32)
        pha = np.array([c.state.phagocytosis_activity for c in self.cells], dtype=np.float32)
        gf = np.array([self.growth_factor], dtype=np.float32)
        nutrient = np.array([self.nutrient], dtype=np.float32)
        return {'telomere': tel, 'damage': dmg, 'phagocytosis': pha, 'growth_factor': gf, 'nutrient': nutrient}


# -----------------------------
# Synthetic dataset generator
# -----------------------------

def generate_synthetic_trajectory(n_steps: int = 100, n_cells: int = 32, seed: int = None):
    rng = np.random.RandomState(seed)
    env = VirtualOOC(n_cells=n_cells)

    # random schedule of perturbations
    perturb_schedule = {}
    if rng.rand() < 0.5:
        # add a single injury
        idxs = rng.choice(n_cells, size=max(1, n_cells // 16), replace=False).tolist()
        perturb_schedule[int(n_steps*0.2)] = {'injury_indices': idxs}
    if rng.rand() < 0.4:
        perturb_schedule[int(n_steps*0.5)] = {'pathogen': float(rng.rand()*0.8)}

    # naive controller: pulses when telomere < threshold and proliferation high
    data = []
    for t in range(n_steps):
        if t in perturb_schedule:
            perturb = perturb_schedule[t]
        else:
            perturb = None

        obs = env.get_observation()
        pulses = []
        for i in range(n_cells):
            tel = obs['telomere'][i]
            dmg = obs['damage'][i]
            # rule-based pulse: boost if telomere below 6 and damage not high
            p = 0.0
            if tel < 6.0 and dmg < 0.7:
                p = min(1.0, (6.0 - tel) / 4.0)
            pulses.append(p)

        snapshots = env.step(pulses, dt=1.0, perturb=perturb)
        data.append({'obs': obs, 'pulses': np.array(pulses, dtype=np.float32)})

    return data


# -----------------------------
# PyTorch regulator model (safe, abstract)
# -----------------------------

class RegulatorNet(nn.Module):
    """A lightweight per-cell regulator with shared weights.

    Input per-cell: telomere, damage, phagocytosis, plus global nutrient/gf.
    Output: telomerase_pulse in [0, 1].
    """

    def __init__(self, hidden_dim: int = 64):
        super().__init__()
        self.per_cell = nn.Sequential(
            nn.Linear(5, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1),
            nn.Sigmoid(),
        )

    def forward(self, tel, dmg, pha, gf, nutrient):
        # tel, dmg, pha: (batch, n_cells)
        # gf, nutrient: (batch, 1)
        b, n = tel.shape
        # tile global signals
        gf_tile = gf.expand(-1, n)
        nut_tile = nutrient.expand(-1, n)
        x = torch.stack([tel, dmg, pha, gf_tile, nut_tile], dim=-1)  # (b,n,5)
        x = x.view(b * n, 5)
        out = self.per_cell(x).view(b, n)
        return out


# -----------------------------
# Training utilities
# -----------------------------

def collate_trajectory(data: List[dict]):
    # data is list of timesteps: each has 'obs' and 'pulses'
    tel = np.stack([d['obs']['telomere'] for d in data], axis=0)
    dmg = np.stack([d['obs']['damage'] for d in data], axis=0)
    pha = np.stack([d['obs']['phagocytosis'] for d in data], axis=0)
    gf = np.stack([d['obs']['growth_factor'] for d in data], axis=0)
    nutrient = np.stack([d['obs']['nutrient'] for d in data], axis=0)
    pulses = np.stack([d['pulses'] for d in data], axis=0)

    # convert to torch tensors and add batch dim (batch=1)
    tel_t = torch.from_numpy(tel).unsqueeze(0)
    dmg_t = torch.from_numpy(dmg).unsqueeze(0)
    pha_t = torch.from_numpy(pha).unsqueeze(0)
    gf_t = torch.from_numpy(gf).to(torch.float32).unsqueeze(0)
    nutrient_t = torch.from_numpy(nutrient).to(torch.float32).unsqueeze(0)
    pulses_t = torch.from_numpy(pulses).unsqueeze(0)
    return tel_t, dmg_t, pha_t, gf_t, nutrient_t, pulses_t


def train_regulator(model: nn.Module,
                    optimizer,
                    data_trajectories: List[List[dict]],
                    epochs: int = 50,
                    device: str = 'cpu'):
    model.to(device)
    loss_fn = nn.MSELoss()

    for epoch in range(epochs):
        total_loss = 0.0
        random.shuffle(data_trajectories)
        for traj in data_trajectories:
            tel_t, dmg_t, pha_t, gf_t, nutrient_t, pulses_t = collate_trajectory(traj)
            tel_t = tel_t.to(device).float()
            dmg_t = dmg_t.to(device).float()
            pha_t = pha_t.to(device).float()
            gf_t = gf_t.to(device).float()
            nutrient_t = nutrient_t.to(device).float()
            pulses_t = pulses_t.to(device).float()

            optimizer.zero_grad()
            # naive per-timestep prediction loop (teacher forcing style)
            pred_seq = []
            seq_len = tel_t.shape[1]
            for t in range(seq_len):
                tel_step = tel_t[:, t, :]
                dmg_step = dmg_t[:, t, :]
                pha_step = pha_t[:, t, :]
                gf_step = gf_t[:, t, :]
                nutrient_step = nutrient_t[:, t, :]
                pred = model(tel_step, dmg_step, pha_step, gf_step, nutrient_step)
                pred_seq.append(pred.unsqueeze(1))

            pred_seq = torch.cat(pred_seq, dim=1)
            loss = loss_fn(pred_seq, pulses_t)
            loss.backward()
            # gradient clipping for stability (important when using higher LR)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()

            total_loss += float(loss.item())

        avg_loss = total_loss / max(1, len(data_trajectories))
        if epoch % max(1, epochs // 10) == 0 or epoch == epochs - 1:
            print(f"Epoch {epoch+1}/{epochs} avg_loss={avg_loss:.6f}")

    return model


# -----------------------------
# End-to-end experiment harness
# -----------------------------

def run_experiment(seed: int = 0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    # 1) Generate dataset: multiple trajectories
    trajectories = [generate_synthetic_trajectory(n_steps=80, n_cells=32, seed=seed + i) for i in range(40)]

    # 2) Initialize model and optimizer with a relatively high LR but stable schedule
    model = RegulatorNet(hidden_dim=64)
    base_lr = 1e-3  # 'higher' for small model - tune as needed
    optimizer = optim.Adam(model.parameters(), lr=base_lr)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50)

    # 3) Train
    model = train_regulator(model, optimizer, trajectories, epochs=30)
    scheduler.step()

    # 4) Quick evaluation on a held-out synthetic trajectory
    test_traj = generate_synthetic_trajectory(n_steps=120, n_cells=32, seed=999)
    tel_t, dmg_t, pha_t, gf_t, nutrient_t, pulses_t = collate_trajectory(test_traj)
    with torch.no_grad():
        pred_seq = []
        for t in range(tel_t.shape[1]):
            tel_step = tel_t[:, t, :].float()
            dmg_step = dmg_t[:, t, :].float()
            pha_step = pha_t[:, t, :].float()
            gf_step = gf_t[:, t, :].float()
            nutrient_step = nutrient_t[:, t, :].float()
            pred = model(tel_step, dmg_step, pha_step, gf_step, nutrient_step)
            pred_seq.append(pred.squeeze(0).cpu().numpy())

    pred_seq = np.stack(pred_seq, axis=0)
    # Evaluate simple metrics: mean squared error per cell over time
    mse = float(np.mean((pred_seq - pulses_t.squeeze(0).numpy()) ** 2))
    print(f"Evaluation MSE on held-out synthetic data: {mse:.6f}")

    # 5) Visualize telomere trajectories for a small subset of cells
    visualize_test_run(test_traj, pred_seq, n_plot=6)


def visualize_test_run(traj: List[dict], pred_seq: np.ndarray, n_plot: int = 6):
    # show telomere length vs time and predicted pulses
    tel = np.stack([d['obs']['telomere'] for d in traj], axis=0)
    pulses = np.stack([d['pulses'] for d in traj], axis=0)
    time = np.arange(tel.shape[0])

    n_cells = tel.shape[1]
    chosen = list(range(min(n_plot, n_cells)))

    fig, axs = plt.subplots(len(chosen), 2, figsize=(10, 2.5 * len(chosen)))
    for i, c in enumerate(chosen):
        axs[i, 0].plot(time, tel[:, c])
        axs[i, 0].set_ylabel(f"Tel {c}")
        axs[i, 0].set_xlabel("Timestep")
        axs[i, 1].plot(time, pulses[:, c])
        axs[i, 1].set_ylabel("Pulse (target)")
        axs[i, 1].set_xlabel("Timestep")

    plt.tight_layout()
    plt.show()


# -----------------------------
# If executed as a script
# -----------------------------

if __name__ == '__main__':
    run_experiment(seed=42)

# === TUNED MODEL VERSION 2 ===
# Improved architecture, optimizer, scheduler
class RegulatorNetV2(nn.Module):
    def __init__(self, input_dim=4, hidden_dim=128, dropout=0.2):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(hidden_dim, 3)
        )
    def forward(self, x): return self.net(x)

model_v2 = RegulatorNetV2()
optimizer_v2 = torch.optim.AdamW(model_v2.parameters(), lr=3e-3, weight_decay=1e-4)
scheduler_v2 = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer_v2, T_0=50)
loss_fn_v2 = nn.MSELoss()


# === HYPERPARAMETER SEARCH MODULE ===
class HyperparameterSearch:
    def __init__(self, model_cls, search_space, device="cpu"):
        self.model_cls = model_cls
        self.search_space = search_space
        self.device = device
        self.results = []

    def sample_params(self):
        import random
        return {k: random.choice(v) for k, v in self.search_space.items()}

    def run_trial(self, X, y):
        params = self.sample_params()
        model = self.model_cls(
            input_dim=params.get("input_dim", 64),
            hidden_dim=params.get("hidden_dim", 128),
            output_dim=params.get("output_dim", 4),
            dropout=params.get("dropout", 0.1)
        ).to(self.device)

        optimizer = torch.optim.AdamW(model.parameters(), lr=params.get("lr", 1e-4))
        loss_fn = nn.MSELoss()

        for epoch in range(params.get("epochs", 5)):
            optimizer.zero_grad()
            pred = model(X)
            loss = loss_fn(pred, y)
            loss.backward()
            optimizer.step()

        final_loss = loss.item()
        self.results.append((params, final_loss))
        return final_loss

    def best(self):
        return sorted(self.results, key=lambda x: x[1])[0] if self.results else None

# Example search space
search_space = {
    "lr": [1e-5, 5e-5, 1e-4, 5e-4],
    "hidden_dim": [64, 128, 256],
    "dropout": [0.0, 0.1, 0.2],
    "epochs": [3, 5, 8]
}


# === VIRTUAL TELOMERASE CONTROLLER ===
class TelomeraseController:
    """A safety-first virtual telomerase controller.

    Modes:
    - 'rule': deterministic rule-based controller that issues pulses only when a
      cell shows clear 'need' (low telomere, high proliferation drive, and not
      too damaged).
    - 'learned': a small neural network that predicts pulse magnitudes but is
      gated by the same 'need' mask so it cannot act when there is no need.

    Safety features:
    - per-cell cumulative activity cap (prevents unlimited stimulation)
    - per-cell cooldown windows between pulses
    - global maximum duty-cycle
    - optional explicit 'emergency only' mode
    """

    def __init__(self,
                 n_cells: int,
                 mode: str = 'rule',
                 cooldown_steps: int = 5,
                 cumulative_cap: float = 10.0,
                 max_pulse: float = 1.0,
                 device: str = 'cpu'):
        assert mode in ('rule', 'learned')
        self.n_cells = n_cells
        self.mode = mode
        self.cooldown_steps = cooldown_steps
        self.cumulative_cap = cumulative_cap
        self.max_pulse = max_pulse
        self.device = device

        # runtime state
        self._cooldown_timer = np.zeros(n_cells, dtype=int)
        self._cumulative_activity = np.zeros(n_cells, dtype=float)

        # learned model (small) if requested
        if self.mode == 'learned':
            self.model = nn.Sequential(
                nn.Linear(5, 64),
                nn.ReLU(),
                nn.Linear(64, 32),
                nn.ReLU(),
                nn.Linear(32, 1),
                nn.Sigmoid()
            )
            self.model.to(device)
        else:
            self.model = None

    def reset_runtime(self):
        self._cooldown_timer[:] = 0
        self._cumulative_activity[:] = 0.0

    def _need_mask(self, tel, prolif, damage, growth_factor, nutrient):
        """Return a boolean mask indicating which cells currently 'need' telomerase.

        Criteria (all configurable heuristics):
        - tel < tel_threshold OR prolif > prolif_threshold
        - damage < damage_threshold (avoid stimulating heavily damaged cells)
        - nutrient > nutrient_threshold (have resources)
        """
        tel_threshold = 6.0
        prolif_threshold = 0.6
        damage_threshold = 0.8
        nutrient_threshold = 0.2

        need = (tel < tel_threshold) | (prolif > prolif_threshold)
        need = need & (damage < damage_threshold)
        need = need & (nutrient > nutrient_threshold)
        return need.astype(np.bool_)

    def compute_pulses(self, obs: dict, return_debug: bool = False):
        """Compute per-cell telomerase pulses based on observation dict from OOC.

        obs keys expected: 'telomere', 'proliferation', 'damage', 'growth_factor', 'nutrient'
        All are numpy arrays with shape (n_cells,) except growth_factor/nutrient can be scalars.
        """
        tel = np.array(obs['telomere'], dtype=float)
        prolif = np.array(obs.get('proliferation', np.zeros(self.n_cells)), dtype=float)
        damage = np.array(obs.get('damage', np.zeros(self.n_cells)), dtype=float)
        pha = np.array(obs.get('phagocytosis', np.zeros(self.n_cells)), dtype=float)
        gf = float(obs.get('growth_factor', 0.0))
        nutrient = float(obs.get('nutrient', 1.0))

        need = self._need_mask(tel, prolif, damage, gf, nutrient)

        pulses = np.zeros(self.n_cells, dtype=float)
        debug = {'need_mask': need.copy(), 'cooldown': self._cooldown_timer.copy(), 'cumulative': self._cumulative_activity.copy()}

        # compute candidate pulses
        if self.mode == 'rule':
            # simple proportional rule: larger pulse when telomere much below threshold
            tel_threshold = 6.0
            deficit = np.clip((tel_threshold - tel) / max(1e-6, tel_threshold), 0.0, 1.0)
            # factor in proliferation drive
            prolif_factor = np.clip(prolif / 1.0, 0.0, 1.0)
            candidate = np.maximum(deficit, prolif_factor * 0.8)
            # scale by (1 - damage) to avoid stimulating damaged cells
            candidate = candidate * (1.0 - damage)

        else:
            # learned mode: create per-cell feature vectors and run model
            with torch.no_grad():
                tel_t = torch.from_numpy(tel).float().unsqueeze(1).to(self.device)
                dmg_t = torch.from_numpy(damage).float().unsqueeze(1).to(self.device)
                pha_t = torch.from_numpy(pha).float().unsqueeze(1).to(self.device)
                gf_t = torch.full((self.n_cells, 1), gf, dtype=torch.float32).to(self.device)
                nut_t = torch.full((self.n_cells, 1), nutrient, dtype=torch.float32).to(self.device)
                feat = torch.cat([tel_t, dmg_t, pha_t, gf_t, nut_t], dim=1)
                out = self.model(feat).squeeze(1).cpu().numpy()
                candidate = np.clip(out, 0.0, 1.0)

        # Apply need mask, cooldown, and cumulative caps
        for i in range(self.n_cells):
            if not need[i]:
                pulses[i] = 0.0
                continue
            # cooldown
            if self._cooldown_timer[i] > 0:
                pulses[i] = 0.0
                self._cooldown_timer[i] -= 1
                continue
            # cumulative cap
            remaining = max(0.0, self.cumulative_cap - self._cumulative_activity[i])
            max_allowed = min(self.max_pulse, remaining)
            planned = float(candidate[i]) * self.max_pulse
            pulse = min(planned, max_allowed)
            # enforce global duty-cycle (example: limit to 0.5 average across cells)
            pulses[i] = float(pulse)
            # update runtime trackers if pulse issued
            if pulses[i] > 1e-6:
                self._cumulative_activity[i] += pulses[i]
                self._cooldown_timer[i] = self.cooldown_steps

        if return_debug:
            return pulses, debug
        return pulses

    def train_learned(self, training_data: List[Tuple[dict, np.ndarray]], epochs: int = 20, lr: float = 1e-3):
        """Train the learned model using supervised examples of safe pulses.

        training_data: list of tuples (obs_dict, target_pulses_array)
        Each target_pulses_array should respect the same safety policy externally.
        """
        assert self.mode == 'learned' and self.model is not None
        optim = torch.optim.AdamW(self.model.parameters(), lr=lr)
        loss_fn = nn.MSELoss()
        for epoch in range(epochs):
            epoch_loss = 0.0
            random.shuffle(training_data)
            for obs, target in training_data:
                tel = np.array(obs['telomere'], dtype=float)
                damage = np.array(obs.get('damage', np.zeros(self.n_cells)), dtype=float)
                pha = np.array(obs.get('phagocytosis', np.zeros(self.n_cells)), dtype=float)
                gf = float(obs.get('growth_factor', 0.0))
                nut = float(obs.get('nutrient', 1.0))

                tel_t = torch.from_numpy(tel).float().unsqueeze(1).to(self.device)
                dmg_t = torch.from_numpy(damage).float().unsqueeze(1).to(self.device)
                pha_t = torch.from_numpy(pha).float().unsqueeze(1).to(self.device)
                gf_t = torch.full((self.n_cells, 1), gf, dtype=torch.float32).to(self.device)
                nut_t = torch.full((self.n_cells, 1), nut, dtype=torch.float32).to(self.device)
                feat = torch.cat([tel_t, dmg_t, pha_t, gf_t, nut_t], dim=1)
                pred = self.model(feat).squeeze(1)
                target_t = torch.from_numpy(np.clip(target, 0.0, 1.0)).float().to(self.device)

                optim.zero_grad()
                loss = loss_fn(pred, target_t)
                loss.backward()
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), 1.0)
                optim.step()
                epoch_loss += float(loss.item())
            if (epoch + 1) % max(1, epochs // 5) == 0:
                print(f"TelomeraseController learned epoch {epoch+1}/{epochs} loss={epoch_loss/len(training_data):.6f}")


# === Integration example: add controller option to VirtualOOC.step ===

_original_step = VirtualOOC.step

def step_with_controller(self, controller: TelomeraseController = None, dt: float = 1.0, perturb: dict = None):
    # If controller provided, ask it for pulses; otherwise rely on older API
    if controller is None:
        return _original_step(self, control_signals=[0.0]*self.n_cells, dt=dt, perturb=perturb)

    obs = self.get_observation()
    # include proliferation as it's tracked in cell state
    prolif = np.array([c.state.proliferation_state for c in self.cells], dtype=float)
    obs['proliferation'] = prolif

    pulses = controller.compute_pulses(obs)
    return _original_step(self, control_signals=pulses.tolist(), dt=dt, perturb=perturb)

VirtualOOC.step_with_controller = step_with_controller

# === Quick demo function to show controller behavior ===
def demo_controller_behavior():
    env = VirtualOOC(n_cells=16)
    controller = TelomeraseController(n_cells=16, mode='rule', cooldown_steps=4, cumulative_cap=6.0)
    env.nutrient = 0.9
    env.growth_factor = 0.2

    history = []
    for t in range(60):
        obs = env.get_observation()
        # artificially lower telomere in a subset at t=5 to trigger need
        if t == 5:
            for i in range(3):
                env.cells[i].state.telomere_length = 4.0
                env.cells[i].state.proliferation_state = 0.9
        pulses, debug = controller.compute_pulses(obs, return_debug=True)
        env.step_with_controller(controller)
        history.append({'t': t, 'pulses': pulses.copy(), 'debug': debug})

    # summarize
    print("Demo summary (first 12 timesteps):")
    for h in history[:12]:
        print(f"t={h['t']} pulses={h['pulses'][:6]} need={h['debug']['need_mask'][:6]}")

# If run as script, run demo
if __name__ == '__main__':
    demo_controller_behavior()


# === DIGITAL TWIN + EVOLUTIONARY PROCESS ===
class DigitalTwin:
    """A compact surrogate model representing a cell's effective behavior.

    The digital twin is parameterized by a small set of interpretable parameters
    (e.g. telomere_trim_rate, division_cost, phagocytosis_efficiency). These
    parameters are evolved to optimize desired system-level objectives.

    NOTE: This is purely computational and models abstract scalars — not lab
    protocols or biological recipes.
    """

    def __init__(self, params: dict = None):
        defaults = {
            'telomere_trim_rate': 0.01,
            'division_cost': 0.5,
            'phagocytosis_efficiency': 0.5,
            'min_telomere': 2.0,
            'max_telomere': 50.0,
        }
        self.params = defaults if params is None else {**defaults, **params}

    def simulate(self, n_steps: int = 100, controller: TelomeraseController = None, perturb_schedule: dict = None, seed: int = None):
        """Simulate a single-cell digital twin interacting with the TelomeraseController.

        Returns a dict of time-series metrics used for fitness evaluation.
        """
        rng = np.random.RandomState(seed)
        # Build a lightweight VirtualCell using these params
        vc = VirtualCell(init_telomere=10.0, min_telomere=self.params['min_telomere'],
                         max_telomere=self.params['max_telomere'], telomere_trim_rate=self.params['telomere_trim_rate'],
                         division_cost=self.params['division_cost'])
        # Minimal environment
        nutrient = 1.0
        gf = 0.1
        history = {'tel': [], 'damage': [], 'pha': [], 'pulse': [], 'prolif': []}

        if perturb_schedule is None:
            perturb_schedule = {}
            # small chance of a transient stress mid-run
            if rng.rand() < 0.5:
                perturb_schedule[int(n_steps * 0.4)] = {'stress': 0.6}

        for t in range(n_steps):
            # build observation for controller
            obs = {'telomere': np.array([vc.state.telomere_length]),
                   'damage': np.array([vc.state.damage]),
                   'phagocytosis': np.array([vc.state.phagocytosis_activity]),
                   'growth_factor': np.array([gf]),
                   'nutrient': np.array([nutrient]),
                   'proliferation': np.array([vc.state.proliferation_state])}

            pulse = 0.0
            if controller is not None:
                p = controller.compute_pulses({'telomere': obs['telomere'], 'damage': obs['damage'], 'phagocytosis': obs['phagocytosis'], 'growth_factor': float(gf), 'nutrient': float(nutrient), 'proliferation': obs['proliferation']})
                pulse = float(p[0])

            # apply a simple local stress perturbation if scheduled
            perturb = None
            if t in perturb_schedule:
                perturb = {'injury_indices': [0], 'pathogen': perturb_schedule[t].get('pathogen', 0.0)}
                # map to VirtualCell.step by converting to env_inputs
                env_inputs = {'growth_factor': gf, 'stress': perturb_schedule[t].get('stress', 0.0), 'nutrient': nutrient}
                snapshot = vc.step(env_inputs, telomerase_pulse=pulse, dt=1.0)
            else:
                obs_env = {'growth_factor': gf, 'stress': 0.0, 'nutrient': nutrient}
                snapshot = vc.step(obs_env, telomerase_pulse=pulse, dt=1.0)

            history['tel'].append(snapshot.telomere_length)
            history['damage'].append(snapshot.damage)
            history['pha'].append(snapshot.phagocytosis_activity)
            history['pulse'].append(pulse)
            history['prolif'].append(snapshot.proliferation_state)

        return history


def fitness_function(history: dict):
    """Compute a scalar fitness from a history dict.

    Objectives (example composition):
    - Keep damage low (lower is better)
    - Keep telomere within a safe band (penalize extremes)
    - Minimize cumulative pulse usage (efficiency)
    - Allow transient proliferation when needed (reward controlled response)
    """
    tel = np.array(history['tel'])
    damage = np.array(history['damage'])
    pulses = np.array(history['pulse'])
    prolif = np.array(history['prolif'])

    # damage score (lower better)
    damage_score = 1.0 - np.mean(damage)

    # telomere stability score: penalize being below min or above max by using variance around midrange
    mid = 8.0
    tel_variance_penalty = 1.0 / (1.0 + np.var(tel - mid))

    # efficiency: prefer lower total pulse while still repairing
    pulse_cost = 1.0 / (1.0 + np.sum(pulses))

    # controlled proliferation: reward limited nonzero proliferation but penalize runaway
    prolif_mean = np.mean(prolif)
    prolif_score = 1.0 - min(1.0, max(0.0, (prolif_mean - 0.2)))  # best near 0.2

    # combine with weights
    fitness = 0.4 * damage_score + 0.25 * tel_variance_penalty + 0.2 * pulse_cost + 0.15 * prolif_score
    return float(fitness)


class EvolutionaryOptimizer:
    """Simple evolutionary strategy to evolve DigitalTwin parameters.

    - Population of parameter dictionaries
    - Tournament selection, Gaussian mutation, occasional crossover
    - Fitness evaluated by simulating the digital twin with a fixed controller
    """

    def __init__(self, pop_size: int = 32, n_generations: int = 40, device: str = 'cpu'):
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.device = device
        self.population = []  # list of (params, fitness)

    def _random_individual(self):
        return {
            'telomere_trim_rate': float(np.clip(np.random.normal(0.01, 0.005), 0.001, 0.05)),
            'division_cost': float(np.clip(np.random.normal(0.5, 0.2), 0.1, 1.5)),
            'phagocytosis_efficiency': float(np.clip(np.random.normal(0.5, 0.2), 0.05, 1.0)),
            'min_telomere': float(np.clip(np.random.normal(2.0, 0.5), 0.5, 5.0)),
            'max_telomere': float(np.clip(np.random.normal(50.0, 5.0), 20.0, 100.0))
        }

    def _mutate(self, params: dict, scale: float = 0.1):
        new = params.copy()
        for k in new.keys():
            if np.random.rand() < 0.5:
                new[k] = float(np.clip(new[k] * (1.0 + np.random.normal(0.0, scale)), 0.0001, 1000.0))
        return new

    def _crossover(self, a: dict, b: dict):
        child = {}
        for k in a.keys():
            child[k] = a[k] if np.random.rand() < 0.5 else b[k]
        return child

    def initialize(self):
        self.population = []
        for _ in range(self.pop_size):
            ind = self._random_individual()
            self.population.append((ind, None))

    def evaluate_population(self, controller: TelomeraseController, n_steps: int = 120):
        evaluated = []
        for params, _ in self.population:
            twin = DigitalTwin(params)
            history = twin.simulate(n_steps=n_steps, controller=controller)
            fit = fitness_function(history)
            evaluated.append((params, fit))
        self.population = evaluated

    def run(self, controller: TelomeraseController, log_every: int = 5):
        self.initialize()
        # evaluate initial
        self.evaluate_population(controller)
        best_history = []
        for gen in range(self.n_generations):
            # sort by fitness desc
            self.population.sort(key=lambda x: x[1], reverse=True)
            best_params, best_fit = self.population[0]
            best_history.append((gen, best_fit, best_params))
            if gen % log_every == 0:
                print(f"Gen {gen}: best_fit={best_fit:.4f}")

            # selection + generate new population
            new_pop = self.population[: max(2, self.pop_size // 8)]  # keep top fraction
            while len(new_pop) < self.pop_size:
                # tournament selection
                a = self.population[np.random.randint(0, len(self.population))][0]
                b = self.population[np.random.randint(0, len(self.population))][0]
                child = self._crossover(a, b) if np.random.rand() < 0.5 else a.copy()
                child = self._mutate(child, scale=0.12)
                new_pop.append((child, None))
            self.population = new_pop
            # evaluate
            self.evaluate_population(controller)

        # final sort
        self.population.sort(key=lambda x: x[1], reverse=True)
        return best_history, self.population[0]


# === Quick Evolve Demo ===
def demo_evolution():
    controller = TelomeraseController(n_cells=1, mode='rule', cooldown_steps=3, cumulative_cap=6.0)
    evo = EvolutionaryOptimizer(pop_size=24, n_generations=20)
    best_history, best_individual = evo.run(controller, log_every=2)
    print('''Evolution completed. Best individual:''')
    print(best_individual)

    # visualize best twin behavior
    best_params, best_fit = best_individual
    twin = DigitalTwin(best_params)
    hist = twin.simulate(n_steps=200, controller=controller, seed=123)

    import matplotlib.pyplot as plt
    t = np.arange(len(hist['tel']))
    plt.figure(figsize=(8,4))
    plt.plot(t, hist['tel'], label='telomere')
    plt.plot(t, hist['damage'], label='damage')
    plt.plot(t, hist['pulse'], label='pulse')
    plt.legend()
    plt.title(f"Best twin simulation fit={best_fit:.4f}")
    plt.xlabel('Timestep')
    plt.show()

# If run as script, run demo_evolution
if __name__ == '__main__':
    print('Running evolution demo...')
    demo_evolution()


# === SYNTHETIC TELOMERE DATASET (OPTION C) ===
# Purpose: generate safe, non-actionable synthetic telomere length datasets
# that match user-specified population statistics or estimated distributions.
# These datasets are designed for training/validation of the digital twin and
# controller models. They do NOT contain any wet-lab protocols or instructions.

import scipy.stats as stats

class SyntheticTelomereDataset:
    """Generator for synthetic telomere data.

    Features:
    - Sample population-level telomere length distributions (log-normal or gamma)
    - Simulate longitudinal attrition curves with stochasticity
    - Include covariates: age, cell_type_effect, stress_events, telomerase_activity_proxy
    - Provide utilities to fit distribution parameters from summary stats (mean, sd, quantiles)
    """

    def __init__(self, rng_seed: int = None):
        self.rng = np.random.RandomState(rng_seed)

    def sample_population(self, n: int = 1000, distribution: str = 'lognormal',
                          mean: float = None, sd: float = None, quantiles: dict = None,
                          cell_type_offsets: dict = None):
        """Sample a cross-sectional population of telomere lengths.

        - distribution: 'lognormal' or 'gamma'
        - specify either mean & sd (on natural scale) OR quantiles (dict: {q: value})
        - cell_type_offsets: dict mapping cell_type -> additive shift (can be negative)
        Returns a dict with 'lengths' and 'metadata' arrays.
        """
        if cell_type_offsets is None:
            cell_type_offsets = {'stem': 1.5, 'somatic': 0.0, 'immune': -0.5}
        cell_types = list(cell_type_offsets.keys())
        types = self.rng.choice(cell_types, size=n, p=[0.2, 0.6, 0.2])

        # Infer distribution parameters
        if quantiles is not None:
            # fit lognormal by matching median and an upper quantile if provided
            if distribution == 'lognormal':
                # use median (q=0.5) and q=0.9 to estimate mu and sigma
                q_med = quantiles.get(0.5)
                q_hi = quantiles.get(0.9)
                if q_med is None or q_hi is None:
                    raise ValueError("Provide quantiles 0.5 and 0.9 for lognormal fitting")
                # Solve: q = exp(mu + sigma * z_q)
                z_med = 0.0
                z_hi = stats.norm.ppf(0.9)
                mu = np.log(q_med)
                sigma = (np.log(q_hi) - np.log(q_med)) / (z_hi - z_med + 1e-12)
            else:
                raise NotImplementedError("Quantile fitting only implemented for lognormal")
        elif mean is not None and sd is not None:
            if distribution == 'lognormal':
                # convert mean, sd (natural scale) to lognormal mu, sigma
                variance = sd ** 2
                mu = np.log((mean ** 2) / np.sqrt(variance + mean ** 2))
                sigma = np.sqrt(np.log(1.0 + variance / (mean ** 2)))
            elif distribution == 'gamma':
                # method of moments for gamma: shape=k, scale=theta
                theta = variance = sd ** 2
                k = (mean ** 2) / variance
                theta = variance / mean
                # we'll sample from gamma(k, theta)
                mu = k
                sigma = theta
            else:
                raise ValueError("Unsupported distribution")
        else:
            # fallback: reasonable defaults for telomere length (arbitrary units)
            mu = np.log(8.0)
            sigma = 0.5

        lengths = np.zeros(n, dtype=float)
        for i in range(n):
            ttype = types[i]
            shift = cell_type_offsets.get(ttype, 0.0)
            if distribution == 'lognormal':
                val = self.rng.lognormal(mean=mu, sigma=sigma)
            else:
                val = self.rng.gamma(shape=mu, scale=sigma)
            lengths[i] = max(0.1, val + shift)

        metadata = {'cell_type': types}
        return {'lengths': lengths, 'metadata': metadata}

    def simulate_longitudinal(self, initial_lengths: np.ndarray, n_steps: int = 50,
                              base_attrition_rate: float = 0.02,
                              telomerase_noise_scale: float = 0.05,
                              stress_event_prob: float = 0.02,
                              stress_multiplier: float = 3.0):
        """Simulate telomere time-series for each initial length.

        Model (discrete steps):
          length_{t+1} = length_t - attrition + telomerase_recovery - trimming_noise
        where attrition = base_attrition_rate * f(age/celltype) :"""
        n_cells = initial_lengths.shape[0]
        ts = np.zeros((n_steps, n_cells), dtype=float)
        damage = np.zeros((n_steps, n_cells), dtype=float)
        tel = initial_lengths.copy().astype(float)

        # simple per-cell modifiers
        age_mod = 1.0 + 0.01 * (self.rng.randint(0, 50, size=n_cells))  # crude age effect
        for t in range(n_steps):
            # stress events
            stress_flags = (self.rng.rand(n_cells) < stress_event_prob).astype(float)
            attrition = base_attrition_rate * age_mod * (1.0 + stress_flags * stress_multiplier)
            telomerase_recovery = self.rng.normal(loc=0.0, scale=telomerase_noise_scale, size=n_cells)
            # trimming to avoid runaway long telomeres
            trim = 0.005 * np.maximum(0.0, tel - 50.0)
            # update
            tel = tel - attrition + telomerase_recovery - trim
            tel = np.maximum(0.05, tel)
            # damage increases if tel drops below threshold
            dmg = np.clip((1.0 - tel / 5.0), 0.0, 1.0)

            ts[t, :] = tel
            damage[t, :] = dmg

        return {'telomere_ts': ts, 'damage_ts': damage}

    def make_training_dataset(self, n_cells: int = 500, n_steps: int = 80, distribution: str = 'lognormal'):
        pop = self.sample_population(n=n_cells, distribution=distribution)
        initial = pop['lengths']
        long = self.simulate_longitudinal(initial, n_steps=n_steps)
        # build a dataset dict ready for ML: per-cell time-series rows
        X = long['telomere_ts'].T  # shape (n_cells, n_steps)
        Y = long['damage_ts'].T
        meta = pop['metadata']
        return {'X': X, 'Y': Y, 'meta': meta}


# Example usage:
if __name__ == '__main__':
    gen = SyntheticTelomereDataset(rng_seed=123)
    ds = gen.make_training_dataset(n_cells=200, n_steps=120)
    print('Dataset shapes:', ds['X'].shape, ds['Y'].shape)

    # quick plot of a few cell trajectories
    import matplotlib.pyplot as plt
    for i in range(6):
        plt.plot(ds['X'][i, :], label=f'cell{i}')
    plt.xlabel('Timestep')
    plt.ylabel('Synthetic telomere length (arb units)')
    plt.title('Sample synthetic telomere trajectories')
    plt.legend()
    plt.show()


# === COMPUTATIONAL DRUG DISCOVERY MODULE (SAFE, NON-ACTIONABLE) ===
# Purpose: provide an in-silico, computational-only pipeline for molecule
# generation, virtual screening, surrogate binding prediction, and ADMET-style
# prioritization. This is strictly computational and does not provide any
# wet-lab or synthesis instructions.

# Dependencies used in this module (python packages):
# - rdkit
# - numpy
# - scikit-learn
# - torch (optional for generative model)

from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors
from rdkit.Chem.QED import qed
from rdkit.Chem import rdMolDescriptors
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

class MoleculeGeneratorSketch:
    """A lightweight SMILES-based generator sketch.

    NOTE: This is a placeholder for a proper generative model (VAE, RNN, or
    diffusion for molecules). For safety and simplicity, this module perturbs
    seed SMILES via RDKit transformations to produce candidates.
    """
    def __init__(self, seed_smiles: list = None, rng_seed: int = 42):
        if seed_smiles is None:
            seed_smiles = ['CCO', 'c1ccccc1', 'CC(=O)O']
        self.seed_smiles = seed_smiles
        self.rng = np.random.RandomState(rng_seed)

    def mutate_smiles(self, smi: str) -> str:
        mol = Chem.MolFromSmiles(smi)
        if mol is None: return smi
        # add a random atom (carbon) to a random heavy atom
        mol = Chem.RWMol(mol)
        n = mol.GetNumAtoms()
        idx = self.rng.randint(0, n)
        new_idx = mol.AddAtom(Chem.Atom('C'))
        mol.AddBond(int(idx), int(new_idx), Chem.BondType.SINGLE)
        Chem.SanitizeMol(mol, catchErrors=True)
        return Chem.MolToSmiles(mol)

    def sample(self, n: int = 100) -> list:
        out = []
        for i in range(n):
            base = self.rng.choice(self.seed_smiles)
            s = self.mutate_smiles(base)
            out.append(s)
        return out

# Feature utilities
def mol_to_fp(mol: Chem.Mol, n_bits: int = 2048):
    return np.array(AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=n_bits), dtype=np.int8)

def compute_basic_props(mol: Chem.Mol):
    return {
        'mol_weight': Descriptors.MolWt(mol),
        'logp': Descriptors.MolLogP(mol),
        'num_h_donors': rdMolDescriptors.CalcNumHBD(mol),
        'num_h_acceptors': rdMolDescriptors.CalcNumHBA(mol),
        'qed': qed(mol)
    }

# Surrogate binding scorer (trainable)
class BindingSurrogate:
    """Train a RandomForest surrogate to predict 'binding score' from fingerprints.

    For safety: training labels should come from non-proprietary public datasets
    or simulated docking scores. In this demo we will create synthetic labels
    using a simple heuristic function.
    """
    def __init__(self):
        self.model = RandomForestRegressor(n_estimators=200, random_state=0)
        self.trained = False

    def synthetic_label(self, mol: Chem.Mol) -> float:
        # a toy function: favor moderate MW, moderate logP, and presence of aromatic rings
        props = compute_basic_props(mol)
        aromatic = sum(1 for a in mol.GetAtoms() if a.GetIsAromatic())
        score = (1.0 / (1.0 + abs(props['mol_weight'] - 300) / 100.0))
        score += (1.0 / (1.0 + abs(props['logp'] - 2.0)))
        score += 0.1 * aromatic
        return float(score)

    def featurize_and_label(self, smiles_list: list):
        X = []
        y = []
        valid_smiles = []
        for s in smiles_list:
            mol = Chem.MolFromSmiles(s)
            if mol is None: continue
            fp = mol_to_fp(mol)
            X.append(fp)
            y.append(self.synthetic_label(mol))
            valid_smiles.append(s)
        X = np.vstack(X)
        y = np.array(y, dtype=float)
        return X, y, valid_smiles

    def fit(self, smiles_list: list):
        X, y, val = self.featurize_and_label(smiles_list)
        if len(y) < 10:
            raise ValueError('Need at least 10 valid molecules to fit surrogate')
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
        self.model.fit(X_train, y_train)
        self.trained = True
        print('Surrogate fitted — test R2:', self.model.score(X_test, y_test))

    def predict(self, smiles_list: list):
        X, _, val = self.featurize_and_label(smiles_list)
        if X.size == 0:
            return []
        preds = self.model.predict(X)
        return list(zip(val, preds))

# ADMET-style lightweight filters (non-actionable predictions)
def admet_filter(smiles_list: list):
    filtered = []
    for s in smiles_list:
        mol = Chem.MolFromSmiles(s)
        if mol is None: continue
        props = compute_basic_props(mol)
        # simple Lipinski-like filter
        if props['mol_weight'] > 500: continue
        if props['num_h_donors'] > 5: continue
        if props['num_h_acceptors'] > 10: continue
        filtered.append((s, props))
    return filtered

# Synthetic accessibility estimate using RDKit SA score recipe (lightweight approximation)
def sa_score(mol: Chem.Mol):
    try:
        from rdkit.Chem import QED as q
        # RDKit doesn't ship SA score by default; use a coarse proxy: 1 / qed
        return max(0.0, 10.0 * (1.0 - q.qed(mol)))
    except Exception:
        return 5.0

# End-to-end virtual screening function
def virtual_screen(seed_smiles: list = None, n_gen: int = 200, top_k: int = 20):
    gen = MoleculeGeneratorSketch(seed_smiles=seed_smiles)
    pool = gen.sample(n=n_gen)
    # basic filtering
    filtered = admet_filter(pool)
    smiles_filtered = [s for s, p in filtered]

    # fit surrogate on filtered pool (synthetic labels for safety)
    surrogate = BindingSurrogate()
    # ensure enough molecules: if not, augment via mutation
    if len(smiles_filtered) < 20:
        extra = gen.sample(n=100)
        smiles_filtered += extra
    surrogate.fit(smiles_filtered)

    preds = surrogate.predict(smiles_filtered)
    # rank by predicted binding score
    preds_sorted = sorted(preds, key=lambda x: x[1], reverse=True)
    top = preds_sorted[:top_k]

    results = []
    for smi, score in top:
        mol = Chem.MolFromSmiles(smi)
        props = compute_basic_props(mol)
        results.append({'smiles': smi, 'score': float(score), 'props': props, 'sa': sa_score(mol)})
    return results

# Example usage:
if __name__ == '__main__':
    print('Running safe virtual screening demo...')
    res = virtual_screen(seed_smiles=['CCO','c1ccccc1','CC(=O)O'], n_gen=300, top_k=10)
    for r in res:
        print(r)

# Safety note:
# This module performs computational chemistry tasks (molecule generation,
# fingerprinting, surrogate modeling) for in-silico exploration only. It
# intentionally avoids any synthetic route suggestions, assay protocols, or
# any guidance on generating or physically obtaining chemical compounds.
# Use only for computational prioritization and hypothesis generation.



# === QUANTUM-INSPIRED LATENT SPACE (COMPLEX / QUBIT-ANALOG) + VISUALS ===
# This module extends the latent state to complex-valued representations
# (amplitude + phase), inspired by qubit state vectors. It remains classical
# simulation (no quantum hardware required).

import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# -----------------------------
# Quantum-Inspired Latent Types
# -----------------------------

class QuantumLatentEncoder(nn.Module):
    """Encodes biological state into complex latent vectors (a + i b).
    Each pair (a_k, b_k) can be interpreted as amplitude/phase components.
    """
    def __init__(self, latent_dim=16):
        super().__init__()
        self.latent_dim = latent_dim
        self.real_net = nn.Sequential(
            nn.Linear(6, 64), nn.GELU(), nn.Linear(64, latent_dim)
        )
        self.imag_net = nn.Sequential(
            nn.Linear(6, 64), nn.GELU(), nn.Linear(64, latent_dim)
        )

    def forward(self, tel, dmg, pha, prolif, gf, nutrient):
        x = torch.stack([tel, dmg, pha, prolif, gf, nutrient], dim=-1)
        a = self.real_net(x)
        b = self.imag_net(x)
        z = torch.complex(a, b)
        # normalize like a quantum state
        norm = torch.clamp(torch.linalg.norm(z, dim=-1, keepdim=True), min=1e-6)
        return z / norm

class QuantumWorldModel(nn.Module):
    """Complex-valued transition model: z_{t+1} = U(z_t, action).
    Implemented via real-valued nets applied to real/imag parts.
    """
    def __init__(self, latent_dim=16):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2*latent_dim + 1, 128), nn.GELU(), nn.Linear(128, 2*latent_dim)
        )

    def forward(self, z, pulse):
        zr = torch.real(z)
        zi = torch.imag(z)
        x = torch.cat([zr, zi, pulse.unsqueeze(-1)], dim=-1)
        out = self.net(x)
        a, b = out.chunk(2, dim=-1)
        z_next = torch.complex(a, b)
        norm = torch.clamp(torch.linalg.norm(z_next, dim=-1, keepdim=True), min=1e-6)
        return z_next / norm

class QuantumPolicy(nn.Module):
    """Maps complex latent to a bounded telomerase intent (0..1).
    Uses magnitude and phase features.
    """
    def __init__(self, latent_dim=16):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2*latent_dim, 64), nn.GELU(), nn.Linear(64, 1), nn.Sigmoid()
        )

    def forward(self, z):
        mag = torch.abs(z)
        phase = torch.angle(z) / np.pi  # normalized phase
        x = torch.cat([mag, phase], dim=-1)
        return self.net(x)

# -----------------------------
# Visualization Utilities
# -----------------------------

def plot_latent_trajectory(z_traj, title='Quantum Latent Trajectory'):
    """Plots magnitude and phase evolution for first 2 dimensions."""
    z = torch.stack(z_traj).detach().cpu()
    mag = torch.abs(z)
    phase = torch.angle(z)

    t = np.arange(len(z_traj))
    plt.figure()
    plt.plot(t, mag[:, 0], label='|z0|')
    plt.plot(t, mag[:, 1], label='|z1|')
    plt.xlabel('time')
    plt.ylabel('magnitude')
    plt.title(title + ' (Magnitude)')
    plt.legend()
    plt.show()

    plt.figure()
    plt.plot(t, phase[:, 0], label='∠z0')
    plt.plot(t, phase[:, 1], label='∠z1')
    plt.xlabel('time')
    plt.ylabel('phase (rad)')
    plt.title(title + ' (Phase)')
    plt.legend()
    plt.show()


def plot_policy_vs_need(needs, intents):
    """Scatter plot of biological need vs policy intent."""
    plt.figure()
    plt.scatter(needs, intents, alpha=0.6)
    plt.xlabel('biological need (proxy)')
    plt.ylabel('policy intent (0..1)')
    plt.title('Policy Intent vs Biological Need')
    plt.show()

# -----------------------------
# Demo: Quantum-Inspired Rollout
# -----------------------------

def demo_quantum_rollout(steps=50):
    enc = QuantumLatentEncoder(latent_dim=8)
    wm = QuantumWorldModel(latent_dim=8)
    pol = QuantumPolicy(latent_dim=8)

    # fake initial biological signals
    tel = torch.rand(1)
    dmg = torch.rand(1)
    pha = torch.rand(1)
    prolif = torch.rand(1)
    gf = torch.rand(1)
    nutrient = torch.rand(1)

    z = enc(tel, dmg, pha, prolif, gf, nutrient)
    z_traj = [z.squeeze(0)]
    intents = []
    needs = []

    for _ in range(steps):
        intent = pol(z).squeeze(0)
        # need proxy: low telomeric reserve & higher damage
        need = torch.clamp(1.0 - tel + dmg, 0.0, 1.0)
        pulse = intent * need  # safety-gated intent (pre hard-gate)

        z = wm(z, pulse)
        z_traj.append(z.squeeze(0))
        intents.append(intent.item())
        needs.append(need.item())

        # evolve fake signals
        tel = torch.clamp(tel + 0.05*pulse - 0.01, 0, 1)
        dmg = torch.clamp(dmg - 0.03*pulse + 0.01, 0, 1)

    plot_latent_trajectory(z_traj, title='Quantum-Inspired Cell Latent')
    plot_policy_vs_need(needs, intents)

# Auto-run demo
if __name__ == '__main__':
    demo_quantum_rollout(steps=60)


# === QUANTUM-INSPIRED EVOLUTIONARY SEARCH (Q-ES) ===
# Evolves policies/world-model parameters using complex-latent fitness rollouts.
# Classical computation with quantum-inspired operators (phase mutation,
# amplitude damping, coherence regularization).

import copy

class QuantumGenome:
    """Wraps a policy (and optionally world model) for evolution."""
    def __init__(self, policy: nn.Module, world_model: nn.Module = None):
        self.policy = copy.deepcopy(policy)
        self.world_model = copy.deepcopy(world_model) if world_model is not None else None

    def parameters(self):
        for p in self.policy.parameters():
            yield p
        if self.world_model is not None:
            for p in self.world_model.parameters():
                yield p


def phase_mutation_(param, sigma=0.05):
    """Quantum-inspired phase mutation: rotate weights in complex plane proxy.
    Implemented as small orthogonal rotations in real-valued parameter space.
    """
    with torch.no_grad():
        noise = torch.randn_like(param) * sigma
        param.add_(noise)


def amplitude_damping_(param, gamma=1e-3):
    """Amplitude damping to prevent runaway weights (coherence loss control)."""
    with torch.no_grad():
        param.mul_(1.0 - gamma)


class QuantumEvolutionarySearch:
    def __init__(
        self,
        encoder: QuantumLatentEncoder,
        world_model: QuantumWorldModel,
        base_policy: QuantumPolicy,
        population_size=20,
        elite_frac=0.2,
        phase_sigma=0.05,
        damping_gamma=1e-3,
        rollout_steps=40,
        device='cpu'
    ):
        self.encoder = encoder.to(device)
        self.world_model = world_model.to(device)
        self.base_policy = base_policy.to(device)
        self.population_size = population_size
        self.elite_frac = elite_frac
        self.phase_sigma = phase_sigma
        self.damping_gamma = damping_gamma
        self.rollout_steps = rollout_steps
        self.device = device

        self.population = [QuantumGenome(self.base_policy, self.world_model)
                           for _ in range(population_size)]

    def _fitness(self, genome: QuantumGenome) -> float:
        """Multi-objective fitness:
        + tissue repair (tel recovery)
        - damage
        - telomerase usage (energy)
        + coherence stability (latent smoothness)
        """
        # initialize fake biological signals
        tel = torch.rand(1, device=self.device)
        dmg = torch.rand(1, device=self.device)
        pha = torch.rand(1, device=self.device)
        prolif = torch.rand(1, device=self.device)
        gf = torch.rand(1, device=self.device)
        nutrient = torch.rand(1, device=self.device)

        z = self.encoder(tel, dmg, pha, prolif, gf, nutrient)
        z_prev = z.clone()

        tel_usage = 0.0
        smoothness = 0.0

        for _ in range(self.rollout_steps):
            intent = genome.policy(z).squeeze(0)
            need = torch.clamp(1.0 - tel + dmg, 0.0, 1.0)
            pulse = intent * need
            tel_usage += pulse.item()

            z = genome.world_model(z, pulse) if genome.world_model is not None else z
            smoothness += torch.linalg.norm(z - z_prev).item()
            z_prev = z

            # evolve signals
            tel = torch.clamp(tel + 0.05*pulse - 0.01, 0, 1)
            dmg = torch.clamp(dmg - 0.03*pulse + 0.01, 0, 1)

        fitness = (
            + tel.item() * 2.0
            - dmg.item() * 2.0
            - 0.5 * tel_usage
            - 0.1 * smoothness
        )
        return float(fitness)

    def _mutate(self, genome: QuantumGenome):
        for p in genome.parameters():
            phase_mutation_(p, self.phase_sigma)
            amplitude_damping_(p, self.damping_gamma)

    def step(self):
        scored = [(self._fitness(g), g) for g in self.population]
        scored.sort(key=lambda x: x[0], reverse=True)

        elite_n = max(1, int(self.elite_frac * self.population_size))
        elites = [copy.deepcopy(g) for _, g in scored[:elite_n]]

        # refill population
        new_pop = elites.copy()
        while len(new_pop) < self.population_size:
            parent = copy.deepcopy(np.random.choice(elites))
            self._mutate(parent)
            new_pop.append(parent)

        self.population = new_pop
        return scored[0][0]


# -----------------------------
# Demo: Quantum Evolution
# -----------------------------

def demo_quantum_evolution(generations=15):
    enc = QuantumLatentEncoder(latent_dim=8)
    wm = QuantumWorldModel(latent_dim=8)
    pol = QuantumPolicy(latent_dim=8)

    qes = QuantumEvolutionarySearch(
        encoder=enc,
        world_model=wm,
        base_policy=pol,
        population_size=24,
        elite_frac=0.25,
        rollout_steps=50
    )

    best_scores = []
    for g in range(generations):
        best = qes.step()
        best_scores.append(best)
        print(f'Gen {g:02d} | best fitness = {best:.3f}')

    plt.figure()
    plt.plot(best_scores)
    plt.xlabel('generation')
    plt.ylabel('best fitness')
    plt.title('Quantum-Inspired Evolution Progress')
    plt.show()

# Auto-run evolution demo
if __name__ == '__main__':
    demo_quantum_evolution(generations=20)



# === ENTANGLED MULTI-CELL EVOLUTION (QUANTUM-INSPIRED) ===
# Cells are coupled via shared latent subspaces (entanglement channels).
# Evolution rewards coordinated, non-pathological tissue-level behavior.

class EntangledCellSystem:
    """Simulates multiple cells with partial latent entanglement."""
    def __init__(self, encoder, world_model, policy, n_cells=4, entangle_strength=0.2, device='cpu'):
        self.encoder = encoder.to(device)
        self.world_model = world_model.to(device)
        self.policy = policy.to(device)
        self.n_cells = n_cells
        self.entangle_strength = entangle_strength
        self.device = device

        # initialize biological states per cell
        self.tel = torch.rand(n_cells, device=device)
        self.dmg = torch.rand(n_cells, device=device)
        self.pha = torch.rand(n_cells, device=device)
        self.prolif = torch.rand(n_cells, device=device)
        self.gf = torch.rand(n_cells, device=device)
        self.nutrient = torch.rand(n_cells, device=device)

        self.z = self.encoder(self.tel, self.dmg, self.pha, self.prolif, self.gf, self.nutrient)

    def entangle(self, z):
        """Simple entanglement: mix global mean latent into each cell."""
        global_z = torch.mean(z, dim=0, keepdim=True)
        return (1 - self.entangle_strength) * z + self.entangle_strength * global_z

    def step(self):
        intents = self.policy(self.z).squeeze(-1)
        needs = torch.clamp(1.0 - self.tel + self.dmg, 0.0, 1.0)
        pulses = intents * needs

        # world-model evolution
        self.z = self.world_model(self.z, pulses)
        # entanglement coupling
        self.z = self.entangle(self.z)

        # update biological proxies
        self.tel = torch.clamp(self.tel + 0.05*pulses - 0.01, 0, 1)
        self.dmg = torch.clamp(self.dmg - 0.03*pulses + 0.01, 0, 1)

        return pulses, needs


class EntangledQuantumEvolution:
    """Evolutionary search where fitness is computed over entangled cell populations."""
    def __init__(self, encoder, world_model, base_policy, n_cells=4, population_size=16, generations=20):
        self.encoder = encoder
        self.world_model = world_model
        self.base_policy = base_policy
        self.n_cells = n_cells
        self.population_size = population_size
        self.generations = generations

        self.population = [QuantumGenome(base_policy, world_model) for _ in range(population_size)]

    def fitness(self, genome):
        system = EntangledCellSystem(
            self.encoder,
            genome.world_model,
            genome.policy,
            n_cells=self.n_cells
        )

        tel_history = []
        dmg_history = []
        sync_penalty = 0.0

        for _ in range(40):
            pulses, needs = system.step()
            tel_history.append(system.tel.mean().item())
            dmg_history.append(system.dmg.mean().item())
            sync_penalty += torch.var(pulses).item()  # penalize desynchronization

        fitness = (
            + np.mean(tel_history)
            - np.mean(dmg_history)
            - 0.5 * sync_penalty
        )
        return fitness

    def run(self):
        best_scores = []
        for g in range(self.generations):
            scored = [(self.fitness(genome), genome) for genome in self.population]
            scored.sort(key=lambda x: x[0], reverse=True)

            best_scores.append(scored[0][0])
            print(f'Entangled Gen {g:02d} | best fitness = {scored[0][0]:.3f}')

            elites = [copy.deepcopy(gm) for _, gm in scored[:max(2, self.population_size//4)]]
            new_pop = elites.copy()

            while len(new_pop) < self.population_size:
                parent = copy.deepcopy(np.random.choice(elites))
                for p in parent.parameters():
                    phase_mutation_(p, sigma=0.04)
                    amplitude_damping_(p, gamma=1e-3)
                new_pop.append(parent)

            self.population = new_pop

        # plot evolution
        plt.figure()
        plt.plot(best_scores)
        plt.xlabel('generation')
        plt.ylabel('best fitness')
        plt.title('Entangled Multi-Cell Evolution')
        plt.show()


# -----------------------------
# Demo: Entangled Multi-Cell Evolution
# -----------------------------

if __name__ == '__main__':
    enc = QuantumLatentEncoder(latent_dim=8)
    wm = QuantumWorldModel(latent_dim=8)
    pol = QuantumPolicy(latent_dim=8)

    eqe = EntangledQuantumEvolution(
        encoder=enc,
        world_model=wm,
        base_policy=pol,
        n_cells=6,
        population_size=18,
        generations=15
    )

    eqe.run()
```

```python
class DigitalCell:
    def __init__(self):
        self.telomere = 1.0
        self.damage = 0.0
        self.divisions = 0
        self.cancer_risk = 0.0
```

```python
def simulate_step(cell, hypothesis):
    # erosion
    cell.telomere -= 0.01

    # damage accumulation
    cell.damage += 0.02

    # hypothesis-controlled repair
    if hypothesis.rules["telomerase_activation"] == "only_if(injury AND stem_cell)":
        if cell.damage > 0.3:
            cell.telomere += 0.05
            cell.damage -= 0.1

    # safety penalties
    if cell.telomere > hypothesis.rules["upper_telomere_bound"]:
        cell.cancer_risk += 0.2
```

```python
def safety_constraints(cell):
    if cell.cancer_risk > 0.3:
        return False
    if cell.telomere > 1.2:
        return False
    return True
```

```python
def evaluate_hypothesis(hypothesis, steps=200):
    cell = DigitalCell()
    for _ in range(steps):
        simulate_step(cell, hypothesis)
        if not safety_constraints(cell):
            hypothesis.score = -1
            return hypothesis.score

    # multi-objective score
    hypothesis.score = (
        + cell.telomere
        - cell.cancer_risk * 2
        - cell.damage
    )
    return hypothesis.score
```
