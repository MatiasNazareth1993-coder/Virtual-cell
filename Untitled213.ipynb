```python
import tensorflow as tf
from tensorflow.keras import layers, models, activations, losses, optimizers
import numpy as np
import random

# -----------------------------
# Candidate Pools
# -----------------------------
ACTIVATIONS = [
    "elu", "exponential", "gelu", "hard_sigmoid", "hard_silu", "hard_swish",
    "leaky_relu", "linear", "log_softmax", "mish", "relu", "relu6",
    "selu", "sigmoid", "silu", "softmax", "softplus", "softsign", "swish", "tanh"
]

OPTIMIZERS = [
    "Adadelta", "Adafactor", "Adagrad", "Adam", "AdamW", "Adamax", "Ftrl",
    "Lion", "Nadam", "RMSprop", "SGD"
]

# -----------------------------
# Utility: instantiate dynamically
# -----------------------------
def get_activation(name):
    try:
        return activations.get(name)
    except:
        return tf.nn.relu  # fallback

def get_optimizer(name, lr=0.001):
    optim_class = getattr(optimizers, name, tf.keras.optimizers.Adam)
    if name in ["Lion", "AdamW", "Adafactor"]:  # optional kwargs
        return optim_class(learning_rate=lr)
    else:
        return optim_class(lr)

# -----------------------------
# Evolutionary Population
# -----------------------------
class Individual:
    def __init__(self):
        self.activation = random.choice(ACTIVATIONS)
        self.optimizer_name = random.choice(OPTIMIZERS)
        self.lr = 10 ** np.random.uniform(-4, -1)  # 0.0001 ~ 0.1
        self.fitness = None

    def mutate(self):
        if random.random() < 0.3:
            self.activation = random.choice(ACTIVATIONS)
        if random.random() < 0.3:
            self.optimizer_name = random.choice(OPTIMIZERS)
        if random.random() < 0.3:
            self.lr = 10 ** np.random.uniform(-4, -1)

    def crossover(self, other):
        child = Individual()
        child.activation = random.choice([self.activation, other.activation])
        child.optimizer_name = random.choice([self.optimizer_name, other.optimizer_name])
        child.lr = random.choice([self.lr, other.lr])
        return child

# -----------------------------
# Evaluation: quick synthetic task
# -----------------------------
def evaluate(individual):
    # Simple dataset (can be replaced with any scientific dataset)
    X = np.random.rand(1000, 10)
    y = np.sum(X, axis=1, keepdims=True)

    model = models.Sequential([
        layers.Input(shape=(10,)),
        layers.Dense(64, activation=get_activation(individual.activation)),
        layers.Dense(1)
    ])
    model.compile(optimizer=get_optimizer(individual.optimizer_name, lr=individual.lr),
                  loss=losses.MeanSquaredError())

    history = model.fit(X, y, epochs=3, batch_size=32, verbose=0)
    individual.fitness = 1.0 / (history.history['loss'][-1] + 1e-8)
    return individual.fitness

# -----------------------------
# Evolutionary Loop
# -----------------------------
def evolutionary_search(generations=5, population_size=10):
    population = [Individual() for _ in range(population_size)]

    for gen in range(generations):
        # Evaluate
        for ind in population:
            if ind.fitness is None:
                evaluate(ind)

        # Sort by fitness
        population.sort(key=lambda x: x.fitness, reverse=True)
        print(f"Gen {gen+1} Best: {population[0].activation}+{population[0].optimizer_name} lr={population[0].lr:.4f} fitness={population[0].fitness:.4f}")

        # Selection & reproduction
        next_gen = population[:population_size//2]  # top 50%
        while len(next_gen) < population_size:
            p1, p2 = random.sample(next_gen, 2)
            child = p1.crossover(p2)
            child.mutate()
            next_gen.append(child)
        population = next_gen

    return population[0]

# -----------------------------
# Run the Evolution
# -----------------------------
best_individual = evolutionary_search(generations=10, population_size=20)
print("\n✅ Best Activation+Optimizer Combo Found:")
print(best_individual.activation, best_individual.optimizer_name, best_individual.lr)
```

```python
# ===========================================================
# 🌌 CD-SciGen++: Cross-Domain Virtual Cell with 3D Telomere Visualization
# ===========================================================
!pip install plotly gradio matplotlib numpy tensorflow pillow --quiet

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
import gradio as gr
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw
import io
import plotly.graph_objects as go

# -----------------------------
# 1️⃣ Virtual Cell & Telomere T-loop
# -----------------------------
class VirtualCell3D:
    def __init__(self, n_cells=50, initial_telomere=15.0, cosmic_factor=1.0):
        self.n_cells = n_cells
        self.telomere = np.random.uniform(initial_telomere-2, initial_telomere+2, (n_cells,1))
        self.division_signal = np.random.uniform(0.5,1.0,(n_cells,1))
        self.phagocytosis = np.random.uniform(0.7,1.0,(n_cells,1))
        self.age = np.zeros((n_cells,1))
        self.cosmic_field = np.random.uniform(0,cosmic_factor,n_cells).reshape(-1,1)

    def update_telomere(self, telomerase_activity, cosmic_influence):
        self.telomere += 0.1*(telomerase_activity + 0.05*cosmic_influence)
        self.telomere = np.clip(self.telomere,5,50)

    def divide_cells(self):
        can_divide = (self.telomere > 10) & (self.division_signal > 0.6)
        self.age += 1
        self.phagocytosis = np.clip(self.phagocytosis + np.random.normal(0,0.01,self.phagocytosis.shape),0,1)
        return np.sum(can_divide)

# -----------------------------
# 2️⃣ Multi-Agent AI
# -----------------------------
def build_agent(input_dim=4, hidden=64):
    inputs = layers.Input(shape=(input_dim,))
    x = layers.Dense(hidden, activation='gelu')(inputs)
    x = layers.Dense(hidden, activation='swish')(x)
    x = layers.Dense(hidden//2, activation='mish')(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)
    model = models.Model(inputs, outputs)
    model.compile(optimizer=optimizers.Adam(0.01), loss='mse')
    return model

def agent_debate(agents, cell):
    X = np.hstack([cell.telomere, cell.division_signal, cell.phagocytosis, cell.cosmic_field])
    proposals = [agent.predict(X,verbose=0) for agent in agents]
    combined = np.mean(proposals,axis=0)
    return np.clip(combined + np.random.normal(0,0.05,combined.shape),0,1)

def train_agents(agents, cell, epochs=2):
    X = np.hstack([cell.telomere, cell.division_signal, cell.phagocytosis, cell.cosmic_field])
    y = np.clip(1 - (cell.telomere/50),0,1)
    for agent in agents:
        agent.fit(X,y,epochs=epochs,verbose=0)

# -----------------------------
# 3️⃣ Simulation Engine
# -----------------------------
def simulate(cycles=10, n_cells=50, n_agents=3, cosmic_factor=1.0):
    cell = VirtualCell3D(n_cells=n_cells, cosmic_factor=cosmic_factor)
    agents = [build_agent() for _ in range(n_agents)]
    div_history, telo_history = [], []

    for _ in range(cycles):
        activity = agent_debate(agents, cell)
        cell.update_telomere(activity, cell.cosmic_field)
        div_history.append(cell.divide_cells())
        telo_history.append(np.mean(cell.telomere))
        train_agents(agents,cell,epochs=2)

    return cell, agents, div_history, telo_history

# -----------------------------
# 4️⃣ Figures & 3D T-loop Visuals
# -----------------------------
def generate_figures(div_history, telo_history):
    # 2D Simulation Plot
    fig, ax1 = plt.subplots(figsize=(5,3))
    ax1.plot(div_history,label="Dividing Cells")
    ax2 = ax1.twinx()
    ax2.plot(telo_history,color='orange',label="Mean Telomere")
    fig.tight_layout()
    buf = io.BytesIO()
    plt.savefig(buf,format='png')
    plt.close(fig)
    buf.seek(0)
    sim_fig = Image.open(buf)

    # 3D T-loop Telomere Visualization
    theta = np.linspace(0,4*np.pi,100)
    z = np.linspace(0,10,100)
    r = 1 + 0.2*np.sin(5*theta)
    x = r*np.sin(theta)
    y = r*np.cos(theta)
    tloop_fig = go.Figure(data=[go.Scatter3d(x=x,y=y,z=z,mode='lines',line=dict(color='blue',width=4))])
    tloop_fig.update_layout(scene=dict(xaxis_title='X',yaxis_title='Y',zaxis_title='Telomere T-loop'),margin=dict(l=0,r=0,b=0,t=0))

    return sim_fig, tloop_fig

# -----------------------------
# 5️⃣ Auto Paper Generator
# -----------------------------
def generate_paper(divs,telos,n_agents,cosmic_factor,cycles):
    immortality = round(np.mean(telos)/50*100,2)
    return f"""
# Cross-Domain Multi-Agent Virtual Immortal Cell

## Abstract
Virtual cells achieve near-immortality using synthetic telomeres, AI telomerase regulation,
phagocytosis cleanup, multi-agent debate, and cross-domain cosmic feedback.

Predicted immortality score: {immortality}%.

## Methods
- {n_agents} AI agents propose telomerase strategies
- Debate + cross-domain fusion improves stability
- Cosmic factor: {cosmic_factor}, Simulation cycles: {cycles}
- Phagocytosis maintains clean intracellular environment

## Results
- Average telomere length: {np.mean(telos):.2f}
- Average cell divisions per cycle: {np.mean(divs):.2f}

## Discussion
Combining biology, physics, and cosmology enables AI agents to discover novel regulation strategies.
Potential applications: aging, cancer, tissue repair, theoretical biology, and cross-domain hypothesis generation.

## Ethics & Safety
Fully in silico; no biological or societal risk.
"""

# -----------------------------
# 6️⃣ Gradio UI
# -----------------------------
def run_cd_scigen3d(cycles,n_cells,n_agents,cosmic_factor):
    cell, agents, div_hist, telo_hist = simulate(cycles,n_cells,n_agents,cosmic_factor)
    sim_fig, tloop_fig = generate_figures(div_hist,telo_hist)
    paper_text = generate_paper(div_hist,telo_hist,n_agents,cosmic_factor,cycles)
    reviews = [f"Reviewer {i+1}: Cross-domain approach promising for autonomous discovery." for i in range(4)]
    decision = "ACCEPT (Virtual Science Journal)"
    return paper_text,reviews,decision,sim_fig,tloop_fig

with gr.Blocks() as app:
    gr.Markdown("# 🌌 CD-SciGen++: Virtual Immortal Cell & 3D Telomere")

    cycles_slider = gr.Slider(5,30,value=10,step=1,label="Simulation Cycles")
    n_cells_slider = gr.Slider(10,200,value=50,step=5,label="Number of Virtual Cells")
    n_agents_slider = gr.Slider(1,5,value=3,step=1,label="Number of AI Agents")
    cosmic_factor_slider = gr.Slider(0.0,2.0,value=1.0,step=0.05,label="Cosmic Feedback Factor")
    run_btn = gr.Button("Generate Science")

    paper = gr.Textbox(label="Paper Draft",lines=20)
    reviews = gr.JSON(label="Peer Reviews")
    decision = gr.Textbox(label="Final Decision")
    sim_fig = gr.Image(label="Simulation Plot")
    tloop_fig = gr.Plot(label="3D Telomere T-loop")

    run_btn.click(run_cd_scigen3d,
                  inputs=[cycles_slider,n_cells_slider,n_agents_slider,cosmic_factor_slider],
                  outputs=[paper,reviews,decision,sim_fig,tloop_fig])

app.launch()
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import threading
import ipywidgets as widgets
from IPython.display import display, clear_output

# -------------------------------
# 1. Core Classes
# -------------------------------
class VirtualTelomere:
    def __init__(self, initial_repeats=1000):
        self.length = initial_repeats
        self.integrity = 100
        self.division_count = 0
        self.history = {'length':[self.length], 'integrity':[self.integrity]}

    def degrade(self):
        # random degradation
        self.length -= random.randint(1,5)
        self.integrity -= random.uniform(0.1,1.0)
        self.length = max(self.length,0)
        self.integrity = max(self.integrity,0)
        self.division_count += 1
        self.history['length'].append(self.length)
        self.history['integrity'].append(self.integrity)

class Telomerase:
    def __init__(self, affinity=0.8, extension=50):
        self.affinity = affinity
        self.extension = extension

    def act(self, telomere):
        repair_amount = int(self.extension * self.affinity)
        telomere.length += repair_amount
        telomere.integrity += repair_amount*0.1
        telomere.integrity = min(telomere.integrity, 100)

class VirtualCell:
    def __init__(self, cell_id, initial_repeats=1000, telomerase_affinity=0.8,
                 telomerase_extension=50, L_min=1600, L_repair_threshold=4000,
                 I_min=30, prediction_window=5):
        self.cell_id = cell_id
        self.telomere = VirtualTelomere(initial_repeats)
        self.telomerase = Telomerase(telomerase_affinity, telomerase_extension)
        self.L_min = L_min
        self.L_repair_threshold = L_repair_threshold
        self.I_min = I_min
        self.pred_window = prediction_window
        self.active = True

    def step(self):
        if not self.active: return
        self.telomere.degrade()
        # Predictive AI intervention
        future_length = self.telomere.length - self.pred_window*3
        if future_length < self.L_min:
            self.telomerase.act(self.telomere)
        if self.telomere.integrity < self.I_min:
            self.active = False

class Drug:
    def __init__(self, name, type_, effect_strength=1.5):
        self.name = name
        self.type_ = type_
        self.effect_strength = effect_strength

    def apply(self, telomerase):
        if self.type_ == 'activator':
            telomerase.affinity = min(1.0, telomerase.affinity*self.effect_strength)
        elif self.type_ == 'inhibitor':
            telomerase.affinity = max(0.0, telomerase.affinity/self.effect_strength)

# -------------------------------
# 2. Advanced Analytics
# -------------------------------
def collect_metrics(cells):
    lengths = [c.telomere.length for c in cells]
    integrities = [c.telomere.integrity for c in cells]
    active_count = sum(c.active for c in cells)
    risks = [max(0,1-l/1600)+max(0,1-i/30) for l,i in zip(lengths,integrities)]
    return {'avg_length':np.mean(lengths),
            'avg_integrity':np.mean(integrities),
            'active_cells':active_count,
            'avg_risk':np.mean(risks),
            'lengths':lengths,'integrities':integrities,'risks':risks}

def plot_metrics(metrics):
    plt.figure(figsize=(16,6))
    plt.subplot(1,3,1)
    plt.plot(metrics['lengths'], label='Telomere Length')
    plt.title('Telomere Length Distribution')
    plt.subplot(1,3,2)
    plt.plot(metrics['integrities'], label='Integrity', color='orange')
    plt.title('Telomere Integrity')
    plt.subplot(1,3,3)
    sns.heatmap(np.array(metrics['risks']).reshape(-1,1), cmap='coolwarm')
    plt.title('Mutation Risk Heatmap')
    plt.show()

# -------------------------------
# 3. Multi-Cell Simulation
# -------------------------------
def run_multi_cell_simulation(num_cells=50, cycles=50, drugs=None, AI_thresholds=None):
    L_min = AI_thresholds['L_min'] if AI_thresholds else 1600
    L_repair = AI_thresholds['L_repair'] if AI_thresholds else 4000
    I_min = AI_thresholds['I_min'] if AI_thresholds else 30
    pred_window = AI_thresholds['pred_window'] if AI_thresholds else 5

    cells=[]
    for i in range(num_cells):
        repeats = random.randint(800,1200)
        affinity = random.uniform(0.7,0.9)
        extension = random.randint(40,60)
        cells.append(VirtualCell(i, repeats, affinity, extension, L_min, L_repair, I_min, pred_window))

    for cycle in range(cycles):
        if drugs:
            for d_cycle, drug, targets in drugs:
                if cycle==d_cycle:
                    for idx in targets:
                        if idx<num_cells:
                            drug.apply(cells[idx].telomerase)
        for cell in cells:
            cell.step()

    metrics = collect_metrics(cells)
    plot_metrics(metrics)
    return metrics, cells

# -------------------------------
# 4. Multi-Drug GA Optimization (Healthy Telomere Range)
# -------------------------------
class Chromosome:
    def __init__(self, num_drugs=3, num_cells=50, cycles=50):
        self.L_min=random.randint(1000,2000)
        self.L_repair=random.randint(2000,5000)
        self.I_min=random.randint(20,50)
        self.pred_window=random.randint(3,10)
        self.drugs=[]
        for _ in range(num_drugs):
            cycle=random.randint(5,cycles-5)
            type_=random.choice(['activator','inhibitor'])
            strength=random.uniform(1,2)
            targets=random.sample(range(num_cells),k=random.randint(1,num_cells//5))
            self.drugs.append((cycle, Drug(f"{type_}_{_}",type_,strength), targets))

def fitness(chrom, target_min, target_max, num_cells=50, cycles=50):
    metrics,_ = run_multi_cell_simulation(num_cells, cycles, chrom.drugs,
                                         AI_thresholds={'L_min':chrom.L_min,'L_repair':chrom.L_repair,
                                                        'I_min':chrom.I_min,'pred_window':chrom.pred_window})
    lengths = np.array(metrics['lengths'])
    deviation = np.where(lengths<target_min,target_min-lengths,0)+np.where(lengths>target_max,lengths-target_max,0)
    mse = np.mean(deviation**2)
    return -mse + metrics['active_cells'] - metrics['avg_risk']*100

def optimize_multi_drug_healthy(target_min=2000,target_max=2500,generations=3,population_size=3):
    best_score=-np.inf
    best_chrom=None
    for gen in range(generations):
        population=[Chromosome() for _ in range(population_size)]
        for chrom in population:
            score = fitness(chrom,target_min,target_max)
            if score>best_score:
                best_score=score
                best_chrom=chrom
        print(f"Generation {gen+1}: Best Fitness = {best_score}")
    return best_chrom,best_score
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import json

# -------------------------------
# 1. Core Classes (same as before)
# -------------------------------
# VirtualTelomere, Telomerase, VirtualCell, Drug, Chromosome
# Use previous code

# Classes from previous cell to avoid re-running it (assuming it was run successfully)
class VirtualTelomere:
    def __init__(self, initial_repeats=1000):
        self.length = initial_repeats
        self.integrity = 100
        self.history = {'length':[self.length],'integrity':[self.integrity]}
    def degrade(self):
        self.length -= random.randint(1,5)
        self.integrity -= random.uniform(0.1,1.0)
        self.length = max(self.length,0)
        self.integrity = max(self.integrity,0)
        self.history['length'].append(self.length)
        self.history['integrity'].append(self.integrity)

class Telomerase:
    def __init__(self, affinity=0.8, extension=50):
        self.affinity = affinity
        self.extension = extension
    def act(self, telomere):
        repair = int(self.extension*self.affinity)
        telomere.length += repair
        telomere.integrity += repair*0.1
        telomere.integrity = min(telomere.integrity,100)

class VirtualCell:
    def __init__(self, cell_id, initial_repeats=1000, telomerase_affinity=0.8,
                 telomerase_extension=50, L_min=1600, I_min=30, pred_window=5):
        self.cell_id = cell_id
        self.telomere = VirtualTelomere(initial_repeats)
        self.telomerase = Telomerase(telomerase_affinity, telomerase_extension)
        self.L_min = L_min
        self.I_min = I_min
        self.pred_window = pred_window
        self.active = True
    def step(self):
        if not self.active: return
        self.telomere.degrade()
        future_length = self.telomere.length - self.pred_window*3
        if future_length < self.L_min:
            self.telomerase.act(self.telomere)
        if self.telomere.integrity < self.I_min:
            self.active = False

class Drug:
    def __init__(self, name, type_, effect_strength=1.5):
        self.name = name
        self.type_ = type_
        self.effect_strength = effect_strength
    def apply(self, telomerase):
        if self.type_=='activator':
            telomerase.affinity = min(1.0, telomerase.affinity*self.effect_strength)
        elif self.type_=='inhibitor':
            telomerase.affinity = max(0.0, telomerase.affinity/self.effect_strength)

class Chromosome:
    def __init__(self, num_drugs=3, num_cells=50, cycles=50):
        self.L_min=random.randint(1000,2000)
        self.I_min=random.randint(20,50)
        self.pred_window=random.randint(3,10)
        self.drugs=[]
        for _ in range(num_drugs):
            cycle=random.randint(5,cycles-5)
            type_=random.choice(['activator','inhibitor'])
            strength=random.uniform(1,2)
            targets=random.sample(range(num_cells), k=random.randint(1,num_cells//5))
            self.drugs.append((cycle, Drug(f"{type_}_{_}",type_,strength), targets))

# -------------------------------
# 2. Evaluate Chromosome + Metrics
# -------------------------------
def evaluate_chromosome_with_risk(chrom, target_min, target_max, num_cells=50, cycles=50):
    cells=[]
    for i in range(num_cells):
        repeats=random.randint(800,1200)
        affinity=random.uniform(0.7,0.9)
        extension=random.randint(40,60)
        cells.append(VirtualCell(i,repeats,affinity,extension,chrom.L_min,chrom.I_min,chrom.pred_window))
    for cycle in range(cycles):
        for d_cycle, drug, targets in chrom.drugs:
            if cycle==d_cycle:
                for idx in targets:
                    if idx<num_cells:
                        drug.apply(cells[idx].telomerase)
        for cell in cells:
            cell.step()
    lengths = np.array([c.telomere.length for c in cells])
    integrities = np.array([c.telomere.integrity for c in cells])
    risks = np.array([max(0,1-l/1600)+max(0,1-i/30) for l,i in zip(lengths, integrities)])
    deviation = np.where(lengths<target_min,target_min-lengths,0)+np.where(lengths>target_max,lengths-target_max,0)
    mse = np.mean(deviation**2)
    fitness = -mse + np.sum([cell.active for cell in cells]) - np.mean(risks)*100
    metrics = {'lengths':lengths,'integrities':integrities,'active_cells':sum([cell.active for cell in cells]),
               'risks':risks,'cells':cells}
    return fitness, metrics

# -------------------------------
# 3. GA Optimization with Risk Visualization + Export
# -------------------------------

def convert_numpy_to_python(obj):
    """Recursively converts NumPy objects to Python native types."""
    if isinstance(obj, np.integer):
        return int(obj)
    elif isinstance(obj, np.floating):
        return float(obj)
    elif isinstance(obj, np.ndarray):
        return obj.tolist()
    elif isinstance(obj, dict):
        return {k: convert_numpy_to_python(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [convert_numpy_to_python(elem) for elem in obj]
    else:
        return obj

def optimize_ga_with_risk_export(target_min=2000,target_max=2500,num_cells=50,cycles=50,
                                 generations=5,population_size=5,num_drugs=3, export_file="GA_results.json"):
    best_score=-np.inf
    best_chrom=None
    fitness_progress=[]
    risk_history=[]

    fig, axes = plt.subplots(3,3, figsize=(18,12))
    plt.ion()

    for gen in range(generations):
        population=[Chromosome(num_drugs,num_cells,cycles) for _ in range(population_size)]
        for chrom in population:
            score, metrics = evaluate_chromosome_with_risk(chrom,target_min,target_max,num_cells,cycles)
            if score>best_score:
                best_score=score
                best_chrom=chrom
                best_metrics=metrics
        fitness_progress.append(best_score)
        risk_history.append(best_metrics['risks'])

        # Drug timeline
        drug_matrix = np.zeros((num_cells, cycles))
        for d_cycle, drug, targets in best_chrom.drugs:
            for t in targets:
                if t<num_cells:
                    drug_matrix[t,d_cycle]=1

        # Clear axes
        for ax in axes.flatten():
            ax.clear()

        # Fitness
        axes[0,0].plot(fitness_progress, marker='o')
        axes[0,0].set_title('Best Fitness per Generation')
        axes[0,0].set_xlabel('Generation')
        axes[0,0].set_ylabel('Fitness')

        # Telomere Length
        axes[0,1].hist(best_metrics['lengths'], bins=20, color='skyblue')
        axes[0,1].axvline(target_min,color='green',linestyle='--')
        axes[0,1].axvline(target_max,color='green',linestyle='--')
        axes[0,1].set_title('Telomere Length (Best Chromosome)')

        # Integrity
        axes[0,2].hist(best_metrics['integrities'], bins=20, color='orange')
        axes[0,2].set_title('Telomere Integrity')

        # Active/Inactive
        axes[1,0].bar(['Active','Inactive'], [best_metrics['active_cells'],num_cells-best_metrics['active_cells']], color=['green','red'])
        axes[1,0].set_ylim(0,num_cells)
        axes[1,0].set_title('Population Status')

        # Mutation Risk Heatmap
        sns.heatmap(np.array(best_metrics['risks']).reshape(-1,1), cmap='coolwarm', cbar=True, ax=axes[1,1])
        axes[1,1].set_title('Mutation Risk Heatmap')

        # Drug Timeline
        axes[1,2].imshow(drug_matrix, aspect='auto', cmap='Reds', interpolation='nearest')
        axes[1,2].set_title('Drug Timeline')
        axes[1,2].set_xlabel('Cycle')
        axes[1,2].set_ylabel('Cell ID')

        # Average Telomere Length
        avg_length = [np.mean([c.telomere.history['length'][t] for c in best_metrics['cells']]) for t in range(cycles)]
        axes[2,0].plot(range(cycles), avg_length, color='blue')
        axes[2,0].axhline(target_min,color='green',linestyle='--')
        axes[2,0].axhline(target_max,color='green',linestyle='--')
        axes[2,0].set_title('Average Telomere Length Over Time')
        axes[2,0].set_xlabel('Cycle')

        # Telomere Integrity Over Time (average)
        avg_integrity = [np.mean([c.telomere.history['integrity'][t] for c in best_metrics['cells']]) for t in range(cycles)]
        axes[2,1].plot(range(cycles), avg_integrity, color='orange')
        axes[2,1].set_title('Average Telomere Integrity Over Time')
        axes[2,1].set_xlabel('Cycle')

        # Risk Over Time
        axes[2,2].plot(range(gen + 1), [np.mean(r) for r in risk_history], color='red', marker='o')
        axes[2,2].set_title('Average Mutation Risk per Generation')
        axes[2,2].set_xlabel('Generation')

        clear_output(wait=True)
        print(f"Generation {gen+1}/{generations} | Best Fitness: {best_score:.2f}")
        plt.pause(0.1)

    plt.ioff()
    plt.show()

    # Export GA Results
    export_data = {
        'best_chromosome': {
            'L_min': best_chrom.L_min,
            'I_min': best_chrom.I_min,
            'pred_window': best_chrom.pred_window,
            'drugs': [{'cycle':d[0],'name':d[1].name,'type':d[1].type_,'strength':d[1].effect_strength,'targets':d[2]} for d in best_chrom.drugs]
        },
        'fitness_progress': fitness_progress,
        'risk_history': [list(r) for r in risk_history],
        'avg_telomere_lengths': avg_length,
        'avg_telomere_integrities': avg_integrity
    }

    # Convert NumPy types to Python native types before dumping
    export_data = convert_numpy_to_python(export_data)

    with open(export_file,'w') as f:
        json.dump(export_data,f, indent=4)

    print(f"GA results exported to {export_file}")
    return best_chrom, best_score

# -------------------------------
# 4. Interactive Widgets
# -------------------------------
generations_widget = widgets.IntSlider(value=5, min=1, max=20, step=1, description='Generations')
population_widget = widgets.IntSlider(value=5, min=2, max=20, step=1, description='Population')
num_cells_widget = widgets.IntSlider(value=50, min=10, max=200, step=10, description='Cells')
cycles_widget = widgets.IntSlider(value=50, min=10, max=200, step=10, description='Cycles')
target_min_widget = widgets.IntSlider(value=2000, min=500, max=5000, step=50, description='Target Min')
target_max_widget = widgets.IntSlider(value=2500, min=500, max=5000, step=50, description='Target Max')
num_drugs_widget = widgets.IntSlider(value=3, min=1, max=10, step=1, description='Drugs')

run_ga_button = widgets.Button(description='Run GA + Risk Visualization + Export')

def run_ga_button_click(b):
    optimize_ga_with_risk_export(target_min_widget.value,target_max_widget.value,
                                 num_cells_widget.value,cycles_widget.value,
                                 generations_widget.value,population_widget.value,
                                 num_drugs_widget.value, export_file="GA_results.json")
run_ga_button.on_click(run_ga_button_click)

display(widgets.VBox([generations_widget, population_widget, num_cells_widget, cycles_widget,
                      target_min_widget, target_max_widget, num_drugs_widget, run_ga_button]))
```

```python
import numpy as np
import random
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display, clear_output
import json

# -------------------------------
# 1. GA + Cell Simulation Functions
# -------------------------------
def simulate_chromosome(chrom, num_cells=50, cycles=50):
    cells=[]
    for i in range(num_cells):
        repeats = np.random.randint(800,1200)
        affinity = np.random.uniform(0.7,0.9)
        extension = np.random.randint(40,60)
        cells.append(VirtualCell(i, repeats, affinity, extension, chrom.L_min, chrom.I_min, chrom.pred_window))

    length_history = np.zeros((num_cells, cycles))
    integrity_history = np.zeros((num_cells, cycles))
    risk_history = np.zeros((num_cells, cycles))
    drug_matrix = np.zeros((num_cells, cycles))

    for cycle in range(cycles):
        # Apply drugs
        for d_cycle, drug, targets in chrom.drugs:
            if cycle == d_cycle:
                for t in targets:
                    if t<num_cells:
                        drug.apply(cells[t].telomerase)
                        drug_matrix[t,cycle] = 1
        # Step cells
        for idx, cell in enumerate(cells):
            cell.step()
            length_history[idx,cycle] = cell.telomere.length
            integrity_history[idx,cycle] = cell.telomere.integrity
            risk_history[idx,cycle] = max(0,1-cell.telomere.length/1600)+max(0,1-cell.telomere.integrity/30)

    # Fitness: combine deviation from healthy range + active cells - risk
    lengths = length_history[:,-1]
    deviation = np.where(lengths<target_min,lengths-target_min,0)+np.where(lengths>target_max,lengths-target_max,0)
    mse = np.mean(deviation**2)
    fitness = -mse + sum([c.active for c in cells]) - np.mean(risk_history[:,-1])*100

    metrics = {
        'length_history': length_history,
        'integrity_history': integrity_history,
        'risk_history': risk_history,
        'drug_matrix': drug_matrix,
        'cells': cells
    }

    return fitness, metrics

# -------------------------------
# 2. Live GA Optimization
# -------------------------------
def run_live_ga(target_min=2200, target_max=2500, num_cells=50, cycles=50,
                generations=5, population_size=5, num_drugs=3, export_file="GA_live.json"):

    best_score = -np.inf
    best_chrom = None
    best_metrics = None

    output = widgets.Output()
    display(output)

    cycle_slider = widgets.IntSlider(value=0, min=0, max=cycles-1, step=1, description='Cycle')
    display(cycle_slider)

    # Store simulation histories for best chromosome
    sim_data = {'length':None,'integrity':None,'risk':None,'drug':None}

    def update_dashboard(change):
        cycle = change['new']
        with output:
            output.clear_output(wait=True)
            fig = go.Figure()

            # Telomere Length
            fig.add_trace(go.Bar(
                x=list(range(num_cells)),
                y=sim_data['length'][:,cycle],
                marker_color='skyblue',
                text=[f"{l:.0f}" for l in sim_data['length'][:,cycle]],
                hoverinfo='text+x',
                name='Telomere Length'
            ))
            fig.add_shape(type='line', x0=-0.5, x1=num_cells-0.5, y0=target_min, y1=target_min,
                          line=dict(color='green', dash='dash'))
            fig.add_shape(type='line', x0=-0.5, x1=num_cells-0.5, y0=target_max, y1=target_max,
                          line=dict(color='green', dash='dash'))

            # Mutation Risk Heatmap
            fig.add_trace(go.Heatmap(
                z=sim_data['risk'][:, :cycle+1],
                x=list(range(cycle+1)),
                y=list(range(num_cells)),
                colorscale='RdBu_r',
                zmin=0,
                zmax=np.max(sim_data['risk']),
                colorbar=dict(title='Mutation Risk'),
                name='Risk'
            ))

            # Drug Timeline Heatmap
            fig.add_trace(go.Heatmap(
                z=sim_data['drug'][:, :cycle+1],
                x=list(range(cycle+1)),
                y=list(range(num_cells)),
                colorscale='Reds',
                showscale=False,
                name='Drug'
            ))

            fig.update_layout(title=f'Live Smart Cell Lab - Cycle {cycle}',
                              height=700, barmode='overlay')
            fig.show()

    cycle_slider.observe(update_dashboard, names='value')

    # Run GA generations
    for gen in range(generations):
        population = [Chromosome(num_drugs, num_cells, cycles) for _ in range(population_size)]
        for chrom in population:
            fitness, metrics = simulate_chromosome(chrom, num_cells, cycles)
            if fitness > best_score:
                best_score = fitness
                best_chrom = chrom
                best_metrics = metrics
                sim_data['length'] = metrics['length_history']
                sim_data['integrity'] = metrics['integrity_history']
                sim_data['risk'] = metrics['risk_history']
                sim_data['drug'] = metrics['drug_matrix']

        with output:
            print(f"Generation {gen+1}/{generations} | Best Fitness: {best_score:.2f}")

    # Export results
    export_data = {
        'best_chromosome': {
            'L_min': best_chrom.L_min,
            'I_min': best_chrom.I_min,
            'pred_window': best_chrom.pred_window,
            'drugs':[{'cycle':d[0],'name':d[1].name,'type':d[1].type_,'strength':d[1].effect_strength,'targets':d[2]} for d in best_chrom.drugs]
        },
        'fitness': best_score
    }
    with open(export_file,'w') as f:
        json.dump(export_data,f, indent=4)
    with output:
        print(f"\nGA results exported to {export_file}")

# -------------------------------
# 3. Run the Live GA Dashboard
# -------------------------------
run_live_ga(target_min=2200, target_max=2500, num_cells=50, cycles=50,
            generations=5, population_size=5, num_drugs=3)
```

```python
# -------------------------------
# 1. Extend Drug Class to Handle Interactions
# -------------------------------
class Drug:
    def __init__(self, name, type_, effect_strength=1.5, interactions=None):
        self.name = name
        self.type_ = type_
        self.effect_strength = effect_strength
        self.interactions = interactions or {}  # dict: {other_drug_name: multiplier}

    def apply(self, telomerase, active_drugs=None):
        multiplier = 1.0
        if active_drugs:
            for d in active_drugs:
                if d.name in self.interactions:
                    multiplier *= self.interactions[d.name]
        if self.type_=='activator':
            telomerase.affinity = min(1.0, telomerase.affinity * self.effect_strength * multiplier)
        elif self.type_=='inhibitor':
            telomerase.affinity = max(0.0, telomerase.affinity / (self.effect_strength * multiplier))

# -------------------------------
# 2. Chromosome with Drug Interactions
# -------------------------------
class Chromosome:
    def __init__(self, num_drugs=3, num_cells=50, cycles=50):
        self.L_min = random.randint(1000,2000)
        self.I_min = random.randint(20,50)
        self.pred_window = random.randint(3,10)
        self.drugs = []
        drug_names = []
        for i in range(num_drugs):
            cycle = random.randint(5, cycles-5)
            type_ = random.choice(['activator','inhibitor'])
            strength = random.uniform(1,2)
            targets = random.sample(range(num_cells), k=random.randint(1,num_cells//5))
            name = f"{type_}_{i}"
            interactions = {}
            # Randomly define interactions with previous drugs
            for dn in drug_names:
                interactions[dn] = random.uniform(0.8,1.2)  # synergistic (>1) or antagonistic (<1)
            drug = Drug(name, type_, strength, interactions)
            self.drugs.append((cycle, drug, targets))
            drug_names.append(name)

# -------------------------------
# 3. Simulation Function considering interactions
# -------------------------------
def simulate_chromosome_interactions(chrom, num_cells=50, cycles=50):
    cells=[]
    for i in range(num_cells):
        repeats = np.random.randint(800,1200)
        affinity = np.random.uniform(0.7,0.9)
        extension = np.random.randint(40,60)
        cells.append(VirtualCell(i,repeats,affinity,extension,chrom.L_min,chrom.I_min,chrom.pred_window))

    length_history = np.zeros((num_cells, cycles))
    integrity_history = np.zeros((num_cells, cycles))
    risk_history = np.zeros((num_cells, cycles))
    drug_matrix = np.zeros((num_cells, cycles))

    for cycle in range(cycles):
        active_drugs = []
        # Apply drugs with interactions
        for d_cycle, drug, targets in chrom.drugs:
            if cycle == d_cycle:
                active_drugs.append(drug)
                for t in targets:
                    if t<num_cells:
                        drug.apply(cells[t].telomerase, active_drugs=active_drugs)
                        drug_matrix[t,cycle] = 1
        # Step cells
        for idx, cell in enumerate(cells):
            cell.step()
            length_history[idx,cycle] = cell.telomere.length
            integrity_history[idx,cycle] = cell.telomere.integrity
            risk_history[idx,cycle] = max(0,1-cell.telomere.length/1600)+max(0,1-cell.telomere.integrity/30)

    # Fitness combining telomere deviation + active cells - mean risk
    lengths = length_history[:,-1]
    deviation = np.where(lengths<target_min,lengths-target_min,0)+np.where(lengths>target_max,lengths-target_max,0)
    mse = np.mean(deviation**2)
    fitness = -mse + sum([c.active for c in cells]) - np.mean(risk_history[:,-1])*100

    metrics = {'length_history':length_history,
               'integrity_history':integrity_history,
               'risk_history':risk_history,
               'drug_matrix':drug_matrix,
               'cells':cells}

    return fitness, metrics

# -------------------------------
# 4. Live GA Dashboard for Multi-Drug Interactions
# -------------------------------
def run_live_ga_interactions(target_min=2200, target_max=2500, num_cells=50, cycles=50,
                             generations=5, population_size=5, num_drugs=3, export_file="GA_interactions.json"):

    best_score = -np.inf
    best_chrom = None
    best_metrics = None
    sim_data = {'length':None,'integrity':None,'risk':None,'drug':None}

    output = widgets.Output()
    display(output)

    cycle_slider = widgets.IntSlider(value=0, min=0, max=cycles-1, step=1, description='Cycle')
    display(cycle_slider)

    def update_dashboard(change):
        cycle = change['new']
        with output:
            output.clear_output(wait=True)
            fig = go.Figure()
            # Telomere Length
            fig.add_trace(go.Bar(
                x=list(range(num_cells)),
                y=sim_data['length'][:,cycle],
                marker_color='skyblue',
                text=[f"{l:.0f}" for l in sim_data['length'][:,cycle]],
                hoverinfo='text+x',
                name='Telomere Length'
            ))
            fig.add_shape(type='line', x0=-0.5, x1=num_cells-0.5, y0=target_min, y1=target_min,
                          line=dict(color='green', dash='dash'))
            fig.add_shape(type='line', x0=-0.5, x1=num_cells-0.5, y0=target_max, y1=target_max,
                          line=dict(color='green', dash='dash'))
            # Mutation Risk
            fig.add_trace(go.Heatmap(
                z=sim_data['risk'][:, :cycle+1],
                x=list(range(cycle+1)),
                y=list(range(num_cells)),
                colorscale='RdBu_r',
                zmin=0,
                zmax=np.max(sim_data['risk']),
                colorbar=dict(title='Mutation Risk')
            ))
            # Drug Timeline
            fig.add_trace(go.Heatmap(
                z=sim_data['drug'][:, :cycle+1],
                x=list(range(cycle+1)),
                y=list(range(num_cells)),
                colorscale='Reds',
                showscale=False
            ))
            fig.update_layout(title=f'Live Smart Cell Lab (Drug Interactions) - Cycle {cycle}',
                              height=700, barmode='overlay')
            fig.show()

    cycle_slider.observe(update_dashboard, names='value')

    # Run GA generations
    for gen in range(generations):
        population = [Chromosome(num_drugs, num_cells, cycles) for _ in range(population_size)]
        for chrom in population:
            fitness, metrics = simulate_chromosome_interactions(chrom, num_cells, cycles)
            if fitness > best_score:
                best_score = fitness
                best_chrom = chrom
                best_metrics = metrics
                sim_data['length'] = metrics['length_history']
                sim_data['integrity'] = metrics['integrity_history']
                sim_data['risk'] = metrics['risk_history']
                sim_data['drug'] = metrics['drug_matrix']
        with output:
            print(f"Generation {gen+1}/{generations} | Best Fitness: {best_score:.2f}")

    # Export results
    export_data = {
        'best_chromosome': {
            'L_min': best_chrom.L_min,
            'I_min': best_chrom.I_min,
            'pred_window': best_chrom.pred_window,
            'drugs':[{'cycle':d[0],'name':d[1].name,'type':d[1].type_,'strength':d[1].effect_strength,'interactions':d[1].interactions,'targets':d[2]} for d in best_chrom.drugs]
        },
        'fitness': best_score
    }
    with open(export_file,'w') as f:
        json.dump(export_data,f, indent=4)
    with output:
        print(f"\nGA results exported to {export_file}")

# -------------------------------
# 5. Run Example
# -------------------------------
run_live_ga_interactions(target_min=2200, target_max=2500,
                         num_cells=50, cycles=50,
                         generations=5, population_size=5, num_drugs=3)
```

```python
import itertools
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import io
import ipywidgets as widgets
from IPython.display import display, clear_output

# -------------------------------
# 1. Function to calculate synergy for multiple drugs
# -------------------------------
def calculate_multi_drug_synergy(drug_list, num_cells=50):
    """
    Calculates predicted effect of multi-drug combination on telomere length and mutation risk
    Returns average telomere increase and average risk reduction
    """
    base_lengths = np.random.randint(800,1200, size=num_cells)
    base_risks = np.random.uniform(0.1,0.5, size=num_cells)

    multiplier = 1.0
    # Apply all pairwise interactions
    for i, d1 in enumerate(drug_list):
        for d2 in drug_list[i+1:]:
            if d2.name in d1.interactions:
                multiplier *= d1.interactions[d2.name]
            if d1.name in d2.interactions:
                multiplier *= d2.interactions[d1.name]

    avg_telomere_increase = np.mean(base_lengths) * 0.05 * multiplier
    avg_risk_reduction = np.mean(base_risks) * 0.1 * multiplier

    return avg_telomere_increase, avg_risk_reduction

# -------------------------------
# 2. Generate all combinations (pairs or triples)
# -------------------------------
def generate_combinations(drugs, combo_size=2):
    return list(itertools.combinations(drugs, combo_size))

# -------------------------------
# 3. Plot synergy heatmap
# -------------------------------
def plot_synergy_heatmap(drugs, combo_size=2):
    combos = generate_combinations(drugs, combo_size)
    tel_increases = []
    risk_reductions = []
    labels = []

    for combo in combos:
        tel_inc, risk_red = calculate_multi_drug_synergy(combo)
        tel_increases.append(tel_inc)
        risk_reductions.append(risk_red)
        labels.append(", ".join([d.name for d in combo]))

    # Normalize for color scaling
    tel_increases_norm = (tel_increases - np.min(tel_increases)) / (np.max(tel_increases) - np.min(tel_increases)+1e-6)
    risk_reductions_norm = (risk_reductions - np.min(risk_reductions)) / (np.max(risk_reductions) - np.min(risk_reductions)+1e-6)

    fig, ax = plt.subplots(1,2, figsize=(18,6))

    # Telomere Increase Heatmap
    im1 = ax[0].bar(range(len(combos)), tel_increases, color=plt.cm.viridis(tel_increases_norm))
    ax[0].set_xticks(range(len(combos)))
    ax[0].set_xticklabels(labels, rotation=45, ha='right')
    ax[0].set_ylabel("Predicted Telomere Increase")
    ax[0].set_title(f"Multi-Drug Combination Synergy (Telomere Increase)")

    # Risk Reduction Heatmap
    im2 = ax[1].bar(range(len(combos)), risk_reductions, color=plt.cm.plasma(risk_reductions_norm))
    ax[1].set_xticks(range(len(combos)))
    ax[1].set_xticklabels(labels, rotation=45, ha='right')
    ax[1].set_ylabel("Predicted Risk Reduction")
    ax[1].set_title(f"Multi-Drug Combination Synergy (Mutation Risk Reduction)")

    plt.tight_layout()
    plt.show()

    return combos, tel_increases, risk_reductions

# -------------------------------
# 4. Example Multi-Drug Explorer
# -------------------------------
def multi_drug_explorer_demo(num_drugs=5, combo_size=2):
    # Generate random drugs with interactions
    drugs = []
    drug_names = []
    for i in range(num_drugs):
        type_ = np.random.choice(['activator','inhibitor'])
        name = f"{type_}_{i}"
        interactions = {dn: np.random.uniform(0.8,1.2) for dn in drug_names}
        drugs.append(Drug(name, type_, effect_strength=np.random.uniform(1,2), interactions=interactions))
        drug_names.append(name)

    print(f"Generated {num_drugs} drugs:")
    for d in drugs:
        print(f"  {d.name}, type: {d.type_}, interactions: {d.interactions}")

    # Plot all combinations
    plot_synergy_heatmap(drugs, combo_size)

# -------------------------------
# 5. Run the Demo
# -------------------------------
multi_drug_explorer_demo(num_drugs=6, combo_size=2)
```

```python
# Real pharmacology-inspired Drug class
class RealDrug:
    def __init__(self, name, type_, IC50=None, EC50=None, Emax=1.0, interactions=None):
        """
        name    : string
        type_   : "inhibitor" | "activator"
        IC50    : drug conc where half-max inhibition occurs (µM) — for inhibitors
        EC50    : drug conc where half-max activation occurs (µM) — for activators
        Emax    : max magnitude of effect (fractional)
        interactions: dict of {other_drug_name: multiplier}
        """
        self.name = name
        self.type_ = type_
        self.IC50 = IC50
        self.EC50 = EC50
        self.Emax = Emax
        self.interactions = interactions or {}

    def effect_multiplier(self, conc):
        """
        Compute dose-response multiplier (0–1) based on IC50/EC50.
        conc is in simulation concentration units (µM).
        """
        if self.type_ == "inhibitor" and self.IC50 is not None:
            # Classic inhibitory Emax model
            return self.Emax * (conc / (self.IC50 + conc))
        elif self.type_ == "activator" and self.EC50 is not None:
            # Activator: saturable activation
            return self.Emax * (conc / (self.EC50 + conc))
        return 0.0

    def apply(self, telomerase, conc=1.0, active_drugs=None):
        """
        Apply effect on telomerase based on conc and drug interactions.
        active_drugs: other RealDrug objects whose presence may modulate effect.
        """
        # Compute base multiplier from this drug
        base_mult = self.effect_multiplier(conc)

        # Apply interactions (synergy / antagonism)
        interact_mult = 1.0
        if active_drugs:
            for other in active_drugs:
                if other.name in self.interactions:
                    interact_mult *= self.interactions[other.name]

        total_effect = base_mult * interact_mult

        if self.type_ == "inhibitor":
            # Reduce telomerase affinity
            telomerase.affinity *= max(0, 1 - total_effect)
        elif self.type_ == "activator":
            # Increase telomerase affinity up to ceiling
            telomerase.affinity = min(1.0, telomerase.affinity + total_effect)
```

```python
import numpy as np
import random

# --- Core Classes (copied for self-containment) ---
class VirtualTelomere:
    def __init__(self, initial_repeats=1000):
        self.length = initial_repeats
        self.integrity = 100
        self.history = {'length':[self.length],'integrity':[self.integrity]}
    def degrade(self):
        self.length -= random.randint(1,5)
        self.integrity -= random.uniform(0.1,1.0)
        self.length = max(self.length,0)
        self.integrity = max(self.integrity,0)
        self.history['length'].append(self.length)
        self.history['integrity'].append(self.integrity)

class Telomerase:
    def __init__(self, affinity=0.8, extension=50):
        self.affinity = affinity
        self.extension = extension
    def act(self, telomere):
        repair = int(self.extension*self.affinity)
        telomere.length += repair
        telomere.integrity += repair*0.1
        telomere.integrity = min(telomere.integrity,100)

class VirtualCell:
    def __init__(self, cell_id, initial_repeats=1000, telomerase_affinity=0.8,
                 telomerase_extension=50, L_min=1600, I_min=30, pred_window=5):
        self.cell_id = cell_id
        self.telomere = VirtualTelomere(initial_repeats)
        self.telomerase = Telomerase(telomerase_affinity, telomerase_extension)
        self.L_min = L_min
        self.I_min = I_min
        self.pred_window = pred_window
        self.active = True
    def step(self):
        if not self.active: return
        self.telomere.degrade()
        future_length = self.telomere.length - self.pred_window*3
        if future_length < self.L_min:
            self.telomerase.act(self.telomere)
        if self.telomere.integrity < self.I_min:
            self.active = False

# Simple Drug class for Chromosome
class Drug:
    def __init__(self, name, type_, effect_strength=1.5):
        self.name = name
        self.type_ = type_
        self.effect_strength = effect_strength
    def apply(self, telomerase):
        if self.type_=='activator':
            telomerase.affinity = min(1.0, telomerase.affinity*self.effect_strength)
        elif self.type_=='inhibitor':
            telomerase.affinity = max(0.0, telomerase.affinity/self.effect_strength)

# Chromosome class definition
class Chromosome:
    def __init__(self, num_drugs=3, num_cells=50, cycles=50):
        self.L_min=random.randint(1000,2000)
        self.I_min=random.randint(20,50)
        self.pred_window=random.randint(3,10)
        self.drugs=[]
        for _ in range(num_drugs):
            cycle=random.randint(5,cycles-5)
            type_=random.choice(['activator','inhibitor'])
            strength=random.uniform(1,2)
            targets=random.sample(range(num_cells), k=random.randint(1,num_cells//5))
            self.drugs.append((cycle, Drug(f"{type_}_{_}",type_,strength), targets))

# Real pharmacology-inspired Drug class
class RealDrug:
    def __init__(self, name, type_, IC50=None, EC50=None, Emax=1.0, interactions=None):
        self.name = name
        self.type_ = type_
        self.IC50 = IC50
        self.EC50 = EC50
        self.Emax = Emax
        self.interactions = interactions or {}

    def effect_multiplier(self, conc):
        if self.type_ == "inhibitor" and self.IC50 is not None:
            return self.Emax * (conc / (self.IC50 + conc))
        elif self.type_ == "activator" and self.EC50 is not None:
            return self.Emax * (conc / (self.EC50 + conc))
        return 0.0

    def apply(self, telomerase, conc=1.0, active_drugs=None):
        base_mult = self.effect_multiplier(conc)
        interact_mult = 1.0
        if active_drugs:
            for other in active_drugs:
                if other.name in self.interactions:
                    interact_mult *= self.interactions[other.name]
        total_effect = base_mult * interact_mult

        if self.type_ == "inhibitor":
            telomerase.affinity *= max(0, 1 - total_effect)
        elif self.type_ == "activator":
            telomerase.affinity = min(1.0, telomerase.affinity + total_effect)

# Example real telomerase drugs with literature‑inspired PD params (from wCq5X98PBP5E)
real_drugs = [
    RealDrug(name="Imetelstat", type_="inhibitor", IC50=0.1, Emax=1.0),   # IC50 ~0.05–0.2 µM
    RealDrug(name="MST-312",   type_="inhibitor", IC50=12.0, Emax=0.8),  # IC50 ~12 µM
    RealDrug(name="EGCG",      type_="inhibitor", IC50=80.0, Emax=0.5),  # weak effect
    RealDrug(name="TA-65",     type_="activator", EC50=5.0, Emax=0.4)    # EC50 ~1–10 µM
]
# Add interaction multipliers (example synergy/antagonism)
real_drugs[0].interactions = {"MST-312": 1.1}   # slight synergy
real_drugs[1].interactions = {"Imetelstat": 0.9} # slight antagonism


class RealDrugChromosome(Chromosome):
    def __init__(self, num_drugs=3, num_cells=50, cycles=50):
        super().__init__(num_drugs, num_cells, cycles)
        # Instead of synthetic drugs, pick from real_drugs pool
        self.drugs = []
        for _ in range(num_drugs):
            d = random.choice(real_drugs)
            d_targets = random.sample(range(num_cells), k=random.randint(1, num_cells//5))
            d_cycle = random.randint(0, cycles-1)
            d_conc  = random.choice([0.1, 0.5, 1.0, 5.0, 10.0])  # potential dosing
            self.drugs.append((d_cycle, d, d_targets, d_conc))
```

```python
class MultiCompartmentPK:
    def __init__(self, drug_obj, dose, k12=0.3, k21=0.2, ke=0.05):
        """
        k12 : plasma → cell rate
        k21 : cell → plasma rate
        ke  : elimination rate from plasma
        """
        self.drug = drug_obj
        self.plasma = dose
        self.cell = 0.0
        self.k12 = k12
        self.k21 = k21
        self.ke = ke

    def step(self):
        # distribution
        to_cell = self.k12 * self.plasma
        to_plasma = self.k21 * self.cell

        self.plasma += to_plasma - to_cell
        self.cell += to_cell - to_plasma

        # elimination (plasma)
        self.plasma *= (1 - self.ke)

    def apply(self, telomerase, active_drugs=None):
        # Use intracellular concentration for PD
        self.drug.apply(telomerase, conc=self.cell, active_drugs=active_drugs)
```

```python
class MultiCompartmentPK:
    def __init__(self, drug_obj, dose, k12=0.3, k21=0.2, ke=0.05):
        """
        k12 : plasma → cell rate
        k21 : cell → plasma rate
        ke  : elimination rate from plasma
        """
        self.drug = drug_obj
        self.plasma = dose
        self.cell = 0.0
        self.k12 = k12
        self.k21 = k21
        self.ke = ke

    def step(self):
        # distribution
        to_cell = self.k12 * self.plasma
        to_plasma = self.k21 * self.cell

        self.plasma += to_plasma - to_cell
        self.cell += to_cell - to_plasma

        # elimination (plasma)
        self.plasma *= (1 - self.ke)

    def apply(self, telomerase, active_drugs=None):
        # Use intracellular concentration for PD
        self.drug.apply(telomerase, conc=self.cell, active_drugs=active_drugs)
```

```python
def run_advanced_simulation(pk_drugs, ai, num_cells=50, cycles=60):
    cells = [
        VirtualCell(i,
            initial_repeats=random.randint(800,1200),
            telomerase_affinity=random.uniform(0.7,0.9),
            telomerase_extension=random.randint(40,60),
            L_min=1600,
            I_min=30,
            pred_window=5
        ) for i in range(num_cells)
    ]

    tel_hist = np.zeros((num_cells, cycles))
    plasma_hist = {d.drug.name: [] for d in pk_drugs}
    cell_hist = {d.drug.name: [] for d in pk_drugs}

    for c in range(cycles):
        avg_tel = np.mean([cell.telomere.length for cell in cells])

        # AI decides dosing
        dosing = ai.decide_dose(avg_tel)
        for pk in pk_drugs:
            if pk.drug.name in dosing:
                pk.plasma += dosing[pk.drug.name]

        # PK + PD
        for pk in pk_drugs:
            pk.step()
            # Pass active_drugs for interaction calculation
            current_active_drugs = [p.drug for p in pk_drugs if p.plasma > 0 or p.cell > 0] # Example: consider drugs with non-zero concentration active
            for cell in cells:
                pk.apply(cell.telomerase, active_drugs=current_active_drugs)
            plasma_hist[pk.drug.name].append(pk.plasma)
            cell_hist[pk.drug.name].append(pk.cell)

        # Cell update
        for i, cell in enumerate(cells):
            cell.step()
            tel_hist[i, c] = cell.telomere.length

    return tel_hist, plasma_hist, cell_hist
```

```python
class TelomereAIController:
    def __init__(self, healthy_min=2200, healthy_max=2500, dose_change_factor=0.5):
        self.healthy_min = healthy_min
        self.healthy_max = healthy_max
        self.dose_change_factor = dose_change_factor

    def decide_dose(self, avg_telomere_length):
        dosing = {}
        # Simple AI logic: if telomere length is too low, activate telomerase (TA-65)
        # If too high or stable, inhibit (Imetelstat)

        # Assuming TA-65 is real_drugs[3] and Imetelstat is real_drugs[0]
        # This needs to be more sophisticated with actual drug names if using generic pk_drugs list

        if avg_telomere_length < self.healthy_min:
            # Increase TA-65 to boost telomeres, decrease Imetelstat if present
            dosing["TA-65"] = 3.0 * self.dose_change_factor # Example dose increase
            dosing["Imetelstat"] = 0.0 # Stop inhibitor if too low
        elif avg_telomere_length > self.healthy_max:
            # Increase Imetelstat to prevent over-extension, decrease TA-65
            dosing["Imetelstat"] = 0.5 * self.dose_change_factor # Example dose increase
            dosing["TA-65"] = 0.0 # Stop activator if too high
        else:
            # Maintain or slightly reduce doses if in healthy range
            dosing["TA-65"] = 0.0  # No new dose, let PK handle existing
            dosing["Imetelstat"] = 0.0 # No new dose

        return dosing
```
