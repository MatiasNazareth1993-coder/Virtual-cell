```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from copy import deepcopy
from typing import List, Dict, Any
import math

# Core activation
class REGELUPlus(nn.Module):
    def __init__(self, alpha=0.5, beta=0.5):
        super().__init__()
        self.alpha = nn.Parameter(torch.tensor(alpha))
        self.beta = nn.Parameter(torch.tensor(beta))

    def forward(self, x):
        alpha = torch.clamp(self.alpha, 0, 1)
        beta = torch.clamp(self.beta, 0, 1)
        pos = alpha * F.gelu(x) + (1-alpha) * F.relu(x)
        neg = beta * F.elu(x) + (1-beta) * F.leaky_relu(x, 0.01)
        return torch.where(x>=0, pos, neg)

# -----------------------------
# Meta HyperFusion Activation (dynamic)
# -----------------------------
class MetaHyperFusionBioAct(nn.Module):
    """Per-channel learnable mix of ReLU, GELU, SiLU, ELU, Swish with scaling."""
    def __init__(self, features=None):
        super().__init__()
        self.features = features
        if features is None:
            self.raw = nn.Parameter(torch.randn(5))
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(features, 5))
            self.scale = nn.Parameter(torch.ones(features))

    def forward(self, x):
        if self.features is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            return r*w[0] + g*w[1] + l*w[2] + e*w[3] + sw*w[4]
        else:
            w = torch.softmax(self.raw, dim=-1)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,)*(x.dim()-2)), -1)
            x_s = x * s
            r = F.relu(x_s); g = F.gelu(x_s); l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s); sw = x_s*torch.sigmoid(x_s)
            out = r*w[...,0] + g*w[...,1] + l*w[...,2] + e*w[...,3] + sw*w[...,4]
            return out

# -----------------------------
# Efficient Block (attention-free)
# -----------------------------
class EfficientBlock(nn.Module):
    def __init__(self, dim, tokens, hidden_dim=None, num_experts=4):
        super().__init__()
        hidden_dim = hidden_dim or dim*2
        self.norm = nn.LayerNorm(dim)
        self.token_mixer = nn.Sequential(
            nn.Linear(tokens, tokens*2),
            MetaHyperFusionBioAct(features=tokens*2), # FIX: Changed features=dim to features=tokens*2
            nn.Linear(tokens*2, tokens)
        )
        self.channel_mixer = nn.Sequential(
            nn.Linear(dim, hidden_dim*2),
            nn.GLU(dim=-1),
            MetaHyperFusionBioAct(features=hidden_dim),
            nn.Linear(hidden_dim, dim)
        )
        self.experts = nn.ModuleList([nn.Linear(dim, dim) for _ in range(num_experts)])
        self.gate = nn.Linear(dim, num_experts)

    def forward(self, x):
        y = x.transpose(1,2)
        y = y + self.token_mixer(y)
        y = y.transpose(1,2)
        y = y + self.channel_mixer(y)

        gate_w = torch.softmax(self.gate(y), dim=-1)
        expert_outs = torch.stack([F.relu(e(y)) for e in self.experts], dim=-1)
        y = (expert_outs * gate_w.unsqueeze(2)).sum(dim=-1)
        return y

# -----------------------------
# Efficient Backbone
# -----------------------------
class EfficientNet(nn.Module):
    def __init__(self, input_dim=2, dim=256, tokens=32, depth=12, out_dim=1):
        super().__init__()
        self.input_dim = input_dim # Store input_dim
        self.embed = nn.Linear(input_dim, dim)
        self.pos_embed = nn.Parameter(torch.randn(1, tokens, dim)*0.02)
        self.blocks = nn.ModuleList([EfficientBlock(dim, tokens) for _ in range(depth)])
        self.norm = nn.LayerNorm(dim)
        self.head = nn.Linear(dim, out_dim)

    def forward(self, x):
        if x.dim()==2:
            x = x.unsqueeze(1).repeat(1, self.pos_embed.size(1), 1)
        x = self.embed(x) + self.pos_embed
        for blk in self.blocks:
            x = blk(x)
        x = self.norm(x)
        x = x.mean(dim=1)
        return self.head(x)

# -----------------------------
# Virtual Cell System (Telomeres + Telomerase + Phagocytosis)
# -----------------------------
class TelomereVirtual(nn.Module):
    def __init__(self, init_length=120.0, vec_len=1, degrade_scale=0.5, repair_scale=1.0):
        super().__init__()
        self.register_buffer("_base_length", torch.full((vec_len,), init_length))
        self.vec_len = vec_len
        self.degrade_scale = degrade_scale
        self.repair_scale = repair_scale

    def initial_state(self, batch=1, device=None):
        dev = device or self._base_length.device
        return self._base_length.unsqueeze(0).repeat(batch, 1).to(dev)

    def degrade(self, tel, rnd_noise=0.1):
        noise = torch.randn_like(tel) * rnd_noise * self.degrade_scale
        return torch.clamp(tel - torch.abs(noise) - (0.1 * self.degrade_scale), 0.0)

    def repair(self, tel, strength):
        return tel + self.repair_scale*strength

class VirtualCellSystem(nn.Module):
    def __init__(self, backbone, tel_len_init=120.0, tel_vec_len=1, device=None):
        super().__init__()
        self.device = device or torch.device("cpu")
        self.backbone = backbone
        self.tel = TelomereVirtual(init_length=tel_len_init, vec_len=tel_vec_len)
        # Dummy modules for now to satisfy param_groups in MetaHybridOptimizer
        # In a full implementation, these would be learnable components of the VCS
        self.tvase = nn.Linear(1,1) # Telomerase control analog
        self.fago = nn.Linear(1,1)  # Phagocytosis control analog


    def initialize_telomeres(self, batch=1):
        return self.tel.initial_state(batch, self.device)

    def division_step(self, x, tel_state: torch.Tensor, policy: dict, agent_action_output: torch.Tensor = None):
        # Ensure tel_state explicitly requires gradients for this step if not already
        if not tel_state.requires_grad:
            tel_state = tel_state.clone().detach().requires_grad_(True)

        # Apply telomere degradation
        tel_state = self.tel.degrade(tel_state, rnd_noise=policy["noise_scale"])

        # Determine repair strength. If agent_action_output is provided, use it.
        if agent_action_output is not None:
            # Use the agent's action output to derive repair strength
            # Assuming the first component of the action is relevant for telomere repair
            raw_strength_signal = agent_action_output[:, :self.tel.vec_len]
            # Process through tvase (Telomerase Control Analog) and scale
            strength = self.tvase(raw_strength_signal.mean(dim=1, keepdim=True)).sigmoid() # Output between 0 and 1
            strength = strength * policy["repair_activity_thresh"] # Scale it
        else:
            # Fallback for _generate_data or if not using agent signal for repair
            strength = torch.tensor(torch.rand(1).item(), device=self.device) * policy["repair_activity_thresh"]

        tel_state = self.tel.repair(tel_state, strength=strength)

        division_rate = 0.05 if tel_state.mean() > policy["division_prob_thresh"] else 0.0 # Placeholder

        # Simulate telomere cleaning/repair based on policy and activity
        cleaned_x = x # For now, no actual cleaning happens here, x remains unchanged

        metrics = {
            "tel_mean_after": tel_state.mean().item(),
            "activity_mean": strength.mean().item() if isinstance(strength, torch.Tensor) else strength,
            "division_rate": division_rate,
        }

        return cleaned_x, tel_state, metrics

# -----------------------------
# TriOptimXAMP: triple optimizer with AMP + gradient clipping
# -----------------------------
class TriOptimXAMP:
    def __init__(self, params, base_lr=3e-4, wd=1e-2, max_grad_norm=1.0, amp=True):
        self.params = list(params)
        self.adam = torch.optim.AdamW(self.params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(self.params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(self.params, lr=base_lr*0.5, momentum=0.9)
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp)
        self.max_grad_norm = max_grad_norm

    def zero_grad(self):
        self.adam.zero_grad(); self.sgd.zero_grad(); self.sgd_m.zero_grad()

    def step(self):
        torch.nn.utils.clip_grad_norm_(self.params, self.max_grad_norm)
        self.adam.step(); self.sgd.step(); self.sgd_m.step()

    def state_dict(self):
        """Returns the state of the optimizers as a dict."""
        return {
            "adam": self.adam.state_dict(),
            "sgd": self.sgd.state_dict(),
            "sgd_m": self.sgd_m.state_dict(),
            "scaler": self.scaler.state_dict(),
        }

    def load_state_dict(self, state_dict):
        """Loads the optimizers state."""
        self.adam.load_state_dict(state_dict["adam"])
        self.sgd.load_state_dict(state_dict["sgd"])
        self.sgd_m.load_state_dict(state_dict["sgd_m"])
        self.scaler.load_state_dict(state_dict["scaler"])

# -----------------------------
# Autonomous Agent (Self-Improving)
# -----------------------------
class AutonomousAgent(nn.Module):
    def __init__(self, backbone, vcs, hidden=256, action_dim=8):
        super().__init__()
        self.backbone = backbone
        self.vcs = vcs
        self.encoder = nn.Sequential(nn.Linear(backbone.head.out_features, hidden),
                                     MetaHyperFusionBioAct(hidden),
                                     nn.Linear(hidden, hidden))
        self.policy = nn.Sequential(nn.Linear(hidden, hidden),
                                    MetaHyperFusionBioAct(hidden),
                                    nn.Linear(hidden, action_dim))
        self.value = nn.Sequential(nn.Linear(hidden, hidden),
                                   MetaHyperFusionBioAct(hidden),
                                   nn.Linear(hidden, 1))

    def observe(self, x):
        emb = self.backbone(x)
        if emb.dim()>2: emb = emb.mean(dim=1)
        return self.encoder(emb)

    def act(self, z):
        mu = self.policy(z)
        noise = torch.randn_like(mu)*0.1
        return torch.tanh(mu + noise)

    def evaluate(self, z, action):
        v = self.value(z).squeeze(-1)
        adv = (action*z).mean(dim=-1)
        return v, v + 0.01*adv

# -----------------------------
# Sequential Meta-Evolution
# -----------------------------
class SequentialMetaEvolution:
    """
    Sequential Self-Improving Agents framework.
    Each generation produces new data, learns from it, and improves itself.
    """

    def __init__(self,
                 backbone: nn.Module,
                 vcs: nn.Module,
                 agent_hidden: int = 256,
                 action_dim: int = 8,
                 device: str = "cpu",
                 generations: int = 5,
                 steps_per_gen: int = 100,
                 batch_size: int = 16):
        self.device = torch.device(device)
        self.backbone = backbone.to(self.device)
        self.vcs = vcs.to(self.device)
        self.agent_hidden = agent_hidden
        self.action_dim = action_dim
        self.generations = generations
        self.steps_per_gen = steps_per_gen
        self.batch_size = batch_size

        # Initialize first agent
        self.agents: List[nn.Module] = []
        self.optimizers: List[Any] = []

    def _init_agent(self):
        agent = AutonomousAgent(backbone=self.backbone, vcs=self.vcs, action_dim=self.action_dim, hidden=self.agent_hidden).to(self.device)
        optim = TriOptimXAMP(agent.parameters(), base_lr=3e-4, wd=1e-2) # Changed to TriOptimXAMP
        return agent, optim

    def _generate_data(self, agent: nn.Module) -> torch.Tensor:
        # abstract safe data generation loop
        x = torch.randn(self.batch_size, self.backbone.input_dim).to(self.device)
        tel = self.vcs.initialize_telomeres(batch=self.batch_size).to(self.device)
        for step in range(self.steps_per_gen):
            z = agent.observe(x)
            action = agent.act(z)
            x = x + 0.01 * action[:, :self.backbone.input_dim]  # safe perturbation
            # In data generation, no need for differentiable telomere updates directly from agent
            cleaned, tel, _ = self.vcs.division_step(x, tel, policy={
                "min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                "division_cost":0.5, "noise_scale":0.03,
                "division_prob_thresh":0.5, "repair_activity_thresh":0.2
            })
        return x.detach()

    def _train_agent_on_data(self, agent: nn.Module, optim: Any, data: torch.Tensor):
        scaler = torch.cuda.amp.GradScaler(enabled=(self.device.type=="cuda"))
        for step in range(self.steps_per_gen):
            optim.zero_grad()
            with torch.cuda.amp.autocast(enabled=(self.device.type=="cuda")):
                z = agent.observe(data)
                action = agent.act(z)

                # Initial tel state for this training step, ensuring it requires grad
                tel_initial_state = self.vcs.initialize_telomeres(batch=data.shape[0])
                tel_initial_state.requires_grad_(True) # Explicitly make it require grad

                cleaned, tel, metrics = self.vcs.division_step(
                    data,
                    tel_initial_state,
                    policy={
                        "min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                        "division_cost":0.5, "noise_scale":0.03,
                        "division_prob_thresh":0.5, "repair_activity_thresh":0.2
                    },
                    agent_action_output=action # Pass the agent's differentiable action
                )
                # Meta-loss: keep tel mean near target and low division
                target_tel = torch.tensor([100.0], device=self.device)
                loss_tel = F.mse_loss(tel.mean().unsqueeze(0), target_tel)
                loss = loss_tel + 0.1 * torch.tensor(metrics["division_rate"], device=self.device)
            scaler.scale(loss).backward()
            torch.nn.utils.clip_grad_norm_(agent.parameters(), max_norm=1.0)  # gradient clipping
            scaler.step(optim.adam) # Changed to optim.adam for TriOptimXAMP. Fixed: This was correct, but optim is TriOptimXAMP, not Adam. Should be scaler.step(optim.adam)
            scaler.update()

    def run(self):
        for gen in range(self.generations):
            print(f"=== Generation {gen} ===")
            agent, optim = self._init_agent()
            self.agents.append(agent)
            self.optimizers.append(optim)

            # Generate data using previous agent if available
            if gen == 0:
                data = torch.randn(self.batch_size, self.backbone.input_dim).to(self.device)
            else:
                data = self._generate_data(self.agents[-2])  # generate new data from previous agent

            # Train current agent on this data
            self._train_agent_on_data(agent, optim, data)

            # Optional: save checkpoint per generation
            torch.save({
                "agent_state": agent.state_dict(),
                "optim_state": optim.state_dict(),
            }, f"agent_gen{gen}.pth")
            print(f"Generation {gen} complete, agent checkpoint saved.")

# -----------------------------
# Usage example
# -----------------------------
if __name__ == "__main__":
    # Using EfficientNet as a backbone since GHU1C_v2 definition is not readily available
    backbone = EfficientNet(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64)
    vcs = VirtualCellSystem(backbone=backbone, device="cpu", tel_len_init=100.0, tel_vec_len=1)
    meta_runner = SequentialMetaEvolution(backbone, vcs, device="cpu", generations=5)
    meta_runner.run()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from copy import deepcopy
from typing import List, Dict, Any
import random
import math

# Core activation
class REGELUPlus(nn.Module):
    def __init__(self, alpha=0.5, beta=0.5):
        super().__init__()
        self.alpha = nn.Parameter(torch.tensor(alpha))
        self.beta = nn.Parameter(torch.tensor(beta))

    def forward(self, x):
        alpha = torch.clamp(self.alpha, 0, 1)
        beta = torch.clamp(self.beta, 0, 1)
        pos = alpha * F.gelu(x) + (1-alpha) * F.relu(x)
        neg = beta * F.elu(x) + (1-beta) * F.leaky_relu(x, 0.01)
        return torch.where(x>=0, pos, neg)

# -----------------------------
# Meta HyperFusion Activation (dynamic)
# -----------------------------
class MetaHyperFusionBioAct(nn.Module):
    """Per-channel learnable mix of ReLU, GELU, SiLU, ELU, Swish with scaling."""
    def __init__(self, features=None):
        super().__init__()
        self.features = features
        if features is None:
            self.raw = nn.Parameter(torch.randn(5))
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(features, 5))
            self.scale = nn.Parameter(torch.ones(features))

    def forward(self, x):
        if self.features is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            return r*w[0] + g*w[1] + l*w[2] + e*w[3] + sw*w[4]
        else:
            w = torch.softmax(self.raw, dim=-1)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,)*(x.dim()-2)), -1)
            x_s = x * s
            r = F.relu(x_s); g = F.gelu(x_s); l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s); sw = x_s*torch.sigmoid(x_s)
            out = r*w[...,0] + g*w[...,1] + l*w[...,2] + e*w[...,3] + sw*w[...,4]
            return out

# -----------------------------
# Efficient Block (attention-free)
# -----------------------------
class EfficientBlock(nn.Module):
    def __init__(self, dim, tokens, hidden_dim=None, num_experts=4):
        super().__init__()
        hidden_dim = hidden_dim or dim*2
        self.norm = nn.LayerNorm(dim)
        self.token_mixer = nn.Sequential(
            nn.Linear(tokens, tokens*2),
            MetaHyperFusionBioAct(features=tokens*2),
            nn.Linear(tokens*2, tokens)
        )
        self.channel_mixer = nn.Sequential(
            nn.Linear(dim, hidden_dim*2),
            nn.GLU(dim=-1),
            MetaHyperFusionBioAct(features=hidden_dim),
            nn.Linear(hidden_dim, dim)
        )
        self.experts = nn.ModuleList([nn.Linear(dim, dim) for _ in range(num_experts)])
        self.gate = nn.Linear(dim, num_experts)

    def forward(self, x):
        y = x.transpose(1,2)
        y = y + self.token_mixer(y)
        y = y.transpose(1,2)
        y = y + self.channel_mixer(y)

        gate_w = torch.softmax(self.gate(y), dim=-1)
        expert_outs = torch.stack([F.relu(e(y)) for e in self.experts], dim=-1)
        y = (expert_outs * gate_w.unsqueeze(2)).sum(dim=-1)
        return y

# -----------------------------
# Efficient Backbone
# -----------------------------
class EfficientNet(nn.Module):
    def __init__(self, input_dim=2, dim=256, tokens=32, depth=12, out_dim=1):
        super().__init__()
        self.input_dim = input_dim
        self.embed = nn.Linear(input_dim, dim)
        self.pos_embed = nn.Parameter(torch.randn(1, tokens, dim)*0.02)
        self.blocks = nn.ModuleList([EfficientBlock(dim, tokens) for _ in range(depth)])
        self.norm = nn.LayerNorm(dim)
        self.head = nn.Linear(dim, out_dim)

    def forward(self, x):
        if x.dim()==2:
            x = x.unsqueeze(1).repeat(1, self.pos_embed.size(1), 1)
        x = self.embed(x) + self.pos_embed
        for blk in self.blocks:
            x = blk(x)
        x = self.norm(x)
        x = x.mean(dim=1)
        return self.head(x)

# -----------------------------
# Virtual Cell System (Telomeres + Telomerase + Phagocytosis)
# -----------------------------
class TelomereVirtual(nn.Module):
    def __init__(self, init_length=120.0, vec_len=1, degrade_scale=0.5, repair_scale=1.0):
        super().__init__()
        self.register_buffer("_base_length", torch.full((vec_len,), init_length))
        self.vec_len = vec_len
        self.degrade_scale = degrade_scale
        self.repair_scale = repair_scale

    def initial_state(self, batch=1, device=None):
        dev = device or self._base_length.device
        return self._base_length.unsqueeze(0).repeat(batch, 1).to(dev)

    def degrade(self, tel, rnd_noise=0.1):
        noise = torch.randn_like(tel) * rnd_noise * self.degrade_scale
        return torch.clamp(tel - torch.abs(noise) - (0.1 * self.degrade_scale), 0.0)

    def repair(self, tel, strength):
        return tel + self.repair_scale*strength

class VirtualCellSystem(nn.Module):
    def __init__(self, backbone, tel_len_init=120.0, tel_vec_len=1, device=None):
        super().__init__()
        self.device = device or torch.device("cpu")
        self.backbone = backbone
        self.tel = TelomereVirtual(init_length=tel_len_init, vec_len=tel_vec_len)
        # Dummy modules for now to satisfy param_groups in MetaHybridOptimizer
        # In a full implementation, these would be learnable components of the VCS
        self.tvase = nn.Linear(1,1) # Telomerase control analog
        self.fago = nn.Linear(1,1)  # Phagocytosis control analog


    def initialize_telomeres(self, batch=1):
        return self.tel.initial_state(batch, self.device)

    def division_step(self, x, tel_state: torch.Tensor, policy: dict, agent_action_output: torch.Tensor = None):
        # Ensure tel_state explicitly requires gradients for this step if not already
        if not tel_state.requires_grad:
            tel_state = tel_state.clone().detach().requires_grad_(True)

        # Apply telomere degradation
        tel_state = self.tel.degrade(tel_state, rnd_noise=policy["noise_scale"])

        # Determine repair strength. If agent_action_output is provided, use it.
        if agent_action_output is not None:
            # Use the agent's action output to derive repair strength
            # Assuming the first component of the action is relevant for telomere repair
            raw_strength_signal = agent_action_output[:, :self.tel.vec_len]
            # Process through tvase (Telomerase Control Analog) and scale
            strength = self.tvase(raw_strength_signal.mean(dim=1, keepdim=True)).sigmoid() # Output between 0 and 1
            strength = strength * policy["repair_activity_thresh"] # Scale it
        else:
            # Fallback for _generate_data or if not using agent signal for repair
            strength = torch.tensor(torch.rand(1).item(), device=self.device) * policy["repair_activity_thresh"]

        tel_state = self.tel.repair(tel_state, strength=strength)

        division_rate = 0.05 if tel_state.mean() > policy["division_prob_thresh"] else 0.0 # Placeholder

        # Simulate telomere cleaning/repair based on policy and activity
        cleaned_x = x # For now, no actual cleaning happens here, x remains unchanged

        metrics = {
            "tel_mean_after": tel_state.mean().item(),
            "activity_mean": strength.mean().item() if isinstance(strength, torch.Tensor) else strength,
            "division_rate": division_rate,
        }

        return cleaned_x, tel_state, metrics

# -----------------------------
# TriOptimXAMP: triple optimizer with AMP + gradient clipping
# -----------------------------
class TriOptimXAMP:
    def __init__(self, params, base_lr=3e-4, wd=1e-2, max_grad_norm=1.0, amp=True):
        self.params = list(params)
        self.adam = torch.optim.AdamW(self.params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(self.params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(self.params, lr=base_lr*0.5, momentum=0.9)
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp)
        self.max_grad_norm = max_grad_norm

    def zero_grad(self):
        self.adam.zero_grad(); self.sgd.zero_grad(); self.sgd_m.zero_grad()

    def step(self):
        torch.nn.utils.clip_grad_norm_(self.params, self.max_grad_norm)
        self.adam.step(); self.sgd.step(); self.sgd_m.step()

    def state_dict(self):
        """Returns the state of the optimizers as a dict."""
        return {
            "adam": self.adam.state_dict(),
            "sgd": self.sgd.state_dict(),
            "sgd_m": self.sgd_m.state_dict(),
            "scaler": self.scaler.state_dict(),
        }

    def load_state_dict(self, state_dict):
        """Loads the optimizers state."""
        self.adam.load_state_dict(state_dict["adam"])
        self.sgd.load_state_dict(state_dict["sgd"])
        self.sgd_m.load_state_dict(state_dict["sgd_m"])
        self.scaler.load_state_dict(state_dict["scaler"])

# -----------------------------
# Autonomous Agent (Self-Improving)
# -----------------------------
class AutonomousAgent(nn.Module):
    def __init__(self, backbone, vcs, hidden=256, action_dim=8):
        super().__init__()
        self.backbone = backbone
        self.vcs = vcs
        self.encoder = nn.Sequential(nn.Linear(backbone.head.out_features, hidden),
                                     MetaHyperFusionBioAct(hidden),
                                     nn.Linear(hidden, hidden))
        self.policy = nn.Sequential(nn.Linear(hidden, hidden),
                                    MetaHyperFusionBioAct(hidden),
                                    nn.Linear(hidden, action_dim))
        self.value = nn.Sequential(nn.Linear(hidden, hidden),
                                   MetaHyperFusionBioAct(hidden),
                                   nn.Linear(hidden, 1))

    def observe(self, x):
        emb = self.backbone(x)
        if emb.dim()>2: emb = emb.mean(dim=1)
        return self.encoder(emb)

    def act(self, z):
        mu = self.policy(z)
        noise = torch.randn_like(mu)*0.1
        return torch.tanh(mu + noise)

    def evaluate(self, z, action):
        v = self.value(z).squeeze(-1)
        adv = (action*z).mean(dim=-1)
        return v, v + 0.01*adv


# -----------------------------
# Meta-Agent: evolves hyperparams & architecture
# -----------------------------
class MetaAutonomousAgent(AutonomousAgent):
    """
    Extends the standard AutonomousAgent to allow:
    - per-generation hyperparameter tuning (lr, hidden size, activation scaling)
    - optional dynamic architecture (depth, hidden dim)
    """
    def __init__(self, backbone: nn.Module, vcs: nn.Module, action_dim: int = 8, hidden: int = 256, act_scale: float = 1.0):
        super().__init__(backbone, vcs, action_dim, hidden)
        self.act_scale = nn.Parameter(torch.tensor(act_scale))  # learnable activation scaling

    def forward_with_scaled_activation(self, x):
        z = self.observe(x)
        scaled_z = z * torch.clamp(self.act_scale, 0.5, 2.0)
        return self.act(scaled_z)

    def mutate_hyperparams(self):
        """
        Randomly mutate hidden size and activation scaling
        """
        # small Gaussian perturbation
        with torch.no_grad():
            self.act_scale += torch.randn_like(self.act_scale) * 0.05
        # hidden size mutation not implemented directly (requires re-init), just a placeholder
        # could implement dynamic layer pruning/expansion in future

# -----------------------------
# Sequential Meta-Evolution with Hyperparameter Evolution
# -----------------------------
class SequentialMetaEvolutionV2:
    """
    Sequential self-improving agents with:
    - per-generation hyperparameter evolution
    - dynamic data generation
    - meta-optimizer with gradient clipping & AMP
    """
    def __init__(self,
                 backbone: nn.Module,
                 vcs: nn.Module,
                 action_dim: int = 8,
                 device: str = "cpu",
                 generations: int = 5,
                 steps_per_gen: int = 100,
                 batch_size: int = 16):
        self.device = torch.device(device)
        self.backbone = backbone.to(self.device)
        self.vcs = vcs.to(self.device)
        self.action_dim = action_dim
        self.generations = generations
        self.steps_per_gen = steps_per_gen
        self.batch_size = batch_size

        self.agents: List[MetaAutonomousAgent] = []
        self.optimizers: List[Any] = []

    def _init_agent(self):
        agent = MetaAutonomousAgent(backbone=self.backbone, vcs=self.vcs, action_dim=self.action_dim, hidden=256, act_scale=1.0).to(self.device)
        optim = TriOptimXAMP(agent.parameters(), base_lr=3e-4, wd=1e-2)
        return agent, optim

    def _generate_data(self, agent: MetaAutonomousAgent) -> torch.Tensor:
        x = torch.randn(self.batch_size, self.backbone.input_dim).to(self.device)
        tel = self.vcs.initialize_telomeres(batch=self.batch_size).to(self.device)
        for step in range(self.steps_per_gen):
            z = agent.observe(x)
            action = agent.act(z)
            x = x + 0.01 * action[:, :self.backbone.input_dim]
            cleaned, tel, _ = self.vcs.division_step(x, tel, policy={
                "min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                "division_cost":0.5, "noise_scale":0.03,
                "division_prob_thresh":0.5, "repair_activity_thresh":0.2
            })
        return x.detach()

    def _train_agent_on_data(self, agent: MetaAutonomousAgent, optim: Any, data: torch.Tensor):
        scaler = torch.cuda.amp.GradScaler(enabled=(self.device.type=="cuda"))
        for step in range(self.steps_per_gen):
            optim.zero_grad()
            with torch.cuda.amp.autocast(enabled=(self.device.type=="cuda")):
                z = agent.observe(data)
                action = agent.act(z)
                cleaned, tel, metrics = self.vcs.division_step(data, self.vcs.initialize_telomeres(batch=data.shape[0]), policy={
                    "min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                    "division_cost":0.5, "noise_scale":0.03,
                    "division_prob_thresh":0.5, "repair_activity_thresh":0.2
                })
                target_tel = torch.tensor([100.0], device=self.device)
                loss_tel = F.mse_loss(tel.mean().unsqueeze(0), target_tel)
                loss = loss_tel + 0.1 * torch.tensor(metrics["division_rate"], device=self.device)
            scaler.scale(loss).backward()
            torch.nn.utils.clip_grad_norm_(agent.parameters(), max_norm=1.0)
            scaler.step(optim.adam)
            scaler.update()

    def _evolve_agent(self, agent: MetaAutonomousAgent):
        """
        Mutate hyperparameters slightly for next generation.
        """
        agent.mutate_hyperparams()

    def run(self):
        for gen in range(self.generations):
            print(f"=== Generation {gen} ===")
            agent, optim = self._init_agent()
            if gen > 0:
                # inherit previous generation weights
                agent.load_state_dict(deepcopy(self.agents[-1].state_dict()))
                # evolve hyperparameters
                self._evolve_agent(agent)
            self.agents.append(agent)
            self.optimizers.append(optim)

            # Generate data
            if gen == 0:
                data = torch.randn(self.batch_size, self.backbone.input_dim).to(self.device)
            else:
                data = self._generate_data(self.agents[-2])

            # Train agent
            self._train_agent_on_data(agent, optim, data)

            # Save checkpoint
            torch.save({
                "agent_state": agent.state_dict(),
                "optim_state": optim.state_dict(),
            }, f"meta_agent_gen{gen}.pth")
            print(f"Generation {gen} complete, checkpoint saved.")

# -----------------------------
# Usage
# -----------------------------
if __name__ == "__main__":
    backbone = EfficientNet(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64)
    vcs = VirtualCellSystem(backbone=backbone, device="cpu", tel_len_init=100.0, tel_vec_len=1)
    meta_runner = SequentialMetaEvolutionV2(backbone, vcs, device="cpu", generations=5)
    meta_runner.run()
```

```python
class VirtualTelomereCell(nn.Module):
    def __init__(self, tel_len=120, seq_dim=16, repair_scale=1.0):
        super().__init__()
        # Learnable telomere sequence embedding (analog of digital genome)
        self.tel_sequence = nn.Parameter(torch.randn(seq_dim))
        self.tel_length = tel_len
        self.repair_scale = repair_scale

    def divide(self, telomerase_activity):
        # Digital division: degrade telomeres + optional repair
        noise = torch.randn_like(self.tel_sequence) * 0.05
        self.tel_sequence = self.tel_sequence - torch.abs(noise)
        self.tel_sequence += telomerase_activity * self.repair_scale
        self.tel_sequence = torch.clamp(self.tel_sequence, 0.0, 10.0)
        return self.tel_sequence

class AGI_VirtualCellAgent(nn.Module):
    def __init__(self, input_dim, hidden_dim=128, action_dim=16):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            REGELUPlus(),
            nn.Linear(hidden_dim, hidden_dim)
        )
        self.policy = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim),
            REGELUPlus(),
            nn.Linear(hidden_dim, action_dim)  # proposes telomerase + repair signals
        )

    def forward(self, obs):
        z = self.encoder(obs)
        return torch.tanh(self.policy(z))

class AGI_MetaSystem:
    def __init__(self, num_cells=16, device='cpu'):
        self.device = device
        self.cells = [VirtualTelomereCell().to(device) for _ in range(num_cells)]
        self.agent = AGI_VirtualCellAgent(input_dim=16).to(device)
        self.optimizer = TriOptimXAMP(self.agent.parameters())

    def step(self):
        # Observe all cells
        obs = torch.stack([c.tel_sequence for c in self.cells])
        actions = self.agent(obs)
        # Apply actions and update cell states
        for i, c in enumerate(self.cells):
            c.divide(actions[i])
        # Compute self-supervised reward
        reward = sum([c.tel_sequence.mean().item() for c in self.cells])
        # Optimize agent to improve telomere longevity
        self.optimizer.zero_grad()
        loss = -torch.tensor(reward, requires_grad=True)
        loss.backward()
        self.optimizer.step()
        return reward
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from typing import List, Dict, Any

# Core activation
class REGELUPlus(nn.Module):
    def __init__(self, alpha=0.5, beta=0.5):
        super().__init__()
        self.alpha = nn.Parameter(torch.tensor(alpha))
        self.beta = nn.Parameter(torch.tensor(beta))

    def forward(self, x):
        alpha = torch.clamp(self.alpha, 0, 1)
        beta = torch.clamp(self.beta, 0, 1)
        pos = alpha * F.gelu(x) + (1-alpha) * F.relu(x)
        neg = beta * F.elu(x) + (1-beta) * F.leaky_relu(x, 0.01)
        return torch.where(x>=0, pos, neg)


# Virtual Cell
class VirtualTelomereCell(nn.Module):
    def __init__(self, seq_dim=16, tel_len_init=120.0, repair_scale=1.0):
        super().__init__()
        self.tel_sequence = nn.Parameter(torch.randn(seq_dim) * 0.1)
        self.tel_length = tel_len_init
        self.repair_scale = repair_scale

    def divide(self, telomerase_activity: torch.Tensor):
        noise = torch.randn_like(self.tel_sequence) * 0.05
        self.tel_sequence.data -= torch.abs(noise)
        self.tel_sequence.data += telomerase_activity * self.repair_scale
        self.tel_sequence.data = torch.clamp(self.tel_sequence.data, 0.0, 10.0)
        return self.tel_sequence
```

```python
class VirtualOOC:
    """
    Multi-cell organ-on-a-chip simulator.
    Each cell has telomeres, can divide, interact, and be 'phagocytosed'.
    The AGI agent observes the system and proposes repair/division strategies.
    """

    def __init__(self, num_cells=16, seq_dim=16, device='cpu', phago_shrink=0.3):
        self.device = device
        self.cells = [VirtualTelomereCell(seq_dim=seq_dim).to(device) for _ in range(num_cells)]
        self.phagocytosis_factor = phago_shrink

    def observe(self) -> torch.Tensor:
        # Observe all cell telomere sequences
        return torch.stack([c.tel_sequence for c in self.cells])

    def phagocytose(self):
        # Remove weak cells or shrink their telomeres
        for c in self.cells:
            shrinkage = torch.rand_like(c.tel_sequence) * self.phagocytosis_factor
            c.tel_sequence.data -= shrinkage
            c.tel_sequence.data = torch.clamp(c.tel_sequence.data, 0.0, 10.0)

    def step(self, actions: torch.Tensor):
        # actions: proposed telomerase repair per cell
        rewards = []
        for i, c in enumerate(self.cells):
            tel_before = c.tel_sequence.clone()
            c.divide(actions[i])
            rewards.append(float(c.tel_sequence.mean() - tel_before.mean()))
        # Random phagocytosis to simulate immune pressure
        self.phagocytose()
        return sum(rewards) / len(rewards)  # mean reward
```

```python
class AGI_OOC_Agent(nn.Module):
    """
    Observes multi-cell environment and proposes repair/division actions
    to maximize telomere longevity and stability.
    """
    def __init__(self, seq_dim=16, hidden_dim=128):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(seq_dim, hidden_dim),
            REGELUPlus(),
            nn.Linear(hidden_dim, hidden_dim),
            REGELUPlus()
        )
        self.policy = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim),
            REGELUPlus(),
            nn.Linear(hidden_dim, seq_dim)
        )

    def forward(self, obs: torch.Tensor) -> torch.Tensor:
        z = self.encoder(obs)
        return torch.tanh(self.policy(z))
```

```python
class AGI_OOC_System:
    """
    Closed-loop self-improving system: cells + OOC + AGI agent.
    Uses TriOptimX for hybrid optimization.
    """

    def __init__(self, num_cells=16, seq_dim=16, device='cpu'):
        self.device = device
        self.ooc = VirtualOOC(num_cells=num_cells, seq_dim=seq_dim, device=device)
        self.agent = AGI_OOC_Agent(seq_dim=seq_dim).to(device)
        self.optimizer = TriOptimX(self.agent.parameters())
        self.seq_dim = seq_dim

    def step(self):
        obs = self.ooc.observe().to(self.device)
        actions = self.agent(obs)
        reward = self.ooc.step(actions)
        # Simple loss: negative reward
        self.optimizer.zero_grad()
        loss = -torch.tensor(reward, device=self.device, requires_grad=True)
        loss.backward()
        self.optimizer.step()
        return reward

    def run(self, steps=100):
        for step in range(steps):
            r = self.step()
            if step % 10 == 0:
                print(f"[Step {step}] Mean telomere stability reward: {r:.4f}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

# -----------------------------
# Hybrid Super Activation: HyperBioAct
# -----------------------------
class HyperBioAct(nn.Module):
    """
    Combines multiple advanced activations with learnable per-channel mixing,
    optional momentum-like smoothing, and scaling.
    Includes: GELU, ReLU, ELU, LeakyReLU, Swish, Mish, Tanh, Sigmoid.
    """

    def __init__(self, channels: int = None, momentum: float = 0.9):
        super().__init__()
        self.channels = channels
        self.momentum = momentum
        if channels is None:
            self.raw = nn.Parameter(torch.randn(8))  # global mixing weights
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(channels, 8))
            self.scale = nn.Parameter(torch.ones(channels))
        # buffer for momentum smoothing
        self.register_buffer("prev_out", None)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x shape: (..., C) or (B, C)
        if self.channels is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            acts = torch.stack([
                F.relu(x_s),
                F.gelu(x_s),
                F.leaky_relu(x_s, 0.02),
                F.elu(x_s),
                x_s * torch.sigmoid(x_s),   # Swish / SiLU
                x_s * torch.tanh(F.softplus(x_s)), # Mish approximation
                torch.tanh(x_s),
                torch.sigmoid(x_s)
            ], dim=-1)
            out = (acts * w).sum(dim=-1)
        else:
            # per-channel
            w = torch.softmax(self.raw, dim=-1)  # (C, 8)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,) * (x.dim() - 2)), -1)
            x_s = x * s
            acts = torch.stack([
                F.relu(x_s),
                F.gelu(x_s),
                F.leaky_relu(x_s, 0.02),
                F.elu(x_s),
                x_s * torch.sigmoid(x_s),
                x_s * torch.tanh(F.softplus(x_s)),
                torch.tanh(x_s),
                torch.sigmoid(x_s)
            ], dim=-1)
            out = (acts * w).sum(dim=-1)

        # momentum smoothing
        if self.prev_out is None:
            self.prev_out = out.detach()
        else:
            self.prev_out = self.momentum * self.prev_out + (1.0 - self.momentum) * out.detach()
        out = 0.5 * out + 0.5 * self.prev_out  # combine instant + momentum

        return out
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MetaHyperFusionBioAct(nn.Module):
    """
    Adaptive multi-activation with per-neuron mixing, scaling, and momentum smoothing.
    Self-optimizing during training.
    """
    def __init__(self, features: int, momentum: float = 0.9, prune_threshold: float = 0.05):
        super().__init__()
        self.features = features
        self.momentum = momentum
        self.prune_threshold = prune_threshold

        # Learnable softmax weights per neuron for 8 activations
        self.raw_weights = nn.Parameter(torch.randn(features, 8))
        # Learnable per-neuron scaling
        self.scale = nn.Parameter(torch.ones(features))
        # Momentum buffer
        self.register_buffer("momentum_out", torch.zeros(1))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x: (B, ..., features)
        shape_orig = x.shape
        x_flat = x.view(-1, self.features)  # flatten leading dims

        # Apply per-neuron scale
        scaled_x = x_flat * torch.clamp(self.scale, 0.1, 10.0)

        # Compute activations
        acts = torch.stack([
            F.relu(scaled_x),
            F.gelu(scaled_x),
            F.leaky_relu(scaled_x, 0.02),
            F.elu(scaled_x),
            scaled_x * torch.sigmoid(scaled_x),            # Swish
            scaled_x * torch.tanh(F.softplus(scaled_x)),   # Mish approximation
            torch.tanh(scaled_x),
            torch.sigmoid(scaled_x)
        ], dim=-1)  # shape: (B*..., features, 8)

        # Softmax weights per neuron
        w = F.softmax(self.raw_weights, dim=-1).unsqueeze(0)  # (1, features, 8)
        out = (acts * w).sum(dim=-1)  # weighted sum over activations

        # Momentum smoothing
        if self.momentum_out.numel() == 1:
            self.momentum_out = out.detach()
        else:
            self.momentum_out = self.momentum * self.momentum_out + (1 - self.momentum) * out.detach()
        out = 0.5 * out + 0.5 * self.momentum_out

        # Pruning very small activation contributions
        w_mask = (w > self.prune_threshold).float()
        out = (acts * w_mask).sum(dim=-1)

        return out.view(*shape_orig)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# -----------------------------
# Meta HyperFusion Activation (dynamic)
# -----------------------------
class MetaHyperFusionBioAct(nn.Module):
    """Per-channel learnable mix of ReLU, GELU, SiLU, ELU, Swish with scaling."""
    def __init__(self, features=None):
        super().__init__()
        self.features = features
        if features is None:
            self.raw = nn.Parameter(torch.randn(5))
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(features, 5))
            self.scale = nn.Parameter(torch.ones(features))

    def forward(self, x):
        if self.features is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            return r*w[0] + g*w[1] + l*w[2] + e*w[3] + sw*w[4]
        else:
            w = torch.softmax(self.raw, dim=-1)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,)*(x.dim()-2)), -1)
            x_s = x * s
            r = F.relu(x_s); g = F.gelu(x_s); l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s); sw = x_s*torch.sigmoid(x_s)
            out = r*w[...,0] + g*w[...,1] + l*w[...,2] + e*w[...,3] + sw*w[...,4]
            return out

# -----------------------------
# Efficient Block (attention-free)
# -----------------------------
class EfficientBlock(nn.Module):
    def __init__(self, dim, tokens, hidden_dim=None, num_experts=4):
        super().__init__()
        hidden_dim = hidden_dim or dim*2
        self.norm = nn.LayerNorm(dim)
        self.token_mixer = nn.Sequential(
            nn.Linear(tokens, tokens*2),
            MetaHyperFusionBioAct(features=tokens*2),
            nn.Linear(tokens*2, tokens)
        )
        self.channel_mixer = nn.Sequential(
            nn.Linear(dim, hidden_dim*2),
            nn.GLU(dim=-1),
            MetaHyperFusionBioAct(features=hidden_dim),
            nn.Linear(hidden_dim, dim)
        )
        self.experts = nn.ModuleList([nn.Linear(dim, dim) for _ in range(num_experts)])
        self.gate = nn.Linear(dim, num_experts)

    def forward(self, x):
        y = x.transpose(1,2)
        y = y + self.token_mixer(y)
        y = y.transpose(1,2)
        y = y + self.channel_mixer(y)

        gate_w = torch.softmax(self.gate(y), dim=-1)
        expert_outs = torch.stack([F.relu(e(y)) for e in self.experts], dim=-1)
        y = (expert_outs * gate_w.unsqueeze(2)).sum(dim=-1)
        return y

# -----------------------------
# Efficient Backbone
# -----------------------------
class EfficientNet(nn.Module):
    def __init__(self, input_dim=2, dim=256, tokens=32, depth=12, out_dim=1):
        super().__init__()
        self.embed = nn.Linear(input_dim, dim)
        self.pos_embed = nn.Parameter(torch.randn(1, tokens, dim)*0.02)
        self.blocks = nn.ModuleList([EfficientBlock(dim, tokens) for _ in range(depth)])
        self.norm = nn.LayerNorm(dim)
        self.head = nn.Linear(dim, out_dim)

    def forward(self, x):
        if x.dim()==2:
            x = x.unsqueeze(1).repeat(1, self.pos_embed.size(1), 1)
        x = self.embed(x) + self.pos_embed
        for blk in self.blocks:
            x = blk(x)
        x = self.norm(x)
        x = x.mean(dim=1)
        return self.head(x)

# -----------------------------
# Virtual Cell System (Telomeres + Telomerase + Phagocytosis)
# -----------------------------
class TelomereVirtual(nn.Module):
    def __init__(self, init_length=120.0, vec_len=1, degrade_scale=0.5, repair_scale=1.0):
        super().__init__()
        self.register_buffer("_base_length", torch.full((vec_len,), init_length))
        self.vec_len = vec_len
        self.degrade_scale = degrade_scale
        self.repair_scale = repair_scale

    def initial_state(self, batch=1, device=None):
        dev = device or self._base_length.device
        return self._base_length.unsqueeze(0).repeat(batch, 1).to(dev)

    def degrade(self, tel, rnd_noise=0.1):
        noise = torch.randn_like(tel) * rnd_noise * self.degrade_scale
        return torch.clamp(tel - torch.abs(noise) - (0.1 * self.degrade_scale), 0.0)

    def repair(self, tel, strength):
        return tel + self.repair_scale*strength

class VirtualCellSystem(nn.Module):
    def __init__(self, backbone, tel_len_init=120.0, tel_vec_len=1, device=None):
        super().__init__()
        self.device = device or torch.device("cpu")
        self.backbone = backbone
        self.tel = TelomereVirtual(init_length=tel_len_init, vec_len=tel_vec_len)

    def initialize_telomeres(self, batch=1):
        return self.tel.initial_state(batch, self.device)

    def step(self, x, tel_state):
        emb = self.backbone(x)
        tel_state = self.tel.degrade(tel_state)
        return emb, tel_state

# -----------------------------
# TriOptimXAMP: triple optimizer with AMP + gradient clipping
# -----------------------------
class TriOptimXAMP:
    def __init__(self, params, base_lr=3e-4, wd=1e-2, max_grad_norm=1.0, amp=True):
        self.params = list(params)
        self.adam = torch.optim.AdamW(self.params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(self.params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(self.params, lr=base_lr*0.5, momentum=0.9)
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp)
        self.max_grad_norm = max_grad_norm

    def zero_grad(self):
        self.adam.zero_grad(); self.sgd.zero_grad(); self.sgd_m.zero_grad()

    def step(self):
        torch.nn.utils.clip_grad_norm_(self.params, self.max_grad_norm)
        self.adam.step(); self.sgd.step(); self.sgd_m.step()

    def state_dict(self):
        """Returns the state of the optimizers as a dict."""
        return {
            "adam": self.adam.state_dict(),
            "sgd": self.sgd.state_dict(),
            "sgd_m": self.sgd_m.state_dict(),
            "scaler": self.scaler.state_dict(),
        }

    def load_state_dict(self, state_dict):
        """Loads the optimizers state."""
        self.adam.load_state_dict(state_dict["adam"])
        self.sgd.load_state_dict(state_dict["sgd"])
        self.sgd_m.load_state_dict(state_dict["sgd_m"])
        self.scaler.load_state_dict(state_dict["scaler"])

# -----------------------------
# Autonomous Agent (Self-Improving)
# -----------------------------
class AutonomousAgent(nn.Module):
    def __init__(self, backbone, vcs, hidden=256, action_dim=8):
        super().__init__()
        self.backbone = backbone
        self.vcs = vcs
        self.encoder = nn.Sequential(nn.Linear(backbone.head.out_features, hidden),
                                     MetaHyperFusionBioAct(hidden),
                                     nn.Linear(hidden, hidden))
        self.policy = nn.Sequential(nn.Linear(hidden, hidden),
                                    MetaHyperFusionBioAct(hidden),
                                    nn.Linear(hidden, action_dim))
        self.value = nn.Sequential(nn.Linear(hidden, hidden),
                                   MetaHyperFusionBioAct(hidden),
                                   nn.Linear(hidden, 1))

    def observe(self, x):
        emb = self.backbone(x)
        if emb.dim()>2: emb = emb.mean(dim=1)
        return self.encoder(emb)

    def act(self, z):
        mu = self.policy(z)
        noise = torch.randn_like(mu)*0.1
        return torch.tanh(mu + noise)

    def evaluate(self, z, action):
        v = self.value(z).squeeze(-1)
        adv = (action*z).mean(dim=-1)
        return v, v + 0.01*adv

# -----------------------------
# Example Run Loop
# -----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
backbone = EfficientNet(input_dim=2, dim=128, tokens=8, depth=4, out_dim=64).to(device)
vcs = VirtualCellSystem(backbone, tel_len_init=80.0, tel_vec_len=1, device=device).to(device)
agent = AutonomousAgent(backbone, vcs, hidden=128, action_dim=128).to(device) # FIX: action_dim to 128
optim = TriOptimXAMP(agent.parameters(), base_lr=2e-4, wd=1e-2, max_grad_norm=1.0, amp=True)

x = torch.randn(8,2).to(device)
tel = vcs.initialize_telomeres(batch=8)

for step in range(10):
    optim.zero_grad()
    with torch.cuda.amp.autocast(enabled=True):
        z = agent.observe(x)
        action = agent.act(z)
        emb, tel = vcs.step(x + 0.01*action[:,:2], tel)
        reward = -torch.abs(tel.mean() - 80.0)
        v, q = agent.evaluate(z, action)
        loss = F.mse_loss(v, reward.expand_as(v)) - 0.5*q.mean()
    optim.scaler.scale(loss).backward()
    optim.scaler.step(optim.adam)
    optim.scaler.update()
```

```python
import torch
import math
from typing import List, Dict, Any

class MetaHybridOptimizer:
    """
    Hybrid optimizer combining multiple optimizer strategies:
    - AdamW: adaptive + weight decay
    - SGD: momentum for smoothing
    - Lion: fast convergence (optional)
    Supports AMP and per-group gradient clipping.
    """

    def __init__(self, params, lr: float = 3e-4, weight_decay: float = 1e-2,
                 clip_norm: float = 1.0, amp: bool = True):
        self.params = list(params)
        self.clip_norm = clip_norm
        self.amp = amp

        # Define internal optimizers
        self.adamw = torch.optim.AdamW(self.params, lr=lr, weight_decay=weight_decay)
        self.sgd = torch.optim.SGD(self.params, lr=lr*0.25, momentum=0.9)
        try:
            from lion_pytorch import Lion
            self.lion = Lion(self.params, lr=lr*0.5)
        except ImportError:
            self.lion = None  # optional

        # AMP GradScaler
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp)

    def zero_grad(self):
        self.adamw.zero_grad()
        self.sgd.zero_grad()
        if self.lion: self.lion.zero_grad()

    def step(self):
        # Gradient clipping per param group
        torch.nn.utils.clip_grad_norm_(self.params, self.clip_norm)

        # Step each optimizer
        self.adamw.step()
        self.sgd.step()
        if self.lion: self.lion.step()

    def backward_and_step(self, loss):
        """Shortcut for mixed-precision safe backward and step"""
        self.zero_grad()
        self.scaler.scale(loss).backward()
        self.step()
        self.scaler.update()

    def state_dict(self) -> Dict[str, Any]:
        state = {
            "adamw": self.adamw.state_dict(),
            "sgd": self.sgd.state_dict(),
            "scaler": self.scaler.state_dict(),
        }
        if self.lion:
            state["lion"] = self.lion.state_dict()
        return state

    def load_state_dict(self, state: Dict[str, Any]):
        self.adamw.load_state_dict(state.get("adamw", {}))
        self.sgd.load_state_dict(state.get("sgd", {}))
        self.scaler.load_state_dict(state.get("scaler", {}))
        if self.lion and "lion" in state:
            self.lion.load_state_dict(state["lion"])
```

```python
import torch
import torch.nn as nn
import math

class QubitTensorLayer(nn.Module):
    """
    Each virtual qubit represents a 2D complex state (alpha, beta), normalized.
    Provides a quantum-inspired expansion of feature space.
    """
    def __init__(self, num_qubits: int, embed_dim: int):
        super().__init__()
        self.num_qubits = num_qubits
        self.embed_dim = embed_dim

        # Represent qubit states as (alpha, beta) amplitudes
        # Shape: (num_qubits, 2*embed_dim) to separate real and imaginary parts
        self.qubit_params = nn.Parameter(torch.randn(num_qubits, 2*embed_dim) * 0.1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x: (B, features)
        # Expand into qubit space
        batch_size = x.shape[0]
        qubits = self.qubit_params.unsqueeze(0).repeat(batch_size, 1, 1)
        # Normalize qubit amplitudes
        alpha = qubits[..., :self.embed_dim]
        beta = qubits[..., self.embed_dim:]
        norm = torch.sqrt(alpha**2 + beta**2 + 1e-8)
        alpha, beta = alpha/norm, beta/norm
        # Combine with input features
        x_expanded = x.unsqueeze(1) * alpha + x.unsqueeze(1) * beta
        # Flatten qubit dimension into feature dimension
        return x_expanded.reshape(batch_size, -1)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class UniversalActivation(nn.Module):
    """
    Dynamically selects or mixes multiple activation functions for each layer.
    Includes: elu, gelu, mish, relu, silu/swish, tanh, sigmoid, hard_swish, etc.
    """
    def __init__(self, act_list=None):
        super().__init__()
        # Default activation functions
        self.act_list = act_list or [
            F.elu, F.gelu, F.relu, F.silu, F.leaky_relu, F.tanh, F.sigmoid,
            lambda x: F.softplus(x), lambda x: F.hardtanh(x), lambda x: F.logsigmoid(x)
        ]
        # Learnable mixture weights
        self.weights = nn.Parameter(torch.ones(len(self.act_list)) / len(self.act_list))

    def forward(self, x):
        out = sum(w * act(x) for w, act in zip(F.softmax(self.weights, dim=0), self.act_list))
        return out
```

```python
class HybridOptimizer:
    """
    Combines multiple optimizers and selects best updates dynamically.
    Includes: SGD, Adam, AdamW, Adadelta, Adafactor, RMSprop, Lion, Nadam...
    """
    def __init__(self, params, lr=1e-3):
        params = list(params) # Convert iterator to list to allow multiple consumption
        self.optimizers = [
            torch.optim.SGD(params, lr=lr, momentum=0.9),
            torch.optim.Adam(params, lr=lr),
            torch.optim.AdamW(params, lr=lr),
            torch.optim.Adadelta(params, lr=lr),
            torch.optim.RMSprop(params, lr=lr),
        ]
        self.weights = torch.ones(len(self.optimizers)) / len(self.optimizers)

    def zero_grad(self):
        for opt in self.optimizers:
            opt.zero_grad()

    def step(self):
        # Compute weighted parameter update
        for w, opt in zip(F.softmax(self.weights, dim=0), self.optimizers):
            if w > 0.01:  # only apply significant weights
                opt.step()
                opt.zero_grad()

    def state_dict(self):
        return {
            f"optimizer_{i}": opt.state_dict() for i, opt in enumerate(self.optimizers)
        }

    def load_state_dict(self, state_dict):
        for i, opt in enumerate(self.optimizers):
            opt.load_state_dict(state_dict[f"optimizer_{i}"])
```

```python
class MultiLoss:
    """
    Dynamically selects or mixes multiple losses for self-optimization.
    Supports MSE, MAE, Huber, Dice, CrossEntropy, Focal Loss, KL divergence, etc.
    """
    def __init__(self):
        self.loss_fns = [
            nn.MSELoss(), nn.L1Loss(), nn.CrossEntropyLoss(),
            nn.KLDivLoss(), nn.HuberLoss(), nn.BCEWithLogitsLoss()
        ]
        self.weights = nn.Parameter(torch.ones(len(self.loss_fns)) / len(self.loss_fns))

    def __call__(self, y_pred, y_true):
        total_loss = sum(w * loss_fn(y_pred, y_true)
                         for w, loss_fn in zip(F.softmax(self.weights, dim=0), self.loss_fns))
        return total_loss
```

```python
class QuantumVirtualCellAgent(nn.Module):
    def __init__(self, input_dim, hidden_dim=512, qubits=64, genome_dim=128):
        super().__init__()
        self.input_layer = nn.Linear(input_dim, hidden_dim)
        self.act = UniversalActivation()
        self.qubit_layer = QubitTensorLayer(num_qubits=qubits, embed_dim=hidden_dim)
        self.hidden = nn.Linear(hidden_dim*qubits, hidden_dim)
        self.genome_out = nn.Linear(hidden_dim, genome_dim)  # virtual telomerase/genome
        self.output_layer = nn.Linear(genome_dim, input_dim)  # reconstruct or predict next state
        self.optimizer_engine = None  # will assign HybridOptimizer externally
        self.loss_engine = MultiLoss()

    def forward(self, x):
        x = self.act(self.input_layer(x))
        x = self.qubit_layer(x)
        x = self.act(self.hidden(x))
        genome_features = self.genome_out(x)
        out = self.output_layer(genome_features)
        return out, genome_features
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SuperCellAGI(nn.Module):
    def __init__(self, input_dim, hidden_dim=512, genome_dim=256, qubits=64):
        super().__init__()
        # Input encoding: cell state + microenvironment
        self.input_layer = nn.Linear(input_dim, hidden_dim)

        # Learnable universal activation combining ReLU, Swish, GELU, Mish
        self.act = UniversalActivation()

        # High-dimensional exploration layer (quantum-inspired)
        self.qubit_layer = QubitTensorLayer(num_qubits=qubits, embed_dim=hidden_dim)

        # Hidden layer
        self.hidden = nn.Linear(hidden_dim * qubits, hidden_dim)

        # Genome output: telomerase, stem cell, super-cell design
        self.genome_out = nn.Linear(hidden_dim, genome_dim)

        # Phenotype: observable cell behavior
        self.output_layer = nn.Linear(genome_dim, input_dim)

        # Optimizer & loss engines
        self.optimizer_engine = HybridOptimizer(self.parameters())
        self.loss_engine = MultiLoss()

        # Memory of synthetic cells and organ-on-a-chip states
        self.synthetic_cells = []
        self.ooc_memory = []

    def forward(self, x):
        x = self.act(self.input_layer(x))
        x = self.qubit_layer(x)
        x = self.act(self.hidden(x))
        genome_features = self.genome_out(x)
        out = self.output_layer(genome_features)
        return out, genome_features
```

```python
    def generate_super_cells(self, n_cells=32, noise=0.05):
        """Autonomously generate new super-cell generations."""
        if len(self.synthetic_cells) == 0:
            x = torch.randn(n_cells, self.input_layer.in_features)
        else:
            x = torch.stack(self.synthetic_cells[-n_cells:])
        y_pred, genome = self.forward(x)

        # Introduce random telomere mutations for exploration
        genome_mutated = genome + noise * torch.randn_like(genome)
        y_pred = self.output_layer(genome_mutated)

        self.synthetic_cells.append(y_pred.detach())
        return y_pred, genome_mutated
```

```python
    def simulate_ooc(self, ooc_input):
        """Virtual organ-on-a-chip simulation with repair & phagocytosis."""
        pred, genome = self.forward(ooc_input)

        # Split genome for telomerase, stem cell, super-cell control
        telomerase = torch.sigmoid(genome[:, :genome.shape[1]//3])
        stem_cells = torch.sigmoid(genome[:, genome.shape[1]//3:2*genome.shape[1]//3])
        super_cell_control = torch.sigmoid(genome[:, 2*genome.shape[1]//3:])

        ooc_state = pred * (telomerase + stem_cells + super_cell_control)
        self.ooc_memory.append(ooc_state.detach())
        return ooc_state
```

```python
    def self_evolve(self, x, iterations=1):
        """Autonomously optimize and evolve super-cell generations."""
        for _ in range(iterations):
            y_pred, genome = self.forward(x)

            # Multi-loss function: stability, proliferation, phagocytosis efficiency
            loss = self.loss_engine(y_pred, x)
            loss.backward()

            # Hybrid optimizer: combines Adam, AdamW, RMSprop dynamically
            self.optimizer_engine.step()
            self.optimizer_engine.zero_grad()

            # Optionally mutate genome for exploration
            with torch.no_grad():
                genome += 0.01 * torch.randn_like(genome)
```

```python
# Example set of real telomere repeat motifs:
REAL_TELOMERE_MOTIFS = [
    "TTAGGG", "TGAGGG", "TTGGGG", "TCAGGG"  # common human telomere motif variants
]

# One‑hot encode motifs for model input:
def encode_telomere_motifs(motifs):
    """
    Simple one‑hot encoding for telomeric repeat motifs.
    This creates an empirical telomere pattern embedding.
    """
    all_bases = ['A','C','G','T']
    encodings = []
    for m in motifs:
        enc = []
        for c in m:
            base_vec = [1 if c == b else 0 for b in all_bases]
            enc.extend(base_vec)
        encodings.append(enc)
    return torch.tensor(encodings, dtype=torch.float32)

# Real telomere pattern embeddings
telomere_pattern_embeddings = encode_telomere_motifs(REAL_TELOMERE_MOTIFS)
```

```python
class EmpiricalTelomereModel(nn.Module):
    """Model empirical telomere repeat composition and length dynamics."""
    def __init__(self, pattern_embeddings):
        super().__init__()
        self.pattern_embeddings = pattern_embeddings
        self.length_predictor = nn.Linear(pattern_embeddings.size(1), 1)

    def forward(self, repeat_composition):
        """
        repeat_composition: a vector representing counts/frequencies of patterns
        """
        # Estimate telomere length (empirical)
        length_est = F.softplus(self.length_predictor(repeat_composition))
        return length_est
```

```python
TARGET_MIN = 5000    # ~5kb in humans
TARGET_MAX = 15000   # ~15kb in humans

def telomere_reward(length_est):
    # maximize being in the real human telomere range
    inside_range = torch.clip((length_est - TARGET_MIN) / (TARGET_MAX - TARGET_MIN), 0, 1)
    return inside_range.mean()
```

```python
import pandas as pd

def load_telomere_lengths(path: str):
    """
    Load a tabular file of sample IDs with telomere length measurements.
    Expected columns: `sample_id`, `telomere_length_bp`
    """
    df = pd.read_csv(path, sep=None, engine='python')  # auto‑detect CSV/TSV
    if 'telomere_length_bp' not in df.columns:
        raise ValueError("Expected a column `telomere_length_bp`")
    tel_lengths = torch.tensor(df['telomere_length_bp'].values, dtype=torch.float32)
    return df, tel_lengths
```

```python
def load_tvr_data(path: str):
    """
    Load telomeric variant repeat counts.
    Columns: sample_id, TTAGGG, TGAGGG, TTGGGG, ...
    """
    df = pd.read_csv(path, sep=None, engine='python')
    motif_cols = [c for c in df.columns if c not in ['sample_id']]
    data = torch.tensor(df[motif_cols].values, dtype=torch.float32)
    # Normalize to frequencies:
    data = data / (data.sum(dim=1, keepdim=True) + 1e-8)
    return df['sample_id'].tolist(), data, motif_cols
```

```python
class EmpiricalTelomereLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.mse = nn.MSELoss()

    def forward(self, predicted_length, true_length):
        # encourage telomere predictions to match observed data
        return self.mse(predicted_length, true_length)
```

```python
class DataAugmentedSuperCellAGI(SuperCellAGI):
    def forward(self, x, tel_repeat_freq=None, expr_data=None):
        # x: original features (e.g., cell state)
        # tel_repeat_freq: real telomeric variant frequencies
        # expr_data: gene expression (TERT, shelterin, other markers)
        features = [x]
        if tel_repeat_freq is not None:
            features.append(tel_repeat_freq)
        if expr_data is not None:
            features.append(expr_data)

        # concatenate real biological features
        x_combined = torch.cat(features, dim=-1)

        # rest of forward pass
        out, genome_features = super().forward(x_combined)
        return out, genome_features
```
