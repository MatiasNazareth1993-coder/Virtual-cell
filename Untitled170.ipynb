```python
class CellDrugDynamics(nn.Module):
    def __init__(self, cell_dim=128, drug_dim=128, hidden=256):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(cell_dim + drug_dim, hidden),
            nn.GELU(),
            nn.Linear(hidden, hidden),
            nn.SiLU(),
            nn.Linear(hidden, cell_dim)
        )

    def forward(self, cell_state, drug_emb):
        x = torch.cat([cell_state, drug_emb], dim=-1)
        return self.net(x)
```

```python
def clinical_phase_score(cell_outcome):
    # cell_outcome: viability, arrest, apoptosis trajectories
    safety  = 1 - np.mean(cell_outcome["apoptosis"])
    efficacy = np.mean(cell_outcome["arrest"])
    stability = np.mean(cell_outcome["viability"])

    # Phase score is a weighted sum
    return 0.2*safety + 0.4*efficacy + 0.4*stability
```

```python
import tensorflow as tf
from tensorflow.keras import layers

class QuantumInspiredMix(layers.Layer):
    """
    Complex-valued mixing with phase encoding
    (quantum-inspired, runs on classical hardware)
    """
    def __init__(self, units):
        super().__init__()
        self.units = units

    def build(self, input_shape):
        self.Wr = self.add_weight(shape=(input_shape[-1], self.units),
                                  initializer="glorot_uniform")
        self.Wi = self.add_weight(shape=(input_shape[-1], self.units),
                                  initializer="glorot_uniform")

    def call(self, x):
        real = tf.matmul(x, self.Wr)
        imag = tf.matmul(x, self.Wi)
        magnitude = tf.sqrt(real**2 + imag**2 + 1e-8)
        phase = tf.math.atan2(imag, real)
        # entanglement-like interaction
        return magnitude * tf.math.cos(phase)
```

```python
def build_multimodal_world_model(
    graph_dim=256, sensor_dim=64, spectral_dim=2048,
    chem_dim=2048, dt_embed_dim=128, latent_dim=512
):
    # Inputs
    graph_in   = layers.Input((graph_dim,), name="cell_graph")
    sensor_in  = layers.Input((None, sensor_dim), name="ooc_sensors")
    spectral_in= layers.Input((None, spectral_dim), name="spectral")
    chem_in    = layers.Input((chem_dim,), name="chem_fp")
    dti_in     = layers.Input((dt_embed_dim,), name="drug_target")

    # Encode temporal signals
    s_enc = layers.Bidirectional(layers.LSTM(128, return_sequences=True))(sensor_in)
    sp_enc= layers.Bidirectional(layers.LSTM(256))(spectral_in)

    # Merge static knowledge
    static = layers.Concatenate()([graph_in, chem_in, dti_in])

    # Quantum-inspired fusion
    fused = layers.Concatenate()([s_enc[:,-1], sp_enc, static])
    fused = QuantumInspiredMix(1024)(fused)

    # Latent world state
    z = layers.Dense(latent_dim, activation="gelu")(fused)

    # Generative heads (new data)
    next_cell_state = layers.Dense(graph_dim, activation="tanh", name="cell_state_gen")(z)
    spectrum_gen    = layers.Dense(spectral_dim, activation="softplus", name="spectrum_gen")(z)
    uptake_curve    = layers.Dense(128, activation="sigmoid", name="uptake_gen")(z)

    # Uncertainty (aleatoric)
    log_var = layers.Dense(latent_dim, name="log_variance")(z)

    return tf.keras.Model(
        inputs=[graph_in, sensor_in, spectral_in, chem_in, dti_in],
        outputs=[next_cell_state, spectrum_gen, uptake_curve, log_var],
        name="Multimodal_Quantum_VirtualCell"
    )
```

```python
class MetaAdaptiveLoss(tf.keras.losses.Loss):
    def __init__(self):
        super().__init__()
        self.alpha = tf.Variable(1.0, trainable=True)
        self.beta  = tf.Variable(1.0, trainable=True)
        self.gamma = tf.Variable(1.0, trainable=True)

    def call(self, y_true, y_pred):
        mse = tf.reduce_mean((y_true - y_pred)**2)
        kl  = tf.reduce_mean(y_pred * tf.math.log(y_pred + 1e-8))
        huber = tf.reduce_mean(tf.keras.losses.huber(y_true, y_pred))
        return self.alpha*mse + self.beta*kl + self.gamma*huber
```

```python
import numpy as np

class CognitiveKernel:
    def __init__(self, world_model, constraints):
        self.world_model = world_model
        self.constraints = constraints

        # Persistent internal state
        self.belief_state = None
        self.goal_state = None
        self.history = []

    # 1. Update beliefs from new observations
    def update_beliefs(self, observations):
        self.belief_state = self.world_model.encode(observations)
        self.history.append(("belief_update", self.belief_state))

    # 2. Imagine futures (counterfactuals)
    def imagine(self, steps=5, branches=4):
        rollouts = []
        for _ in range(branches):
            latent = self.belief_state
            traj = []
            for _ in range(steps):
                latent = self.world_model.predict(latent)
                traj.append(latent)
            rollouts.append(traj)
        return rollouts

    # 3. Synthesize goals (not predefined rewards)
    def synthesize_goals(self, rollouts):
        # Example: minimize instability, maximize novelty
        scores = []
        for traj in rollouts:
            stability = self.constraints.stability_score(traj)
            novelty = self.constraints.novelty_score(traj)
            scores.append(novelty - stability)
        self.goal_state = np.argmax(scores)
        return self.goal_state

    # 4. Select next experiment/action
    def select_action(self, rollouts):
        return rollouts[self.goal_state][0]

    # 5. Meta-adjust learning objectives
    def meta_update(self):
        self.world_model.adjust_loss(self.constraints)
```

```python
class StabilityConstraints:
    def stability_score(self, trajectory):
        diffs = [np.linalg.norm(trajectory[i+1]-trajectory[i])
                 for i in range(len(trajectory)-1)]
        return np.mean(diffs)

    def novelty_score(self, trajectory):
        return np.var(trajectory)
```

```python
import pandas as pd
import torch
from rdkit import Chem
from rdkit.Chem import AllChem

def smiles_to_fp(smiles, n_bits=2048):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    fp = AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=n_bits)
    return torch.tensor(list(fp), dtype=torch.float32)

def load_drug_dataset(path="data/drugs.csv"):
    df = pd.read_csv(path)
    data = []
    for _, row in df.iterrows():
        fp = smiles_to_fp(row["smiles"])
        if fp is not None:
            data.append({
                "fp": fp,
                "activity": row["activity"],
                "toxicity": row["toxicity"]
            })
    return data
```

```python
import torch
import torch.nn as nn

class AgingVirtualCell(nn.Module):
    def __init__(self, dim=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim, 128),
            nn.GELU(),
            nn.Linear(128, 64),
            nn.GELU(),
            nn.Linear(64, 6)
        )

    def forward(self, latent):
        state = self.net(latent) # state will be (batch_size, 6)

        # biologically constrained outputs - access features (columns)
        telomere = torch.sigmoid(state[:, 0])
        proliferation = torch.sigmoid(state[:, 1])
        immune = torch.sigmoid(state[:, 2])
        dna_damage = torch.sigmoid(state[:, 3])
        senescence = torch.sigmoid(state[:, 4])
        regeneration = torch.sigmoid(state[:, 5])

        return torch.stack([
            telomere,
            proliferation,
            immune,
            dna_damage,
            senescence,
            regeneration
        ], dim=-1)
```

```python
def aging_safety_certificate(state):
    tel, prol, imm, dna, sen, regen = state

    # Core constraints
    if prol > 0.6:
        return 0.0  # cancer risk

    if tel > 0.9 and sen < 0.2:
        return 0.1  # immortalization risk

    if dna > 0.5:
        return 0.2  # genomic instability

    # Desired aging profile
    score = (
        0.4 * tel +
        0.4 * regen +
        0.2 * (1 - sen)
    )

    return float(score)
```

```python
import torch
import torch.nn as nn

class AgingLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.w_tel = nn.Parameter(torch.tensor(1.0))
        self.w_regen = nn.Parameter(torch.tensor(1.0))
        self.w_sen = nn.Parameter(torch.tensor(1.0))
        self.w_prol = nn.Parameter(torch.tensor(1.0))

    def forward(self, state):
        # state is expected to be a tensor of shape (batch_size, 6)
        tel = state[:, 0]
        prol = state[:, 1]
        imm = state[:, 2]
        dna = state[:, 3]
        sen = state[:, 4]
        regen = state[:, 5]

        loss = (
            - self.w_tel * tel.mean() + # Encourage longer telomeres
            - self.w_regen * regen.mean() + # Encourage regeneration
            + self.w_sen * sen.mean() + # Discourage senescence
            + self.w_prol * prol.mean() # Discourage proliferation
        )
        return loss.mean()
```

```python
import torch
import torch.nn as nn

class DNAEncoder(nn.Module):
    def __init__(self, vocab=5, dim=128):
        super().__init__()
        self.embed = nn.Embedding(vocab, dim)
        encoder_layer = nn.TransformerEncoderLayer(d_model=dim, nhead=4, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=4)

    def forward(self, x):
        x = self.embed(x) # (batch_size, seq_len, embed_dim)
        x = self.transformer(x) # (batch_size, seq_len, embed_dim)
        return x.mean(dim=1) # Average over sequence length -> (batch_size, embed_dim)
```

```python
class TelomereEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(4, 64),
            nn.GELU(),
            nn.Linear(64, 64)
        )

    def forward(self, metrics):
        return self.net(metrics)
```

```python
class TelemetryEncoder(nn.Module):
    def __init__(self, input_dim=6):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 64, batch_first=True)

    def forward(self, x):
        _, (h, _) = self.lstm(x)
        return h[-1]
```

```python
class MultimodalFusion(nn.Module):
    def __init__(self):
        super().__init__()
        self.fuse = nn.Sequential(
            nn.Linear(256, 128),
            nn.GELU(),
            nn.Linear(128, 64)
        )

    def forward(self, dna, telo, tele):
        combined = torch.cat([dna, telo, tele], dim=-1)
        return self.fuse(combined)
```

```python
import numpy as np

def causal_matrix(states):
    n = states.shape[1]
    mat = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            mat[i, j] = np.corrcoef(states[:, i], states[:, j])[0, 1]

    return np.nan_to_num(mat)
```

```python
import numpy as np

def phagocytosis_clearance(
    debris_level,
    macrophage_activity,
    receptor_sensitivity
):
    """
    Models immune waste removal capacity
    """
    clearance = macrophage_activity * receptor_sensitivity
    remaining = max(debris_level - clearance, 0.0)
    return remaining
```

```python
def telomere_update(
    telomere_length,
    replication_rate,
    telomerase_activity
):
    shortening = 0.02 * replication_rate
    extension = 0.03 * telomerase_activity
    new_length = telomere_length - shortening + extension
    return max(min(new_length, 1.0), 0.0)
```

```python
def cancer_attractor(state):
    """
    Rejects unstable immortal-like regimes
    """
    telomerase = state["Telomerase"]
    replication = state["Replication"]
    immune = state["Immune"]
    dna = state["DNA"]

    if telomerase > 0.85 and replication > 0.8 and immune < 0.4:
        return True
    if dna < 0.4:
        return True
    return False
```

```python
def apply_drug(state, drug):
    """
    drug = [telomerase_mod, immune_mod, replication_mod]
    """
    state["Telomerase"] *= (1 + drug[0])
    state["Immune"] *= (1 + drug[1])
    state["Replication"] *= (1 + drug[2])
    return state
```

```python
class DrugGenerator(nn.Module):
    """
    Generates drug perturbations conditioned on cell state
    """
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(8, 64),
            nn.SiLU(),
            nn.Linear(64, 64),
            nn.SiLU(),
            nn.Linear(64, 3),
            nn.Tanh()  # limits perturbation range
        )

    def forward(self, cell_state):
        return 0.2 * self.net(cell_state)  # safe drug scale
```

```python
# app/transformer_model.py
import torch
import torch.nn as nn

class DrugTransformer(nn.Module):
    def __init__(self,
                 cell_dim=10,
                 drug_dim=8,
                 d_model=128,
                 nhead=8,
                 num_layers=4):
        super().__init__()

        self.cell_embed = nn.Linear(cell_dim, d_model)
        self.drug_embed = nn.Linear(drug_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            batch_first=True
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.output = nn.Linear(d_model, drug_dim)

    def forward(self, cell_state, drug_seed):
        """
        cell_state: [B, cell_dim]
        drug_seed:  [B, drug_dim]
        """
        cell_token = self.cell_embed(cell_state).unsqueeze(1)
        drug_token = self.drug_embed(drug_seed).unsqueeze(1)

        tokens = torch.cat([cell_token, drug_token], dim=1)
        latent = self.transformer(tokens)

        return self.output(latent[:, -1])

# Write the content to app/transformer_model.py
with open('app/transformer_model.py', 'w') as f:
    f.write('''import torch
import torch.nn as nn

class DrugTransformer(nn.Module):
    def __init__(
        self, cell_dim=10, drug_dim=8, d_model=128, nhead=8, num_layers=4
    ):
        super().__init__()

        self.cell_embed = nn.Linear(cell_dim, d_model)
        self.drug_embed = nn.Linear(drug_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, batch_first=True
        )

        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        self.output = nn.Linear(d_model, drug_dim)

    def forward(self, cell_state, drug_seed):
        cell_token = self.cell_embed(cell_state).unsqueeze(1)
        drug_token = self.drug_embed(drug_seed).unsqueeze(1)

        tokens = torch.cat([cell_token, drug_token], dim=1)
        latent = self.transformer(tokens)

        return self.output(latent[:, -1])
''')
```

```python
# app/train_transformer.py
import torch
from torch.utils.data import DataLoader
from app.datasets import DrugCellDataset
from app.transformer_model import DrugTransformer

def train_model(epochs=20, batch_size=64):
    dataset = DrugCellDataset()
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

    model = DrugTransformer()
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4)
    loss_fn = torch.nn.MSELoss()

    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for cell, drug in loader:
            seed = torch.rand_like(drug)

            pred = model(cell, seed)
            loss = loss_fn(pred, drug)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        print(f"Epoch {epoch+1} | Loss: {total_loss/len(loader):.4f}")

    torch.save(model.state_dict(), "app/drug_transformer.pt")

train_model()
```

```python
# chem/mol_graph.py
!pip install torch-geometric
from rdkit import Chem
import torch
from torch_geometric.data import Data

def mol_to_graph(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None

    x = []
    for atom in mol.GetAtoms():
        x.append([
            atom.GetAtomicNum(),
            atom.GetDegree(),
            atom.GetFormalCharge(),
            atom.GetHybridization().real
        ])

    edge_index = []
    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        edge_index += [[i, j], [j, i]]

    return Data(
        x=torch.tensor(x, dtype=torch.float),
        edge_index=torch.tensor(edge_index, dtype=torch.long).t()
    )
```

```python
# chem/gnn_encoder.py
import torch
import torch.nn as nn
from torch_geometric.nn import GCNConv, global_mean_pool

class MolecularGNN(nn.Module):
    def __init__(self, in_dim=4, hidden=64):
        super().__init__()
        self.conv1 = GCNConv(in_dim, hidden)
        self.conv2 = GCNConv(hidden, hidden)
        self.out = nn.Linear(hidden, hidden)

    def forward(self, data):
        x, edge_index, batch = data.x, data.edge_index, data.batch
        x = torch.relu(self.conv1(x, edge_index))
        x = torch.relu(self.conv2(x, edge_index))
        return self.out(global_mean_pool(x, batch))
```

```python
import tensorflow as tf

def build_quantum_autoencoder(input_dim, latent_dim=32):
    encoder = tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dense(latent_dim, activation="tanh")
    ])

    decoder = tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dense(input_dim, activation="linear")
    ])

    autoenc = tf.keras.Model(
        inputs=tf.keras.Input(shape=(input_dim,)),
        outputs=decoder(encoder.output)
    )
    autoenc.compile(optimizer="adam", loss="mse")

    return encoder, decoder, autoenc
```

```python
import tensorflow as tf
import numpy as np

class QuantumVariationalLayer(tf.keras.layers.Layer):
    def __init__(self, n_qubits=8):
        super().__init__()
        self.n_qubits = n_qubits

    def build(self, input_shape):
        self.theta = self.add_weight(
            shape=(input_shape[-1], self.n_qubits),
            initializer="glorot_uniform",
            trainable=True
        )

    def call(self, x):
        # Simulated RX / RY rotations
        z = tf.matmul(x, self.theta)
        return tf.concat([
            tf.sin(z),
            tf.cos(z)
        ], axis=-1)
```

```python
import pennylane as qml
import tensorflow as tf
import numpy as np

n_qubits = 4
dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev, interface="tf")
def quantum_circuit(inputs, weights, num_layers_qc): # Added num_layers_qc
    # Data encoding
    for i in range(n_qubits):
        qml.RY(inputs[i], wires=i)

    # Variational layers
    for k in tf.range(num_layers_qc): # Use num_layers_qc to control loop
        layer = weights[k] # Access layer directly by index
        for i in range(n_qubits):
            qml.RX(layer[i], wires=i)
        for i in range(n_qubits - 1):
            qml.CNOT(wires=[i, i + 1])

    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]
```

```python
class QuantumLayer(tf.keras.layers.Layer):
    def __init__(self, n_layers=2):
        super().__init__()
        self.n_layers = n_layers

    def build(self, input_shape):
        self.weights_q = self.add_weight(
            shape=(self.n_layers, n_qubits),
            initializer="random_normal",
            trainable=True
        )

    def call(self, inputs):
        return tf.vectorized_map(
            lambda x: quantum_circuit(x, self.weights_q, self.n_layers), # Pass self.n_layers
            inputs
        )

    def compute_output_shape(self, input_shape):
        # input_shape is (batch_size, n_qubits) for 'inputs'
        # The quantum_circuit returns n_qubits expectation values for each input
        return (input_shape[0], n_qubits)
```

```python
def mcts(state, policy_net, value_net, depth=10):
    best_action = None
    best_value = -np.inf

    # Convert state dict to tensor for policy_net
    state_tensor = tf.convert_to_tensor([list(state.values())], dtype=tf.float32)

    # policy_net returns raw logits, apply softmax to get probabilities for sampling
    policy_logits = policy_net(state_tensor)[0]
    # Ensure policy_logits is a list or array of floats for sample_actions
    # Assuming sample_actions can handle tf.Tensor, or it needs conversion to numpy
    # If policy_net.output is already probabilities, remove activation='softmax' from Dense(5, ...) and apply tf.nn.softmax here

    # Note: organ_policy is tf.keras.Sequential with Dense(5), so its output is logits, not probabilities directly.
    # We need to apply softmax before sampling actions.
    policy_probs = tf.nn.softmax(policy_logits).numpy()

    # Use policy_probs to sample actions, assuming sample_actions expects probabilities or a policy representation
    # For this dummy setup, sample_actions doesn't use policy_output directly for sampling, but creates dummy actions.
    # The error is in calling policy_net(state) with a dict, not how sample_actions uses the output.

    # Let's assume sample_actions takes a dummy argument for now if it's not actually using policy_output for logic
    # Or, if sample_actions is meant to interpret the policy output:

    # Revised loop: sample actions based on policy_probs
    # Since sample_actions is a dummy that just returns random actions, we'll keep it simple
    # However, for correct MCTS, actions would be chosen based on policy_probs.
    # For now, let's just make the call to policy_net(state_tensor) correct.

    # This part needs to be revised if sample_actions is meant to be intelligent
    # For the immediate fix of ValueError from policy_net(state)
    # The error originates from policy_net(state) -> state being a dict.

    # Let's adjust sample_actions to take a tensor input if it's meant to derive actions from it
    # Given sample_actions returns random actions, we can just call it.
    # The original for loop: `for action in sample_actions(policy_net(state)):`
    # This implies policy_net(state) produces an object that sample_actions can iterate over.
    # Given organ_policy returns a tensor, sample_actions(tensor) might not be directly iterable.

    # Re-evaluating the flow:
    # 1. Convert state to tensor for policy_net call.
    # 2. Call policy_net to get logits/values.
    # 3. Use policy_output (e.g., logits) to sample actions. The dummy `sample_actions` does not use its input `policy_output` to sample.
    #    It simply returns random actions. So `sample_actions(policy_logits)` is syntactically fine.

    for action in sample_actions(policy_logits):
        s = state.copy() # Use a copy to avoid modifying the original state dictionary in inner loop
        for _ in range(depth):
            s = simulate_transition(s, action)

        # Convert state dict to tensor for value_net
        v = value_net(tf.convert_to_tensor([list(s.values())], dtype=tf.float32))[0][0]
        if v > best_value:
            best_value = v
            best_action = action

    return best_action
```

```python
class GraphConv(tf.keras.layers.Layer):
    def __init__(self, units):
        super().__init__()
        self.dense = tf.keras.layers.Dense(units, activation="gelu")

    def call(self, x, adj):
        return self.dense(tf.matmul(adj, x))
```

```python
class MolecularEncoder(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.gc1 = GraphConv(64)
        self.gc2 = GraphConv(128)
        self.pool = tf.keras.layers.GlobalAveragePooling1D()

    def call(self, atoms, adj):
        x = self.gc1(atoms, adj)
        x = self.gc2(x, adj)
        return self.pool(tf.expand_dims(x, 0))
```

```python
class TelomereController(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.net = tf.keras.Sequential([
            tf.keras.layers.Dense(128, activation="elu"),
            tf.keras.layers.Dense(64, activation="gelu"),
            tf.keras.layers.Dense(3, activation="tanh")
        ])

    def call(self, z):
        return self.net(z)
```

```python
time_steps = 120

# Initial telomere length (bp)
telomere = [8000]

# AI intervention signal (0 = off, 1 = on)
telomerase_signal = []

for t in range(1, time_steps):
    loss = np.random.randint(50, 120)  # bp lost per division
    current = telomere[-1] - loss

    # AI rule: intervene only near danger zone
    if current < 3500:
        current += np.random.randint(200, 400)  # controlled elongation
        telomerase_signal.append(1)
    else:
        telomerase_signal.append(0)

    telomere.append(current)

telomere = np.array(telomere)
telomerase_signal = np.array(telomerase_signal)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.set_xlim(0, time_steps)
ax.set_ylim(1500, 8500)

ax.set_title("AI-Regulated Telomere Dynamics")
ax.set_xlabel("Time step")
ax.set_ylabel("Telomere length (bp)")

# Biological thresholds
ax.axhline(2000, color="red", linestyle="--", label="Senescence threshold")
ax.axhline(8000, color="green", linestyle=":", label="Youthful telomere")

line, = ax.plot([], [], lw=2)
dots = ax.scatter([], [], color="orange", s=50, label="AI intervention")

ax.legend(loc="lower right")

def update(frame):
    x = np.arange(frame)
    y = telomere[:frame]
    line.set_data(x, y)

    # Mark intervention points
    intervention_x = np.where(telomerase_signal[:frame] == 1)[0]
    intervention_y = telomere[intervention_x]
    dots.set_offsets(np.c_[intervention_x, intervention_y])

    return line, dots

anim = FuncAnimation(
    fig,
    update,
    frames=time_steps,
    interval=120,
    blit=True
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

steps = 150
telomere = np.linspace(8000, 3000, steps)

# Constraint parameters
senescence_bp = 3500
cancer_ceiling = 0.75
```

```python
def quantum_policy(t):
    # Raw logits
    repair = np.exp(-(t - senescence_bp)**2 / (2 * 600**2))
    maintain = np.exp(-(t - 5000)**2 / (2 * 800**2))
    suppress = np.exp(-(t - 8000)**2 / (2 * 700**2))

    state = np.array([repair, maintain, suppress])
    return state / np.linalg.norm(state)
```

```python
states = np.array([quantum_policy(t) for t in telomere])

P_repair = states[:, 0]**2
P_maintain = states[:, 1]**2
P_suppress = states[:, 2]**2

# Enforce cancer safety
P_repair = np.clip(P_repair, 0, cancer_ceiling)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(telomere, P_repair, label="|Repair⟩² (Telomerase ON)", linewidth=2)
plt.plot(telomere, P_maintain, label="|Maintain⟩²", linewidth=2)
plt.plot(telomere, P_suppress, label="|Suppress⟩² (Cancer-safe)", linewidth=2)

plt.axvline(senescence_bp, color="red", linestyle="--", alpha=0.6)
plt.axhline(cancer_ceiling, color="orange", linestyle="--", alpha=0.6)

plt.xlabel("Telomere Length (bp)")
plt.ylabel("Decision Probability")
plt.title("Quantum-Inspired Telomerase Decision Manifold")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
