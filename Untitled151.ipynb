```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from torch.utils.data import TensorDataset, DataLoader

# Placeholder for GHU1C_v2 (Full definition not provided in notebook state)
class GHU1C_v2(nn.Module):
    def __init__(self, input_dim, dim, tokens, depth, out_dim):
        super().__init__()
        self.input_dim = input_dim
        self.dim = dim
        self.tokens = tokens
        self.depth = depth
        self.out_dim = out_dim
        self.proj = nn.Linear(input_dim, dim)
        self.blocks = nn.ModuleList([nn.Linear(dim, dim) for _ in range(depth)]) # Simplified
        self.head = nn.Linear(dim, out_dim) # Simplified

    def forward(self, x):
        x = self.proj(x)
        for block in self.blocks:
            x = block(x)
        return self.head(x)

# -----------------------------
# Activation: HyperFusionActX_v3 (from cell kaC6i69adFQk)
# -----------------------------
class HyperFusionActX_v3(nn.Module):
    def __init__(self, channels=None):
        super().__init__()
        self.channels = channels
        if channels is None:
            self.raw = nn.Parameter(torch.randn(5))
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(channels, 5))
            self.scale = nn.Parameter(torch.ones(channels))

    def forward(self, x):
        if self.channels is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            out = w[0]*r + w[1]*g + w[2]*l + w[3]*e + w[4]*sw
            return out
        else:
            w = torch.softmax(self.raw, dim=-1)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,)*(x.dim()-2)), -1)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            out = w[...,0]*r + w[...,1]*g + w[...,2]*l + w[...,3]*e + w[...,4]*sw
            return out

# -----------------------------
# Tri-Optimizer AMP + Gradient Clipping (from cell -pEshqVFd2Di)
# -----------------------------
class TriOptimXAMP:
    """
    Tri-optimizer with AdamW + SGD + SGD-Momentum, supporting AMP and per-group gradient clipping.
    """
    def __init__(self, params, base_lr=3e-4, wd=1e-2, clip_grad=1.0, amp_enabled=True, device='cpu'):
        self.device = device
        self.clip_grad = clip_grad
        self.amp_enabled = amp_enabled

        # Extract all actual parameters for gradient clipping
        self.all_params = []
        for group in params:
            self.all_params.extend(list(group['params'])) # Ensure it's a list for extend

        self.scaler = torch.cuda.amp.GradScaler(enabled=self.amp_enabled)

        # Create three optimizers, passing the original parameter groups
        self.adam = torch.optim.AdamW(params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(params, lr=base_lr*0.5, momentum=0.9)

    def zero_grad(self):
        self.adam.zero_grad()
        self.sgd.zero_grad()
        self.sgd_m.zero_grad()

    def backward(self, loss):
        """Scales loss and backpropagates safely with AMP if enabled."""
        if self.amp_enabled:
            self.scaler.scale(loss).backward()
        else:
            loss.backward()

    def step_amp(self):
        """Performs optimizer step with AMP scaling and gradient clipping"""
        if self.clip_grad > 0:
            # Clip gradients using the flat list of all parameters
            torch.nn.utils.clip_grad_norm_(self.all_params, self.clip_grad)
        if self.amp_enabled:
            self.scaler.step(self.adam)
            self.scaler.step(self.sgd)
            self.scaler.step(self.sgd_m)
            self.scaler.update()
        else:
            self.adam.step()
            self.sgd.step()
            self.sgd_m.step()

    def state_dict(self):
        return {
            "adam": self.adam.state_dict(),
            "sgd": self.sgd.state_dict(),
            "sgd_m": self.sgd_m.state_dict(),
            "scaler": self.scaler.state_dict() if self.amp_enabled else None
        }

    def load_state_dict(self, sd):
        self.adam.load_state_dict(sd.get("adam", {}))
        self.sgd.load_state_dict(sd.get("sgd", {}))
        self.sgd_m.load_state_dict(sd.get("sgd_m", {}))
        if self.amp_enabled and "scaler" in sd and sd["scaler"] is not None:
            self.scaler.load_state_dict(sd["scaler"])

# -----------------------------
# Virtual Cell System components (from cell kaC6i69adFQk)
# -----------------------------
class TelomereVirtual(nn.Module):
    def __init__(self, init_length=120.0, vec_len=1, degrade_scale=0.5, repair_scale=1.0):
        super().__init__()
        self.register_buffer('_base_length', torch.full((vec_len,), float(init_length)))
        self.degrade_scale = degrade_scale
        self.repair_scale = repair_scale

    def initial_state(self, batch=1, device=None):
        dev = device or self._base_length.device
        return self._base_length.unsqueeze(0).repeat(batch,1).to(dev)

    def step_update(self, tel, do_repair=None, srr_strength=None, noise_scale=0.1):
        noise = torch.randn_like(tel) * noise_scale * self.degrade_scale
        tel_new = torch.clamp(tel - torch.abs(noise) - 0.1*self.degrade_scale, min=0.0)
        if do_repair is not None:
            mask = do_repair.bool().float()
            tel_new += mask * self.repair_scale * srr_strength
        return torch.clamp(tel_new, min=0.0)

class TelomeraseController(nn.Module):
    def __init__(self, emb_dim=512, hidden=128):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(emb_dim+1, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden,2))

    def forward(self, emb, tel_scalar):
        x = torch.cat([emb, tel_scalar], dim=-1)
        out = self.net(x)
        activity = torch.sigmoid(out[...,0:1])
        budget = F.softplus(out[...,1:2])
        return activity, budget

class PhagocytosisVirtual(nn.Module):
    def __init__(self, emb_dim=512, shrink=0.4):
        super().__init__()
        self.proj = nn.Linear(emb_dim, emb_dim)
        self.act = HyperFusionActX_v3(emb_dim)
        self.shrink = shrink

    def forward(self, emb):
        est = self.act(self.proj(emb))
        return emb - self.shrink * torch.tanh(est)

# Placeholder for VirtualCellSystem (Full definition not provided in notebook state)
class VirtualCellSystem(nn.Module):
    def __init__(self, backbone, device, tel_len_init, tel_vec_len):
        super().__init__()
        self.backbone = backbone
        self.device = device
        self.tel_len_init = tel_len_init
        self.tel_vec_len = tel_vec_len
        # Assuming emb_dim is backbone's out_dim based on other usages
        self.tvase = TelomeraseController(emb_dim=backbone.out_dim, hidden=128)
        self.phagocytosis = PhagocytosisVirtual(emb_dim=backbone.out_dim)
        self.telomere = TelomereVirtual(init_length=tel_len_init, vec_len=tel_vec_len)

    def initialize_telomeres(self, batch):
        return self.telomere.initial_state(batch=batch, device=self.device)

    def division_step(self, x, tel, policy):
        # Simplified logic for demonstration
        emb = self.backbone(x)
        cleaned_emb = self.phagocytosis(emb)

        # Telomere update logic (simplified)
        # For a full implementation, you'd use policy parameters for do_repair, srr_strength etc.
        tel_new = self.telomere.step_update(tel, srr_strength=torch.ones_like(tel), noise_scale=policy["noise_scale"])

        # Placeholder metrics
        metrics = {
            "tel_mean_after": tel_new.mean().item(),
            "division_rate": torch.rand(1).item() * policy["division_prob_thresh"]
        }
        return cleaned_emb, tel_new, metrics

    def get_embedding(self, x):
        return self.backbone(x)

# -----------------------------
# Autonomous Agent (from cell kaC6i69adFQk)
# -----------------------------
class AutonomousAgent(nn.Module):
    def __init__(self, backbone, vcs, action_dim=8, hidden=256):
        super().__init__()
        self.backbone = backbone
        self.vcs = vcs
        self.encoder = nn.Sequential(nn.Linear(backbone.out_dim, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, hidden))
        self.policy = nn.Sequential(nn.Linear(hidden, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, action_dim))
        self.value = nn.Sequential(nn.Linear(hidden, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, 1))
        # New decoder head to project encoded features back to original embedding dimension
        self.decoder_head = nn.Linear(hidden, backbone.out_dim)

    def observe(self, x):
        emb = self.backbone(x)
        if emb.dim()>2: emb = emb.mean(dim=1)
        return self.encoder(emb)

    def act(self, z):
        mu = self.policy(z)
        noise = torch.randn_like(mu)*0.1
        return torch.tanh(mu+noise)

    def evaluate(self, z, action):
        v = self.value(z).squeeze(-1)
        adv = (action*z).mean(dim=-1)
        return v, v + 0.01*adv


# -----------------------------
# Función para generar datos con agentes autónomos
# -----------------------------
def generate_data_with_agents(agent: AutonomousAgent, vcs: VirtualCellSystem,
                              num_batches: int = 50, batch_size: int = 16, steps: int = 10,
                              device: str = "cpu"):
    dev = torch.device(device)
    agent.to(dev)
    vcs.to(dev)
    data_records = []

    for b in range(num_batches):
        # Estado inicial aleatorio
        x = torch.randn(batch_size, 2).to(dev)
        tel = vcs.initialize_telomeres(batch=batch_size).to(dev)

        for step in range(steps):
            # Observa y actúa
            z = agent.observe(x)
            action = agent.act(z)
            x = x + 0.01 * action[:, :2]  # perturbación segura

            # Simulación de célula virtual
            cleaned, tel, metrics = vcs.division_step(
                x, tel,
                policy={
                    "min_tel_threshold": 5.0,
                    "max_tel_threshold": 1000.0,
                    "division_cost": 0.5,
                    "noise_scale": 0.03,
                    "division_prob_thresh": 0.5,
                    "repair_activity_thresh": 0.2
                }
            )

            # Guardar datos simulados
            data_records.append({
                "inputs": x.detach().cpu(),
                "cleaned_emb": cleaned.detach().cpu(),
                "telomeres": tel.detach().cpu(),
                "metrics": metrics
            })

    return data_records

# -----------------------------
# Inicialización de backbone, VCS y agente
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
backbone = GHU1C_v2(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64).to(device)
vcs = VirtualCellSystem(backbone=backbone, device=device, tel_len_init=100.0, tel_vec_len=1).to(device)
agent = AutonomousAgent(backbone=backbone, vcs=vcs, action_dim=8, hidden=256).to(device)

# Optimizador seguro con AMP, triple ruta y gradient clipping
param_groups = [{"params": agent.parameters(), "lr": 3e-4, "weight_decay": 1e-2}]
optim = TriOptimXAMP(param_groups, base_lr=3e-4, wd=1e-2, clip_grad=1.0, amp_enabled=True, device=device)

# -----------------------------
# Generar dataset
# -----------------------------
dataset_records = generate_data_with_agents(agent, vcs,
                                           num_batches=100,
                                           batch_size=16,
                                           steps=10,
                                           device=device)

# Convertir a TensorDataset para PyTorch
inputs = torch.cat([r["inputs"] for r in dataset_records], dim=0)
cleaned = torch.cat([r["cleaned_emb"] for r in dataset_records], dim=0)
tel = torch.cat([r["telomeres"] for r in dataset_records], dim=0)
virtual_dataset = TensorDataset(inputs, cleaned, tel)

print(f"Datos generados: {len(virtual_dataset)} muestras")
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
import torch.nn.functional as F

# -----------------------------
# Multi-agent sequential self-improving system
# -----------------------------
class MultiAgentSelfImproving:
    def __init__(self, num_agents: int, device="cpu",
                 batch_size=16, steps_per_batch=10, max_grad_norm=1.0):
        self.device = torch.device(device)
        self.num_agents = num_agents
        self.batch_size = batch_size
        self.steps_per_batch = steps_per_batch
        self.max_grad_norm = max_grad_norm

        # Inicializamos agentes y VCS independientes
        self.agents = []
        for _ in range(num_agents):
            backbone = GHU1C_v2(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64).to(self.device)
            vcs = VirtualCellSystem(backbone=backbone, device=self.device, tel_len_init=100.0, tel_vec_len=1).to(self.device)
            agent = AutonomousAgent(backbone=backbone, vcs=vcs, action_dim=8, hidden=256).to(self.device)
            optim = TriOptimXAMP([{"params": agent.parameters(), "lr":3e-4, "weight_decay":1e-2}],
                                 base_lr=3e-4, wd=1e-2, clip_grad=self.max_grad_norm)
            self.agents.append({"backbone": backbone, "vcs": vcs, "agent": agent, "optim": optim})

        self.scaler = torch.cuda.amp.GradScaler(enabled=(self.device.type=="cuda"))

    def generate_data(self, agent_dict, num_batches=10):
        """Genera datos simulados por un agente específico."""
        records = []
        for b in range(num_batches):
            x = torch.randn(self.batch_size, 2, device=self.device)
            tel = agent_dict["vcs"].initialize_telomeres(batch=self.batch_size).to(self.device)
            for step in range(self.steps_per_batch):
                z = agent_dict["agent"].observe(x)
                action = agent_dict["agent"].act(z)
                x = x + 0.01 * action[:, :2]
                cleaned, tel, metrics = agent_dict["vcs"].division_step(
                    x, tel,
                    policy={
                        "min_tel_threshold":5.0,
                        "max_tel_threshold":1000.0,
                        "division_cost":0.5,
                        "noise_scale":0.03,
                        "division_prob_thresh":0.5,
                        "repair_activity_thresh":0.2
                    }
                )
                records.append({"inputs": x.detach(), "cleaned": cleaned.detach(), "tel": tel.detach()})
        return records

    def train_agent_on_records(self, agent_dict, records, epochs=1):
        """Entrena al agente usando datos compartidos de todos los agentes."""
        inputs = torch.cat([r["inputs"] for r in records], dim=0)
        cleaned = torch.cat([r["cleaned"] for r in records], dim=0)
        dataset = TensorDataset(inputs, cleaned)
        loader = DataLoader(dataset, batch_size=self.batch_size, shuffle=True)

        agent = agent_dict["agent"]
        optim = agent_dict["optim"]

        for epoch in range(epochs):
            for x_batch, cleaned_batch in loader:
                x_batch = x_batch.to(self.device)
                cleaned_batch = cleaned_batch.to(self.device)
                with torch.cuda.amp.autocast(enabled=(self.device.type=="cuda")):
                    z = agent.observe(x_batch)
                    # Use the new decoder_head to project z back to the dimension of cleaned_batch
                    pred_cleaned = agent.decoder_head(z)
                    loss = F.mse_loss(pred_cleaned, cleaned_batch)
                optim.zero_grad()
                optim.backward(loss)
                optim.step_amp()

    def run_cycles(self, cycles=5, num_batches_per_agent=10, train_epochs=1):
        """Ejecuta ciclos secuenciales de generación y entrenamiento multi-agente."""
        for c in range(cycles):
            print(f"\n=== Cycle {c+1}/{cycles} ===")
            all_records = []
            # Cada agente genera datos localmente
            for idx, agent_dict in enumerate(self.agents):
                records = self.generate_data(agent_dict, num_batches=num_batches_per_agent)
                print(f"Agent {idx} generated {len(records)} records.")
                all_records.extend(records)
            # Todos los agentes se entrenan con el dataset combinado
            for idx, agent_dict in enumerate(self.agents):
                self.train_agent_on_records(agent_dict, all_records, epochs=train_epochs)
                print(f"Agent {idx} trained on shared dataset.")

# -----------------------------
# Ejemplo de ejecución
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
multi_agent_system = MultiAgentSelfImproving(num_agents=3, device=device,
                                            batch_size=16, steps_per_batch=10, max_grad_norm=1.0)

multi_agent_system.run_cycles(cycles=5, num_batches_per_agent=10, train_epochs=1)
```

```python
class MultiAgentInteractive:
    def __init__(self, num_agents: int, device="cpu",
                 batch_size=16, steps_per_batch=10, max_grad_norm=1.0):
        self.device = torch.device(device)
        self.num_agents = num_agents
        self.batch_size = batch_size
        self.steps_per_batch = steps_per_batch
        self.max_grad_norm = max_grad_norm

        self.agents = []
        for _ in range(num_agents):
            backbone = GHU1C_v2(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64).to(self.device)
            vcs = VirtualCellSystem(backbone=backbone, device=self.device, tel_len_init=100.0, tel_vec_len=1).to(self.device)
            agent = AutonomousAgent(backbone=backbone, vcs=vcs, action_dim=8, hidden=256).to(self.device)
            optim = TriOptimXAMP([{"params": agent.parameters(), "lr":3e-4, "weight_decay":1e-2}],
                                 base_lr=3e-4, wd=1e-2, clip_grad=self.max_grad_norm)
            self.agents.append({"backbone": backbone, "vcs": vcs, "agent": agent, "optim": optim})

        self.scaler = torch.cuda.amp.GradScaler(enabled=(self.device.type=="cuda"))

    def interaction_step(self):
        """Cada agente observa a todos los demás y genera acciones sobre el mundo compartido."""
        global_state = torch.randn(self.batch_size, 2, device=self.device)  # estado inicial compartido
        tel_states = [a["vcs"].initialize_telomeres(batch=self.batch_size).to(self.device) for a in self.agents]

        records = []

        for step in range(self.steps_per_batch):
            actions = []
            # Cada agente genera su acción basada en el estado global
            for idx, agent_dict in enumerate(self.agents):
                z = agent_dict["agent"].observe(global_state)
                action = agent_dict["agent"].act(z)
                actions.append(action)

            # Actualización del estado global: mezcla de acciones de todos los agentes
            action_stack = torch.stack(actions, dim=0)
            combined_action = action_stack.mean(dim=0)
            global_state = global_state + 0.01 * combined_action[:, :2]  # efecto abstracto seguro

            # Actualizar VCS de cada agente
            for idx, agent_dict in enumerate(self.agents):
                cleaned, tel_states[idx], metrics = agent_dict["vcs"].division_step(
                    global_state, tel_states[idx],
                    policy={
                        "min_tel_threshold":5.0,
                        "max_tel_threshold":1000.0,
                        "division_cost":0.5,
                        "noise_scale":0.03,
                        "division_prob_thresh":0.5,
                        "repair_activity_thresh":0.2})
                records.append({"inputs": global_state.detach(),
                                "cleaned": cleaned.detach(),
                                "tel": tel_states[idx].detach()})
        return records

    def train_agents(self, records, epochs=1):
        """Todos los agentes se entrenan con los datos emergentes compartidos."""
        inputs = torch.cat([r["inputs"] for r in records], dim=0)
        cleaned = torch.cat([r["cleaned"] for r in records], dim=0)
        dataset = TensorDataset(inputs, cleaned)
        loader = DataLoader(dataset, batch_size=self.batch_size, shuffle=True)

        for agent_dict in self.agents:
            agent = agent_dict["agent"]
            optim = agent_dict["optim"]
            for epoch in range(epochs):
                for x_batch, cleaned_batch in loader:
                    x_batch = x_batch.to(self.device)
                    cleaned_batch = cleaned_batch.to(self.device)
                    with torch.cuda.amp.autocast(enabled=(self.device.type=="cuda")):
                        z = agent.observe(x_batch)
                        pred_cleaned = agent.decoder_head(z)
                        loss = F.mse_loss(pred_cleaned, cleaned_batch)
                    optim.zero_grad()
                    optim.backward(loss)
                    optim.step_amp()

    def run_cycles(self, cycles=5, steps_per_cycle=10, train_epochs=1):
        for c in range(cycles):
            print(f"\n=== Cycle {c+1}/{cycles} ===")
            records = self.interaction_step()
            self.train_agents(records, epochs=train_epochs)
            print(f"Cycle {c+1} completed. Agents trained on {len(records)} emergent records.")


# -----------------------------
# Ejecución
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
interactive_system = MultiAgentInteractive(num_agents=3, device=device,
                                           batch_size=16, steps_per_batch=10, max_grad_norm=1.0)
interactive_system.run_cycles(cycles=5, steps_per_cycle=10, train_epochs=1)
```

```python
class SelfEvolvingMultiAgent:
    def __init__(self, num_agents: int, device="cpu", batch_size=16,
                 steps_per_batch=10, max_grad_norm=1.0):
        self.device = torch.device(device)
        self.num_agents = num_agents
        self.batch_size = batch_size
        self.steps_per_batch = steps_per_batch
        self.max_grad_norm = max_grad_norm

        self.agents = []
        for _ in range(num_agents):
            backbone = GHU1C_v2(input_dim=2, dim=256, tokens=16, depth=8, out_dim=64).to(self.device)
            vcs = VirtualCellSystem(backbone=backbone, device=self.device, tel_len_init=100.0, tel_vec_len=1).to(self.device)
            agent = AutonomousAgent(backbone=backbone, vcs=vcs, action_dim=8, hidden=256).to(self.device)
            # Hyperparametros adaptativos iniciales
            lr = 3e-4
            weight_decay = 1e-2
            optim = TriOptimXAMP([{"params": agent.parameters(), "lr": lr, "weight_decay": weight_decay}],
                                 base_lr=lr, wd=weight_decay, clip_grad=self.max_grad_norm)
            self.agents.append({
                "backbone": backbone,
                "vcs": vcs,
                "agent": agent,
                "optim": optim,
                "lr": lr,
                "wd": weight_decay
            })

        self.scaler = torch.cuda.amp.GradScaler(enabled=(self.device.type=="cuda"))

    def curiosity_reward(self, tel_state, tel_prev):
        """Recompensa por explorar estados de telómeros menos predecibles."""
        return ((tel_state - tel_prev).abs().mean(dim=-1)).detach()

    def interaction_step(self):
        global_state = torch.randn(self.batch_size, 2, device=self.device)
        tel_states = [a["vcs"].initialize_telomeres(batch=self.batch_size).to(self.device) for a in self.agents]
        tel_prev_states = [tel.clone() for tel in tel_states]
        records = []

        for step in range(self.steps_per_batch):
            actions = []
            for idx, agent_dict in enumerate(self.agents):
                z = agent_dict["agent"].observe(global_state)
                action = agent_dict["agent"].act(z)
                actions.append(action)

            combined_action = torch.stack(actions, dim=0).mean(dim=0)
            global_state = global_state + 0.01 * combined_action[:, :2]

            for idx, agent_dict in enumerate(self.agents):
                cleaned, tel_states[idx], metrics = agent_dict["vcs"].division_step(
                    global_state, tel_states[idx],
                    policy={
                        "min_tel_threshold":5.0,
                        "max_tel_threshold":1000.0,
                        "division_cost":0.5,
                        "noise_scale":0.03,
                        "division_prob_thresh":0.5,
                        "repair_activity_thresh":0.2})
                reward_curiosity = self.curiosity_reward(tel_states[idx], tel_prev_states[idx])
                tel_prev_states[idx] = tel_states[idx].clone()
                records.append({"inputs": global_state.detach(),
                                "cleaned": cleaned.detach(),
                                "tel": tel_states[idx].detach(),
                                "reward_curiosity": reward_curiosity})
        return records

    def meta_update_hyperparams(self, agent_dict, reward_mean):
        """Ajuste adaptativo de learning rate y weight decay según desempeño."""
        factor = 1.0 + 0.1 * (reward_mean - 0.05)
        new_base_lr = max(1e-5, min(1e-3, agent_dict["lr"] * factor))
        new_wd = max(1e-4, min(1e-1, agent_dict["wd"] * (2.0 - factor)))

        agent_dict["lr"] = new_base_lr
        agent_dict["wd"] = new_wd

        optim_wrapper = agent_dict["optim"]
        # Update AdamW's parameter groups
        for pg in optim_wrapper.adam.param_groups:
            pg["lr"] = new_base_lr
            pg["weight_decay"] = new_wd
        # Update SGD's parameter groups
        for pg in optim_wrapper.sgd.param_groups:
            pg["lr"] = new_base_lr * 0.25 # Maintain original ratio
            pg["weight_decay"] = new_wd
        # Update SGD_M's parameter groups
        for pg in optim_wrapper.sgd_m.param_groups:
            pg["lr"] = new_base_lr * 0.5 # Maintain original ratio
            pg["weight_decay"] = new_wd

    def train_agents(self, records, epochs=1):
        inputs = torch.cat([r["inputs"] for r in records], dim=0)
        cleaned = torch.cat([r["cleaned"] for r in records], dim=0)
        rewards_curiosity = torch.cat([r["reward_curiosity"] for r in records], dim=0)
        dataset = TensorDataset(inputs, cleaned, rewards_curiosity)
        loader = DataLoader(dataset, batch_size=self.batch_size, shuffle=True)

        for agent_dict in self.agents:
            agent = agent_dict["agent"]
            optim = agent_dict["optim"]
            for epoch in range(epochs):
                for x_batch, cleaned_batch, reward_batch in loader:
                    x_batch = x_batch.to(self.device)
                    cleaned_batch = cleaned_batch.to(self.device)
                    reward_batch = reward_batch.to(self.device)

                    with torch.cuda.amp.autocast(enabled=(self.device.type=="cuda")):
                        z = agent.observe(x_batch)
                        pred_cleaned = agent.decoder_head(z)
                        loss = F.mse_loss(pred_cleaned, cleaned_batch) - 0.1 * reward_batch.mean()
                    optim.zero_grad()
                    optim.backward(loss) # Use TriOptimXAMP's backward method
                    optim.step_amp()    # Use TriOptimXAMP's step_amp method

            # meta-optimización de hyperparams por agente
            self.meta_update_hyperparams(agent_dict, reward_batch.mean().item())

    def run_cycles(self, cycles=10, steps_per_cycle=10, train_epochs=1):
        for c in range(cycles):
            print(f"\n=== Cycle {c+1}/{cycles} ===")
            records = self.interaction_step()
            self.train_agents(records, epochs=train_epochs)
            print(f"Cycle {c+1} completed. Agents self-improved with emergent dataset.")

# -----------------------------
# Ejecución
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
self_evolving_system = SelfEvolvingMultiAgent(num_agents=3, device=device,
                                              batch_size=16, steps_per_batch=10, max_grad_norm=1.0)
self_evolving_system.run_cycles(cycles=5, steps_per_cycle=10, train_epochs=1)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from torch.utils.tensorboard import SummaryWriter

# -----------------------------
# Activation: HyperFusionActX_v3
# -----------------------------
class HyperFusionActX_v3(nn.Module):
    def __init__(self, channels=None):
        super().__init__()
        self.channels = channels
        if channels is None:
            self.raw = nn.Parameter(torch.randn(5))
            self.scale = nn.Parameter(torch.tensor(1.0))
        else:
            self.raw = nn.Parameter(torch.randn(channels, 5))
            self.scale = nn.Parameter(torch.ones(channels))

    def forward(self, x):
        if self.channels is None:
            w = torch.softmax(self.raw, dim=0)
            s = torch.clamp(self.scale, 0.1, 10.0)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            out = w[0]*r + w[1]*g + w[2]*l + w[3]*e + w[4]*sw
            return out
        else:
            w = torch.softmax(self.raw, dim=-1)
            s = torch.clamp(self.scale, 0.1, 10.0).reshape(1, *((1,)*(x.dim()-2)), -1)
            x_s = x * s
            r = F.relu(x_s)
            g = F.gelu(x_s)
            l = F.leaky_relu(x_s, 0.02)
            e = F.elu(x_s)
            sw = x_s * torch.sigmoid(x_s)
            out = w[...,0]*r + w[...,1]*g + w[...,2]*l + w[...,3]*e + w[...,4]*sw
            return out

# -----------------------------
# TriOptimX with AMP + gradient clipping
# -----------------------------
class TriOptimX:
    def __init__(self, params, base_lr=3e-4, wd=1e-2, grad_clip=1.0, amp_enabled=True):
        self.params = list(params)
        self.adam = torch.optim.AdamW(self.params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(self.params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(self.params, lr=base_lr*0.5, momentum=0.9)
        self.grad_clip = grad_clip
        self.amp_enabled = amp_enabled
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp_enabled)

    def zero_grad(self):
        self.adam.zero_grad(); self.sgd.zero_grad(); self.sgd_m.zero_grad()

    def step(self):
        # gradient clipping
        total_norm = torch.nn.utils.clip_grad_norm_(self.params, self.grad_clip)
        # step with AMP if enabled
        self.scaler.step(self.adam)
        self.scaler.step(self.sgd)
        self.scaler.step(self.sgd_m)
        self.scaler.update()

# -----------------------------
# Virtual Cell System (telómeros + telomerasa + phagocytosis)
# -----------------------------
class TelomereVirtual(nn.Module):
    def __init__(self, init_length=120.0, vec_len=1, degrade_scale=0.5, repair_scale=1.0):
        super().__init__()
        self.register_buffer('_base_length', torch.full((vec_len,), float(init_length)))
        self.degrade_scale = degrade_scale
        self.repair_scale = repair_scale

    def initial_state(self, batch=1, device=None):
        dev = device or self._base_length.device
        return self._base_length.unsqueeze(0).repeat(batch,1).to(dev)

    def step_update(self, tel, do_repair=None, srr_strength=None, noise_scale=0.1):
        noise = torch.randn_like(tel) * noise_scale * self.degrade_scale
        tel_new = torch.clamp(tel - torch.abs(noise) - 0.1*self.degrade_scale, min=0.0)
        if do_repair is not None:
            mask = do_repair.bool().float()
            tel_new += mask * self.repair_scale * srr_strength
        return torch.clamp(tel_new, min=0.0)

class TelomeraseController(nn.Module):
    def __init__(self, emb_dim=512, hidden=128):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(emb_dim+1, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden,2))

    def forward(self, emb, tel_scalar):
        x = torch.cat([emb, tel_scalar], dim=-1)
        out = self.net(x)
        activity = torch.sigmoid(out[...,0:1])
        budget = F.softplus(out[...,1:2])
        return activity, budget

class PhagocytosisVirtual(nn.Module):
    def __init__(self, emb_dim=512, shrink=0.4):
        super().__init__()
        self.proj = nn.Linear(emb_dim, emb_dim)
        self.act = HyperFusionActX_v3(emb_dim)
        self.shrink = shrink

    def forward(self, emb):
        est = self.act(self.proj(emb))
        return emb - self.shrink * torch.tanh(est)

# -----------------------------
# Autonomous Agent with Meta-Evolution
# -----------------------------
class AutonomousAgent(nn.Module):
    def __init__(self, backbone, vcs, action_dim=8, hidden=256):
        super().__init__()
        self.backbone = backbone
        self.vcs = vcs
        self.encoder = nn.Sequential(nn.Linear(backbone.dim, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, hidden))
        self.policy = nn.Sequential(nn.Linear(hidden, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, action_dim))
        self.value = nn.Sequential(nn.Linear(hidden, hidden), HyperFusionActX_v3(hidden), nn.Linear(hidden, 1))

    def observe(self, x):
        emb = self.backbone(x)
        if emb.dim()>2: emb = emb.mean(dim=1)
        return self.encoder(emb)

    def act(self, z):
        mu = self.policy(z)
        noise = torch.randn_like(mu)*0.1
        return torch.tanh(mu+noise)

    def evaluate(self, z, action):
        v = self.value(z).squeeze(-1)
        adv = (action*z).mean(dim=-1)
        return v, v + 0.01*adv

class MetaEvolvingAgent:
    def __init__(self, agents_list):
        self.agents = agents_list

    def mutate_architecture(self, agent_dict):
        agent = agent_dict['agent']
        for blk in agent.backbone.blocks:
            blk.router.weight.data += 0.05*torch.randn_like(blk.router.weight)
        for module in agent.modules():
            if isinstance(module, HyperFusionActX_v3):
                module.raw.data += 0.05*torch.randn_like(module.raw)
                module.scale.data *= 1.0+0.01*torch.randn_like(module.scale)

    def run_meta_evolution(self, cycles=10, steps_per_cycle=10, train_epochs=1):
        for c in range(cycles):
            print(f'Meta-Evolution Cycle {c+1}/{cycles}')
            for agent_dict in self.agents:
                self.mutate_architecture(agent_dict)
            print(f'Cycle {c+1} completed. Agents evolved.')
```

```python
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from copy import deepcopy

# -----------------------------
# Meta-Evolving Autonomous Agent
# -----------------------------
class MetaEvolvingAgent:
    """
    Agent manager that maintains a population of autonomous agents,
    allows self-improvement via sequential training, mutation, and policy refinement.
    """
    def __init__(self, agent_class, backbone, vcs, pop_size=4, device="cpu", action_dim=8, hidden=256):
        self.device = torch.device(device)
        self.agents = [agent_class(backbone=deepcopy(backbone), vcs=deepcopy(vcs),
                                   action_dim=action_dim, hidden=hidden).to(self.device)
                       for _ in range(pop_size)]
        self.pop_size = pop_size
        self.optimizers = [TriOptimX(agent.parameters(), base_lr=3e-4, wd=1e-2) for agent in self.agents]
        self.history = []

    def generate_data(self, steps=50, batch_size=8):
        """
        Sequentially run each agent, let them act on environment,
        collect states and resulting metrics as new data.
        """
        data_list, metrics_list = [], []
        for idx, agent in enumerate(self.agents):
            vcs = agent.vcs
            x = torch.randn(batch_size, 2).to(self.device)
            tel = vcs.initialize_telomeres(batch=batch_size).to(self.device)
            for step in range(steps):
                z = agent.observe(x)
                action = agent.act(z)
                x = x + 0.01 * action[:, :2]
                cleaned, tel, metrics = vcs.division_step(x, tel, policy={
                    "min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                    "division_cost":0.5, "noise_scale":0.03,
                    "division_prob_thresh":0.5, "repair_activity_thresh":0.2
                })
                data_list.append(deepcopy(x.cpu()))
                metrics_list.append(metrics)
        return data_list, metrics_list

    def train_agents(self, data_list, epochs=5, batch_size=8, grad_clip=1.0):
        """
        Self-improvement loop: train each agent sequentially on generated data
        """
        dataset = TensorDataset(torch.cat(data_list))
        loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)
        for idx, agent in enumerate(self.agents):
            optimizer = self.optimizers[idx]
            agent.train()
            for epoch in range(epochs):
                for batch in loader:
                    x_batch = batch[0].to(self.device)
                    optimizer.zero_grad()
                    # compute embedding and fake loss: keep telomeres stable
                    z = agent.observe(x_batch)
                    cleaned, tel, metrics = agent.vcs.division_step(x_batch,
                                                                    agent.vcs.initialize_telomeres(batch=x_batch.size(0)).to(self.device),
                                                                    policy={"min_tel_threshold":5.0, "max_tel_threshold":1000.0,
                                                                            "division_cost":0.5, "noise_scale":0.03,
                                                                            "division_prob_thresh":0.5, "repair_activity_thresh":0.2})
                    loss = F.mse_loss(tel, torch.full_like(tel, 100.0))
                    loss.backward()
                    torch.nn.utils.clip_grad_norm_(agent.parameters(), grad_clip)
                    optimizer.step()
            print(f"Agent {idx} trained for {epochs} epochs")

    def mutate_agents(self, mutation_rate=0.05):
        """
        Apply small random mutations to agent parameters to explore new policies
        """
        for agent in self.agents:
            with torch.no_grad():
                for p in agent.parameters():
                    if torch.rand(1).item() < mutation_rate:
                        p.add_(torch.randn_like(p) * mutation_rate)
        print("Mutation step applied to all agents.")

    def sequential_self_improvement(self, cycles=5, steps_per_cycle=50, epochs=3):
        """
        Full loop: generate data -> train -> mutate -> repeat
        """
        for cycle in range(cycles):
            print(f"\n=== Self-Improvement Cycle {cycle+1}/{cycles} ===")
            data_list, metrics_list = self.generate_data(steps=steps_per_cycle)
            self.train_agents(data_list, epochs=epochs)
            self.mutate_agents()
            self.history.append(metrics_list)
```

```python
class TriOptimXAMP:
    """
    Tri-optimizer with AdamW + SGD + SGD-Momentum, supporting AMP and per-group gradient clipping.
    """
    def __init__(self, params, base_lr=3e-4, wd=1e-2, clip_grad=1.0, amp_enabled=True, device='cpu'):
        self.device = device
        self.clip_grad = clip_grad
        self.amp_enabled = amp_enabled
        self.params = list(params)
        self.scaler = torch.cuda.amp.GradScaler(enabled=self.amp_enabled)

        # Create three optimizers
        self.adam = torch.optim.AdamW(self.params, lr=base_lr, weight_decay=wd)
        self.sgd = torch.optim.SGD(self.params, lr=base_lr*0.25)
        self.sgd_m = torch.optim.SGD(self.params, lr=base_lr*0.5, momentum=0.9)

    def zero_grad(self):
        self.adam.zero_grad()
        self.sgd.zero_grad()
        self.sgd_m.zero_grad()

    def step(self):
        # Gradient clipping
        if self.clip_grad > 0:
            torch.nn.utils.clip_grad_norm_(self.params, self.clip_grad)
        # Step each optimizer sequentially
        self.adam.step()
        self.sgd.step()
        self.sgd_m.step()

    def backward(self, loss):
        """
        Scales loss and backpropagates safely with AMP if enabled.
        """
        if self.amp_enabled:
            self.scaler.scale(loss).backward()
        else:
            loss.backward()

    def step_amp(self):
        """
        Performs optimizer step with AMP scaling
        """
        if self.amp_enabled:
            self.scaler.step(self.adam)
            self.scaler.step(self.sgd)
            self.scaler.step(self.sgd_m)
            self.scaler.update()
        else:
            self.adam.step()
            self.sgd.step()
            self.sgd_m.step()

    def state_dict(self):
        return {
            "adam": self.adam.state_dict(),
            "sgd": self.sgd.state_dict(),
            "sgd_m": self.sgd_m.state_dict(),
            "scaler": self.scaler.state_dict() if self.amp_enabled else None
        }

    def load_state_dict(self, sd):
        self.adam.load_state_dict(sd.get("adam", {}))
        self.sgd.load_state_dict(sd.get("sgd", {}))
        self.sgd_m.load_state_dict(sd.get("sgd_m", {}))
        if self.amp_enabled and "scaler" in sd and sd["scaler"] is not None:
            self.scaler.load_state_dict(sd["scaler"])
```

```python
# -----------------------------
# Sequential Self-Improving Agent Loop
# -----------------------------
import os
from torch.utils.tensorboard import SummaryWriter

class SequentialMetaAgent:
    """
    Loop de agentes secuenciales auto-mejorantes.
    Cada "agente" entrena sobre sus observaciones y genera nueva data para el siguiente.
    """
    def __init__(self, agent_model, device='cuda', base_lr=3e-4, clip_grad=1.0, amp_enabled=True, log_dir='runs/meta_agent'):
        self.device = device
        self.agent = agent_model.to(self.device)
        self.optimizer = TriOptimXAMP(
            self.agent.parameters(),
            base_lr=base_lr,
            clip_grad=clip_grad,
            amp_enabled=amp_enabled,
            device=device
        )
        self.writer = SummaryWriter(log_dir)
        self.global_step = 0

    def step_agent(self, x_batch, target_tel=100.0):
        self.optimizer.zero_grad()
        with torch.cuda.amp.autocast(enabled=self.optimizer.amp_enabled):
            # Observe and encode
            z = self.agent.observe(x_batch)
            # Virtual cell division step
            cleaned, tel, metrics = self.agent.vcs.division_step(
                x_batch,
                self.agent.vcs.initialize_telomeres(batch=x_batch.size(0)),
                policy={
                    "min_tel_threshold": 5.0, "max_tel_threshold": 1000.0,
                    "division_cost": 0.5, "noise_scale": 0.03,
                    "division_prob_thresh": 0.5,
                    "repair_activity_thresh": 0.2
                }
            )
            # Loss: mantener telómeros cerca de target
            loss = F.mse_loss(tel, torch.full_like(tel, target_tel))

        # Backprop y optimizer step
        self.optimizer.backward(loss)
        self.optimizer.step_amp()

        # Logging
        self.writer.add_scalar('loss', loss.item(), self.global_step)
        self.writer.add_scalar('tel_mean', metrics['tel_mean_after'], self.global_step)
        self.writer.add_scalar('division_rate', metrics['division_rate'], self.global_step)
        self.global_step += 1

        # Generar nueva data para el siguiente agente (abstract safe perturbation)
        x_next = cleaned + 0.01 * torch.randn_like(cleaned)
        return x_next, metrics, loss.item()

    def run_loop(self, init_batch, steps=100, target_tel=100.0):
        x = init_batch.to(self.device)
        for step in range(steps):
            x, metrics, loss_val = self.step_agent(x, target_tel=target_tel)
            if step % 10 == 0:
                print(f"Step {step}: tel_mean={metrics['tel_mean_after']:.3f}, division_rate={metrics['division_rate']:.3f}, loss={loss_val:.4f}")
        print("Sequential meta-agent loop finished.")

    def save_checkpoint(self, path='meta_agent_ckpt.pth'):
        state = {
            'agent_state': self.agent.state_dict(),
            'optimizer_state': self.optimizer.state_dict(),
            'global_step': self.global_step
        }
        torch.save(state, path)
        print(f"Checkpoint saved at {path}")

    def load_checkpoint(self, path='meta_agent_ckpt.pth'):
        state = torch.load(path, map_location=self.device)
        self.agent.load_state_dict(state['agent_state'])
        self.optimizer.load_state_dict(state['optimizer_state'])
        self.global_step = state.get('global_step', 0)
        print(f"Checkpoint loaded from {path}")
```

```python
# -----------------------------
# Meta-Loop de Auto-Mejora Multi-Agente
# -----------------------------
class MetaLoopAgents:
    """
    Meta-loop de agentes secuenciales:
    Cada generación aprende del output del anterior y mejora políticas de telómeros.
    """
    def __init__(self, agent_fn, generations=5, batch_size=16, steps_per_agent=100,
                 device='cuda', base_lr=3e-4, clip_grad=1.0, amp_enabled=True, log_dir='runs/meta_loop'):
        """
        agent_fn: función que retorna un agente listo (AutonomousAgent)
        generations: cuántos agentes secuenciales
        batch_size: tamaño de batch inicial
        steps_per_agent: pasos de entrenamiento por agente
        """
        self.device = device
        self.generations = generations
        self.batch_size = batch_size
        self.steps_per_agent = steps_per_agent
        self.agent_fn = agent_fn
        self.agents = []
        self.meta_logs = SummaryWriter(log_dir)

    def run_generation(self, init_batch, gen_idx):
        agent_model = self.agent_fn().to(self.device)
        meta_agent = SequentialMetaAgent(agent_model, device=self.device, base_lr=3e-4,
                                         clip_grad=1.0, amp_enabled=True,
                                         log_dir=f'runs/meta_loop/gen_{gen_idx}')
        self.agents.append(meta_agent)

        print(f"\n=== Generación {gen_idx} iniciando ===")
        x_next = init_batch.to(self.device)
        meta_agent.run_loop(x_next, steps=self.steps_per_agent)
        return meta_agent.agent.vcs.get_embedding(x_next)

    def run_meta_loop(self, init_batch=None):
        """
        Ejecuta todas las generaciones secuenciales, cada una aprendiendo de la anterior.
        """
        if init_batch is None:
            init_batch = torch.randn(self.batch_size, 2).to(self.device)

        current_batch = init_batch
        for gen in range(self.generations):
            # Ejecuta la generación y obtiene embedding para la siguiente
            current_batch = self.run_generation(current_batch, gen)
        print("\n=== Meta-loop completo: todas las generaciones finalizadas ===")

    def save_all_checkpoints(self, folder='meta_loop_ckpts'):
        os.makedirs(folder, exist_ok=True)
        for idx, meta_agent in enumerate(self.agents):
            path = os.path.join(folder, f'meta_agent_gen_{idx}.pth')
            meta_agent.save_checkpoint(path)
        print(f"Todos los checkpoints guardados en {folder}")
```

```python
class MetaEvoAgentSystem:
    """
    Sistema de agentes meta-evolutivos:
    Cada generación no solo aprende, sino que modifica su backbone y activaciones
    para optimizar rendimiento y estabilidad del VirtualCellSystem.
    """
    def __init__(self, agent_fn, generations=5, batch_size=16, steps_per_agent=100,
                 device='cuda', base_lr=3e-4, clip_grad=1.0, amp_enabled=True, log_dir='runs/meta_evo'):
        self.device = device
        self.generations = generations
        self.batch_size = batch_size
        self.steps_per_agent = steps_per_agent
        self.agent_fn = agent_fn
        self.agents = []
        self.meta_logs = SummaryWriter(log_dir)

    def evolve_backbone(self, backbone):
        """
        Aplica pequeñas mutaciones evolutivas al backbone:
        - Añade/quita bloques
        - Ajusta dimensiones internas
        - Ajusta tokens
        """
        new_depth = max(2, int(backbone.blocks.__len__() * (1 + 0.1 * (torch.randn(1).item()))))
        new_tokens = max(4, int(backbone.tokens * (1 + 0.05 * (torch.randn(1).item()))))
        new_dim = max(64, int(backbone.dim * (1 + 0.05 * (torch.randn(1).item()))))

        # Reconstruir backbone con los nuevos parámetros
        new_backbone = GHU1C_v2(input_dim=backbone.input_dim, dim=new_dim, tokens=new_tokens, depth=new_depth, out_dim=backbone.head.out_features).to(self.device)
        return new_backbone

    def optimize_activation(self, agent):
        """
        Ajusta dinámicamente los parámetros de activaciones HyperFusionActX_v3
        para cada canal, buscando curvaturas más eficientes y mejor momentum.
        """
        for module in agent.modules():
            if isinstance(module, HyperFusionActX_v3):
                module.scale.data *= (1.0 + 0.01 * torch.randn_like(module.scale))
                module.raw.data += 0.01 * torch.randn_like(module.raw)

    def run_generation(self, init_batch, gen_idx):
        agent_model = self.agent_fn().to(self.device)
        # Mutar backbone antes de entrenar
        agent_model.backbone = self.evolve_backbone(agent_model.backbone)
        # Crear wrapper de entrenamiento con AMP y gradient clipping
        meta_agent = SequentialMetaAgent(agent_model, device=self.device, base_lr=3e-4,
                                         clip_grad=1.0, amp_enabled=True,
                                         log_dir=f'runs/meta_evo/gen_{gen_idx}')
        self.agents.append(meta_agent)

        print(f"\n=== Generación {gen_idx} iniciando ===")
        x_next = init_batch.to(self.device)

        # Entrenamiento con auto-optimización de activaciones
        for step in range(self.steps_per_agent):
            meta_agent.train_step(x_next)
            if step % 10 == 0:
                self.optimize_activation(meta_agent.agent)

        # Output embeddings para la siguiente generación
        return meta_agent.agent.vcs.get_embedding(x_next)

    def run_meta_evolution(self, init_batch=None):
        if init_batch is None:
            init_batch = torch.randn(self.batch_size, 2).to(self.device)

        current_batch = init_batch
        for gen in range(self.generations):
            current_batch = self.run_generation(current_batch, gen)
        print("\n=== Meta-Evolución completa ===")

    def save_all_checkpoints(self, folder='meta_evo_ckpts'):
        os.makedirs(folder, exist_ok=True)
        for idx, meta_agent in enumerate(self.agents):
            path = os.path.join(folder, f'meta_agent_gen_{idx}.pth')
            meta_agent.save_checkpoint(path)
        print(f"Todos los checkpoints guardados en {folder}")
```

```python
class SequentialMetaAgent:
    """
    Wrapper de entrenamiento seguro con AMP y gradient clipping,
    mantiene per-group TriOptimX y logging.
    """
    def __init__(self, agent: AutonomousAgent, device='cuda', base_lr=3e-4, clip_grad=1.0, amp_enabled=True, log_dir='runs/meta_agent'):
        self.agent = agent.to(device)
        self.device = device
        self.clip_grad = clip_grad
        self.scaler = torch.cuda.amp.GradScaler(enabled=amp_enabled)
        self.opt = TriOptimX(self.agent.parameters(), base_lr=base_lr)
        self.writer = SummaryWriter(log_dir)
        self.step_idx = 0

    def train_step(self, x: torch.Tensor):
        self.agent.train()
        x = x.to(self.device)
        self.opt.zero_grad()

        with torch.cuda.amp.autocast(enabled=self.scaler.is_enabled()):
            z = self.agent.observe(x)
            action = self.agent.act(z)
            cleaned, tel, metrics = self.agent.vcs.division_step(x, self.agent.vcs.initialize_telomeres(x.shape[0]),
                                                                 policy={"min_tel_threshold":5.0,
                                                                         "max_tel_threshold":1000.0,
                                                                         "division_cost":0.5,
                                                                         "noise_scale":0.03,
                                                                         "division_prob_thresh":0.5,
                                                                         "repair_activity_thresh":0.2})
            # Reward proxy: estabilidad telómero y bajo rate de división
            reward = -abs(metrics["tel_mean_after"] - 100.0) - 0.1 * metrics["division_rate"]
            v, q = self.agent.evaluate(z, action)
            value_loss = F.mse_loss(v, torch.tensor([reward], device=self.device).expand_as(v))
            policy_loss = -q.mean()
            loss = value_loss + 0.5 * policy_loss

        self.scaler.scale(loss).backward()
        # Gradient clipping por grupo
        torch.nn.utils.clip_grad_norm_(self.agent.parameters(), self.clip_grad)
        self.scaler.step(self.opt)
        self.scaler.update()

        if self.step_idx % 10 == 0:
            self.writer.add_scalar("reward", reward, self.step_idx)
            self.writer.add_scalar("tel_mean_after", metrics["tel_mean_after"], self.step_idx)
            self.writer.add_scalar("division_rate", metrics["division_rate"], self.step_idx)
        self.step_idx += 1
        return metrics

    def save_checkpoint(self, path: str):
        torch.save({
            "agent_state": self.agent.state_dict(),
            "opt_state": self.opt.state_dict(),
        }, path)
```

```python
class SequentialMetaEvolution:
    def __init__(self, agent_fn, generations=5, batch_size=16, steps_per_agent=100, device='cuda', base_lr=3e-4):
        self.agent_fn = agent_fn
        self.generations = generations
        self.batch_size = batch_size
        self.steps_per_agent = steps_per_agent
        self.device = device
        self.base_lr = base_lr
        self.agents = []

    def mutate_backbone(self, backbone: GHU1C_v2):
        """Mutación evolutiva de backbone: profundidad, tokens, dimensión"""
        new_depth = max(2, int(backbone.blocks.__len__() * (1 + 0.1 * torch.randn(1).item())))
        new_tokens = max(4, int(backbone.tokens * (1 + 0.05 * torch.randn(1).item())))
        new_dim = max(64, int(backbone.dim * (1 + 0.05 * torch.randn(1).item())))
        return GHU1C_v2(input_dim=backbone.input_dim, dim=new_dim, tokens=new_tokens,
                        depth=new_depth, out_dim=backbone.head.out_features).to(self.device)

    def optimize_activations(self, agent: AutonomousAgent):
        """Ajuste de HyperFusionActX_v3 por canal"""
        for module in agent.modules():
            if isinstance(module, HyperFusionActX_v3):
                module.scale.data *= (1.0 + 0.01 * torch.randn_like(module.scale))
                module.raw.data += 0.01 * torch.randn_like(module.raw)

    def run_generation(self, init_batch: torch.Tensor, gen_idx: int):
        agent_model = self.agent_fn().to(self.device)
        # Mutar backbone antes de entrenar
        agent_model.backbone = self.mutate_backbone(agent_model.backbone)
        meta_agent = SequentialMetaAgent(agent_model, device=self.device, base_lr=self.base_lr,
                                         clip_grad=1.0, amp_enabled=True, log_dir=f'runs/meta_evo/gen_{gen_idx}')
        self.agents.append(meta_agent)

        print(f"\n=== Generación {gen_idx} ===")
        current_batch = init_batch.to(self.device)
        for step in range(self.steps_per_agent):
            metrics = meta_agent.train_step(current_batch)
            if step % 10 == 0:
                self.optimize_activations(meta_agent.agent)
        # Generar embeddings para la siguiente generación
        next_batch = meta_agent.agent.vcs.get_embedding(current_batch)
        return next_batch

    def run(self, init_batch=None):
        if init_batch is None:
            init_batch = torch.randn(self.batch_size, 2).to(self.device)
        current_batch = init_batch
        for gen in range(self.generations):
            current_batch = self.run_generation(current_batch, gen)
        print("\n=== Meta-Evolución completa ===")

    def save_all_checkpoints(self, folder='meta_evo_ckpts'):
        os.makedirs(folder, exist_ok=True)
        for idx, meta_agent in enumerate(self.agents):
            path = os.path.join(folder, f'meta_agent_gen_{idx}.pth')
            meta_agent.save_checkpoint(path)
        print(f"Todos los checkpoints guardados en {folder}")
```
