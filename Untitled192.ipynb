```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
```

```python
@keras.utils.register_keras_serializable()
class ElectroChemicalActivation(layers.Layer):
    def __init__(self, alpha=1.0, beta=0.5, gamma=0.1):
        super().__init__()
        self.alpha = alpha   # electrical strength
        self.beta = beta     # chemical smoothness
        self.gamma = gamma   # oscillatory coupling

    def call(self, x):
        gelu = tf.nn.gelu(x)
        mish = x * tf.tanh(tf.nn.softplus(x))
        osc = tf.sin(self.gamma * x)

        # neuromodulator gate
        gate = tf.sigmoid(self.alpha * x)

        return gate * (gelu + self.beta * mish + osc)
```

```python
class NeuroModulatedAdam(tf.keras.optimizers.AdamW):
    def __init__(self, lr=1e-3, dopamine_scale=0.1):
        super().__init__(learning_rate=lr)
        self.dopamine_scale = dopamine_scale

    def apply_gradients(self, grads_and_vars, dopamine=1.0):
        modulated = []
        for g, v in grads_and_vars:
            if g is not None:
                g = g * (1 + self.dopamine_scale * dopamine)
                g = tf.sign(g) * tf.sqrt(tf.abs(g) + 1e-8)
            modulated.append((g, v))
        super().apply_gradients(modulated)
```

```python
class BrainRegionRules(layers.Layer):
    def call(self, state, region_id):
        damage = state[:,0]
        energy = state[:,1]
        novelty = state[:,2]

        if region_id == "PFC":
            return tf.sigmoid(1 - damage)
        if region_id == "Hippocampus":
            return tf.sigmoid(novelty * energy)
        if region_id == "Sensory":
            return tf.sigmoid(energy - damage)
        return tf.ones_like(damage)
```

```python
class BrainAtlasEmbedding(layers.Layer):
    def __init__(self, regions=6, dim=16):
        super().__init__()
        self.emb = layers.Embedding(regions, dim)

    def call(self, region_ids):
        return self.emb(region_ids)
```

```python
cell_state_input = keras.Input(shape=(3,), name="cell_state")
x_cell = layers.Dense(32, activation="swish")(cell_state_input)
```

```python
sensory_input = keras.Input(shape=(64,), name="sensory")
region_input = keras.Input(shape=(), dtype="int32")

atlas_emb = BrainAtlasEmbedding()(region_input)

x = layers.Dense(128)(sensory_input)
x = ElectroChemicalActivation()(x)
x = layers.Concatenate()([x, atlas_emb, x_cell])
x = layers.Dense(128)(x)
x = ElectroChemicalActivation()(x)

output = layers.Dense(10, activation="softmax")(x)

model = keras.Model(
    inputs=[sensory_input, cell_state_input, region_input],
    outputs=output
)
```

```python
def neuro_bio_loss(y_true, y_pred, cell_state):
    ce = keras.losses.categorical_crossentropy(y_true, y_pred)
    energy = tf.reduce_mean(cell_state[:,1])
    stability = tf.reduce_mean(tf.square(y_pred))
    diversity = -tf.reduce_mean(tf.math.log(tf.math.reduce_std(y_pred, axis=0)+1e-6))
    return ce + 0.1*energy + 0.1*stability + 0.05*diversity
```

```python
class NeuroModulatedAdam(tf.keras.optimizers.AdamW):
    def __init__(self, lr=1e-3, dopamine_scale=0.1):
        super().__init__(learning_rate=lr)
        self.dopamine_scale = dopamine_scale

    def apply_gradients(self, grads_and_vars, dopamine=1.0):
        modulated = []
        for g, v in grads_and_vars:
            if g is not None:
                # Convert IndexedSlices to a dense tensor if necessary
                if isinstance(g, tf.IndexedSlices):
                    g = tf.convert_to_tensor(g)
                g = g * (1 + self.dopamine_scale * dopamine)
                g = tf.sign(g) * tf.sqrt(tf.abs(g) + 1e-8)
            modulated.append((g, v))
        super().apply_gradients(modulated)

optimizer = NeuroModulatedAdam()

model.compile(
    optimizer=optimizer,
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# dummy run
x = np.random.randn(64,64)
cell = np.random.rand(64,3)
regions = np.random.randint(0,6,size=(64,))
y = keras.utils.to_categorical(np.random.randint(0,10,size=64),10)

model.fit([x, cell, regions], y, epochs=2)
```

```python
BASE_OPS = [
    tf.nn.relu,
    tf.nn.gelu,
    tf.nn.silu,
    tf.nn.tanh,
    tf.nn.sigmoid,
    tf.nn.softplus,
    tf.math.sin,
    tf.math.cos,
    tf.math.exp,
]

COMBINE_OPS = [
    lambda a, b: a + b,
    lambda a, b: a * b,
    lambda a, b: a - b,
]
```

```python
class ActivationGenome:
    def __init__(self, depth=3):
        self.depth = depth
        self.ops = [random.choice(BASE_OPS) for _ in range(depth)]
        self.combine = random.choice(COMBINE_OPS)
        self.alpha = tf.Variable(tf.random.uniform([], 0.5, 2.0))

    def __call__(self, x):
        out = self.ops[0](x)
        for op in self.ops[1:]:
            out = self.combine(out, op(x))
        return out * self.alpha
```

```python
def mutate(genome, rate=0.3):
    if random.random() < rate:
        genome.ops[random.randint(0, genome.depth-1)] = random.choice(BASE_OPS)
    if random.random() < rate:
        genome.combine = random.choice(COMBINE_OPS)
    if random.random() < rate:
        genome.alpha.assign(tf.clip_by_value(
            genome.alpha + tf.random.normal([], 0.1), 0.1, 5.0
        ))
```

```python
def evaluate_activation(genome):
    x = tf.random.normal((1024,))

    with tf.GradientTape() as tape:
        tape.watch(x)
        y = genome(x)
        loss = tf.reduce_mean(y**2)

    grad = tape.gradient(loss, x)

    stability = tf.reduce_mean(tf.abs(grad))
    variance = tf.math.reduce_std(y)

    # Lower instability, higher expressiveness
    fitness = variance / (stability + 1e-6)
    return fitness.numpy()
```

```python
def evolve_activations(generations=20, population=30):
    pop = [ActivationGenome() for _ in range(population)]

    for g in range(generations):
        scores = [(evaluate_activation(p), p) for p in pop]
        scores.sort(key=lambda x: x[0], reverse=True)

        print(f"Gen {g} | Best fitness:", scores[0][0])

        survivors = [p for _, p in scores[:population//3]]

        new_pop = survivors.copy()
        while len(new_pop) < population:
            parent = random.choice(survivors)
            child = ActivationGenome()
            child.ops = parent.ops.copy()
            child.combine = parent.combine
            child.alpha.assign(parent.alpha)
            mutate(child)
            new_pop.append(child)

        pop = new_pop

    return pop[0]
```

```python
class EvolvedActivation(tf.keras.layers.Layer):
    def __init__(self, genome):
        super().__init__()
        self.genome = genome

    def call(self, x):
        return self.genome(x)
```

```python
best_genome = evolve_activations()

x = tf.keras.layers.Dense(128)(sensory_input)
x = EvolvedActivation(best_genome)(x)
```

```python
import tensorflow as tf
import numpy as np
import random
```

```python
class OptimizerGenome:
    def __init__(self):
        self.lr = tf.Variable(10 ** tf.random.uniform([], -4, -2))
        self.beta1 = tf.Variable(tf.random.uniform([], 0.7, 0.99))
        self.beta2 = tf.Variable(tf.random.uniform([], 0.9, 0.999))
        self.alpha_sign = tf.Variable(tf.random.uniform([], 0.0, 1.0))
        self.alpha_var = tf.Variable(tf.random.uniform([], 0.0, 1.0))
        self.stability_eps = tf.Variable(tf.random.uniform([], 1e-8, 1e-4))
```

```python
class EvolvedOptimizer(tf.keras.optimizers.Optimizer):
    def __init__(self, genome, name="EvolvedOptimizer", **kwargs):
        super().__init__(name, **kwargs)
        self.g = genome

    def build(self, var_list):
        self.m = [self.add_variable_from_reference(v, "m") for v in var_list]
        self.v = [self.add_variable_from_reference(v, "v") for v in var_list]

    def update_step(self, grad, var, idx):
        m = self.m[idx]
        v = self.v[idx]

        # Momentum (electrical)
        m.assign(self.g.beta1 * m + (1 - self.g.beta1) * grad)

        # Variance (chemical)
        v.assign(self.g.beta2 * v + (1 - self.g.beta2) * tf.square(grad))

        # Hybrid update
        update = (
            m / (tf.sqrt(v) + self.g.stability_eps)
            + self.g.alpha_sign * tf.sign(grad)
            + self.g.alpha_var * grad
        )

        var.assign_sub(self.g.lr * update)
```

```python
def evaluate_optimizer(lr, beta1, beta2, alpha_sign, alpha_var, stability_eps):
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation="gelu"),
        tf.keras.layers.Dense(1)
    ])

    # Create a dummy genome object from the passed float values for EvolvedOptimizer's __init__
    # This object's variables are not tf.Variable, but just Python floats
    class DummyGenome:
        def __init__(self, lr, beta1, beta2, alpha_sign, alpha_var, stability_eps):
            self.lr = tf.constant(lr) # EvolvedOptimizer expects a tf.Variable, so convert float to tf.constant
            self.beta1 = tf.constant(beta1)
            self.beta2 = tf.constant(beta2)
            self.alpha_sign = tf.constant(alpha_sign)
            self.alpha_var = tf.constant(alpha_var)
            self.stability_eps = tf.constant(stability_eps)

    dummy_genome = DummyGenome(lr, beta1, beta2, alpha_sign, alpha_var, stability_eps)
    opt = EvolvedOptimizer(dummy_genome)
    model.compile(optimizer=opt, loss="mse")

    x = np.random.randn(256, 10)
    y = np.sum(x, axis=1, keepdims=True)

    history = model.fit(x, y, epochs=5, verbose=0)
    loss = history.history["loss"][-1]

    # Stability proxy (Lyapunov-like)
    stability = sum(tf.reduce_mean(tf.abs(v)) for v in model.trainable_variables)

    return float(loss + 1e-4 * stability)
```

```python
def mutate(genome, rate=0.3):
    if random.random() < rate:
        genome.lr.assign(genome.lr * tf.exp(tf.random.normal([], 0, 0.2)))
    if random.random() < rate:
        genome.beta1.assign(tf.clip_by_value(
            genome.beta1 + tf.random.normal([], 0, 0.05), 0.5, 0.999))
    if random.random() < rate:
        genome.beta2.assign(tf.clip_by_value(
            genome.beta2 + tf.random.normal([], 0, 0.05), 0.8, 0.9999))
    if random.random() < rate:
        genome.alpha_sign.assign(tf.clip_by_value(
            genome.alpha_sign + tf.random.normal([], 0, 0.1), 0.0, 1.0))
```

```python
def evolve_optimizers(gens=10, pop_size=12):
    pop = [OptimizerGenome() for _ in range(pop_size)]

    for g in range(gens):
        # Extract float values from genome before passing to evaluate_optimizer
        scored = [
            (evaluate_optimizer(
                p.lr.numpy(), p.beta1.numpy(), p.beta2.numpy(),
                p.alpha_sign.numpy(), p.alpha_var.numpy(), p.stability_eps.numpy()
            ), p) for p in pop
        ]
        scored.sort(key=lambda x: x[0])

        print(f"Gen {g} | Best loss+stability:", scored[0][0])

        survivors = [p for _, p in scored[:pop_size // 3]]

        new_pop = survivors.copy()
        while len(new_pop) < pop_size:
            parent = random.choice(survivors)
            child = OptimizerGenome()
            for attr in vars(parent):
                getattr(child, attr).assign(getattr(parent, attr))
            mutate(child)
            new_pop.append(child)

        pop = new_pop

    return scored[0][1]
```

```python
import tensorflow as tf
import numpy as np
import random
```

```python
class VirtualCell:
    def __init__(self, batch=256):
        self.telomere = tf.random.uniform([batch, 1], 0.6, 1.0)
        self.damage   = tf.random.uniform([batch, 1], 0.0, 0.3)
        self.metab    = tf.random.uniform([batch, 1], 0.3, 0.8)
        self.need_div = tf.random.uniform([batch, 1], 0.0, 1.0)

    def step(self, telomerase):
        # telomerase acts preferentially on short telomeres
        short = tf.nn.relu(0.8 - self.telomere)
        self.telomere += telomerase * short
        self.telomere = tf.clip_by_value(self.telomere, 0.3, 1.2)

        # damage dynamics
        self.damage += 0.02 * self.metab - 0.05 * telomerase
        self.damage = tf.clip_by_value(self.damage, 0.0, 1.0)

        # metabolism fluctuates
        self.metab += tf.random.normal(self.metab.shape, 0, 0.02)
        self.metab = tf.clip_by_value(self.metab, 0.0, 1.0)
```

```python
def build_controller():
    return tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation="gelu"),
        tf.keras.layers.Dense(64, activation="mish"),
        tf.keras.layers.Dense(1, activation="sigmoid")  # telomerase ∈ [0,1]
    ])
```

```python
class LossGenome:
    def __init__(self):
        self.w_mse   = tf.Variable(tf.random.uniform([], 0.1, 1.0))
        self.w_tel   = tf.Variable(tf.random.uniform([], 0.1, 1.0))
        self.w_dmg   = tf.Variable(tf.random.uniform([], 0.1, 1.0))
        self.w_met   = tf.Variable(tf.random.uniform([], 0.0, 0.5))
        self.w_lyap  = tf.Variable(tf.random.uniform([], 0.01, 0.2))
```

```python
def evolved_bio_loss(genome, cell, telomerase, prev_state):
    tel_target = 0.9

    mse = tf.reduce_mean((cell.telomere - tel_target)**2)
    tel_homeo = tf.reduce_mean(tf.nn.relu(cell.telomere - 1.05)**2)
    damage_penalty = tf.reduce_mean(tf.nn.relu(cell.damage - 0.4)**2)
    energy = tf.reduce_mean(telomerase * cell.metab)

    # Lyapunov-like stability
    V_prev = tf.reduce_mean(tf.square(prev_state))
    V_now  = tf.reduce_mean(tf.square(
        tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)
    ))
    lyap = tf.nn.relu(V_now - V_prev)

    return (
        genome.w_mse  * mse +
        genome.w_tel  * tel_homeo +
        genome.w_dmg  * damage_penalty +
        genome.w_met  * energy +
        genome.w_lyap * lyap
    )
```

```python
def mutate_loss(genome, rate=0.4):
    for w in [genome.w_mse, genome.w_tel, genome.w_dmg, genome.w_met, genome.w_lyap]:
        if random.random() < rate:
            w.assign(tf.clip_by_value(
                w * tf.exp(tf.random.normal([], 0, 0.3)), 1e-3, 5.0))
```

```python
def evaluate_loss(genome):
    cell = VirtualCell()
    model = build_controller()
    opt = tf.keras.optimizers.Adam(1e-3)

    prev_state = tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)

    for _ in range(10):
        with tf.GradientTape() as tape:
            state = tf.concat([cell.telomere, cell.damage, cell.metab, cell.need_div], axis=1)
            telomerase = model(state)
            cell.step(telomerase)
            loss = evolved_bio_loss(genome, cell, telomerase, prev_state)

        grads = tape.gradient(loss, model.trainable_variables)
        opt.apply_gradients(zip(grads, model.trainable_variables))
        prev_state = tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)

    # fitness: low loss + low damage + stable telomeres
    return float(loss + tf.reduce_mean(cell.damage))
```

```python
def evolve_losses(gens=8, pop=10):
    population = [LossGenome() for _ in range(pop)]

    for g in range(gens):
        scored = [(evaluate_loss(l), l) for l in population]
        scored.sort(key=lambda x: x[0])

        print(f"Gen {g} | Best loss score:", scored[0][0])

        elites = [l for _, l in scored[:pop//3]]
        new_pop = elites.copy()

        while len(new_pop) < pop:
            parent = random.choice(elites)
            child = LossGenome()
            for k in vars(parent):
                getattr(child, k).assign(getattr(parent, k))
            mutate_loss(child)
            new_pop.append(child)

        population = new_pop

    return scored[0][1]
```

```python
best_loss = evolve_losses()

# Train longer with the discovered bio-loss
cell = VirtualCell()
model = build_controller()
opt = tf.keras.optimizers.Adam(1e-3)

prev_state = tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)

for step in range(100):
    with tf.GradientTape() as tape:
        state = tf.concat([cell.telomere, cell.damage, cell.metab, cell.need_div], axis=1)
        telomerase = model(state)
        cell.step(telomerase)
        loss = evolved_bio_loss(best_loss, cell, telomerase, prev_state)

    grads = tape.gradient(loss, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))
    prev_state = tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)

    if step % 10 == 0:
        print(step, "loss", float(loss),
              "tel", float(tf.reduce_mean(cell.telomere)),
              "dmg", float(tf.reduce_mean(cell.damage)))
```

```python
def hippocampus_loss(genome, cell, telomerase, prev_state):
    # Encourage plasticity & regeneration
    plasticity = tf.reduce_mean(tf.square(tf.nn.relu(0.8 - cell.telomere)))
    energy     = tf.reduce_mean(cell.metab * telomerase)
    damage     = tf.reduce_mean(tf.nn.relu(cell.damage - 0.5)**2)

    return (
        genome.w_tel * plasticity +
        0.3 * genome.w_met * energy +
        0.5 * genome.w_dmg * damage
    )


def cortex_loss(genome, cell, telomerase, prev_state):
    # Encourage stability & precise control
    mse = tf.reduce_mean((cell.telomere - 0.9)**2)
    damage = tf.reduce_mean(tf.nn.relu(cell.damage - 0.3)**2)

    V_prev = tf.reduce_mean(tf.square(prev_state))
    V_now  = tf.reduce_mean(tf.square(
        tf.concat([cell.telomere, cell.damage, cell.metab], axis=1)
    ))
    lyap = tf.nn.relu(V_now - V_prev)

    return (
        genome.w_mse * mse +
        genome.w_dmg * damage +
        genome.w_lyap * lyap
    )
```

```python
class BrainTissue:
    def __init__(self):
        self.hpc = VirtualCell()
        self.ctx = VirtualCell()
```

```python
ACTIVATIONS = ["relu", "gelu", "mish", "silu", "tanh"]
OPTIMIZERS  = ["Adam", "AdamW", "Lion", "RMSprop"]
```

```python
def build_controller(activation):
    return tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation=activation),
        tf.keras.layers.Dense(64, activation=activation),
        tf.keras.layers.Dense(1, activation="sigmoid")
    ])
```

```python
def make_optimizer(name, lr):
    if name == "Adam":
        return tf.keras.optimizers.Adam(lr)
    if name == "AdamW":
        return tf.keras.optimizers.AdamW(lr)
    if name == "Lion":
        return tf.keras.optimizers.experimental.Lion(lr)
    if name == "RMSprop":
        return tf.keras.optimizers.RMSprop(lr)
```

```python
class NeuroAgent:
    def __init__(self):
        self.activation = random.choice(ACTIVATIONS)
        self.opt_name   = random.choice(OPTIMIZERS)
        self.lr         = 10 ** random.uniform(-4, -2)
        self.loss_genome = LossGenome()

    def mutate(self):
        if random.random() < 0.3:
            self.activation = random.choice(ACTIVATIONS)
        if random.random() < 0.3:
            self.opt_name = random.choice(OPTIMIZERS)
        if random.random() < 0.4:
            self.lr *= float(tf.exp(tf.random.normal([], 0, 0.4)))
        mutate_loss(self.loss_genome)
```

```python
def evaluate_agent(agent):
    brain = BrainTissue()
    model = build_controller(agent.activation)
    opt = make_optimizer(agent.opt_name, agent.lr)

    prev_h = tf.concat([brain.hpc.telomere, brain.hpc.damage, brain.hpc.metab], axis=1)
    prev_c = tf.concat([brain.ctx.telomere, brain.ctx.damage, brain.ctx.metab], axis=1)

    for _ in range(10):
        with tf.GradientTape() as tape:
            # Hippocampus
            s_h = tf.concat([brain.hpc.telomere, brain.hpc.damage, brain.hpc.metab, brain.hpc.need_div], axis=1)
            t_h = model(s_h)
            brain.hpc.step(t_h)
            L_h = hippocampus_loss(agent.loss_genome, brain.hpc, t_h, prev_h)

            # Cortex
            s_c = tf.concat([brain.ctx.telomere, brain.ctx.damage, brain.ctx.metab, brain.ctx.need_div], axis=1)
            t_c = model(s_c)
            brain.ctx.step(t_c)
            L_c = cortex_loss(agent.loss_genome, brain.ctx, t_c, prev_c)

            loss = L_h + L_c

        grads = tape.gradient(loss, model.trainable_variables)
        opt.apply_gradients(zip(grads, model.trainable_variables))

        prev_h = tf.concat([brain.hpc.telomere, brain.hpc.damage, brain.hpc.metab], axis=1)
        prev_c = tf.concat([brain.ctx.telomere, brain.ctx.damage, brain.ctx.metab], axis=1)

    # Fitness = low loss + low damage + region balance
    return float(
        loss +
        tf.reduce_mean(brain.hpc.damage) +
        tf.reduce_mean(brain.ctx.damage)
    )
```

```python
def evolve_agents(gens=6, pop=8):
    population = [NeuroAgent() for _ in range(pop)]

    for g in range(gens):
        scored = [(evaluate_agent(a), a) for a in population]
        scored.sort(key=lambda x: x[0])

        print(f"Gen {g} | best fitness {scored[0][0]:.4f}",
              "| act", scored[0][1].activation,
              "| opt", scored[0][1].opt_name)

        elites = [a for _, a in scored[:pop//2]]
        new_pop = elites.copy()

        while len(new_pop) < pop:
            parent = random.choice(elites)
            child = NeuroAgent()
            child.activation = parent.activation
            child.opt_name   = parent.opt_name
            child.lr         = parent.lr
            child.loss_genome = parent.loss_genome
            child.mutate()
            new_pop.append(child)

        population = new_pop

    return scored[0][1]
```

```python
def hippocampus_loss(y_true, y_pred):
    # penalize instability
    return tf.reduce_mean(tf.square(y_true - y_pred))

def cortex_loss(y_pred):
    # encourage diversity
    batch_flat = tf.reshape(y_pred, [tf.shape(y_pred)[0], -1])
    cov = tf.matmul(batch_flat, batch_flat, transpose_a=True)
    return -tf.linalg.trace(cov)

def brain_region_loss(y_true, y_pred, region="cortex"):
    if region == "hippocampus":
        return hippocampus_loss(y_true, y_pred)
    elif region == "cortex":
        return cortex_loss(y_pred)
    else:
        return tf.keras.losses.MeanSquaredError()(y_true, y_pred)
```

```python
import torch, torch.nn as nn
import numpy as np
from dataclasses import dataclass

device = "cuda" if torch.cuda.is_available() else "cpu"

@dataclass
class CellState:
    telomere: float
    damage: float
    metabolism: float
    p53: float
    alive: bool = True

@dataclass
class FieldState:
    nutrient: float
    toxin: float
    growth_factor: float
```

```python
class BiologicalRules:
    @staticmethod
    def telomerase_allowed(cell: CellState):
        return cell.telomere < 0.6 and cell.damage < 0.4 and cell.p53 < 0.7

    @staticmethod
    def apoptosis(cell: CellState):
        return cell.damage > 0.85 or cell.p53 > 0.9

    @staticmethod
    def phagocytosis_signal(cell: CellState):
        return not cell.alive
```

```python
class MicroEnvironment:
    def __init__(self):
        self.field = FieldState(
            nutrient=np.random.uniform(0.4, 0.9),
            toxin=np.random.uniform(0.0, 0.3),
            growth_factor=np.random.uniform(0.3, 0.8),
        )

    def step(self):
        self.field.nutrient *= 0.99
        self.field.toxin *= 0.98
        self.field.growth_factor *= 0.995
```

```python
def update_cell(cell: CellState, field: FieldState, telomerase_signal):
    if not cell.alive:
        return cell

    # Damage dynamics
    cell.damage += 0.02 * field.toxin - 0.01 * field.nutrient
    cell.damage = np.clip(cell.damage, 0, 1)

    # Telomere erosion
    cell.telomere -= 0.01

    # Telomerase regulation
    if telomerase_signal and BiologicalRules.telomerase_allowed(cell):
        cell.telomere += 0.03

    # p53 response
    cell.p53 = cell.damage

    if BiologicalRules.apoptosis(cell):
        cell.alive = False

    return cell
```

```python
def generate_trajectory(T=50):
    env = MicroEnvironment()
    cell = CellState(
        telomere=1.0,
        damage=np.random.rand()*0.2,
        metabolism=np.random.rand(),
        p53=0.0
    )

    traj = []
    for t in range(T):
        tel_signal = np.random.rand() > 0.5
        cell = update_cell(cell, env.field, tel_signal)
        env.step()
        traj.append([
            cell.telomere,
            cell.damage,
            cell.metabolism,
            cell.p53,
            float(cell.alive)
        ])
    return np.array(traj)
```

```python
class WorldModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(5, 128), nn.GELU(),
            nn.Linear(128, 64)
        )
        self.transition = nn.Sequential(
            nn.Linear(64, 128), nn.GELU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.GELU(),
            nn.Linear(128, 5)
        )

    def forward(self, x):
        z = self.encoder(x)
        z_next = z + self.transition(z)
        return self.decoder(z_next)

wm = WorldModel().to(device)
```

```python
@torch.no_grad()
def dream_rollout(seed_state, steps=40):
    x = torch.tensor(seed_state, dtype=torch.float32).to(device)
    dreams = []
    for _ in range(steps):
        x = wm(x)
        dreams.append(x.cpu().numpy())
    return np.array(dreams)
```

```python
class InvariantNet(nn.Module):
    def __init__(self, state_dim=5):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, 128),
            nn.GELU(),
            nn.Linear(128, 64),
            nn.GELU(),
            nn.Linear(64, 3)  # discovers 3 invariants
        )

    def forward(self, x):
        return self.net(x)
```

```python
def invariance_loss(inv_net, trajectory):
    x = torch.tensor(trajectory[:-1], dtype=torch.float32).to(device)
    x_next = torch.tensor(trajectory[1:], dtype=torch.float32).to(device)

    I1 = inv_net(x)
    I2 = inv_net(x_next)

    return ((I2 - I1) ** 2).mean()
```

```python
def emergent_reward(inv_net, trajectory):
    with torch.no_grad():
        loss = invariance_loss(inv_net, trajectory)
    return -loss.item()  # lower change = higher reward
```

```python
def novelty_metric(trajectory):
    traj = torch.tensor(trajectory, dtype=torch.float32)
    return traj.std(dim=0).mean().item()
```

```python
def total_self_reward(inv_net, trajectory, alpha=1.0, beta=0.2):
    return (
        alpha * emergent_reward(inv_net, trajectory)
        + beta * novelty_metric(trajectory)
    )
```
